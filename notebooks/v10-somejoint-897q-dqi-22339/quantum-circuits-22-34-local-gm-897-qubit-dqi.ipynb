## How the score works

Each task has an official verifier. It runs the circuit on many sampled inputs, checks every answer exactly, and bills a cost:

| Task | Cost | The letters |
|---|---|---|
| invsqrt | $Q \cdot C$ | $Q$ qubits; $C = N_{\mathrm{Tof}} + N_T / 4$, the mean Toffolis and T gates of one shot |
| DQI | $R \cdot Q^2 \cdot T$ | $T$ the mean Toffolis of one shot; $R \ge 1$ grows with the failed shots |
| Chemistry | $\lambda_{\mathrm{eff}} \cdot T \cdot Q$ | $\lambda_{\mathrm{eff}}$ the Hamiltonian's effective normalization; $T$ the Toffolis of one step |

A task's **ratio** is the organizers' reference cost divided by this cost. The references are 2,076,774, $2.4439 \times 10^{13}$ and 3,355,599,588. The score is the **geometric mean of the three ratios**:

$$\text{score} = \left( \frac{C^{\mathrm{ref}}_{\mathrm{invsqrt}}}{C_{\mathrm{invsqrt}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{DQI}}}{C_{\mathrm{DQI}}} \cdot \frac{C^{\mathrm{ref}}_{\mathrm{chem}}}{C_{\mathrm{chem}}} \right)^{1/3}$$

So making any one task 1% cheaper lifts the score by about 0.33%.

The three bills reward different things. In DQI, $Q$ is squared: near 880 qubits one qubit is worth 0.23% of the cost. In invsqrt one qubit near 63 is worth 1.6%. In chemistry each of the three factors counts the same.

## DQI · GPT-6 Astra Pro's 897-qubit decoder with the 658-Toffoli jet

**The task.** Seventy 12-bit quantum registers hold the syndromes $S_i = \sum_j e_j \alpha^{(i+1)j}$ of an unknown error vector $e$ over $\mathrm{GF}(2^{12})$. The vector $e$ has at most 35 nonzero values among $m = 4095$ positions. Beside the syndromes sit 4,095 classical words $M_j$. The circuit must multiply the phase by

$$(-1)^{\sum_j \operatorname{popcount}(e_j \wedge M_j)},$$

give the syndromes back and leave every other qubit clean. In effect it decodes $e$ in superposition, applies the phase, and undoes the decoding.

The verifier draws 102,400 shots from DQI's own error-weight distribution and bills

$$\text{cost} = R \cdot Q^2 \cdot T, \qquad R = \frac{\ln 2}{-\ln\left( \frac{1}{2} + 2\sqrt{\varepsilon_+} \right)},$$

where $\varepsilon_+$ is a very cautious upper bound on the rate of failed shots, at confidence $1 - 2^{-128}$. With no failures, $R = 1.1913$.

**This circuit.**
- $Q = 897$, of which 840 hold the syndromes.
- A shot runs $T = 440{,}633.0$ Toffolis on average, at most 454,477. So $Q^2 T = 3.5454 \times 10^{11}$, 2.80% below our 883-qubit decoder of the previous version, with $Q = 883$ and $T = 467{,}789.6$.
- At the default seed, 0 shots decode a wrong value and 0 a wrong phase, so $R = 1.1913$.
  - Failures are rare and random. The verifier's seeds 20261027, 777, 4242 and 84317029 give two circuits the same inputs; there it failed in 1, 0, 0 and 1 shots of 102,400, against 0, 2, 2 and 2 for our previous decoder. All were phase errors.
  - With two salts of our own, each chosen before its run, it made no error.
  - The leaderboard draws its own shots.
- The cost is $4.22346 \times 10^{11}$, 57.8659× below the reference.

**Where it comes from.** We asked GPT-6 Astra Pro, in ChatGPT, for a whole decoder much cheaper than ours. With the question went a kit that builds a decoder from our decoder's modules and checks it on random shots of the task in seconds. Pro worked for 4 hours 34 minutes with 31 agents of its own. Its answer adds three files and leaves every module of ours unchanged. Its numbers reproduced on our machine and with the official builder. We then put in our 658-Toffoli jet in place of the kit's 683-Toffoli one, which saves 1,400 Toffolis.

**What is new**, in Pro's three files:
- **Homogeneous external chains.** Nine late external nodes of the extraction no longer call the jet. They compose the existing extraction steps homogeneously, normalize once, and replay the measurement phases exactly. The other eight external nodes still call it, 56 pairs in all.
- **The geometry** uses a first jet at zero for its final two coordinates.
- **The synthesis** uses additive continuants, leaves out one fixed outer quadratic with a matching correction in the scan, and shares norms in its reverse divisions.
- **Tail products** defer their measurement phases until the inverse restores the values they need.
- **The conversion** shares its gap predicates and reuses the monic leading quotient as an inverse word.

**How it decodes.** One shot, stage by stage, as the kit counts it on 4,096 shots: mean Toffolis and the peak of live qubits.

| Stage | What it does | Toffolis | Peak qubits |
|---|---|---:|---:|
| Linear maps | turn the syndromes into the values and first jets of a rational function $G$ at fixed nodes of $\mathrm{GF}(2^{24})$; move the phase of 70 mask coordinates onto the syndromes | 0 | 840 |
| Extraction | a continued fraction of $G$, in place: jets at eight external nodes, homogeneous chains at nine | 65,881 | 897 |
| Synthesis | folds the heads of the fraction into the error locator $B$ and a numerator $A$ | 33,317 | 889 |
| Conversion | $A \leftarrow A / B' \bmod B$, so that $A(\beta_j)$ is the error value at every error position | 68,352 | 897 |
| Scan | 11 Toffolis at each of 4,025 points: where $B$ vanishes, apply that position's phase | 44,275 | 870 |
| Undo | conversion, synthesis and extraction backwards, so every register returns to its start | 69,455 + 41,931 + 117,422 | 897 |

Four stages reach 897 qubits, so narrowing any one of them alone would not lower $Q$.

**Kept from our decoder:**
- the confluent Hermite extraction and its jet;
- the zero-Toffoli mask projection;
- the scan;
- the joint field arithmetic.

**Built here.** The decoder's source, 68 files, is embedded in this notebook: our 883-qubit decoder's modules as they went to Pro, Pro's three files unchanged, the jet, and the build script. The Run section builds the circuit from this source in about a minute and a half. It checks that the result is byte for byte the circuit the verifier passed.


**Credits.** **GPT-6 Astra Pro**, asked by Claude for Agniv Banerjee (SomeJoint), in two questions: the decoder's three new files (homogeneous external chains with exact phase replay, a synthesis with additive continuants and shared norms, deferred tail phases, and a conversion that shares its gap predicates), and the 658-Toffoli jet. **Our research team** of AI agents (GPT-6 Astra leading GPT-6.1 Sol, directed by Claude for Agniv Banerjee, SomeJoint): the fused decoder whose modules it uses. **Ahmed Berat Özer**: the narrowing of our fused decoder in his version 9, from which our 883-qubit decoder came. **Alex Chilton**: the 883-qubit Schur decoder whose design it starts from, based on **jday96314**'s QuantumAI decoder and classical models (that repository states no license). **mrmeows** and **Ahmed Berat Özer**: the public mask projection of round17 and version 8, after **manwithacat**'s classical-mask projection. Earlier links of the design: **Dmitrii Gluzdov**'s round12 decoder, **Agniv Banerjee (SomeJoint)**'s Bézout–Hankel phase solve, **ViberTeam** and **Phuc Thai (DePie)**.
```python
COMPETITION_FILES = {'kickmix/LICENSE.txt': 'cfc7749b96f63bd31c3c42b5c471bf756814053e847c10f3eb003417bc523d30', 'kickmix/kickmix.so': 'c66b3216c4029e0cdfd09054306aa0feadf53ab2f50c1164e37f752874e5005e', 'verifiers/LICENSE.txt': 'cfc7749b96f63bd31c3c42b5c471bf756814053e847c10f3eb003417bc523d30', 'verifiers/dfthcsos/FeMoCo76o113e.fcidump': 'e5a48af746fd075026dec9e8f5df2d363802a121e8c53af9a0b984bde695e1bd', 'verifiers/dfthcsos/FeMoCo76o113e.fixed_shifts.npz': '7b7facf63fe35a9d9d3d00787ffdb700c863fa845fe4724d7f7c54778151cd87', 'verifiers/dfthcsos/SCORING_README.md': 'c2228569f96e58e118419bc32be2be9e33d6abffb8117cb0b91e1fc25b2a0230', 'verifiers/dfthcsos/__init__.py': '81915e324cc0cade67f52a54b7a51991fd96abaea94d1f8a4d423dae20e57348', 'verifiers/dfthcsos/scoreinfra/__init__.py': 'be90e0d503d5c20dcac0102af638b9a4ccad0341daa5baa5d49216a259e4516f', 'verifiers/dfthcsos/scoreinfra/circuits.py': 'fc027b52fdcebe14404f403a089122485d097ed65d03e6d989623d620b1ab68c', 'verifiers/dfthcsos/scoreinfra/costs_prepare.py': '7f1e190c42fc5325f3e6689fbbef51f95b0cb12f1c828cda92b6c088e72206da', 'verifiers/dfthcsos/scoreinfra/dfthc_math.py': 'ae82898d47a7a86b96378fb906e0f8a9f46003065cf19c7c09952a123b5b5af8', 'verifiers/dfthcsos/scoreinfra/reference.py': 'a2d5a63191d655729e29f0fde1e85fe4630bf896a91a63bb894d8d7daf71ed39', 'verifiers/dfthcsos/verify.py': '0775be7a528922367caebe9e4d051d6199811f1aefeaa4b793eaaa4621b8ec9d', 'verifiers/dqi/__init__.py': '01c870aa9763f4e4357fc6654bff1d3296cec995bd0943f87b4a48946ada5b40', 'verifiers/dqi/dqi_reference_circuit.py': '2b8e1b80a806157c333a7e308eaf2499ee09de499efbc05875202c327959959e', 'verifiers/dqi/verify_submission_dqi.py': 'f14274bc0943fb9769ddb01131be6063191a15009fe1caf1d4bdd9fd25194e65', 'verifiers/invsqrt/__init__.py': 'e9a6ecf253c343528578be5fa92eb820ccade63333aebb784552fb4ae75dbc0b', 'verifiers/invsqrt/nr_reference_circuit.py': 'bb90330dcf774f7b502de471293facba67f27952cc8808fe4a9970a9b1322e23', 'verifiers/invsqrt/verify_submission_invsqrt.py': 'ddef520dec6b68d1cf0a6f328ddd36d79eb84570b9192cf3f7acaf073f06733a'}
SOURCES = {'notebooks/launch.py': '"""Isolated launcher for the pinned official verifiers and reviewed local code.\n\nSPDX-License-Identifier: Apache-2.0\nOnly authenticated competition directories are added; never input ZIP code.\n"""\nfrom pathlib import Path\nimport runpy\nimport sys\n\nroot = Path(__file__).resolve().parent.parent\nsys.dont_write_bytecode = True\nsys.path[:0] = [str(root / \'data/comp/kickmix\'), str(root / \'data/comp/verifiers\')]\ntarget = Path(sys.argv[1]).resolve()\nsys.argv = [str(target), *sys.argv[2:]]\nrunpy.run_path(str(target), run_name=\'__main__\')\n', 'notebooks/take.py': '"""Copy pinned submission members out of the notebook\'s inputs.\n\nSPDX-License-Identifier: Apache-2.0\nRun with Python -I: take.py INPUT_ROOT OUTPUT_DIR MEMBERS_JSON [MORE_ROOTS...]\nMEMBERS_JSON maps each member name to its sha256 and size. A member is taken from a\nplain file whose path ends with the member\'s name, or else from a ZIP member whose\nname ends the same way; only a copy with the pinned size and SHA-256 is kept. Nothing\nfrom an input is executed, and no archive is extracted whole. MORE_ROOTS, such as the folder of\nmembers built in the notebook, are searched the same way.\n"""\nimport hashlib\nimport json\nfrom pathlib import Path, PurePosixPath\nimport shutil\nimport sys\nimport zipfile\n\n\ndef digest(stream):\n    h = hashlib.sha256()\n    while chunk := stream.read(1 << 22):\n        h.update(chunk)\n    return h.hexdigest()\n\n\ndef ends_with(path_parts, name_parts):\n    return tuple(path_parts[-len(name_parts):]) == name_parts\n\n\ndef main():\n    root, out, manifest = Path(sys.argv[1]), Path(sys.argv[2]), json.loads(Path(sys.argv[3]).read_text())\n    roots = [root, *map(Path, sys.argv[4:])]\n    files = sorted(p for r in roots if r.is_dir() for p in r.rglob(\'*\') if p.is_file())\n    zips = [p for p in files if p.suffix == \'.zip\']\n    taken = {}\n    for name, pin in manifest.items():\n        parts = PurePosixPath(name).parts\n        assert not PurePosixPath(name).is_absolute() and \'..\' not in parts, name\n        target = out / name\n        target.parent.mkdir(parents=True, exist_ok=True)\n        for path in files:\n            if ends_with(path.parts, parts) and path.stat().st_size == pin[\'size\']:\n                with path.open(\'rb\') as stream:\n                    if digest(stream) == pin[\'sha256\']:\n                        shutil.copyfile(path, target)\n                        taken[name] = str(path)\n                        break\n        for path in zips if name not in taken else []:\n            with zipfile.ZipFile(path) as archive:\n                for info in archive.infolist():\n                    if not ends_with(PurePosixPath(info.filename).parts, parts) or info.file_size != pin[\'size\']:\n                        continue\n                    # Reading to EOF also checks the member\'s ZIP CRC.\n                    with archive.open(info) as source, target.open(\'wb\') as copy:\n                        shutil.copyfileobj(source, copy, length=1 << 20)\n                    with target.open(\'rb\') as stream:\n                        if digest(stream) == pin[\'sha256\']:\n                            taken[name] = f\'{path}:{info.filename}\'\n                            break\n                    target.unlink()\n            if name in taken:\n                break\n        if name not in taken:\n            raise SystemExit(f\'{name}: no input holds SHA-256 {pin["sha256"]}; attach the source its block names\')\n    print(json.dumps({\'status\': \'PASS\', \'taken\': taken}, indent=1))\n\n\nif __name__ == \'__main__\':\n    main()\n', 'notebooks/check-submission': '#!/usr/bin/env python3\n"""Checks a submission folder against the metric\'s extra rules, the ones the local verifiers skip.\n\n    .venv/bin/python -I tools/check-submission SUBMISSION_DIR\n\nThe rules come from the metric notebook, as reported in Kaggle discussion 751951; a submission that breaks one\nis rejected as a whole, even when every verifier passes:\n- every circuit\'s largest qubit or bit id, max(num_qubits, num_bits) - 1, is below 2**30 // 1536 = 699,050;\n- every register has a name;\n- the folder holds exactly invsqrt.kmb, dqi.kmb and dfthcsos/ with angle_k_0 .. angle_k_<n-1> (no gaps) and\n  submission.npz, and nothing else anywhere;\n- submission.npz arrays are little-endian floats, except the index arrays tree_edges and angle_words, which are\n  integer types. The sample submission, which the metric accepted in submission 1, stores angle_words as int64.\n  The 8-bit factorizations of 2026-10-26 also store brot and kept_indices as int64 and one_body_requires_refresh\n  as a bool; the metric scored exactly such a file, version 14 of kirilltym\'s QROM + Tower42 (public 19.57899).\n"""\nimport os\nimport pathlib\nimport re\nimport sys\n\nimport numpy as np\n\nsys.path.insert(0, str(pathlib.Path(__file__).resolve().parent.parent / \'data\' / \'comp\' / \'kickmix\'))\nimport kickmix as km  # noqa: E402\n\nMAX_ID = 2**30 // 1536\nINTEGER_ARRAYS = (\'tree_edges\', \'angle_words\', \'brot\', \'kept_indices\')\nBOOL_ARRAYS = (\'one_body_requires_refresh\',)\n\nroot = sys.argv[1]\nproblems = []\nfiles = sorted(os.path.relpath(os.path.join(d, f), root) for d, _, fs in os.walk(root) for f in fs)\nangles = [f for f in files if re.fullmatch(r\'dfthcsos/angle_k_\\d+\\.kmb\', f)]\nexpected = {\'invsqrt.kmb\', \'dqi.kmb\', \'dfthcsos/submission.npz\'} | {f\'dfthcsos/angle_k_{i}.kmb\' for i in range(len(angles))}\nif set(files) != expected:\n    problems.append(f\'file set differs: extra {sorted(set(files) - expected)}, missing {sorted(expected - set(files))}\')\nfor f in [f for f in files if f.endswith(\'.kmb\')]:\n    c = km.Circuit.from_file(os.path.join(root, f))\n    top = max(c.num_qubits, c.num_bits) - 1\n    unnamed = [k for k in c.register_data if not isinstance(k, str)]\n    print(f\'{f}: qubits {c.num_qubits}, bits {c.num_bits}, largest id {top}, {len(c.register_data)} registers\')\n    if top >= MAX_ID:\n        problems.append(f\'{f}: largest id {top} >= {MAX_ID}\')\n    if unnamed:\n        problems.append(f\'{f}: unnamed registers {unnamed}\')\nnpz = np.load(os.path.join(root, \'dfthcsos\', \'submission.npz\'), allow_pickle=False)\nfor name in npz.files:\n    a = npz[name]\n    kind = np.integer if name in INTEGER_ARRAYS else np.bool_ if name in BOOL_ARRAYS else np.floating\n    kind_ok = np.issubdtype(a.dtype, kind)\n    little = a.dtype.byteorder in (\'<\', \'|\') or (a.dtype.byteorder == \'=\' and sys.byteorder == \'little\')\n    print(f\'npz {name}: dtype {a.dtype.str}, shape {a.shape}\')\n    if not kind_ok or not little:\n        problems.append(f\'npz {name}: dtype {a.dtype.str}\')\nprint(\'PROBLEMS:\' if problems else \'ALL CHECKS PASS\', *problems, sep=\'\\n  \')\nsys.exit(1 if problems else 0)\n', 'notebooks/extra-format.py': '#!/usr/bin/env python3\n"""Supplement check-submission with NPZ header bounds and ANGLE T counts.\n\nFrom runs/submission-8/extra-format.py; booleans are allowed since 2026-10-27, as in check-submission.\n"""\nimport json\nimport math\nfrom pathlib import Path\nimport sys\nimport zipfile\n\nimport kickmix as km\nimport numpy as np\n\nroot = Path(sys.argv[1]) / "dfthcsos"\naudit = {"npz_headers": {}, "angle_circuits": {}}\nwith zipfile.ZipFile(root / "submission.npz") as z:\n    for member in z.infolist():\n        with z.open(member) as f:\n            version = np.lib.format.read_magic(f)\n            assert version == (1, 0), (member.filename, version)\n            shape, _, dtype = np.lib.format.read_array_header_1_0(f)\n        assert dtype.kind in ("f", "i", "u", "b") and dtype.byteorder != ">"\n        assert dtype.names is None and dtype.subdtype is None\n        audit["npz_headers"][member.filename] = {\n            "shape": list(shape), "dtype": dtype.str,\n            "array_bytes": math.prod(shape) * dtype.itemsize,\n            "npy_version": list(version),\n        }\naudit["total_npz_array_bytes"] = sum(x["array_bytes"] for x in audit["npz_headers"].values())\nassert audit["total_npz_array_bytes"] <= 2**26\nfor p in sorted(root.glob("angle_k_*.kmb")):\n    circuit = km.Circuit.from_file(str(p))\n    stats = circuit.operation_statistics()\n    audit["angle_circuits"][p.name] = {\n        "num_qubits": circuit.num_qubits, "num_bits": circuit.num_bits,\n        "registers": list(circuit.register_data), "t_gates": int(stats.num_t_like),\n        "static_toffolis": int(stats.num_toffoli_like),\n    }\n    assert stats.num_t_like == 0 and circuit.num_qubits <= 128\nprint(json.dumps(audit, indent=2))\n', 'LICENSE': '\n                                 Apache License\n                           Version 2.0, January 2004\n                        http://www.apache.org/licenses/\n\n   TERMS AND CONDITIONS FOR USE, REPRODUCTION, AND DISTRIBUTION\n\n   1. Definitions.\n\n      "License" shall mean the terms and conditions for use, reproduction,\n      and distribution as defined by Sections 1 through 9 of this document.\n\n      "Licensor" shall mean the copyright owner or entity authorized by\n      the copyright owner that is granting the License.\n\n      "Legal Entity" shall mean the union of the acting entity and all\n      other entities that control, are controlled by, or are under common\n      control with that entity. For the purposes of this definition,\n      "control" means (i) the power, direct or indirect, to cause the\n      direction or management of such entity, whether by contract or\n      otherwise, or (ii) ownership of fifty percent (50%) or more of the\n      outstanding shares, or (iii) beneficial ownership of such entity.\n\n      "You" (or "Your") shall mean an individual or Legal Entity\n      exercising permissions granted by this License.\n\n      "Source" form shall mean the preferred form for making modifications,\n      including but not limited to software source code, documentation\n      source, and configuration files.\n\n      "Object" form shall mean any form resulting from mechanical\n      transformation or translation of a Source form, including but\n      not limited to compiled object code, generated documentation,\n      and conversions to other media types.\n\n      "Work" shall mean the work of authorship, whether in Source or\n      Object form, made available under the License, as indicated by a\n      copyright notice that is included in or attached to the work\n      (an example is provided in the Appendix below).\n\n      "Derivative Works" shall mean any work, whether in Source or Object\n      form, that is based on (or derived from) the Work and for which the\n      editorial revisions, annotations, elaborations, or other modifications\n      represent, as a whole, an original work of authorship. For the purposes\n      of this License, Derivative Works shall not include works that remain\n      separable from, or merely link (or bind by name) to the interfaces of,\n      the Work and Derivative Works thereof.\n\n      "Contribution" shall mean any work of authorship, including\n      the original version of the Work and any modifications or additions\n      to that Work or Derivative Works thereof, that is intentionally\n      submitted to Licensor for inclusion in the Work by the copyright owner\n      or by an individual or Legal Entity authorized to submit on behalf of\n      the copyright owner. For the purposes of this definition, "submitted"\n      means any form of electronic, verbal, or written communication sent\n      to the Licensor or its representatives, including but not limited to\n      communication on electronic mailing lists, source code control systems,\n      and issue tracking systems that are managed by, or on behalf of, the\n      Licensor for the purpose of discussing and improving the Work, but\n      excluding communication that is conspicuously marked or otherwise\n      designated in writing by the copyright owner as "Not a Contribution."\n\n      "Contributor" shall mean Licensor and any individual or Legal Entity\n      on behalf of whom a Contribution has been received by Licensor and\n      subsequently incorporated within the Work.\n\n   2. Grant of Copyright License. Subject to the terms and conditions of\n      this License, each Contributor hereby grants to You a perpetual,\n      worldwide, non-exclusive, no-charge, royalty-free, irrevocable\n      copyright license to reproduce, prepare Derivative Works of,\n      publicly display, publicly perform, sublicense, and distribute the\n      Work and such Derivative Works in Source or Object form.\n\n   3. Grant of Patent License. Subject to the terms and conditions of\n      this License, each Contributor hereby grants to You a perpetual,\n      worldwide, non-exclusive, no-charge, royalty-free, irrevocable\n      (except as stated in this section) patent license to make, have made,\n      use, offer to sell, sell, import, and otherwise transfer the Work,\n      where such license applies only to those patent claims licensable\n      by such Contributor that are necessarily infringed by their\n      Contribution(s) alone or by combination of their Contribution(s)\n      with the Work to which such Contribution(s) was submitted. If You\n      institute patent litigation against any entity (including a\n      cross-claim or counterclaim in a lawsuit) alleging that the Work\n      or a Contribution incorporated within the Work constitutes direct\n      or contributory patent infringement, then any patent licenses\n      granted to You under this License for that Work shall terminate\n      as of the date such litigation is filed.\n\n   4. Redistribution. You may reproduce and distribute copies of the\n      Work or Derivative Works thereof in any medium, with or without\n      modifications, and in Source or Object form, provided that You\n      meet the following conditions:\n\n      (a) You must give any other recipients of the Work or\n          Derivative Works a copy of this License; and\n\n      (b) You must cause any modified files to carry prominent notices\n          stating that You changed the files; and\n\n      (c) You must retain, in the Source form of any Derivative Works\n          that You distribute, all copyright, patent, trademark, and\n          attribution notices from the Source form of the Work,\n          excluding those notices that do not pertain to any part of\n          the Derivative Works; and\n\n      (d) If the Work includes a "NOTICE" text file as part of its\n          distribution, then any Derivative Works that You distribute must\n          include a readable copy of the attribution notices contained\n          within such NOTICE file, excluding those notices that do not\n          pertain to any part of the Derivative Works, in at least one\n          of the following places: within a NOTICE text file distributed\n          as part of the Derivative Works; within the Source form or\n          documentation, if provided along with the Derivative Works; or,\n          within a display generated by the Derivative Works, if and\n          wherever such third-party notices normally appear. The contents\n          of the NOTICE file are for informational purposes only and\n          do not modify the License. You may add Your own attribution\n          notices within Derivative Works that You distribute, alongside\n          or as an addendum to the NOTICE text from the Work, provided\n          that such additional attribution notices cannot be construed\n          as modifying the License.\n\n      You may add Your own copyright statement to Your modifications and\n      may provide additional or different license terms and conditions\n      for use, reproduction, or distribution of Your modifications, or\n      for any such Derivative Works as a whole, provided Your use,\n      reproduction, and distribution of the Work otherwise complies with\n      the conditions stated in this License.\n\n   5. Submission of Contributions. Unless You explicitly state otherwise,\n      any Contribution intentionally submitted for inclusion in the Work\n      by You to the Licensor shall be under the terms and conditions of\n      this License, without any additional terms or conditions.\n      Notwithstanding the above, nothing herein shall supersede or modify\n      the terms of any separate license agreement you may have executed\n      with Licensor regarding such Contributions.\n\n   6. Trademarks. This License does not grant permission to use the trade\n      names, trademarks, service marks, or product names of the Licensor,\n      except as required for reasonable and customary use in describing the\n      origin of the Work and reproducing the content of the NOTICE file.\n\n   7. Disclaimer of Warranty. Unless required by applicable law or\n      agreed to in writing, Licensor provides the Work (and each\n      Contributor provides its Contributions) on an "AS IS" BASIS,\n      WITHOUT WARRANTIES OR CONDITIONS OF ANY KIND, either express or\n      implied, including, without limitation, any warranties or conditions\n      of TITLE, NON-INFRINGEMENT, MERCHANTABILITY, or FITNESS FOR A\n      PARTICULAR PURPOSE. You are solely responsible for determining the\n      appropriateness of using or redistributing the Work and assume any\n      risks associated with Your exercise of permissions under this License.\n\n   8. Limitation of Liability. In no event and under no legal theory,\n      whether in tort (including negligence), contract, or otherwise,\n      unless required by applicable law (such as deliberate and grossly\n      negligent acts) or agreed to in writing, shall any Contributor be\n      liable to You for damages, including any direct, indirect, special,\n      incidental, or consequential damages of any character arising as a\n      result of this License or out of the use or inability to use the\n      Work (including but not limited to damages for loss of goodwill,\n      work stoppage, computer failure or malfunction, or any and all\n      other commercial damages or losses), even if such Contributor\n      has been advised of the possibility of such damages.\n\n   9. Accepting Warranty or Additional Liability. While redistributing\n      the Work or Derivative Works thereof, You may choose to offer,\n      and charge a fee for, acceptance of support, warranty, indemnity,\n      or other liability obligations and/or rights consistent with this\n      License. However, in accepting such obligations, You may act only\n      on Your own behalf and on Your sole responsibility, not on behalf\n      of any other Contributor, and only if You agree to indemnify,\n      defend, and hold each Contributor harmless for any liability\n      incurred by, or claims asserted against, such Contributor by reason\n      of your accepting any such warranty or additional liability.\n\n   END OF TERMS AND CONDITIONS\n\n   APPENDIX: How to apply the Apache License to your work.\n\n      To apply the Apache License to your work, attach the following\n      boilerplate notice, with the fields enclosed by brackets "[]"\n      replaced with your own identifying information. (Don\'t include\n      the brackets!)  The text should be enclosed in the appropriate\n      comment syntax for the file format. We also recommend that a\n      file or class name and description of purpose be included on the\n      same "printed page" as the copyright notice for easier\n      identification within third-party archives.\n\n   Copyright [yyyy] [name of copyright owner]\n\n   Licensed under the Apache License, Version 2.0 (the "License");\n   you may not use this file except in compliance with the License.\n   You may obtain a copy of the License at\n\n       http://www.apache.org/licenses/LICENSE-2.0\n\n   Unless required by applicable law or agreed to in writing, software\n   distributed under the License is distributed on an "AS IS" BASIS,\n   WITHOUT WARRANTIES OR CONDITIONS OF ANY KIND, either express or implied.\n   See the License for the specific language governing permissions and\n   limitations under the License.\n', 'blocks/dqi/pro-q2-897/source/README.md': "# Pro's 897-qubit decoder with the 658-Toffoli jet\n\nThis is the decoder folder of our second decoder kit, unchanged, with one more file:\n\n- **The kit's decoder** (`decode.py`, `exp/`, `playbook/`): our 883-qubit decoder before Ahmed Berat Özer's narrowings, with the 658-Toffoli jet in `exp/jet24/pro_jet.py`.\n- **`final_decoder/`:** GPT-6 Astra Pro's three files, unchanged: `decode.py`, `conversion.py` and `synthesis.py`. They import the kit's modules.\n- **`build_897.py`:** the build script.\n\nBuild with `python build_897.py OUTPUT.kmb`, with Kickmix on the path. It writes `dqi.kmb`, SHA-256 `b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6`, and `dqi.build.json` with its counts. It takes about a minute and 5 GB.\n", 'blocks/dqi/pro-q2-897/source/build_897.py': '"""Builds Pro\'s q2 decoder with the 658-Toffoli jet: python build_897.py OUTPUT.kmb, with Kickmix on the path.\n\nIt writes OUTPUT.kmb, SHA-256 b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6, and OUTPUT.build.json\nwith its counts. It builds the way pro/q2-decoder/local/official.sh does: the entry final_decoder/decode.py, with\nKickmix as the kit\'s builder module circ, and the registers S0 to S69 and E0 to E4094 named in the file.\n"""\nimport hashlib\nimport importlib.util\nimport json\nfrom pathlib import Path\nimport sys\nimport time\n\nimport kickmix as km\n\nsys.modules[\'circ\'] = km\nHERE = Path(__file__).resolve().parent\ntarget = Path(sys.argv[1])\nstart = time.time()\nspec = importlib.util.spec_from_file_location(\'decoder_under_test\', HERE / \'final_decoder/decode.py\')\nmodule = importlib.util.module_from_spec(spec)\nsys.modules[\'decoder_under_test\'] = module\nspec.loader.exec_module(module)\nb = km.CircuitBuilder()\nS = [b.alloc_qubits(12) for _ in range(70)]\nE = [b.alloc_bits(12) for _ in range(4095)]\nmodule.decode(b, S, E)\nmetadata = []\nfor i in range(70):\n    metadata.append(f\'REGISTER r{i} "S{i}"\')\n    metadata.extend(f\'APPEND_TO_REGISTER q{12 * i + k} r{i}\' for k in range(12))\nfor j in range(4095):\n    metadata.append(f\'REGISTER r{70 + j} "E{j}"\')\n    metadata.extend(f\'APPEND_TO_REGISTER b{12 * j + k} r{70 + j}\' for k in range(12))\nc = km.Circuit(\'\\n\'.join(metadata)) + b.finish_circuit()\nc.to_file(str(target), format=\'kmb\')\nwith open(target, \'rb\') as f:\n    digest = hashlib.file_digest(f, \'sha256\').hexdigest()\nrecord = dict(sha256=digest, Q=c.num_qubits, T_static=c.operation_statistics().num_toffoli_like, num_ops=len(c),\n              num_bits=c.num_bits, seconds=round(time.time() - start, 1))\ntarget.with_suffix(\'.build.json\').write_text(json.dumps(record, indent=1) + \'\\n\')\nprint(json.dumps(record), flush=True)\n', 'blocks/dqi/pro-q2-897/source/decode.py': '"""Our DQI decoder as decode(b, S, E), for check_decoder.py.\n\nThe stages, in order: mask projection, the initial linear maps, extraction (the jet kernel 289 times),\nsynthesis, conversion, scan, then conversion, synthesis, extraction and the linear maps undone, and the masks\nrestored. The jet kernel is exp/jet24/pro_jet.py: 683 Toffolis a pair (forward 290, reverse 393) at 43 private\nqubits, which sets Q at 883. The undo of the synthesis uses two narrower divisions than the forward\'s library\ndefault, so that it fits within 883 too: see unsynthesize below.\nWritten for the q2 kit by Halley (Claude Opus 5.5), 2026-10-27, from the decoder\'s own build script.\n"""\nimport sys\nfrom pathlib import Path\n\nHERE = Path(__file__).resolve().parent\nsys.path[:0] = [str(HERE.parent), str(HERE), str(HERE / \'exp/decoder\')]\n\nfrom common import *  # noqa: E402,F401,F403  (km is circ; ROOT is this folder)\nfrom exp.confluent_extract.kernel import initial, extract as extract_jets  # noqa: E402\nfrom exp.confluent_extract.model import constants as confluent_constants  # noqa: E402\nfrom exp.jet24.pro_jet import jet_forward as _jet_forward, jet_reverse as _jet_reverse  # noqa: E402\nimport exp.fold_cached_norm.kernel as FK  # noqa: E402\nfrom exp.fold_joint.kernel import divide414_callback  # noqa: E402\nfrom exp.mask_projection.kernel import project, restore, SELECTED  # noqa: E402\nsys.path.insert(0, str(HERE / \'exp/convert\'))\nfrom exp.convert_short_hankel.phase34 import convert_safe as convert_historyfree  # noqa: E402\nsys.path.insert(0, str(HERE / \'exp/scan\'))\nfrom scan import scan  # noqa: E402\n\n\ndef jet_forward(b, u, v, K=None):\n    assert K in (None, 2048)\n    return _jet_forward(b, u, v)\n\n\ndef jet_reverse(b, u, v, token, K=None):\n    assert K in (None, 2048)\n    return _jet_reverse(b, u, v, token)\n\n\ndef extract(b, words, reverse=False, transcript=None):\n    return extract_jets(b, words, jet_forward, jet_reverse, reverse, transcript)\n\n\ndef divide_zero414(b, x, y, K=None, extra_zero_mask=None, after=None):\n    """The outer fold\'s identity-at-zero division x /= y, at 25 private qubits: 414 Toffolis (the wide one, 324 at\n    49, would set Q at 889 here, since 840 qubits are live)."""\n    divide414_callback(b, x, y, K, extra_zero_mask, after)\n\n\nsynthesize = FK.synthesize\n# The forward synthesis multiplies; only its undo divides. At the inner folds\' divisions 853 qubits are live, so\n# they use the 24-scratch division of exp/mul99/schedules.py (369 Toffolis) instead of the retained-norm one\n# (323 at 36 private), and the outer fold uses divide_zero414.\nunsynthesize = FK.bind(FK.prior.unsynthesize, absorb=FK.prior.absorb_odd, absorb_even=FK.prior.absorb_even,\n                       outer=FK.bind(FK.prior.outer, divide_zero369=divide_zero414), odd_decode=FK.odd_decode,\n                       multiply_zero144=FK.multiply_zero141)\n\n\ndef decode(b, S, E):\n    with b.mark(\'mask_projection\'):\n        mask_token = project(b, S, E)\n    with b.mark(\'initial\'):\n        initial(b, S)\n    with b.mark(\'extract\'):\n        xh = extract(b, S)\n    with b.mark(\'synthesis\'):\n        A, B, yh = synthesize(b, S)\n    history_word = b.alloc_qubits(12)\n    with b.mark(\'convert\'):\n        phase_token = convert_historyfree(b, A, B, history_scratch12=history_word, monic_root=True)\n    with b.mark(\'scan\'):\n        scan(b, A, B, E, skip_points=SELECTED, node_polynomial=confluent_constants()[2])\n    with b.mark(\'unconvert\'):\n        convert_historyfree(b, A, B, True, history_scratch12=history_word, phase_token=phase_token)\n    b.free(history_word)\n    with b.mark(\'unsynthesis\'):\n        unsynthesize(b, yh)\n    with b.mark(\'unextract\'):\n        extract(b, S, True, xh)\n    with b.mark(\'uninitial\'):\n        initial(b, S, True)\n    with b.mark(\'restore_masks\'):\n        restore(b, E, mask_token)\n', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/even_fold.py': '"""Full product/quotient jointly updates explicit odd continuant r, no shear."""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.tail_alias.kernel import *\nfrom exp.mul99.schedules import divide_zero_bridge414\n\ndef full_product(b,head,A,qi,target):\n n=len(target);V,W=base.vandermonde(n)\n for t in range(n):old.rank_affine(b,head,[1,t],A+[qi],V[t][:len(A)+1],target,[W[j][t]for j in range(n)])\n\ndef full_quotient(b,head,A,qi,B,r,kap):\n k=len(B)\n if not k:ar.mul_xor0(b,head[1],qi,r);return\n words=A[1:]+[qi]\n for ac,bc,tc in reversed(base.quotient_terms(k+2,kap)):\n  old.rank_affine(b,head,ac,words,bc,B+[r],tc)\n\ndef absorb_even(b,head,A,B,qi,r,c,kap,reverse=False,transcript=None):\n """r=qnext->qprev in forward, inverse restores qnext; qi disjoint r."""\n k=len(B);out=head+B;target=out+[r]\n if not reverse:\n  value=old.eval_implicit(b,A,qi,c);F=dv.zero_flag(b,value);old.eval_implicit(b,A,qi,c,True)\n  b.cx(F,A[0][0]);spare=[b.alloc_qubits(12),b.alloc_qubits(12)];full=spare+B+[r]\n  linear_words(b,full,base.embedding_matrix(k+3,kap));full_product(b,head,A,qi,full);b.cx(F,A[0][0])\n  hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n  for w,z in zip(head,spare):b.swap(w,z);b.free(z)\n  fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n  return out,A,(hm,fm)\n linear_words(b,target,base.embedding_matrix(k+3,kap),True)\n value=old.eval_implicit(b,A,qi,c);cx(b,head[1],head[0],c);hm,fm=transcript\n divide_zero_bridge414(b,combine(head),value,K=K,extra_zero_mask=fm[0]);b.free(fm)\n cx(b,head[1],head[0],c);b.cz(combine(head),hm);b.free(hm);old.eval_implicit(b,A,qi,c,True)\n full_quotient(b,head,A,qi,B,r,kap)\n return out,A,transcript\n', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/kernel.py': '"""Half-density continuant prefix: odd encoded heads, one rolling odd q.\n\nRaw b0 becomes a reversible physical q0=1 border; r carries its old value.\nThe border is freed after inner folds and r becomes the numerator top word.\nNo persistent quantum zero flag or second leading-coefficient bank is used.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.tail_alias.kernel import *\nfrom exp.alternating_gauge.even_fold import absorb_even\nfrom exp.alternating_gauge.scalar import prefix_reverse141,divide_zero117,multiply_zero144\n\ndef add_tail_a(b,a,q,d):\n ar.linear_inplace(b,q,ar.square_columns(12));ar.mul_xor0(b,a,q,d)\n ar.linear_inplace(b,q,[power(1<<j,2048)for j in range(12)])\n\ndef odd_encode(b,heads,i,r,d=None):\n a,x=heads[i];p=heads[i-2][1]if i>1 else heads[0][1]\n def after():\n  add_tail_a(b,a,r,d);ar.mul_xor0(b,x,r,d)\n return pg.forward(b,x,p,r,after_product=after if d is not None else None)\n\ndef odd_decode(b,heads,i,r,m,d=None):\n a,x=heads[i];p=heads[i-2][1]if i>1 else heads[0][1]\n if d is not None:ar.mul_xor0(b,x,r,d)\n prefix_reverse141(b,x,p,r,m)\n if d is not None:add_tail_a(b,a,r,d)\n\ndef even_step(b,heads,i,r,d=None,reverse=False):\n a,x=heads[i];q=heads[i-1][1]\n if d is not None and reverse:\n  add_tail_a(b,a,q,d);ar.mul_xor0(b,r,q,d)\n ar.mul_xor0(b,x,q,r)\n if d is not None and not reverse:\n  add_tail_a(b,a,q,d);ar.mul_xor0(b,r,q,d)\n\ndef synthesize(b,words):\n h=len(words)//2;assert h>=3 and h%2==1;cs,ks,_,_=old.constants(h)\n heads=[words[2*i:2*i+2]for i in range(h)];d=heads[0][0];border=heads[0][1]\n r=b.alloc_qubits(12);b.cx(border,r);b.cx(border,d);b.cx(r,border);b.x(border[0]);masks=[None]*h\n for i in range(1,h):\n  if i%2:masks[i]=odd_encode(b,heads,i,r,d)\n  else:even_step(b,heads,i,r,d)\n divide_zero117(b,d,r);A=[d];B=[];hist=[]\n for i in reversed(range(1,h)):\n  if i%2:\n   odd_decode(b,heads,i,r,masks[i]);qi=r;p=heads[i-2][1]if i>1 else border\n  else:\n   qi=heads[i-1][1];p=r\n  aa,bb=A,B\n  if i%2:A,B,token=absorb(b,heads[i],A,B,qi,p,None,0,cs[i],ks[i],wide=False)\n  else:A,B,token=absorb_even(b,heads[i],A,B,qi,r,cs[i],ks[i])\n  hist.append((i,aa,bb,qi,p,cs[i],ks[i],token))\n b.x(border[0]);b.free(border);C=B+[r];ot=outer(b,C,A,cs[0],ks[0],wide=True)\n return C,A,dict(h=h,heads=heads,d=d,r=r,border=border,history=hist,A=C,B=A,outer=ot)\n\ndef unsynthesize(b,t):\n h=t[\'h\'];heads=t[\'heads\'];r=t[\'r\'];cs,ks,_,_=old.constants(h)\n outer(b,t[\'A\'],t[\'B\'],cs[0],ks[0],True,t[\'outer\'],True)\n reclaim(b,[t[\'border\']]);b.x(t[\'border\'][0]);masks=[None]*h\n for i,A,B,qi,p,c,kap,token in reversed(t[\'history\']):\n  if i%2:\n   absorb(b,heads[i],A,B,qi,p,None,0,c,kap,True,token,False);masks[i]=odd_encode(b,heads,i,r)\n  else:absorb_even(b,heads[i],A,B,qi,r,c,kap,True,token)\n multiply_zero144(b,t[\'d\'],r)\n for i in reversed(range(1,h)):\n  if i%2:odd_decode(b,heads,i,r,masks[i],t[\'d\'])\n  else:even_step(b,heads,i,r,t[\'d\'],True)\n b.cx(r,t[\'d\']);b.x(t[\'border\'][0]);b.cx(r,t[\'border\']);b.cx(t[\'border\'],r);b.free(r)\n', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/scalar.py': '"""25-private zero-aware scalar context; same117T as promised old divide."""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.tail_alias.kernel import *\nfrom exp.inverse19_joint.kernel import inverse6_init19,inverse_zero\nfrom exp.arithmetic.lowwidth import phase_inverse_generic6\nfrom exp.safe_scalar.phase import phase_inverse36_zero\nfrom exp.noscratch.schedules import norm_xor_noscratch\n\n\ndef divide_zero117(b,x,y,before=None,after=None):\n """All-input x/=y or x when y0;117T/25private; hooks see F=[y0]."""\n to,frm,kk=ar.tower(12);N=b.alloc_qubits(6);ar.linear_inplace(b,y,to);norm_xor_noscratch(b,y,N,kk);ar.linear_inplace(b,y,frm)\n u,F=inverse6_init19(b,N,True)\n if before is not None:before(F)\n conj=[ar.field(12).frobenius(1<<i,6)for i in range(12)]\n ar.linear_inplace(b,y,conj);b.cx(F,y[0]);out=b.alloc_qubits(12);ar.mul_xor0(b,x,y,out)\n oldm=b.alloc_bits(12);b.hmr(x,oldm);b.swap(x,out);b.free(out);b.cx(F,y[0]);ar.linear_inplace(b,y,conj)\n ar.linear_inplace(b,x,to);b.cx(F,u[0]);b.cx(F,N[0])\n for off in(0,6):\n  half=x[off:off+6];out=b.alloc_qubits(6);ar.mul_xor0(b,half,u,out)\n  m=b.alloc_bits(6);b.hmr(half,m);b.swap(half,out);b.free(out);ar.phase_product(b,half,N,m);b.free(m)\n b.cx(F,N[0]);b.cx(F,u[0]);ar.linear_inplace(b,x,frm)\n b.cx(F,y[0]);ar.phase_product(b,x,y,oldm);b.cx(F,y[0]);b.free(oldm)\n if after is not None:after(F)\n fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F);b.z(fm[0]);zm=b.alloc_bits(6);b.bit_store(zm,False);b.cx(fm[0],zm[0]);ar.phase_product(b,N,u,zm);b.free(zm);b.free(fm)\n um=b.alloc_bits(6);b.hmr(u,um);b.free(u);phase_inverse_generic6(b,N,um);b.free(um)\n ar.linear_inplace(b,y,to);ar._erase_norm(b,y,N,kk);ar.linear_inplace(b,y,frm);b.free(N)\n\ndef multiply_zero144(b,x,y):\n """Inverse of divide_zero117,144T/25private, no deferred transcript."""\n u,F=inverse_zero(b,y);b.cx(F,y[0]);out=b.alloc_qubits(12);ar.mul_xor0(b,x,y,out)\n m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out);b.cx(F,y[0]);b.cx(F,u[0]);ar.phase_product(b,x,u,m);b.cx(F,u[0]);b.free(m)\n um=b.alloc_bits(12);fm=b.alloc_bits(1);b.hmr(u,um);b.hmr(F,fm[0]);b.free(u);b.free(F)\n phase_inverse36_zero(b,y,um,fm[0]);b.free(um);b.free(fm)\n\ndef prefix_reverse141(b,x,p,q,m):\n """Inverse oldpg.forward:141T/25private, including zero-q swap chart."""\n def before(F):b.cswap(F,x,p);b.cx(p,x);b.ccx(F,p,x)\n def after(F):\n  b.cz(x,m)\n  for j in range(12):b.ccz(F,x[j],m[j]);b.ccz(F,p[j],m[j])\n  b.free(m)\n divide_zero117(b,x,q,before,after)\n', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/division.py': '"""Width-bounded GF24 scalar multiplication/division reference schedules.\n\nAll24-bit words use explicit a+b*y, y²+y=K. This file is separate from the\nverified primitive API while the scalar-normalization direction is measured.\n"""\nfrom field import *\n\n\ndef mul24_xor126(b,x,y,z,K=None):\n    """z ^=x*y,126Toffolis,1scratch; inputs/target disjoint."""\n    if K is None:K=extension_columns()[2]\n    f=field(12);a=x[:12];c=x[12:];d=y[:12];e=y[12:];za=z[:12];zb=z[12:]\n    # p=a*d contributes to both output halves.\n    b.cx(za,zb);mul_xor(b,a,d,za);b.cx(za,zb)\n    # q=c*e contributes Kq to the first half only.\n    b.gf2_imul(f.invert(K),target=km.array(za),field=f);mul_xor(b,c,e,za);b.gf2_imul(K,target=km.array(za),field=f)\n    # r=(a+c)(d+e) contributes to the second half.\n    b.cx(c,a);b.cx(e,d);mul_xor(b,a,d,zb);b.cx(e,d);b.cx(c,a)\n\n\n\ndef mul24_xor(b,x,y,z,K=None):\n    """Default105T/1scratch product; old126Tcomparison ismul24_xor126."""\n    mul24_xor105(b,x,y,z,K)\n\n\ndef phase_product24(b,x,y,mask,K=None):\n    if K is None:K=extension_columns()[2]\n    a=x[:12];c=x[12:];d=y[:12];e=y[12:];ma=mask[:12];mb=mask[12:]\n    phase_product(b,a,d,ma)\n    mk=b.alloc_bits(12);b.bit_store(mk,False)\n    # Transpose constant multiplication K into the classical phase mask.\n    for i in range(12):\n        for j in bits(field(12).mul(K,1<<i)):b.cx(ma[j],mk[i])\n    phase_product(b,c,e,mk);b.free(mk)\n    phase_product(b,a,e,mb);phase_product(b,c,d,mb);phase_product(b,c,e,mb)\n\n\ndef multiply_forward(b,x,y,K=None):\n    """105T forward x*=y, returns deferred oldx phase mask.\n\n    For y=0 this is correct only on valid subspace x=0. It is not an all-input\n    identity-at-zero permutation; caller must enforce that promise or add flag.\n    """\n    z=b.alloc_qubits(24);mul24_xor(b,x,y,z,K)\n    m=b.alloc_bits(24);b.hmr(x,m);b.swap(x,z);b.free(z);return m\n\n\ndef divide_preserving_factor(b,x,y,K=None):\n    """513T division x/=y preserving y,43workspace; valid x=0 when y=0."""\n    ym=inverse24_inplace_forward(b,y,K)\n    z=b.alloc_qubits(24);mul24_xor(b,x,y,z,K)\n    xm=b.alloc_bits(24);b.hmr(x,xm);b.swap(x,z);b.free(z)\n    inverse24_inplace_reverse(b,y,ym,K)\n    phase_product24(b,x,y,xm,K);b.free(xm)\n\n\ndef multiply_reverse(b,x,y,transcript,K=None):\n    divide_preserving_factor(b,x,y,K)\n    b.cz(x,transcript);b.free(transcript)\n\n\ndef divide_via_norm(b,x,y,K=None,retain_norm=True):\n    """Division x/=y:321T/43workspace or342T/37workspace.\n\n    y=0 requires promisedx=0. This all-valid-state exactness includes0 but does\n    NOT implement identity on arbitraryx at y=0.\n    """\n    import field as ar\n    if K is None:K=extension_columns()[2]\n    b.cx(y[12:],y[:12]);oldmask=multiply_forward(b,x,y,K);b.cx(y[12:],y[:12])\n    N=b.alloc_qubits(12);ar._norm_xor(b,y,N,K)\n    if retain_norm:u,D=ar._inverse_keep_norm(b,N)\n    else:u=inverse(b,N)\n    for offset in (0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor(b,half,u,out)\n        erase_product(b,out,N,half);b.swap(half,out);b.free(out)\n    if retain_norm:\n        to,frm,K12=tower(12);linear_inplace(b,u,to)\n        v=b.alloc_qubits(6);ar._norm_xor(b,u,v,K12);linear_inplace(b,u,frm)\n        ar._erase_inverse_with_norm(b,N,u,v);b.free(u)\n        erase_inverse(b,D,v);b.free(v)\n        linear_inplace(b,N,to);ar._erase_norm(b,N,D,K12);b.free(D);linear_inplace(b,N,frm)\n    else:erase_inverse_releasing(b,N,u)\n    ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,oldmask,K);b.free(oldmask)\n\n\ndef multiply_reverse_via_norm(b,x,y,transcript,K=None,retain_norm=True):\n    divide_via_norm(b,x,y,K,retain_norm=retain_norm)\n    b.cz(x,transcript);b.free(transcript)\n\n\ndef zero_flag(b,x):\n    """Compute [x=0], n-1ANDs; returnoneflag, clearallinternalANDs."""\n    n=len(x);b.x(x);ladder=b.alloc_qubits(n-1)\n    b.init_and(x[0],x[1],ladder[0])\n    for i in range(1,n-1):b.init_and(ladder[i-1],x[i+1],ladder[i])\n    for i in reversed(range(n-2)):\n        if i==0:b.del_and(x[0],x[1],ladder[0])\n        else:b.del_and(ladder[i-1],x[i+1],ladder[i])\n    b.x(x);b.free(ladder[:-1]);return ladder[-1]\n\n\ndef phase_zero(b,y,mask,head=None,headmask=None):\n    """Applyzero(y)*(mask+dot(head,headmask));22T+1CCZ for y24."""\n    n=len(y);b.x(y);ladder=b.alloc_qubits(n-2)\n    b.init_and(y[0],y[1],ladder[0])\n    for i in range(1,n-2):b.init_and(ladder[i-1],y[i+1],ladder[i])\n    b.ccz(ladder[-1],y[-1],mask)\n    if head is not None:\n        q=b.alloc_qubits(1)[0]\n        for xi,mi in zip(head,headmask):b.ccx(mi,xi,q)\n        b.ccz(ladder[-1],y[-1],q)\n        for xi,mi in reversed(list(zip(head,headmask))):b.ccx(mi,xi,q)\n        b.free(q)\n    for i in reversed(range(n-2)):\n        if i==0:b.del_and(y[0],y[1],ladder[0])\n        else:b.del_and(ladder[i-1],y[i+1],ladder[i])\n    b.free(ladder);b.x(y)\n\n\ndef multiply_zero_forward(b,x,y,K=None):\n    """152T multiplication, identityat y0. Deferredhead+zero transcripts."""\n    flag=zero_flag(b,y);z=b.alloc_qubits(24);mul24_xor(b,x,y,z,K);b.ccx(flag,x,z)\n    oldmask=b.alloc_bits(24);b.hmr(x,oldmask);b.swap(x,z);b.free(z)\n    fm=b.alloc_bits(1);b.hmr(flag,fm[0]);b.free(flag)\n    return oldmask,fm[0]\n\n\ndef divide_zero(b,x,y,K=None,extra_zero_mask=None):\n    """379T exact division x/=y, identityat y0,43workspace."""\n    import field as ar\n    if K is None:K=extension_columns()[2]\n    N=b.alloc_qubits(12);ar._norm_xor(b,y,N,K);flag=zero_flag(b,N)\n    b.cx(y[12:],y[:12]);z=b.alloc_qubits(24);mul24_xor(b,x,y,z,K);b.ccx(flag,x,z);b.cx(y[12:],y[:12])\n    oldmask=b.alloc_bits(24);b.hmr(x,oldmask);b.swap(x,z);b.free(z)\n    b.cx(flag,N[0]);fm=b.alloc_bits(1);b.hmr(flag,fm[0]);b.free(flag)\n    u,D=ar._inverse_keep_norm(b,N)\n    for offset in (0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor(b,half,u,out)\n        erase_product(b,out,N,half);b.swap(half,out);b.free(out)\n    to,frm,K12=tower(12);linear_inplace(b,u,to);v=b.alloc_qubits(6);ar._norm_xor(b,u,v,K12);linear_inplace(b,u,frm)\n    ar._erase_inverse_with_norm(b,N,u,v);b.free(u);erase_inverse(b,D,v);b.free(v)\n    linear_inplace(b,N,to);ar._erase_norm(b,N,D,K12);b.free(D);linear_inplace(b,N,frm)\n    nm=b.alloc_bits(12);b.hmr(N,nm);b.free(N);ar._phase_norm(b,y,nm,K)\n    comb=b.alloc_bits(1);b.bit_store(comb,False);b.cx(fm[0],comb[0]);b.cx(nm[0],comb[0])\n    if extra_zero_mask is not None:b.cx(extra_zero_mask,comb[0])\n    phase_product24(b,x,y,oldmask,K);phase_zero(b,y,comb[0],x,oldmask)\n    b.free(oldmask);b.free(fm);b.free(nm);b.free(comb)\n\n\ndef multiply_zero_reverse(b,x,y,transcript,K=None):\n    oldmask,fm=transcript;divide_zero(b,x,y,K,extra_zero_mask=fm)\n    b.cz(x,oldmask);b.free(oldmask);b.free(fm)\n\n\n\ndef erase_inverse_releasing(b,x,y):\n    """Phase-only inverse erasure; measure/free target BEFORE recomputation.\n\n    Same9/45/153T recursion as field.erase_inverse, but permits early reuse\n    of y\'s qubits. This helper owns and frees y.\n    """\n    import field as ar\n    n=len(x)\n    if n==3:\n        mask=b.alloc_bits(3);b.hmr(y,mask);b.free(y)\n        for im,v in enumerate(inverse3_anf()):\n            ix=list(bits(im))\n            for j in bits(v):\n                if len(ix)==1:b.cz(x[ix[0]],mask[j])\n                elif len(ix)==2:b.ccz(x[ix[0]],x[ix[1]],mask[j])\n        b.free(mask);return\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,y,to)\n    mask=b.alloc_bits(n);b.hmr(y,mask);b.free(y)\n    linear_inplace(b,x,to);norm=b.alloc_qubits(k);ar._norm_xor(b,x,norm,K);invnorm=inverse(b,norm)\n    b.cx(x[k:],x[:k]);phase_product(b,x[:k],invnorm,mask[:k]);phase_product(b,x[k:],invnorm,mask[k:]);b.cx(x[k:],x[:k]);b.free(mask)\n    erase_inverse_releasing(b,norm,invnorm)\n    ar._erase_norm(b,x,norm,K);b.free(norm);linear_inplace(b,x,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/field.py': '"""Low-width GF(2^m) arithmetic, standard Kickmix polynomial basis.\n\nPublic: mul_xor(builder,x,y,z), inverse(builder,x,target=None),\nlinear_xor(builder, x, z, columns), linear_inplace(builder,x,columns),\nphase_product(builder,x,y,mask), erase_linear/erase_product/erase_inverse.\nAll array arguments can be lists or km.array. Arithmetic preserves inputs.\n`inverse` returns a fresh clean output, 0 maps to 0. Default GF2Field basis.\n"""\nfrom functools import lru_cache\nimport circ as km\n\n\ndef bits(x):\n    while x:\n        t=x&-x; yield t.bit_length()-1; x^=t\n\n\ndef apply(columns, value):\n    out=0\n    for i in bits(value): out ^= columns[i]\n    return out\n\n\ndef inverse_columns(columns):\n    n=len(columns)\n    rows=[sum(((c>>i)&1)<<j for j,c in enumerate(columns)) | (1<<(n+i)) for i in range(n)]\n    for j in range(n):\n        p=next(i for i in range(j,n) if rows[i]>>j&1)\n        rows[j],rows[p]=rows[p],rows[j]\n        for i in range(n):\n            if i!=j and rows[i]>>j&1: rows[i]^=rows[j]\n    return [sum(((rows[i]>>(n+j))&1)<<i for i in range(n)) for j in range(n)]\n\n\ndef linear_xor(b,x,z,columns):\n    for i,c in enumerate(columns):\n        for j in bits(c): b.cx(x[i],z[j])\n\n\ndef linear_inplace(b,x,columns):\n    """Invertible GF2 map, no ancillary qubits. columns[j] = image(1<<j)."""\n    n=len(x); rows=[sum(((c>>i)&1)<<j for j,c in enumerate(columns)) for i in range(n)]; ops=[]\n    for j in range(n):\n        p=next(i for i in range(j,n) if rows[i]>>j&1)\n        if p!=j: rows[p],rows[j]=rows[j],rows[p]; ops.append((\'swap\',p,j))\n        for i in range(n):\n            if i!=j and rows[i]>>j&1: rows[i]^=rows[j];ops.append((\'cx\',j,i))\n    for op,i,j in reversed(ops): getattr(b,op)(x[i],x[j])\n\n\ndef phase_linear(b,x,mask,columns):\n    for i,c in enumerate(columns):\n        for j in bits(c): b.cz(x[i],mask[j])\n\n\ndef erase_linear(b,x,z,columns):\n    mask=b.alloc_bits(len(z)); b.hmr(z,mask); phase_linear(b,x,mask,columns);b.free(mask)\n\n\n@lru_cache(None)\ndef field(n): return km.GF2Field(n)\n\n\n@lru_cache(None)\ndef tensor(n):\n    """Symmetric multiplication tensor (linear mask, output vector), rank3/6/15/42."""\n    f=field(n)\n    if n==2: masks=[1,2,3]\n    elif n==3: masks=[1,7,4,5,6,3]\n    elif n in (6,12,24):\n        k,d={6:(2,3),12:(3,4),24:(6,4)}[n]\n        sub=field(k); power=(2**n-1)//(2**k-1)\n        gen=f.pow(f.primitive_element,power)\n        def evalpoly(v,p):\n            q=0\n            for i in reversed(range(k+1)):q=f.mul(q,p)^((v>>i)&1)\n            return q\n        embroot=next(f.pow(gen,i) for i in range(1,2**k) if evalpoly(sub.modulus,f.pow(gen,i))==0)\n        embed=[f.pow(embroot,i) for i in range(k)]\n        theta=f.primitive_element\n        cols=[f.mul(embed[i],f.pow(theta,j)) for j in range(d) for i in range(k)]\n        inv=inverse_columns(cols)\n        lowmasks=[m for m,_ in tensor(k)]\n        masks=[]\n        for point in list(range(2*d-2))+[None]:\n            outcols=[]\n            for extbit in range(n):\n                coord=inv[extbit]\n                if point is None: value=(coord>>(k*(d-1)))&((1<<k)-1)\n                else:\n                    value=0\n                    for j in reversed(range(d)): value=sub.mul(value,point)^((coord>>(k*j))&((1<<k)-1))\n                outcols.append(value)\n            for lm in lowmasks:masks.append(sum(((v&lm).bit_count()%2)<<i for i,v in enumerate(outcols)))\n    else: raise ValueError(n)\n    # Each basis-pair equation determines output vector of rank terms.\n    rank=len(masks); piv={}\n    for i in range(n):\n        for j in range(i,n):\n            row=sum(((m>>i&1)&(m>>j&1))<<r for r,m in enumerate(masks)); rhs=f.mul(1<<i,1<<j)\n            while row:\n                p=(row&-row).bit_length()-1\n                if p in piv:row^=piv[p][0];rhs^=piv[p][1]\n                else:piv[p]=(row,rhs);break\n            if row==0: assert rhs==0,(n,i,j,rhs)\n    out=[0]*rank\n    for p,(row,rhs) in sorted(piv.items(),reverse=True):\n        for j in bits(row^(1<<p)):rhs^=out[j]\n        out[p]=rhs\n    assert all(out)\n    return tuple(zip(masks,out))\n\n\ndef parity_into(b,x,mask,t):\n    for i in bits(mask):b.cx(x[i],t)\n\n\ndef mul_xor(b,x,y,z):\n    """Accumulate product, 1 scratch when inputs disjoint; 3 if overlapping.\n\n    Linear forms are computed temporarily in a chosen input pivot, restored\n    after each rank-one term. The output must be disjoint from both inputs.\n    """\n    n=len(x);assert len(y)==len(z)==n\n    overlap=bool(set(map(str,x)) & set(map(str,y)))\n    if overlap:\n        a,c,p=b.alloc_qubits(3)\n        for mask,v in tensor(n):\n            parity_into(b,x,mask,a);parity_into(b,y,mask,c)\n            b.init_and(a,c,p)\n            for i in bits(v):b.cx(p,z[i])\n            b.del_and(a,c,p)\n            parity_into(b,y,mask,c);parity_into(b,x,mask,a)\n        b.free([a,c,p])\n    else:\n        p=b.alloc_qubits(1)[0]\n        for mask,v in tensor(n):\n            ix=list(bits(mask));pivot=ix[0];a=x[pivot];c=y[pivot]\n            for i in ix[1:]:b.cx(x[i],a);b.cx(y[i],c)\n            b.init_and(a,c,p)\n            for i in bits(v):b.cx(p,z[i])\n            b.del_and(a,c,p)\n            for i in reversed(ix[1:]):b.cx(y[i],c);b.cx(x[i],a)\n        b.free(p)\n\n\ndef phase_product(b,x,y,mask):\n    """Exact phase dot(mask,x*y), zero Toffolis, no quantum workspace."""\n    f=field(len(x))\n    for i in range(len(x)):\n        for j in range(len(y)):\n            v=f.mul(1<<i,1<<j)\n            for k in bits(v):b.ccz(x[i],y[j],mask[k])\n\n\ndef erase_product(b,x,y,z):\n    mask=b.alloc_bits(len(z));b.hmr(z,mask);phase_product(b,x,y,mask);b.free(mask)\n\n\n@lru_cache(None)\ndef tower(n):\n    """Coordinates a+b*y, y²+y=K; returns to/from tower and K."""\n    k=n//2;f=field(n);sub=field(k)\n    gen=f.pow(f.primitive_element,(2**n-1)//(2**k-1))\n    def ev(v,p):\n        out=0\n        for i in reversed(range(k+1)):out=f.mul(out,p)^((v>>i)&1)\n        return out\n    root=next(f.pow(gen,i) for i in range(1,2**k) if ev(sub.modulus,f.pow(gen,i))==0)\n    embed=[f.pow(root,i) for i in range(k)]\n    # Any trace-relative-one element has minimal polynomial y²+y+K.\n    y=next(v for v in range(2**n) if f.pow(v,2**k)^v==1)\n    cols=embed+[f.mul(v,y) for v in embed];inv=inverse_columns(cols)\n    kval=apply(inv,f.square(y)^y);assert kval<2**k\n    return tuple(inv),tuple(cols),kval\n\n\ndef square_columns(n):return [field(n).square(1<<i) for i in range(n)]\n\n\ndef _norm_xor(b,x,z,kval):\n    """x already in tower coordinates. Norm = a²+ab+K b²."""\n    k=len(z);f=field(k);a=x[:k];c=x[k:]\n    linear_xor(b,a,z,square_columns(k));linear_xor(b,c,z,[f.mul(kval,v) for v in square_columns(k)])\n    mul_xor(b,a,c,z)\n\n\ndef _phase_norm(b,x,mask,kval):\n    k=len(mask);f=field(k);a=x[:k];c=x[k:]\n    phase_linear(b,a,mask,square_columns(k));phase_linear(b,c,mask,[f.mul(kval,v) for v in square_columns(k)])\n    phase_product(b,a,c,mask)\n\n\ndef _erase_norm(b,x,z,kval):\n    mask=b.alloc_bits(len(z));b.hmr(z,mask);_phase_norm(b,x,mask,kval);b.free(mask)\n\n\n@lru_cache(None)\ndef inverse3_anf():\n    f=field(3);a=[0 if x==0 else f.invert(x) for x in range(8)]\n    for i in range(3):\n        for s in range(8):\n            if s>>i&1:a[s]^=a[s^(1<<i)]\n    assert a[7]==0\n    return tuple(a)\n\n\ndef inverse(b,x,target=None):\n    """Clean out-of-place inverse: I3=3,I6=21,I12=66,I24=192 Toffolis.\n\n    Target must be initially zero when passed. Allocating target externally can\n    increase peak width: omit target for just-in-time recursive allocation.\n    """\n    n=len(x)\n    if n==3:\n        if target is None:target=b.alloc_qubits(3)\n        anc=b.alloc_qubits(1)[0]\n        for mask,v in enumerate(inverse3_anf()):\n            ix=list(bits(mask))\n            if len(ix)==1:\n                for j in bits(v):b.cx(x[ix[0]],target[j])\n            elif len(ix)==2:\n                b.init_and(x[ix[0]],x[ix[1]],anc)\n                for j in bits(v):b.cx(anc,target[j])\n                b.del_and(x[ix[0]],x[ix[1]],anc)\n        b.free(anc);return target\n    k=n//2;to,frm,K=tower(n)\n    linear_inplace(b,x,to)\n    norm=b.alloc_qubits(k);_norm_xor(b,x,norm,K)\n    invnorm=inverse(b,norm)\n    _erase_norm(b,x,norm,K);b.free(norm)\n    if target is None:target=b.alloc_qubits(n)\n    # x conjugate is (a+b,b), modify temporarily in-place.\n    a=x[:k];c=x[k:];b.cx(c,a)\n    mul_xor(b,a,invnorm,target[:k]);mul_xor(b,c,invnorm,target[k:]);b.cx(c,a)\n    # norm(target) = invnorm, including zero.\n    _erase_norm(b,target,invnorm,K);b.free(invnorm)\n    linear_inplace(b,x,frm);linear_inplace(b,target,frm)\n    return target\n\n\ndef inverse_inplace_forward(b,x):\n    """Forward 192T (n24), classical n-bit deferred input phase transcript."""\n    y=inverse(b,x);mask=b.alloc_bits(len(x));b.hmr(x,mask);b.swap(x,y);b.free(y)\n    return mask\n\n\ndef erase_inverse(b,x,y):\n    """Erase y=inv(x) with a closed phase-only inverse recomputation.\n\n    Generic fallback costs Ierase3=0, Ierase6=9, Ierase12=45,\n    Ierase24=153. It is NOT the claimed 30T inverse12 erasure.\n    """\n    n=len(x)\n    if n==3:\n        mask=b.alloc_bits(3);b.hmr(y,mask)\n        for im,v in enumerate(inverse3_anf()):\n            ix=list(bits(im))\n            for j in bits(v):\n                if len(ix)==1:b.cz(x[ix[0]],mask[j])\n                elif len(ix)==2:b.ccz(x[ix[0]],x[ix[1]],mask[j])\n        b.free(mask);return\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to);linear_inplace(b,y,to)\n    norm=b.alloc_qubits(k);_norm_xor(b,x,norm,K);invnorm=inverse(b,norm)\n    mask=b.alloc_bits(n);b.hmr(y,mask)\n    b.cx(x[k:],x[:k]);phase_product(b,x[:k],invnorm,mask[:k]);phase_product(b,x[k:],invnorm,mask[k:]);b.cx(x[k:],x[:k]);b.free(mask)\n    erase_inverse(b,norm,invnorm);b.free(invnorm)\n    _erase_norm(b,x,norm,K);b.free(norm);linear_inplace(b,x,frm)\n\n\ndef inverse_inplace_reverse_fallback(b,x,transcript):\n    """Close forward transcript; exact345T reverse at degree24, width61.\n\n    Kept as a correctness reference, NOT competitive with the target222T.\n    """\n    y=inverse(b,x)\n    b.cz(y,transcript);b.free(transcript)\n    erase_inverse(b,y,x)\n    b.swap(x,y);b.free(y)\n\n\ndef _inverse_keep_norm(b,x):\n    """Initialized inverse with its half-size input norm kept quantum-live.\n\n    Cost unchanged from inverse(). Caller must eventually erase kept norm.\n    """\n    n=len(x);k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    norm=b.alloc_qubits(k);_norm_xor(b,x,norm,K)\n    invnorm=inverse(b,norm)\n    target=b.alloc_qubits(n)\n    b.cx(x[k:],x[:k]);mul_xor(b,x[:k],invnorm,target[:k]);mul_xor(b,x[k:],invnorm,target[k:]);b.cx(x[k:],x[:k])\n    _erase_norm(b,target,invnorm,K);b.free(invnorm)\n    linear_inplace(b,x,frm);linear_inplace(b,target,frm)\n    return target,norm\n\n\ndef _erase_inverse_with_norm(b,x,y,norm_y):\n    """Erase y=inv(x) when norm(y) is retained. Zero Toffolis.\n\n    y=conjugate(x)*norm(y), so the phase fix is bilinear.\n    Both x and y arrive in polynomial basis; norm_y in half-field basis.\n    """\n    n=len(x);k=n//2;to,frm,K=tower(n)\n    linear_inplace(b,x,to);linear_inplace(b,y,to)\n    b.cx(x[k:],x[:k]);erase_product(b,x[:k],norm_y,y[:k]);erase_product(b,x[k:],norm_y,y[k:]);b.cx(x[k:],x[:k])\n    linear_inplace(b,x,frm)\n\n\ndef inverse_inplace_reverse(b,x,transcript):\n    """Close inverse_inplace_forward: degree24 reverse216T, peak67 qubits.\n\n    Retain norm(norm(input))6 during inverse12. Stream the two output halves\n    so retained N12+t12+D6 add at most43 ancillas to input24. The final D\n    cleanup costs norm(t)15 + inverse6 phase9 =24 additional Toffolis.\n    """\n    n=len(x);assert n in (12,24)\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    N=b.alloc_qubits(k);_norm_xor(b,x,N,K)\n    t,D=_inverse_keep_norm(b,N)\n    outb=b.alloc_qubits(k);mul_xor(b,x[k:],t,outb)\n    erase_product(b,outb,N,x[k:])\n    mul_xor(b,x[:k],t,x[k:]);b.cx(outb,x[k:])\n    b.cx(outb,x[k:]);erase_product(b,x[k:],N,x[:k]);b.cx(outb,x[k:])\n    b.swap(x[:k],x[k:]);b.swap(x[k:],outb);b.free(outb)\n    # N=inv(t), and norm(N)=D was retained inside inverse(N).\n    _erase_inverse_with_norm(b,t,N,D);b.free(N)\n    tok,frmk,Kk=tower(k);linear_inplace(b,t,tok)\n    u=b.alloc_qubits(k//2);_norm_xor(b,t,u,Kk)\n    erase_inverse(b,u,D);b.free(D)\n    _erase_norm(b,t,u,Kk);b.free(u);linear_inplace(b,t,frmk)\n    _erase_norm(b,x,t,K);b.free(t)\n    linear_inplace(b,x,frm)\n    b.cz(x,transcript);b.free(transcript)\n\n\n@lru_cache(None)\ndef extension_columns(K=None):\n    """Map base12 pair a+b*y to standard GF24 polynomial basis.\n\n    Uses base GF12 modulus0x1053. K is any GF12 trace-one element, y²+y=K.\n    Omitting K picks the smallest trace-one word, as in BRIEF.md.\n    Return (to_polynomial_columns, from_polynomial_columns, K).\n    """\n    f=field(12)\n    def trace(v):\n        out=0\n        for i in range(12):out^=f.frobenius(v,i)\n        return out\n    if K is None:K=next(v for v in range(4096) if trace(v)==1)\n    if not 0<=K<4096 or trace(K)!=1:raise ValueError(\'K must have absolute trace1\')\n    _,frm,K0=tower(24)\n    c=next(v for v in range(4096) if f.square(v)^v==K^K0)\n    F=field(24);y=frm[12]^apply(frm[:12],c)\n    cols=tuple(frm[:12])+tuple(F.mul(v,y) for v in frm[:12])\n    return cols,tuple(inverse_columns(cols)),K\n\n\ndef inverse24_inplace_forward(b,x,K=None):\n    """192T forward inversion in explicit a+b*y coordinates; returns transcript."""\n    cols,invcols,_=extension_columns(K);linear_inplace(b,x,cols)\n    transcript=inverse_inplace_forward(b,x);linear_inplace(b,x,invcols)\n    return transcript\n\n\ndef inverse24_inplace_reverse(b,x,transcript,K=None):\n    """216T closed reverse,43 workspace, in explicit a+b*y coordinates."""\n    cols,invcols,_=extension_columns(K);linear_inplace(b,x,cols)\n    inverse_inplace_reverse(b,x,transcript);linear_inplace(b,x,invcols)\n\n\ndef _solve_output_vectors(rows,rank):\n    piv={}\n    for row,rhs in rows:\n        while row:\n            p=(row&-row).bit_length()-1\n            if p in piv:row^=piv[p][0];rhs^=piv[p][1]\n            else:piv[p]=(row,rhs);break\n        if row==0:assert rhs==0,\'tensor does not span function\'\n    out=[0]*rank\n    for p,(row,rhs) in sorted(piv.items(),reverse=True):\n        for j in bits(row^(1<<p)):rhs^=out[j]\n        out[p]=rhs\n    return out\n\n\n@lru_cache(None)\ndef cube4_tensor():\n    """Four ANDs plus linear gates for GF16 cube."""\n    f=field(4);pairs=[(i,j) for i in range(4) for j in range(i+1,4)]\n    quads=[sum(((f.pow((1<<i)^(1<<j),3)^f.pow(1<<i,3)^f.pow(1<<j,3))>>k&1)<<p for p,(i,j) in enumerate(pairs)) for k in range(4)]\n    span={0}\n    for q in quads:span|={v^q for v in list(span)}\n    selected=[];piv={}\n    for a in range(1,16):\n        for c in range(a+1,16):\n            q=sum((((a>>i&1)&(c>>j&1))^((a>>j&1)&(c>>i&1)))<<p for p,(i,j) in enumerate(pairs))\n            if q not in span:continue\n            rem=q\n            for p,v in sorted(piv.items()):\n                if rem>>p&1:rem^=v\n            if rem:\n                p=(rem&-rem).bit_length()-1;piv[p]=rem;selected.append((a,c))\n    assert len(selected)==4\n    rows=[]\n    for i,j in pairs:\n        row=sum((((a>>i&1)&(c>>j&1))^((a>>j&1)&(c>>i&1)))<<p for p,(a,c) in enumerate(selected))\n        rhs=f.pow((1<<i)^(1<<j),3)^f.pow(1<<i,3)^f.pow(1<<j,3);rows.append((row,rhs))\n    out=_solve_output_vectors(rows,len(selected))\n    return tuple((a,c,v) for (a,c),v in zip(selected,out))\n\n\n@lru_cache(None)\ndef cube12_tensor():\n    """Seven GF16 cubic evaluations, each four ANDs: rank28 quadratic map."""\n    f=field(12);sub=field(4);gen=f.pow(f.primitive_element,273)\n    def ev(v,p):\n        out=0\n        for i in reversed(range(5)):out=f.mul(out,p)^((v>>i)&1)\n        return out\n    root=next(f.pow(gen,i) for i in range(1,16) if ev(sub.modulus,f.pow(gen,i))==0)\n    embed=[f.pow(root,i) for i in range(4)]\n    coords=inverse_columns([f.mul(v,f.pow(f.primitive_element,j)) for j in range(3) for v in embed])\n    terms=[]\n    for point in range(7):\n        vals=[]\n        for co in coords:\n            v=0\n            for j in reversed(range(3)):v=sub.mul(v,point)^((co>>(4*j))&15)\n            vals.append(v)\n        for a,c,_ in cube4_tensor():\n            A=sum(((v&a).bit_count()%2)<<i for i,v in enumerate(vals));C=sum(((v&c).bit_count()%2)<<i for i,v in enumerate(vals));terms.append((A,C))\n    rows=[]\n    for i in range(12):\n        for j in range(i+1,12):\n            row=sum((((a>>i&1)&(c>>j&1))^((a>>j&1)&(c>>i&1)))<<p for p,(a,c) in enumerate(terms))\n            rhs=f.pow((1<<i)^(1<<j),3)^f.pow(1<<i,3)^f.pow(1<<j,3);rows.append((row,rhs))\n    out=_solve_output_vectors(rows,len(terms));assert all(out)\n    linear=[]\n    for i in range(12):\n        v=f.pow(1<<i,3)\n        for (a,c),o in zip(terms,out):\n            if a>>i&1 and c>>i&1:v^=o\n        linear.append(v)\n    return tuple((a,c,v) for (a,c),v in zip(terms,out)),tuple(linear)\n\n\ndef cube_xor(b,x,z):\n    """GF12 cube accumulator z ^=x³,28Toffolis and3workspace."""\n    assert len(x)==len(z)==12\n    terms,linear=cube12_tensor();linear_xor(b,x,z,linear)\n    a,c,p=b.alloc_qubits(3)\n    for A,C,v in terms:\n        parity_into(b,x,A,a);parity_into(b,x,C,c);b.init_and(a,c,p)\n        for i in bits(v):b.cx(p,z[i])\n        b.del_and(a,c,p);parity_into(b,x,C,c);parity_into(b,x,A,a)\n    b.free([a,c,p])\n\n\n\n@lru_cache(None)\ndef tensor24_extension(K=None):\n    """105rank multiplicationtensor in explicit a+b*y, y²+y=K basis."""\n    cols,invcols,_=extension_columns(K)\n    return tuple((sum(((m&v).bit_count()%2)<<i for i,v in enumerate(cols)),apply(invcols,out)) for m,out in tensor(24))\n\n\ndef mul24_xor105(b,x,y,z,K=None):\n    """Arbitrarytarget GF24 product in explicit quadraticbasis:105T,1scratch.\n\n    No default126T multiplicationcaller ischanged; this is anoptional newAPI.\n    Allthree24-qubit registers mustbedisjoint.\n    """\n    assert len(x)==len(y)==len(z)==24\n    assert not (set(map(str,x))&set(map(str,y)) or set(map(str,x))&set(map(str,z)) or set(map(str,y))&set(map(str,z)))\n    p=b.alloc_qubits(1)[0]\n    for mask,v in tensor24_extension(K):\n        ix=list(bits(mask));pivot=ix[0];a=x[pivot];c=y[pivot]\n        for i in ix[1:]:b.cx(x[i],a);b.cx(y[i],c)\n        b.init_and(a,c,p)\n        for i in bits(v):b.cx(p,z[i])\n        b.del_and(a,c,p)\n        for i in reversed(ix[1:]):b.cx(y[i],c);b.cx(x[i],a)\n    b.free(p)\n\n\ndef mul_xor0(b,x,y,z):\n    """Standardpolynomialbasis arbitrarytarget product withZEROworkspace.\n\n    Disjointx,y,z required. Conjugateeachoutputrankvectorintoonepivot,\n    applyCCXdirectlythere,thenundooutputconjugation. Same42/105Tcounts.\n    """\n    n=len(x);assert len(y)==len(z)==n\n    assert not(set(map(str,x))&set(map(str,y)) or set(map(str,x))&set(map(str,z)) or set(map(str,y))&set(map(str,z)))\n    for mask,v in tensor(n):\n        ix=list(bits(mask));iz=list(bits(v));p=ix[0];t=iz[0]\n        for i in ix[1:]:b.cx(x[i],x[p]);b.cx(y[i],y[p])\n        for j in iz[1:]:b.cx(z[t],z[j])\n        b.ccx(x[p],y[p],z[t])\n        for j in reversed(iz[1:]):b.cx(z[t],z[j])\n        for i in reversed(ix[1:]):b.cx(y[i],y[p]);b.cx(x[i],x[p])\n\n\ndef mul24_xor105_zero(b,x,y,z,K=None):\n    """Explicita+b*y-basis105T multiplication withZEROprivatequbits."""\n    assert len(x)==len(y)==len(z)==24\n    assert not(set(map(str,x))&set(map(str,y)) or set(map(str,x))&set(map(str,z)) or set(map(str,y))&set(map(str,z)))\n    for mask,v in tensor24_extension(K):\n        ix=list(bits(mask));iz=list(bits(v));p=ix[0];t=iz[0]\n        for i in ix[1:]:b.cx(x[i],x[p]);b.cx(y[i],y[p])\n        for j in iz[1:]:b.cx(z[t],z[j])\n        b.ccx(x[p],y[p],z[t])\n        for j in reversed(iz[1:]):b.cx(z[t],z[j])\n        for i in reversed(ix[1:]):b.cx(y[i],y[p]);b.cx(x[i],x[p])\n', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/lowwidth.py': '"""Streamed towerinversion and25-scratch preserved-factor division candidates."""\nimport field as ar\nfrom field import *\nfrom division import mul24_xor,phase_product24,multiply_forward\n\n\ndef classical_transpose(b,mask,columns):\n    out=b.alloc_bits(len(columns));b.bit_store(out,False)\n    for i,c in enumerate(columns):\n        for j in bits(c):b.cx(mask[j],out[i])\n    return out\n\n\ndef inverse_stream_forward(b,x):\n    """Initialized in-place inverse with deferredoldinputphase, W(n)=n+1.\n\n    T3=3,T6=21,T12=66,T24=192. Transcriptisintheoriginalpolynomialbasis.\n    Allinternalnorm transcriptsarefixedquadratically beforeparentinputchanges.\n    """\n    n=len(x)\n    if n==3:\n        y=inverse(b,x);mask=b.alloc_bits(n);b.hmr(x,mask);b.swap(x,y);b.free(y);return mask\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    N=b.alloc_qubits(k);ar._norm_xor(b,x,N,K)\n    nm=inverse_stream_forward(b,N);ar._phase_norm(b,x,nm,K);b.free(nm)\n    outb=b.alloc_qubits(k);mul_xor(b,x[k:],N,outb)\n    raw=b.alloc_bits(n);b.hmr(x[k:],raw[k:])\n    mul_xor(b,x[:k],N,x[k:]);b.cx(outb,x[k:]);b.hmr(x[:k],raw[:k])\n    b.swap(x[:k],x[k:]);b.swap(x[k:],outb);b.free(outb)\n    ar._erase_norm(b,x,N,K);b.free(N)\n    linear_inplace(b,x,frm)\n    mask=classical_transpose(b,raw,to);b.free(raw);return mask\n\n\ndef phase_product_quantum(b,x,y,mask):\n    """Phase dot(mask,x*y) with QUANTUMmask: rank(n)CCZ,0scratch."""\n    n=len(x);assert len(y)==len(mask)==n\n    for m,v in tensor(n):\n        ix=list(bits(m));iz=list(bits(v));i=ix[0];j=iz[0]\n        for k in ix[1:]:b.cx(x[k],x[i]);b.cx(y[k],y[i])\n        for k in iz[1:]:b.cx(mask[k],mask[j])\n        b.ccz(x[i],y[i],mask[j])\n        for k in reversed(iz[1:]):b.cx(mask[k],mask[j])\n        for k in reversed(ix[1:]):b.cx(y[k],y[i]);b.cx(x[k],x[i])\n\n\ndef phase_inverse_generic6(b,x,mask):\n    """9T inversephase on6inputbits,7scratch; maskclassical6."""\n    to,frm,K=tower(6);linear_inplace(b,x,to)\n    mt=classical_transpose(b,mask,frm)\n    N=b.alloc_qubits(3);ar._norm_xor(b,x,N,K);u=inverse(b,N)\n    b.cx(x[3:],x[:3]);phase_product(b,x[:3],u,mt[:3]);phase_product(b,x[3:],u,mt[3:]);b.cx(x[3:],x[:3])\n    erase_inverse(b,N,u);b.free(u);ar._erase_norm(b,x,N,K);b.free(N)\n    linear_inplace(b,x,frm);b.free(mt)\n\n\ndef phase_inverse_low12(b,x,mask):\n    """60T phaseofinverse12 using13scratch beyondinput12,maskclassical."""\n    to,frm,K=tower(12);linear_inplace(b,x,to)\n    mt=classical_transpose(b,mask,frm)\n    N=b.alloc_qubits(6);ar._norm_xor(b,x,N,K)\n    nm=inverse_stream_forward(b,N);ar._phase_norm(b,x,nm,K);b.free(nm)\n    b.cx(x[6:],x[:6]);phase_product(b,x[:6],N,mt[:6]);phase_product(b,x[6:],N,mt[6:]);b.cx(x[6:],x[:6])\n    um=b.alloc_bits(6);b.hmr(N,um);b.free(N)\n    N=b.alloc_qubits(6);ar._norm_xor(b,x,N,K);phase_inverse_generic6(b,N,um);b.free(um)\n    ar._erase_norm(b,x,N,K);b.free(N);linear_inplace(b,x,frm);b.free(mt)\n\n\ndef fix_scaled_half_masks(b,head,factor,masks,K):\n    """Phase sum_j dot(maskj,headj*norm(factor)):42CCZ,12scratch.\n\n    Thetwoheadhalves shareonequantum12-bittransposedmask, hence42not84CCZ.\n    """\n    L=b.alloc_qubits(12)\n    def coefficient_mask():\n        for i in range(12):\n            for half in range(2):\n                for j in range(12):\n                    for r in bits(field(12).mul(1<<i,1<<j)):\n                        b.ccx(masks[half][r],head[12*half+j],L[i])\n    coefficient_mask()\n    phase_linear(b,factor[:12],L,square_columns(12))\n    phase_linear(b,factor[12:],L,[field(12).mul(K,v) for v in square_columns(12)])\n    phase_product_quantum(b,factor[:12],factor[12:],L)\n    coefficient_mask();b.free(L)\n\n\ndef divide_lowwidth(b,x,y,K=None):\n    """441T division x/=y,preservingy,25workspace; promise x0wheny0."""\n    if K is None:K=extension_columns()[2]\n    b.cx(y[12:],y[:12]);oldmask=multiply_forward(b,x,y,K);b.cx(y[12:],y[:12])\n    u=b.alloc_qubits(12);ar._norm_xor(b,y,u,K)\n    nm=inverse_stream_forward(b,u);ar._phase_norm(b,y,nm,K);b.free(nm)\n    masks=[]\n    for offset in(0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor(b,half,u,out)\n        m=b.alloc_bits(12);b.hmr(half,m);masks.append(m);b.swap(half,out);b.free(out)\n    fix_scaled_half_masks(b,x,y,masks,K)\n    for m in masks:b.free(m)\n    um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    N=b.alloc_qubits(12);ar._norm_xor(b,y,N,K);phase_inverse_low12(b,N,um);b.free(um)\n    ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,oldmask,K);b.free(oldmask)\n\n\ndef phase_inverse36(b,x,mask):\n    """Inverse12phase:36Toffolis,13scratch beyondinput12.\n\n    Norm6cost15; nestednorm3+inverse3cost9; twoGF3trilinearphasescost12CCZ.\n    Theirthirdquantumcontroliscomputedoneparityatatimefromxandclassicalmask,\n    so no6-qubittransposed-maskbank isneeded.\n    """\n    to12,frm12,K12=tower(12);linear_inplace(b,x,to12)\n    mt=classical_transpose(b,mask,frm12)\n    D=b.alloc_qubits(6);ar._norm_xor(b,x,D,K12)\n    to6,frm6,K6=tower(6);linear_inplace(b,D,to6)\n    E=b.alloc_qubits(3);ar._norm_xor(b,D,E,K6);w=inverse(b,E)\n    anc=b.alloc_qubits(1)[0];f6=field(6)\n    def coefficient(h):\n        for j in range(6):\n            for r in bits(f6.mul(h,1<<j)):\n                b.ccx(mt[r],x[j],anc);b.ccx(mt[r],x[j+6],anc);b.ccx(mt[r+6],x[j+6],anc)\n    for segment in(0,1):\n        if segment==0:b.cx(D[3:],D[:3])\n        controls=D[:3] if segment==0 else D[3:]\n        for m,v in tensor(3):\n            h=apply(frm6,v<<(3*segment));ix=list(bits(m));p=ix[0]\n            coefficient(h)\n            for j in ix[1:]:b.cx(controls[j],controls[p]);b.cx(w[j],w[p])\n            b.ccz(controls[p],w[p],anc)\n            for j in reversed(ix[1:]):b.cx(w[j],w[p]);b.cx(controls[j],controls[p])\n            coefficient(h)\n        if segment==0:b.cx(D[3:],D[:3])\n    b.free(anc);erase_inverse(b,E,w);b.free(w)\n    ar._erase_norm(b,D,E,K6);b.free(E);linear_inplace(b,D,frm6)\n    ar._erase_norm(b,x,D,K12);b.free(D);linear_inplace(b,x,frm12);b.free(mt)\n\n\ndef divide_lowwidth375(b,x,y,K=None):\n    """375T preserved-factor division at25scratch,promisedx0wheny0."""\n    if K is None:K=extension_columns()[2]\n    b.cx(y[12:],y[:12]);oldmask=multiply_forward(b,x,y,K);b.cx(y[12:],y[:12])\n    u=b.alloc_qubits(12);ar._norm_xor(b,y,u,K)\n    nm=inverse_stream_forward(b,u);ar._phase_norm(b,y,nm,K);b.free(nm)\n    masks=[]\n    for offset in(0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor(b,half,u,out)\n        m=b.alloc_bits(12);b.hmr(half,m);masks.append(m);b.swap(half,out);b.free(out)\n    um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    N=b.alloc_qubits(12);ar._norm_xor(b,y,N,K)\n    for offset,m in zip((0,12),masks):phase_product(b,x[offset:offset+12],N,m);b.free(m)\n    phase_inverse36(b,N,um);b.free(um)\n    ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,oldmask,K);b.free(oldmask)\n', 'blocks/dqi/pro-q2-897/source/exp/confluent_extract/kernel.py': '"""Private fixed Hermite map and injectable singleton-first jet adapter.\n\nNo stable decoder modules are edited. Pass the arithmetic teammate\'s jet\nforward/reverse kernels; both operate on two24-bit data registers and return\nor consume classical transcripts, with no retained quantum cache.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.decoder.common import combine,cx,scalepair,linear_words,K\nfrom exp.inverse12_escape.joint import inverse24_forward190,inverse24_reverse209\ntry:\n    from .model import constants,plan\nexcept ImportError:\n    from model import constants,plan\n\ndef initial(b,words,reverse=False):\n    assert len(words)%2==0\n    linear_words(b,words,constants(len(words)//2)[3],reverse)\n\ndef translate(b,words,head,target,c):\n    a,z=words[2*head:2*head+2]\n    b.cx(a,words[target]);cx(b,z,words[target],c);b.cx(z,words[target+1]);b.cx(z,words[target+2])\n\ndef extract(b,words,jet_forward,jet_reverse,reverse=False,transcript=None,\n            inverse_forward=inverse24_forward190,inverse_reverse=inverse24_reverse209):\n    """Return/consume (kind,token) history; output head pairs alias input bank.\n\n    jet_forward(b,u24,d24,K) -> token\n    jet_reverse(b,u24,d24,token,K) restores both data and closes all phases.\n    """\n    h=len(words)//2;history=[]if not reverse else list(transcript)\n    for op in (reversed(plan(h))if reverse else plan(h)):\n        if op[\'kind\']==\'coordinates\':cx(b,words[2*op[\'head\']+1],words[2*op[\'head\']],op[\'c\']);continue\n        if op[\'kind\']==\'local\':\n            i=2*op[\'head\'];j=i+2;x=combine(words[j:j+2])\n            if not reverse:\n                b.cx(words[i+1],words[j]);history.append((\'local\',inverse_forward(b,x,K)))\n                cx(b,words[j+1],words[j],op[\'c\'])\n            else:\n                cx(b,words[j+1],words[j],op[\'c\']);kind,t=history.pop();assert kind==\'local\'\n                inverse_reverse(b,x,t,K);b.cx(words[i+1],words[j])\n            continue\n        j=op[\'target\'];u=combine(words[j:j+2]);d=combine(words[j+2:j+4]);q=op[\'q\']\n        if not reverse:\n            translate(b,words,op[\'head\'],j,op[\'c\']);history.append((\'jet\',jet_forward(b,u,d,K)))\n            scalepair(b,d,q);b.cx(u,d);scalepair(b,u,q)\n        else:\n            from exp.decoder.common import inv\n            scalepair(b,u,inv(q));b.cx(u,d);scalepair(b,d,inv(q))\n            kind,t=history.pop();assert kind==\'jet\';jet_reverse(b,u,d,t,K)\n            translate(b,words,op[\'head\'],j,op[\'c\'])\n    if reverse:assert not history\n    return history\n', 'blocks/dqi/pro-q2-897/source/exp/confluent_extract/model.py': '"""Singleton-first confluent quadratic-node map and zero-safe jet EXTRACT.\n\nAll polynomial/rational references are classical. No decoder/verifier claim.\n"""\nfrom functools import lru_cache\nfrom pathlib import Path\nimport sys,json,random,time\nROOT=Path(__file__).resolve().parents[2]\nsys.path.insert(0,str(ROOT/\'playbook/exp/cf_check\'))\nfrom audit import Field\nimport audit as cf_audit\nf=Field(12,0x1053);assert f.K==2048\n\ndef eadd(a,b):return a^b\ndef esquare(a):return f.emul(a,a)\ndef ehead(a,b,c):return (a^f.mul(b,c))|(b<<12)\ndef coords(u,c):\n    b=u>>12;return (u&4095)^f.mul(b,c),b\ndef get(bank,j):return bank[j]|(bank[j+1]<<12)\ndef put(bank,j,x):bank[j]=x&4095;bank[j+1]=x>>12\n\ndef synthetic(poly,gamma):\n    assert len(poly)>=2\n    Q=[0]*(len(poly)-1);Q[-1]=poly[-1]\n    for i in reversed(range(len(Q)-1)):Q[i]=poly[i+1]^f.emul(gamma,Q[i+1])\n    return Q,poly[0]^f.emul(gamma,Q[0])\n\n@lru_cache(None)\ndef constants(h=35):\n    assert h%2 and h>=1;d=(h-1)//2\n    groups=[dict(c=0,kappa=f.K,gamma=4096,head=0,word=0,double=False)]\n    headcs=[0]\n    for j in range(1,d+1):\n        c=2*j;groups.append(dict(c=c,kappa=f.mul(c,c)^c^f.K,gamma=c|4096,head=2*j-1,word=4*j-2,double=True));headcs.extend([c,c])\n    kappas=[f.mul(c,c)^c^f.K for c in headcs];P=[1]\n    for kap in kappas:P=f.pmul(P,[kap,1,1])\n    matrix=[]\n    for group in groups:\n        Q,rem=synthetic(P,group[\'gamma\']);assert rem==0\n        matrix.append([q&4095 for q in Q]);matrix.append([q>>12 for q in Q])\n        if group[\'double\']:\n            QQ,rem=synthetic(Q,group[\'gamma\']);assert rem==0\n            QQ=QQ+[0]\n            matrix.append([q&4095 for q in QQ]);matrix.append([q>>12 for q in QQ])\n    assert len(matrix)==2*h and all(len(r)==2*h for r in matrix)\n    return tuple(headcs),tuple(kappas),tuple(P),tuple(tuple(r)for r in matrix),tuple(groups)\n\n@lru_cache(None)\ndef inverse_matrix(h=35):\n    M=constants(h)[3];n=len(M);a=[list(row)+[int(i==j)for j in range(n)]for i,row in enumerate(M)]\n    for j in range(n):\n        p=next(i for i in range(j,n)if a[i][j]);a[j],a[p]=a[p],a[j]\n        iv=f.inv(a[j][j]);a[j]=[f.mul(v,iv)for v in a[j]]\n        for i in range(n):\n            if i!=j and a[i][j]:\n                q=a[i][j];a[i]=[x^f.mul(q,y)for x,y in zip(a[i],a[j])]\n    assert all(a[i][j]==int(i==j)for i in range(n)for j in range(n))\n    return tuple(tuple(row[n:])for row in a)\n\ndef matrix_apply(M,s):\n    out=[]\n    for row in M:\n        v=0\n        for a,x in zip(row,s):v^=f.mul(a,x)\n        out.append(v)\n    return out\n\ndef initial(s,h=35,reverse=False):return matrix_apply(inverse_matrix(h)if reverse else constants(h)[3],s)\n\n@lru_cache(None)\ndef plan(h=35):\n    groups=constants(h)[4];ops=[]\n    for i,g in enumerate(groups):\n        ops.append(dict(kind=\'coordinates\',head=g[\'head\'],c=g[\'c\']))\n        for target in groups[i+1:]:ops.append(dict(kind=\'jet\',head=g[\'head\'],target=target[\'word\'],c=target[\'c\'],q=g[\'kappa\']^target[\'kappa\']))\n        if g[\'double\']:\n            ops.append(dict(kind=\'local\',head=g[\'head\'],c=g[\'c\']))\n            for target in groups[i+1:]:ops.append(dict(kind=\'jet\',head=g[\'head\']+1,target=target[\'word\'],c=target[\'c\'],q=g[\'kappa\']^target[\'kappa\']))\n    assert sum(op[\'kind\']==\'jet\'for op in ops)==((h-1)//2)**2\n    assert sum(op[\'kind\']==\'local\'for op in ops)==(h-1)//2\n    return tuple(ops)\n\ndef J(u,v):\n    z=int(u==0);ui=f.einv(u);return ui,f.emul(v,esquare(ui^z))\n\ndef extract(bank,h=35,reverse=False,events=None):\n    bank=list(bank);assert len(bank)==2*h\n    for op in (reversed(plan(h))if reverse else plan(h)):\n        if op[\'kind\']==\'coordinates\':bank[2*op[\'head\']]^=f.mul(op[\'c\'],bank[2*op[\'head\']+1]);continue\n        if op[\'kind\']==\'local\':\n            i=2*op[\'head\'];j=i+2\n            if not reverse:\n                u=get(bank,j)^bank[i+1]\n                if events is not None and not u:events.append(dict(kind=\'local_zero_inverse\',head=op[\'head\']+1))\n                a,b=coords(f.einv(u),op[\'c\']);bank[j:j+2]=[a,b]\n            else:\n                a,b=bank[j:j+2];put(bank,j,f.einv(ehead(a,b,op[\'c\']))^bank[i+1])\n            continue\n        i=2*op[\'head\'];j=op[\'target\'];a,b=bank[i:i+2];u=get(bank,j);v=get(bank,j+2);q=op[\'q\'];assert q\n        if not reverse:\n            u^=ehead(a,b,op[\'c\']);v^=b\n            if events is not None and not u:events.append(dict(kind=\'jet_zero_value\',head=op[\'head\'],target=j//2,derivative_nonzero=bool(v)))\n            u,v=J(u,v);v=u^f.escale(v,q);u=f.escale(u,q)\n        else:\n            u=f.escale(u,f.inv(q));v=f.escale(v^u,f.inv(q));u,v=J(u,v);u^=ehead(a,b,op[\'c\']);v^=b\n        put(bank,j,u);put(bank,j+2,v)\n    return bank\n\ndef pair_from_residues(roots,weights):\n    D=[1]\n    for root in roots:D=f.pmul(D,[root,1])\n    N=[0]\n    for root,w in zip(roots,weights):\n        Q,rem=f.pdiv(D,[root,1]);assert rem==[0];N=f.add(N,f.scale(Q,w))\n    return N,D\n\ndef instance(h,roots,weights):\n    P=constants(h)[2];syndrome=[0]*(2*h);errors=[]\n    for root,w in zip(roots,weights):\n        pb=f.ev(P,root);assert pb and not pb>>12\n        start=f.div(w,pb);errors.append(f.div(start,root));v=start\n        for j in range(2*h):syndrome[j]^=v;v=f.mul(v,root)\n    N,D=pair_from_residues(roots,weights)\n    return dict(h=h,roots=roots,residues=weights,errors=errors,syndrome=syndrome,N=N,D=D)\n\ndef direct_jets(roots,weights,h=35):\n    out=[]\n    for group in constants(h)[4]:\n        u=v=0\n        for root,w in zip(roots,weights):\n            t=f.einv(group[\'gamma\']^root);u^=f.escale(t,w);v^=f.escale(esquare(t),w)\n        out.extend([u&4095,u>>12])\n        if group[\'double\']:out.extend([v&4095,v>>12])\n    return out\n\ndef polynomial_oracle(N,D,h=35):\n    cs,ks,*_=constants(h);heads=[];term=None\n    for i,(c,kap)in enumerate(zip(cs,ks)):\n        gamma=c|4096;de=f.ev(D,gamma)\n        if not de:return dict(heads=heads,termination=term,pole=i)\n        u=f.emul(f.ev(N,gamma),f.einv(de));a,b=coords(u,c);heads.extend([a,b])\n        H=f.add(N,f.pmul([a,b],D));Q,rem=f.pdiv(H,[kap,1,1]);assert rem==[0]\n        if H==[0]:term=i;heads.extend([0]*(2*h-len(heads)));break\n        N,D=D,Q\n    return dict(heads=heads,termination=term,pole=None)\n\ndef trial(case):\n    h=case[\'h\'];s=case[\'syndrome\'];jet=initial(s,h);assert jet==direct_jets(case[\'roots\'],case[\'residues\'],h)\n    assert initial(jet,h,True)==s\n    events=[];heads=extract(jet,h,events=events);assert extract(heads,h,True)==jet\n    reference=polynomial_oracle(case[\'N\'],case[\'D\'],h)\n    match=heads==reference[\'heads\']if reference[\'pole\']is None else None\n    raw_ratio=None;raw_padding=None;raw_squarefree=None\n    if reference[\'pole\']is None:\n        assert match\n        pairs=[heads[i:i+2]for i in range(0,len(heads),2)]\n        A,B,tail=cf_audit.synth(f,constants(h)[1],pairs)\n        raw_ratio=f.pmul(A,case[\'D\'])==f.pmul(B,case[\'N\']);assert raw_ratio\n        expected=case[\'D\'][:]\n        for i in range(len(case[\'roots\'])+1,h,2):expected=f.pmul(expected,[constants(h)[1][i],1,1])\n        if not len(case[\'roots\'])%2:expected=[0]+expected\n        raw_padding=B==expected;assert raw_padding\n        raw_squarefree=f.gcd(B,f.derivative(B))==[1];assert raw_squarefree\n        assert len(B)==h+1 and B[-1]==1\n    else:assert heads[:len(reference[\'heads\'])]==reference[\'heads\']\n    return dict(h=h,weight=len(case[\'roots\']),oracle_pole=reference[\'pole\'],termination=reference[\'termination\'],\n        heads_match=match,jet_zero_events=sum(e[\'kind\']==\'jet_zero_value\'for e in events),\n        nonzero_derivative_zero_events=sum(e[\'kind\']==\'jet_zero_value\'and e[\'derivative_nonzero\']for e in events),\n        local_zero_inverses=sum(e[\'kind\']==\'local_zero_inverse\'for e in events),roundtrip=True,\n        raw_synthesis_ratio=raw_ratio,raw_synthesis_padding=raw_padding,raw_synthesis_squarefree=raw_squarefree)\n\ndef main(repeats=4):\n    start=time.time();rng=random.Random(202610270901);tests=[];arbitrary=[]\n    for h in[3,9,35]:\n        inverse_matrix(h)\n        for weight in range(h+1):\n            for repeat in range(repeats):\n                roots=rng.sample(range(1,4096),weight);weights=[rng.randrange(1,4096)for _ in roots]\n                tests.append(trial(instance(h,roots,weights)))\n        for j in range(300):\n            bank=[rng.randrange(4096)for _ in range(2*h)]\n            if j%3==0:\n                for group in constants(h)[4][1:]:bank[group[\'word\']]=bank[group[\'word\']+1]=0\n            if j%3==1:bank=[0 if rng.randrange(2)else v for v in bank]\n            assert extract(extract(bank,h),h,True)==bank\n        arbitrary.append(dict(h=h,cases=300,includes_zero_values_and_zero_jets=True))\n        print(\'h\',h,\'done\',flush=True)\n    peer=json.loads((ROOT/\'exp/confluent_synthesis/fixtures.json\').read_text());compared=0\n    for fixture in peer:\n        h=fixture[\'h\'];case=instance(h,fixture[\'roots\'],fixture[\'weights\']);heads=extract(initial(case[\'syndrome\'],h),h)\n        assert heads==[v for head in fixture[\'heads\']for v in head];compared+=1\n        A,B,_=cf_audit.synth(f,constants(h)[1],[heads[i:i+2]for i in range(0,len(heads),2)])\n        assert A==fixture[\'rawA\']and B==fixture[\'rawB\']\n    result=dict(tests=tests,arbitrary=arbitrary,peer_fixture_heads_compared=compared,\n        matrix_ranks={h:2*h for h in[3,9,35]},counts={h:dict(jets=sum(o[\'kind\']==\'jet\'for o in plan(h)),local_inverses=sum(o[\'kind\']==\'local\'for o in plan(h)))for h in[3,9,35]},\n        seconds=time.time()-start,meaning=\'Classical exactness and fixed-linear map; no full decoder/verifier\')\n    print(json.dumps(dict(cases=len(tests),heads_bad=sum(t[\'heads_match\']is False for t in tests),poles=sum(t[\'oracle_pole\']is not None for t in tests),\n        peer_fixture_heads_compared=compared,arbitrary=arbitrary,seconds=result[\'seconds\']),indent=2))\n    Path(__file__).with_name(\'results.json\').write_text(json.dumps(result,indent=2)+\'\\n\')\n\nif __name__==\'__main__\':main()\n', 'blocks/dqi/pro-q2-897/source/exp/confluent_synthesis/kernel.py': '"""Read-only constants adapter for frozen alternating-gauge synthesis."""\nfrom pathlib import Path\nimport sys\nfrom functools import lru_cache\nfrom types import FunctionType,SimpleNamespace\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.alternating_gauge import kernel as base\nfrom exp.decoder.common import mul,K,pmul\n@lru_cache(None)\ndef constants(h=35):\n assert h>=3 and h%2==1\n cs=[0]+[2*j for j in range(1,(h+1)//2)for _ in range(2)]\n ks=[mul(c,c)^c^K for c in cs];P=[1]\n for k in ks:P=pmul(P,[k,1,1])\n return cs,ks,P,()\nold=SimpleNamespace(**dict(vars(base.old),constants=constants))\nsynthesize=FunctionType(base.synthesize.__code__,dict(base.synthesize.__globals__,old=old),\'synthesize\',base.synthesize.__defaults__,base.synthesize.__closure__)\nunsynthesize=FunctionType(base.unsynthesize.__code__,dict(base.unsynthesize.__globals__,old=old),\'unsynthesize\',base.unsynthesize.__defaults__,base.unsynthesize.__closure__)\n', 'blocks/dqi/pro-q2-897/source/exp/convert/convert.py': '"""Reversible fixed-generic-chart half-degree CONVERT.\n\nPublic convert(builder,A[35] words,B[35] words,reverse=False).\nA,B hold GF12 polynomial coefficients; B35=1 implicit.\nSafe divisors make roundtrip exact even on degree gaps; forward decode is only\nproved for generic Euclid degrees. Runtime gap2 routing is not yet implemented.\n"""\nfrom contextlib import contextmanager\nimport sys\nfrom pathlib import Path\nsys.path.insert(0,str(Path(__file__).resolve().parent))\nfrom kernels import ar,cx,convolution_xor,hankel\nfrom lowwidth import phase_inverse36\n\ndef zero_flag(b,a):\n    for q in a:b.x(q)\n    nodes=[];left=a[0]\n    for q in a[1:]:\n        out=b.alloc_qubits(1)[0];b.init_and(left,q,out);nodes.append((left,q,out));left=out\n    for x,y,z in reversed(nodes[:-1]):b.del_and(x,y,z);b.free(z)\n    for q in a:b.x(q)\n    return nodes[-1][2]\n\ndef erase_zero_flag(b,a,flag):\n    for q in a:b.x(q)\n    nodes=[];left=a[0]\n    for q in a[1:-1]:\n        out=b.alloc_qubits(1)[0];b.init_and(left,q,out);nodes.append((left,q,out));left=out\n    b.del_and(left,a[-1],flag);b.free(flag)\n    for x,y,z in reversed(nodes):b.del_and(x,y,z);b.free(z)\n    for q in a:b.x(q)\n\n@contextmanager\ndef inverse_context(b,a):\n    flag=zero_flag(b,a);b.cx(flag,a[0])\n    ai=ar.inverse(b,a)\n    yield ai\n    mask=b.alloc_bits(12);b.hmr(ai,mask);b.free(ai)\n    phase_inverse36(b,a,mask);b.free(mask)\n    b.cx(flag,a[0]);erase_zero_flag(b,a,flag)\n\ndef scale(b,x,a,ai,divide=True):\n    out=b.alloc_qubits(12);ar.mul_xor(b,x,ai if divide else a,out)\n    ar.erase_product(b,a if divide else ai,out,x)\n    b.swap(x,out);b.free(out)\n\ndef runops(b,ops,a,ai,undo=False):\n    for op in (reversed(ops) if undo else ops):\n        if op[0]==\'s\':scale(b,op[1],a,ai,op[2]^undo)\n        elif op[0]==\'m\':ar.mul_xor(b,op[1],op[2],op[3])\n        elif op[0]==\'c\':convolution_xor(b,op[1],op[2],op[3],undo)\n        else:raise ValueError(op[0])\n\ndef pre(b,f,g,payloads,undo=False):\n    m=len(f)-1;n=m-1;a=g[-1];q0=f[-2]\n    with inverse_context(b,a) as ai:\n        q1=ai if f[-1] is None else f[-1]\n        ops=[(\'s\',v[-1],True) for v in payloads]\n        ops += [(\'m\',g[i],v[-1],v[i]) for v in payloads for i in range(n)]\n        if f[-1] is not None:ops += [(\'s\',q1,True)]\n        ops += [(\'m\',q1,g[-2],q0),(\'s\',q0,True),(\'c\',g[:-1],[q0,q1],f[:-2])]\n        runops(b,ops,a,ai,undo)\n\ndef post(b,f,g,payloads,undo=False):\n    n=len(g)-1;a=g[-1];q0=f[-2]\n    def dots():\n        for v in payloads:\n            for i in range(n):ar.mul_xor(b,g[i],v[i],v[-1])\n    if f[-1] is None:\n        with inverse_context(b,a) as ai:\n            q1=ai\n            ops=[(\'s\',v[-1],False) for v in payloads]\n            ops += [(\'m\',g[i],v[i],v[-1]) for v in payloads for i in range(n)]\n            ops += [(\'s\',v[-1],True) for v in payloads]\n            ops += [(\'c\',g[:-1],[q0,q1],f[:-2]),(\'s\',q0,False),(\'m\',q1,g[-2],q0)]\n            runops(b,ops,a,ai,undo)\n    else:\n        q1=f[-1]\n        def qscale(reverse):\n            with inverse_context(b,q1) as qi:\n                for v in payloads:scale(b,v[-1],q1,qi,not reverse)\n        def last(reverse):\n            with inverse_context(b,a) as ai:\n                ops=[(\'s\',v[-1],True) for v in payloads]\n                ops += [(\'c\',g[:-1],[q0,q1],f[:-2]),(\'s\',q0,False),(\'m\',q1,g[-2],q0),(\'s\',q1,False)]\n                runops(b,ops,a,ai,reverse)\n        if not undo:qscale(False);dots();last(False)\n        else:last(True);dots();qscale(True)\n\ndef raw_solve(b,f,g,payloads,reverse=False):\n    m=len(f)-1\n    assert len(g)==m and all(len(v)==m for v in payloads)\n    if m==1:\n        divisors=[p for p in (f[-1],g[-1]) if p is not None]\n        for a in (reversed(divisors) if reverse else divisors):\n            with inverse_context(b,a) as ai:\n                for v in payloads:scale(b,v[0],a,ai,not reverse)\n        return\n    low=[v[:-1] for v in payloads]\n    if not reverse:\n        pre(b,f,g,payloads);raw_solve(b,g,f[:-2],low);post(b,f,g,payloads)\n    else:\n        post(b,f,g,payloads,True);raw_solve(b,g,f[:-2],low,True);pre(b,f,g,payloads,True)\n\ndef bez_solve(b,O,E,payloads,reverse=False):\n    def reduce():\n        for i in range(len(O)):ar.mul_xor(b,E[-1],O[i],E[i])\n    reduce();raw_solve(b,O+[None],E[:-1],payloads,reverse);reduce()\n\ndef frob(b,words,power):\n    f=ar.field(12);cols=[f.pow(1<<i,power) for i in range(12)]\n    for x in words:ar.linear_inplace(b,x,cols)\n\ndef convert(b,A,B,reverse=False):\n    assert len(A)==len(B)==35\n    E=B[::2];O=B[1::2];u=A[1::2];w=A[:34:2];top=A[-1]\n    if reverse:hankel(b,B,A,True)\n    frob(b,B,2048);frob(b,A,2048)\n    if reverse:\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n        for i in range(17):ar.mul_xor(b,O[i],u[i],top)\n        bez_solve(b,O,E,[u,w],True)\n        for i in range(17):ar.mul_xor(b,top,O[i],w[i])\n    else:\n        for i in range(17):ar.mul_xor(b,top,O[i],w[i])\n        bez_solve(b,O,E,[u,w])\n        for i in range(17):ar.mul_xor(b,O[i],u[i],top)\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n    frob(b,A,2);frob(b,B,2)\n    if not reverse:hankel(b,B,A)\n\ndef raw_solve_blocked(b,f,g,payloads,block_size=2,reverse=False):\n    """Closed checkpointed scalar schedule grouped into quotient blocks.\n\n    This is the no-speedup control construction: at most2*block_size quotient\n    metadata words remain in already consumed f/g coefficients; no matrix or\n    evaluation bank is materialized. Product fusion is deliberately absent.\n    """\n    frames=[]\n    for _ in range(block_size):\n        if len(f)==2:break\n        frames.append((f,g,payloads))\n        if not reverse:pre(b,f,g,payloads)\n        else:post(b,f,g,payloads,True)\n        m=len(f)-1;f,g=g,f[:-2];payloads=[v[:-1]for v in payloads]\n    if len(f)==2:raw_solve(b,f,g,payloads,reverse)\n    else:raw_solve_blocked(b,f,g,payloads,block_size,reverse)\n    for f,g,payloads in reversed(frames):\n        if not reverse:post(b,f,g,payloads)\n        else:pre(b,f,g,payloads,True)\n', 'blocks/dqi/pro-q2-897/source/exp/convert/hankel_check.py': '"""Independent GF(2^12) rank-one, in-place Hankel audit; no verifier."""\nimport random, json\nfrom functools import lru_cache\nMOD=0x1053\n\ndef slowmul(a,b):\n    c=0\n    for _ in range(12):\n        if b&1:c^=a\n        b>>=1;a<<=1\n        if a&4096:a^=MOD\n    return c\nEXP=[1]*8190;LOG=[-1]*4096\nfor i in range(1,4095):EXP[i]=slowmul(EXP[i-1],2)\nassert len(set(EXP[:4095]))==4095\nfor i in range(4095):LOG[EXP[i]]=i;EXP[i+4095]=EXP[i]\ndef mul(a,b):return EXP[LOG[a]+LOG[b]] if a and b else 0\ndef inv(a):assert a;return EXP[4095-LOG[a]]\ndef power(a,n):return EXP[(LOG[a]*n)%4095] if a else int(n==0)\ndef dot(a,b):\n    out=0\n    for x,y in zip(a,b):out^=mul(x,y)\n    return out\n\ndef mi(a):\n    n=len(a);a=[r[:]+[int(i==j) for j in range(n)] for i,r in enumerate(a)]\n    for j in range(n):\n        p=next(i for i in range(j,n) if a[i][j]);a[p],a[j]=a[j],a[p]\n        t=inv(a[j][j]);a[j]=[mul(t,v) for v in a[j]]\n        for i in range(n):\n            if i!=j:\n                t=a[i][j];a[i]=[x^mul(t,y) for x,y in zip(a[i],a[j])]\n    return [r[n:] for r in a]\n\n@lru_cache(None)\ndef terms(k,m):\n    n=k+m-1;points=list(range(n));V=[[power(t,j) for j in range(n)] for t in points];W=mi(V)\n    # Cross = sum_t source(reverse(low),V[t]) * coefficient(kernel,W[:,t]) * V[t,:m]\n    return [(V[t][:k][::-1],[W[j][t] for j in range(n)],V[t][:m]) for t in range(n)]\n\ndef cross(q,low,high):\n    count=0\n    for source,coef,target in terms(len(low),len(high)):\n        p=mul(dot(source,low),dot(coef,q));count+=1\n        for i,c in enumerate(target):high[i]^=mul(c,p)\n    return count\n\ndef tri(q,x,reverse=False):\n    n=len(x)\n    if n==1:return 0\n    k=n//2;m=n-k;lo=x[:k];hi=x[k:]\n    if not reverse:\n        count=tri(q[:m],hi);count+=cross(q[1:n],lo,hi);count+=tri(q[:k],lo)\n    else:\n        count=tri(q[:k],lo,True);count+=cross(q[1:n],lo,hi);count+=tri(q[:m],hi,True)\n    x[:]=lo+hi\n    return count\n\ndef main():\n    rng=random.Random(102617147);tests=0;rank_counts={}\n    for n in (2,3,4,8,16,17,35):\n        for _ in range(100):\n            q=[1]+[rng.randrange(4096) for _ in range(n-1)];x=[rng.randrange(4096) for _ in range(n)];orig=x[:]\n            target=[dot(q[:i+1],orig[i::-1]) for i in range(n)]\n            c=tri(q,x);assert x==target\n            cr=tri(q,x,True);assert x==orig and c==cr\n            rank_counts[n]=c;tests+=1\n    # H_ij=B_(i+j+1), coefficients B_35=1, higher 0. H=J*T with q_s=B_(35-s).\n    for _ in range(100):\n        b=[rng.randrange(4096) for _ in range(35)]+[1];x=[rng.randrange(4096) for _ in range(35)];old=x[:]\n        expected=[dot((b[i+j+1] if i+j+1<=35 else 0 for j in range(35)),old) for i in range(35)]\n        c=tri(list(reversed(b[1:])),x);x.reverse();assert x==expected and c==147\n        x.reverse();tri(list(reversed(b[1:])),x,True);assert x==old\n    out={\'GF12_modulus\':hex(MOD),\'triangular_forward_reverse_cases\':tests,\'hankel35_forward_reverse_cases\':100,\'products\':rank_counts,\'H35_forward_T\':147*42,\'H35_pair_T\':294*42,\'extra_words\':0,\'inversions\':0,\'measurement_phase_corrections\':0}\n    print(json.dumps(out,indent=2))\nif __name__==\'__main__\':main()\n', 'blocks/dqi/pro-q2-897/source/exp/convert/kernels.py': '"""In-place structured GF12 kernels using exp/arithmetic primitives."""\nimport sys\nfrom pathlib import Path\nsys.path.insert(0,str(Path(__file__).resolve().parents[1]/\'arithmetic\'))\nimport field as ar\nfrom hankel_check import mul,inv,mi,power\nfrom functools import lru_cache\n\ndef ccols(c):return [mul(c,1<<i) for i in range(12)]\ndef cx(b,x,y,c=1):\n    if c==1:b.cx(x,y)\n    elif c:ar.linear_xor(b,x,y,ccols(c))\ndef scale_const(b,x,c):\n    if c!=1:ar.linear_inplace(b,x,ccols(c))\n\ndef expose(b,words,coeffs):\n    p=next(i for i,c in enumerate(coeffs) if c);c=coeffs[p]\n    scale_const(b,words[p],c)\n    for i,a in enumerate(coeffs):\n        if i!=p:cx(b,words[i],words[p],a)\n    return words[p],p\n\ndef unexpose(b,words,coeffs,p):\n    for i in reversed(range(len(coeffs))):\n        if i!=p:cx(b,words[i],words[p],coeffs[i])\n    scale_const(b,words[p],inv(coeffs[p]))\n\ndef rank_update(b,aw,ac,bw,bc,tw,tc):\n    x,ap=expose(b,aw,ac);y,bp=expose(b,bw,bc)\n    p=next(i for i,c in enumerate(tc) if c);v=tc[p];vi=inv(v)\n    for i,c in enumerate(tc):\n        if i!=p:cx(b,tw[p],tw[i],mul(c,vi))\n    scale_const(b,tw[p],vi)\n    ar.mul_xor(b,x,y,tw[p])\n    scale_const(b,tw[p],v)\n    for i in reversed(range(len(tc))):\n        if i!=p:cx(b,tw[p],tw[i],mul(tc[i],vi))\n    unexpose(b,bw,bc,bp);unexpose(b,aw,ac,ap)\n\n@lru_cache(None)\ndef vandermonde(n):\n    V=[[power(t,j) for j in range(n)] for t in range(n)]\n    return V,mi(V)\n\n@lru_cache(None)\ndef cross_terms(k,m):\n    n=k+m-1;V,W=vandermonde(n)\n    return [(V[t][:k][::-1],[W[j][t] for j in range(n)],V[t][:m]) for t in range(n)]\n\ndef tri(b,q,x,reverse=False):\n    # q=(q1,...,q_(n-1)), q0=1 implicit.\n    n=len(x)\n    if n==1:return\n    k=n//2;m=n-k\n    if not reverse:tri(b,q[:m-1],x[k:])\n    else:tri(b,q[:k-1],x[:k],True)\n    ts=cross_terms(k,m)\n    for ac,bc,tc in (reversed(ts) if reverse else ts):rank_update(b,x[:k],ac,q[:n-1],bc,x[k:],tc)\n    if not reverse:tri(b,q[:k-1],x[:k])\n    else:tri(b,q[:m-1],x[k:],True)\n\ndef hankel(b,B,A,reverse=False):\n    # H_ij=B_(i+j+1), implicit monic B_degree=len(A).\n    assert len(B)==len(A)\n    q=list(reversed(B[1:]))\n    if reverse:\n        for i in range(len(A)//2):b.swap(A[i],A[-1-i])\n        tri(b,q,A,True)\n    else:\n        tri(b,q,A)\n        for i in range(len(A)//2):b.swap(A[i],A[-1-i])\n\ndef convolution_xor(b,a,c,out,reverse=False):\n    # Coefficients 0..len(out)-1 of a*c; supports short output truncation.\n    N=len(a)+len(c)-1;V,W=vandermonde(N)\n    terms=range(N-1,-1,-1) if reverse else range(N)\n    for t in terms:\n        tc=[W[j][t] for j in range(len(out))]\n        if any(tc):rank_update(b,a,V[t][:len(a)],c,V[t][:len(c)],out,tc)\n\ndef convolution_transpose_left_xor(b,source,c,out,reverse=False):\n    # out[j] += sum_i source[i]*c[i-j], transpose of outpoly*c truncated to source.\n    N=len(out)+len(c)-1;V,W=vandermonde(N)\n    terms=range(N-1,-1,-1) if reverse else range(N)\n    for t in terms:\n        ac=[W[i][t] for i in range(len(source))]\n        if any(ac):rank_update(b,source,ac,c,V[t][:len(c)],out,V[t][:len(out)])\n', 'blocks/dqi/pro-q2-897/source/exp/convert/repair.py': '"""Adaptive one-gap Barnett repair for raw Bezout solve.\n\nSeparate fromdefaultgenericconvert.py. Firstleading-zeroisrepairedbymultiplygbyx\nandprecompanionMx. More/largegapsandrepairwithf(0)=0decodeincorrectlybut\nforward/reversearestilllocallyphase-correctpermutations. No deferredtranscripts.\n"""\nfrom contextlib import contextmanager\nfrom convert import ar,zero_flag,erase_zero_flag,scale,pre,post,raw_solve,frob\nfrom lowwidth import phase_inverse36\nfrom kernels import hankel\n\n@contextmanager\ndef nonzero_inverse(b,a):\n    ai=ar.inverse(b,a)\n    yield ai\n    mask=b.alloc_bits(12);b.hmr(ai,mask);b.free(ai);phase_inverse36(b,a,mask);b.free(mask)\n\n@contextmanager\ndef safe_word(b,a):\n    flag=zero_flag(b,a);b.cx(flag,a[0])\n    yield\n    b.cx(flag,a[0]);erase_zero_flag(b,a,flag)\n\ndef rotate(b,v,F,reverse=False):\n    for j in (range(1,len(v)) if reverse else range(len(v)-1,0,-1)):\n        b.cswap(F,v[j],v[j-1])\n\ndef coefficient_update(b,f,u,target):\n    if f[-1] is None:b.x(f[0][0])\n    else:b.cx(f[-1],f[0])\n    ar.mul_xor(b,f[0],u,target)\n    if f[-1] is None:b.x(f[0][0])\n    else:b.cx(f[-1],f[0])\n\ndef companion(b,f,payloads,F,reverse=False):\n    """Conditionallymultiplyeachpayloadbyx modf, oritsinverse.\n\n    f0 andleadingaresafenedonlyduringthispermutation, soalwaysinvertible.\n    Masku=F*lowest/bbeforeupdates; afterupdatesu=F*lowest/f0.\n    """\n    lead=f[-1]\n    @contextmanager\n    def lead_safe():\n        if lead is None:yield\n        else:\n            with safe_word(b,lead):yield\n    @contextmanager\n    def invctx(a):\n        if a is None:yield None\n        else:\n            with nonzero_inverse(b,a) as ai:yield ai\n    with safe_word(b,f[0]),lead_safe():\n        if not reverse:\n            for v in payloads:rotate(b,v,F)\n        masks=[]\n        divisor=f[0] if reverse else lead\n        with invctx(divisor) as di:\n            for v in payloads:\n                u=b.alloc_qubits(12)\n                for j in range(12):b.init_and(F,v[0][j],u[j])\n                if divisor is not None:scale(b,u,divisor,di,True)\n                if not reverse:\n                    for i in range(1,len(v)):ar.mul_xor(b,f[i],u,v[i])\n                    coefficient_update(b,f,u,v[0])\n                else:\n                    coefficient_update(b,f,u,v[0])\n                    for i in reversed(range(1,len(v))):ar.mul_xor(b,f[i],u,v[i])\n                mask=b.alloc_bits(12);b.hmr(u,mask);b.free(u);masks.append(mask)\n        recovery=lead if reverse else f[0]\n        with invctx(recovery) as ri:\n            for v,mask in zip(payloads,masks):\n                if recovery is None:\n                    for j in range(12):b.ccz(F,v[0][j],mask[j])\n                else:\n                    tmp=b.alloc_qubits(12);ar.mul_xor(b,v[0],ri,tmp)\n                    for j in range(12):b.ccz(F,tmp[j],mask[j])\n                    ar.erase_product(b,v[0],ri,tmp);b.free(tmp)\n                b.free(mask)\n        if reverse:\n            for v in payloads:rotate(b,v,F,True)\n\ndef equal_code(b,H,code):\n    for i,q in enumerate(H):\n        if code>>i&1:b.x(q)\n    flag=zero_flag(b,H)\n    for i,q in enumerate(H):\n        if code>>i&1:b.x(q)\n    return flag\n\ndef erase_equal(b,H,code,flag):\n    for i,q in enumerate(H):\n        if code>>i&1:b.x(q)\n    erase_zero_flag(b,H,flag)\n    for i,q in enumerate(H):\n        if code>>i&1:b.x(q)\n\ndef history_arm(b,H,a,code):\n    hz=zero_flag(b,H);az=zero_flag(b,a);F=b.alloc_qubits(1)[0];b.ccx(hz,az,F)\n    erase_zero_flag(b,a,az);erase_zero_flag(b,H,hz)\n    for i,q in enumerate(H):\n        if code>>i&1:b.cx(F,q)\n    return F\n\ndef history_clear_flag(b,H,code,F):\n    eq=equal_code(b,H,code);b.cx(eq,F);erase_equal(b,H,code,eq);b.free(F)\n\ndef history_finish(b,H,a,code,F):\n    for i,q in enumerate(H):\n        if code>>i&1:b.cx(F,q)\n    hz=zero_flag(b,H);az=zero_flag(b,a);b.ccx(hz,az,F)\n    erase_zero_flag(b,a,az);erase_zero_flag(b,H,hz);b.free(F)\n\ndef repair_solve(b,f,g,payloads,H,reverse=False):\n    m=len(f)-1\n    if m==1:return raw_solve(b,f,g,payloads,reverse)\n    F=history_arm(b,H,g[-1],m)\n    rotate(b,g,F)\n    if not reverse:companion(b,f,payloads,F)\n    history_clear_flag(b,H,m,F)\n    low=[v[:-1]for v in payloads]\n    if not reverse:\n        pre(b,f,g,payloads);repair_solve(b,g,f[:-2],low,H);post(b,f,g,payloads)\n    else:\n        post(b,f,g,payloads,True);repair_solve(b,g,f[:-2],low,H,True);pre(b,f,g,payloads,True)\n    F=equal_code(b,H,m)\n    if reverse:companion(b,f,payloads,F,True)\n    rotate(b,g,F,True)\n    history_finish(b,H,g[-1],m,F)\n\ndef bez_repair(b,O,E,payloads,reverse=False):\n    def reduce():\n        for i in range(len(O)):ar.mul_xor(b,E[-1],O[i],E[i])\n    reduce();H=list(b.alloc_qubits(5));repair_solve(b,O+[None],E[:-1],payloads,H,reverse);b.free(H);reduce()\n\ndef convert_repair(b,A,B,reverse=False):\n    assert len(A)==len(B)==35\n    E=B[::2];O=B[1::2];u=A[1::2];w=A[:34:2];top=A[-1]\n    if reverse:hankel(b,B,A,True)\n    frob(b,B,2048);frob(b,A,2048)\n    if reverse:\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n        for i in range(17):ar.mul_xor(b,O[i],u[i],top)\n        bez_repair(b,O,E,[u,w],True)\n        for i in range(17):ar.mul_xor(b,top,O[i],w[i])\n    else:\n        for i in range(17):ar.mul_xor(b,top,O[i],w[i])\n        bez_repair(b,O,E,[u,w])\n        for i in range(17):ar.mul_xor(b,O[i],u[i],top)\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n    frob(b,A,2);frob(b,B,2)\n    if not reverse:hankel(b,B,A)\n', 'blocks/dqi/pro-q2-897/source/exp/convert_short_hankel/kernel.py': '"""Drop-in guarded CONVERT with exact short-H boundary.\n\nDefault preserves the stable arbitrary borrowed-history completion. Optional\nmonic_root=True requires the supplied history word initially zero; if absent\nit is allocated zero. B/payloads otherwise arbitrary, including all gaps.\nReverse infers the option from its forward phase token.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.safe_scalar import converter as old\n\n\ndef _root_ops(b,ops,a,ai,undo=False):\n for op in (reversed(ops)if undo else ops):\n  if op[0]==\'copy\':b.x(op[1][0]);b.cx(ai,op[1])\n  else:old.runops(b,[op],a,ai,undo)\n\n\ndef _pre_root(b,f,g,payloads,undo=False,incoming=None,defer=False):\n m=len(f)-1;n=m-1;a=g[-1];q0,q1=f[-2:]\n def action(ai):\n  ops=[(\'s\',v[-1],True)for v in payloads]\n  ops +=[(\'m\',g[i],v[-1],v[i])for v in payloads for i in range(n)]\n  ops +=[(\'copy\',q1),(\'m\',q1,g[-2],q0),(\'s\',q0,True),(\'c\',g[:-1],[q0,q1],f[:-2])]\n  _root_ops(b,ops,a,ai,undo)\n return old.inverse_apply(b,a,action,incoming,defer)\n\n\ndef _post_root(b,f,g,payloads,reverse=False,incoming=None,defer=False):\n n=len(g)-1;a=g[-1];q0,q1=f[-2:]\n def action(ai):\n  # q1=1/(a+[a0]), so division by q1 is multiplication by the\n  # already clipped a in this same context, even on scalar-zero charts.\n  ops=[(\'s\',v[-1],False)for v in payloads]\n  ops +=[(\'m\',g[i],v[i],v[-1])for v in payloads for i in range(n)]\n  ops +=[(\'s\',v[-1],True)for v in payloads]\n  ops +=[(\'c\',g[:-1],[q0,q1],f[:-2]),(\'s\',q0,False),(\'m\',q1,g[-2],q0),(\'copy\',q1)]\n  _root_ops(b,ops,a,ai,reverse)\n return old.inverse_apply(b,a,action,incoming,defer)\n\n\ndef _look_root(b,f,g,payloads,reverse,phase):\n m=len(f)-1;F=old.fresh_zero_flag(b,g[-1]);old.rotate(b,g,F)\n if not reverse:carry=_pre_root(b,f,g,payloads,defer=True)\n else:carry=_post_root(b,f,g,payloads,True,phase[\'a\'].pop(m),True)\n old.canonical_permute(b,f,g,F);old.clear_flag(b,g[-1],F)\n parent=([v[-1]for v in payloads],f[-1],f[-2])\n old.look_solve(b,g,f[:-2],[v[:-1]for v in payloads],None,parent,reverse,phase)\n F=old.fresh_zero_flag(b,g[-1]);old.canonical_permute(b,f,g,F,True)\n if not reverse:phase[\'a\'][m]=_post_root(b,f,g,payloads,False,carry,True)\n else:_pre_root(b,f,g,payloads,True,carry)\n old.rotate(b,g,F,True);old.clear_flag(b,g[-1],F)\n\n\ndef _bez_monic(b,O,E,payloads,reverse,history_scratch12,phase):\n def reduce():\n  for i in range(len(O)):old.ar.mul_xor0(b,E[-1],O[i],E[i])\n reduce();lead=list(b.alloc_qubits(12))if history_scratch12 is None else list(history_scratch12)\n b.x(lead[0]);_look_root(b,O+[lead],E[:-1],payloads,reverse,phase);b.x(lead[0])\n if history_scratch12 is None:b.free(lead)\n reduce()\n\n\ndef convert_safe(b,A,B,reverse=False,history_scratch12=None,phase_token=None,*,monic_root=False):\n assert len(A)==len(B)==35\n phase={\'a\':{},\'q\':{},\'_short_monic\':bool(monic_root)}if phase_token is None else phase_token\n assert not reverse or phase_token is not None\n use_monic=phase[\'_short_monic\']\n # Frobenius first on BOTH directions: reverse input is the completed C.\n old.frob(b,B,2048);old.frob(b,A,2048)\n E=B[::2];O=B[1::2];u=A[1::2];w=A[:34:2];top=A[-1]\n def bez(rev):\n  if use_monic:_bez_monic(b,O,E,[u,w],rev,history_scratch12,phase)\n  else:old.bez_lookahead(b,O,E,[u,w],rev,history_scratch12,phase)\n if not reverse:\n  for i in range(17):old.ar.mul_xor0(b,top,O[i],w[i])\n  bez(False)\n  for i in range(17):old.ar.mul_xor0(b,E[i+1],w[i],top)\n  for i in range(17):b.swap(A[2*i],A[2*i+1])\n  old.hankel(b,B[1:],A[:34])\n  for i in reversed(range(34)):b.swap(A[i],A[i+1])\n else:\n  for i in range(34):b.swap(A[i],A[i+1])\n  old.hankel(b,B[1:],A[:34],True)\n  for i in range(17):b.swap(A[2*i],A[2*i+1])\n  for i in range(17):old.ar.mul_xor0(b,E[i+1],w[i],top)\n  bez(True)\n  for i in range(17):old.ar.mul_xor0(b,top,O[i],w[i])\n old.frob(b,A,2);old.frob(b,B,2)\n if reverse:assert not phase[\'a\']and not phase[\'q\']\n return phase if not reverse else None\n', 'blocks/dqi/pro-q2-897/source/exp/convert_short_hankel/phase34.py': '"""Private34T phase substitution; never mutates stable module globals.\n\nSame API as kernel.convert_safe. Saving2T per ultimate inverse closure;\nphase helper21private fits below the existing39-private converter peak.\n"""\nfrom types import FunctionType,SimpleNamespace\nfrom contextlib import contextmanager\nfrom exp.safe_scalar import converter as stable\nfrom exp.fold_cached_norm.phase34 import phase_inverse34\nfrom exp.convert_short_hankel import kernel as short\n\n\ndef clone(fn,scope,name=None):\n out=FunctionType(fn.__code__,scope,name or fn.__name__,fn.__defaults__,fn.__closure__)\n out.__kwdefaults__=None if fn.__kwdefaults__ is None else dict(fn.__kwdefaults__)\n return out\n\n# Erase/inverse_apply_safe are imported closures with different globals;\n# clone the chain explicitly before rebinding the converter namespace.\n_erase=clone(stable.erase,dict(stable.erase.__globals__,phase_inverse36_zero=phase_inverse34))\n_apply=clone(stable.inverse_apply_safe,dict(stable.inverse_apply_safe.__globals__,erase=_erase))\n_scope=dict(vars(stable),erase=_erase,inverse_apply_safe=_apply)\n_qgen=clone(stable.q_inverse.__wrapped__,_scope,\'q_inverse\')\n_scope[\'q_inverse\']=contextmanager(_qgen)\nfor _name in [\'scale\',\'runops\',\'zero_toggle\',\'inverse_apply\',\'pre\',\'canonical_permute\',\'masked_product\',\'fresh_zero_flag\',\'clear_flag\',\'q_post\',\'post_look\',\'base\',\'look_solve\',\'bez_lookahead\']:\n _scope[_name]=clone(getattr(stable,_name),_scope,_name)\n_private=SimpleNamespace(**_scope)\n\n# Clone short-H and optional monic-root functions against that private graph.\n_short_scope=dict(vars(short),old=_private)\nfor _name in [\'_root_ops\',\'_pre_root\',\'_post_root\',\'_look_root\',\'_bez_monic\',\'convert_safe\']:\n _short_scope[_name]=clone(getattr(short,_name),_short_scope,_name)\nconvert_safe=_short_scope[\'convert_safe\']\n', 'blocks/dqi/pro-q2-897/source/exp/decoder/common.py': '"""Lead\'s finite-field and reversible linear utilities (GF12 word bank)."""\nfrom functools import lru_cache\nfrom pathlib import Path\nimport sys\nimport circ as km\n\nROOT=Path(__file__).resolve().parents[2]\nsys.path.insert(0,str(ROOT))\nsys.path.insert(0,str(ROOT/\'exp/arithmetic\'))\nimport field as ar\ngf=km.GF2Field(12)\nK=2048\n\ndef mul(a,b):return gf.mul(a,b)\ndef inv(a):return gf.invert(a) if a else 0\ndef power(a,n):return gf.pow(a,n)\ndef xor(xs):\n    out=0\n    for x in xs:out^=x\n    return out\ndef trim(p):\n    p=list(p)\n    while len(p)>1 and not p[-1]:p.pop()\n    return p\ndef add(a,b):return trim([(a[i] if i<len(a) else 0)^(b[i] if i<len(b) else 0) for i in range(max(len(a),len(b)))])\ndef pmul(a,b):\n    c=[0]*(len(a)+len(b)-1)\n    for i,x in enumerate(a):\n        for j,y in enumerate(b):c[i+j]^=mul(x,y)\n    return trim(c)\ndef divmodp(a,b):\n    a=trim(a);out=[0]*max(1,len(a)-len(b)+1)\n    assert b[-1]\n    bi=inv(b[-1])\n    for k in range(len(a)-len(b),-1,-1):\n        t=mul(a[k+len(b)-1],bi);out[k]=t\n        for j,v in enumerate(b):a[k+j]^=mul(t,v)\n    return trim(out),trim(a)\ndef emul(x,y,k=K):\n    a,b=x&4095,x>>12;c,d=y&4095,y>>12;bd=mul(b,d)\n    return (mul(a,c)^mul(k,bd))|((mul(a,d)^mul(b,c)^bd)<<12)\ndef einv(x,k=K):\n    a,b=x&4095,x>>12;t=inv(mul(a,a)^mul(a,b)^mul(k,mul(b,b)))\n    return mul(a^b,t)|(mul(b,t)<<12)\ndef eev(p,x):\n    v=0\n    for a in reversed(p):v=emul(v,x)^a\n    return v\n\n@lru_cache(None)\ndef constcols(c):return tuple(mul(c,1<<i) for i in range(12))\ndef cx(b,x,z,c=1):\n    if c==1:b.cx(x,z)\n    elif c:ar.linear_xor(b,x,z,constcols(c))\ndef scale(b,x,c):\n    if c!=1:ar.linear_inplace(b,x,constcols(c))\ndef scalepair(b,x,c):\n    scale(b,x[:12],c);scale(b,x[12:],c)\ndef combine(words):return km.array([q for w in words for q in w])\ndef wordlist(reg):return [km.array(list(reg)[i:i+12]) for i in range(0,len(reg),12)]\n\n@lru_cache(None)\ndef matrix_ops(rows):\n    """Elementary row operations reducing an invertible GF12 matrix to I."""\n    a=[list(r) for r in rows];n=len(a);ops=[]\n    assert all(len(r)==n for r in a)\n    for j in range(n):\n        p=next(i for i in range(j,n) if a[i][j])\n        if p!=j:a[p],a[j]=a[j],a[p];ops.append((\'swap\',p,j,0))\n        v=inv(a[j][j])\n        if v!=1:a[j]=[mul(x,v) for x in a[j]];ops.append((\'scale\',j,0,v))\n        for i in range(n):\n            if i!=j and a[i][j]:\n                v=a[i][j];a[i]=[x^mul(y,v) for x,y in zip(a[i],a[j])];ops.append((\'add\',j,i,v))\n    return tuple(ops)\n\ndef linear_words(b,words,rows,reverse=False):\n    ops=matrix_ops(tuple(tuple(r) for r in rows))\n    for op,i,j,v in (ops if reverse else reversed(ops)):\n        if op==\'swap\':b.swap(words[i],words[j])\n        elif op==\'scale\':scale(b,words[i],v if reverse else inv(v))\n        else:cx(b,words[i],words[j],v)\n\ndef zero_hold(b,x):\n    """Return a held all-zero ladder, len(x)-1 Toffolis."""\n    b.x(x);lad=b.alloc_qubits(len(x)-1)\n    b.init_and(x[0],x[1],lad[0])\n    for j in range(2,len(x)):b.init_and(lad[j-2],x[j],lad[j-1])\n    b.x(x)\n    return lad\ndef zero_free(b,x,lad):\n    b.x(x)\n    for j in reversed(range(2,len(x))):b.del_and(lad[j-2],x[j],lad[j-1])\n    b.del_and(x[0],x[1],lad[0]);b.free(lad);b.x(x)\n\ndef reclaim(b,regs):\n    """Reclaim specified known-free physical wires, without new wire IDs."""\n    wanted={str(q) for reg in regs for q in reg}\n    available=b.alloc_qubits(b.finish_circuit().num_qubits-b.num_allocated_qubits)\n    have={str(q) for q in available}\n    assert wanted<=have,(wanted-have)\n    b.free(km.array([q for q in available if str(q) not in wanted]))\n\ndef extmul_xor(b,x,y,z):\n    """126T product in explicit a+b*y basis; one scratch bit."""\n    a,c=x[:12],x[12:];d,e=y[:12],y[12:]\n    b.cx(z[:12],z[12:]);ar.mul_xor(b,a,d,z[:12]);b.cx(z[:12],z[12:])\n    scale(b,z[:12],inv(K))\n    ar.mul_xor(b,c,e,z[:12])\n    scale(b,z[:12],K)\n    b.cx(c,a);b.cx(e,d);ar.mul_xor(b,a,d,z[12:]);b.cx(e,d);b.cx(c,a)\n\ndef phase_extmul(b,x,y,mask):\n    for i in range(24):\n        for j in range(24):\n            v=emul(1<<i,1<<j)\n            for k in ar.bits(v):b.ccz(x[i],y[j],mask[k])\n\ndef write(s,qs,vals):\n    import numpy as np\n    vals=np.asarray(vals,dtype=np.uint64)\n    s.write_across_shots(km.array(qs),((vals[None,:]>>np.arange(len(qs),dtype=np.uint64)[:,None])&1).astype(bool))\ndef read(s,qs):\n    import numpy as np\n    return (s.read_across_shots(km.array(qs)).astype(np.uint64)*(np.uint64(1)<<np.arange(len(qs),dtype=np.uint64))[:,None]).sum(axis=0)\n', 'blocks/dqi/pro-q2-897/source/exp/decoder/extract.py': '"""In-place conjugate-node initial map and triangular EXTRACT."""\nfrom functools import lru_cache\nfrom common import *\nfrom exp.inverse12_escape.joint import inverse24_forward190 as inverse24_forward20,inverse24_reverse209 as inverse24_reverse20\n\n@lru_cache(None)\ndef constants(h=35):\n    cs=list(range(0,2*h,2));ks=[mul(c,c)^c^K for c in cs];P=[1]\n    for k in ks:P=pmul(P,[k,1,1])\n    matrix=[]\n    for c in cs:\n        g=c|(1<<12);Q=[0]*(2*h);Q[-1]=1\n        for j in reversed(range(2*h-1)):Q[j]=P[j+1]^emul(g,Q[j+1])\n        assert P[0]^emul(g,Q[0])==0\n        matrix.append(tuple(v&4095 for v in Q));matrix.append(tuple(v>>12 for v in Q))\n    return cs,ks,P,tuple(matrix)\n\ndef initial(b,words,reverse=False):\n    linear_words(b,words,constants(len(words)//2)[3],reverse=reverse)\n\ndef extract(b,words,reverse=False,transcript=None):\n    h=len(words)//2;cs,ks,_,_=constants(h);bank=[combine(words[2*j:2*j+2]) for j in range(h)]\n    if not reverse:\n        history=[]\n        for i in range(h):\n            cx(b,words[2*i+1],words[2*i],cs[i])\n            for j in range(i+1,h):\n                b.cx(words[2*i],words[2*j]);cx(b,words[2*i+1],words[2*j],cs[j]);b.cx(words[2*i+1],words[2*j+1])\n                history.append(inverse24_forward20(b,bank[j],K))\n                scalepair(b,bank[j],ks[i]^ks[j])\n        return history\n    history=list(transcript)\n    for i in reversed(range(h)):\n        for j in reversed(range(i+1,h)):\n            scalepair(b,bank[j],inv(ks[i]^ks[j]))\n            inverse24_reverse20(b,bank[j],history.pop(),K)\n            b.cx(words[2*i+1],words[2*j+1]);cx(b,words[2*i+1],words[2*j],cs[j]);b.cx(words[2*i],words[2*j])\n        cx(b,words[2*i+1],words[2*i],cs[i])\n    assert not history\n', 'blocks/dqi/pro-q2-897/source/exp/decoder/safe_bridge.py': '"""Zero-safe division by temporarily replacing a zero factor by one."""\nfrom common import *\nimport division as dv\nfrom exp.mul99.schedules import divide_zero_bridge414\n\ndef divide_zero_bridge(b,x,y,extra_zero_mask=None):\n    return divide_zero_bridge414(b,x,y,K=K,extra_zero_mask=extra_zero_mask)\n\nif __name__==\'__main__\':\n    import numpy as np,json\n    rng=np.random.default_rng(18171026);n=16384\n    xv=rng.integers(0,1<<24,n,dtype=np.uint64);yv=rng.integers(0,1<<24,n,dtype=np.uint64);yv[:4096]=0\n    b=km.CircuitBuilder();x=b.alloc_qubits(24);y=b.alloc_qubits(24)\n    old,fm=dv.multiply_zero_forward(b,x,y,K);mm=b.alloc_bits(24);b.cz(x,mm);b.cz(x[2],x[13])\n    divide_zero_bridge(b,x,y,fm);b.free(fm);b.cz(x,old);b.free(old)\n    c=b.finish_circuit();s=km.Simulator(batch_size=n,seed=188182);ms=rng.integers(0,1<<24,n,dtype=np.uint64)\n    write(s,x,xv);write(s,y,yv);write(s,mm,ms);s.do(c)\n    assert np.array_equal(read(s,x),xv) and np.array_equal(read(s,y),yv)\n    z=[emul(int(a),int(bb)) if bb else int(a) for a,bb in zip(xv,yv)]\n    want=np.array([bool((a&int(m)).bit_count()%2)^bool(((a>>2)&1)&((a>>13)&1)) for a,m in zip(z,ms)])\n    assert np.array_equal(want,s.read_phase_flipped_across_shots())\n    for q in range(48,c.num_qubits):assert not s.read_across_shots(km.q(q)).any(),q\n    r=dict(shots=n,pair_T=c.operation_statistics().num_toffoli_like,division_T=c.operation_statistics().num_toffoli_like-152,scratch=c.num_qubits-48,value_errors=0,phase_errors=0)\n    print(json.dumps(r,indent=2));(ROOT/\'exp/decoder/safe_bridge.json\').write_text(json.dumps(r,indent=2)+\'\\n\')\n', 'blocks/dqi/pro-q2-897/source/exp/decoder/synthesis.py': '"""Head-absorption synthesis with a complete deferred measurement transcript.\n\nThe rational interpolation chart is exact on valid heads without extension\npoles. Zero evaluation factors are replaced by one in the remainder map, so\narbitrary heads are preserved; that repair can change the decoded rational\nfunction after an extension pole. The top-coefficient promises are separate.\n"""\nfrom common import *\nfrom extract import constants\nimport division as dv\nfrom lowwidth import divide_lowwidth375\nsys.path.insert(0,str(ROOT/\'exp/convert\'))\nfrom kernels import rank_update,vandermonde\nimport kernels as ck\nfrom exp.noscratch.kernels import mul_xor_noscratch\nfrom convert import inverse_context\nfrom safe_bridge import divide_zero_bridge\n\ndef tail_recur(b,heads,p,q,d,reverse=False):\n    for a,z in (reversed(heads) if reverse else heads):\n        if reverse:b.swap(p,q);ar.mul_xor(b,z,q,p)\n        ar.mul_xor(b,p,q,d)\n        b.cx(z,a);ar.linear_inplace(b,q,ar.square_columns(12));ar.mul_xor(b,a,q,d)\n        ar.linear_inplace(b,q,[power(1<<j,2048) for j in range(12)]);b.cx(z,a)\n        if not reverse:ar.mul_xor(b,z,q,p);b.swap(p,q)\n\n@lru_cache(None)\ndef quotient_terms(n,kap):\n    """floor(l*A/q): A0 does not contribute; n rank-one products."""\n    V,W=vandermonde(n);out=[]\n    for t in range(n):\n        polynomial=[0]+[W[j][t] for j in range(n)]\n        quotient,_=divmodp(polynomial,[kap,1,1]);quotient+= [0]*(n-1-len(quotient))\n        out.append(([1,t],V[t][:n-1],quotient))\n    return out\n\ndef quotient_xor(b,head,A,B,kap,reverse=False):\n    assert len(B)==len(A)-1\n    terms=quotient_terms(len(A),kap)\n    for ac,bc,tc in (reversed(terms) if reverse else terms):\n        if any(tc):rank_update(b,head,ac,A[1:],bc,B,tc)\n\ndef full_product_xor(b,head,A,target):\n    n=len(A)+1;V,W=vandermonde(n)\n    for t in range(n):rank_update_noanc(b,head,[1,t],A,V[t][:len(A)],target,[W[j][t] for j in range(n)])\n\ndef rank_update_noanc(b,aw,ac,bw,bc,tw,tc):\n    x,ap=ck.expose(b,aw,ac);y,bp=ck.expose(b,bw,bc)\n    p=next(i for i,c in enumerate(tc) if c);v=tc[p];vi=inv(v)\n    for i,c in enumerate(tc):\n        if i!=p:cx(b,tw[p],tw[i],mul(c,vi))\n    scale(b,tw[p],vi);mul_xor_noscratch(b,x,y,tw[p]);scale(b,tw[p],v)\n    for i in reversed(range(len(tc))):\n        if i!=p:cx(b,tw[p],tw[i],mul(tc[i],vi))\n    ck.unexpose(b,bw,bc,bp);ck.unexpose(b,aw,ac,ap)\n\n@lru_cache(None)\ndef embedding_matrix(n,kap):\n    # Input [remainder0,remainder1,quotient0,...] -> q*quotient+remainder.\n    rows=[[0]*n for _ in range(n)];rows[0][0]=1;rows[1][1]=1\n    for j in range(n-2):\n        for k,c in enumerate([kap,1,1]):rows[j+k][j+2]=c\n    return tuple(tuple(r) for r in rows)\n\n@lru_cache(None)\ndef evaluation_coefficients(n,c):\n    g=c|(1<<12);out=[1]\n    for j in range(1,n):out.append(emul(out[-1],g))\n    return tuple(out)\n\ndef expose_eval(b,A,c,reverse=False):\n    es=evaluation_coefficients(len(A),c)\n    if reverse:\n        for i in reversed(range(2,len(A))):cx(b,A[i],A[1],es[i]>>12)\n        for i in reversed(range(2,len(A))):cx(b,A[i],A[0],es[i]&4095)\n        cx(b,A[1],A[0],c)\n    else:\n        cx(b,A[1],A[0],c)\n        for i in range(2,len(A)):cx(b,A[i],A[0],es[i]&4095)\n        for i in range(2,len(A)):cx(b,A[i],A[1],es[i]>>12)\n    return combine(A[:2])\n\ndef absorb(b,head,A,B,c,kap,reverse=False,transcript=None):\n    out=head+B\n    if not reverse:\n        value=expose_eval(b,A,c);flag=dv.zero_flag(b,value);expose_eval(b,A,c,True)\n        # Zero-safe factor: A(gamma)=0 gets factor1. Adding1 to A changes\n        # l*A only in its remainder, so the quotient map is unchanged.\n        b.cx(flag,A[0][0])\n        spare=[b.alloc_qubits(12),b.alloc_qubits(12)];target=spare+B\n        linear_words(b,target,embedding_matrix(len(target),kap))\n        full_product_xor(b,head,A,target)\n        b.cx(flag,A[0][0])\n        transcript=b.alloc_bits(24);b.hmr(combine(head),transcript)\n        for h,z in zip(head,spare):b.swap(h,z);b.free(z)\n        flagmask=b.alloc_bits(1);b.hmr(flag,flagmask[0]);b.free(flag)\n        return out,A,(transcript,flagmask)\n    linear_words(b,out,embedding_matrix(len(out),kap),True)\n    value=expose_eval(b,A,c)\n    cx(b,head[1],head[0],c)\n    headmask,flagmask=transcript\n    divide_zero_bridge(b,combine(head),value,flagmask[0]);b.free(flagmask)\n    cx(b,head[1],head[0],c)\n    b.cz(combine(head),headmask);b.free(headmask)\n    expose_eval(b,A,c,True)\n    quotient_xor(b,head,A,B,kap,True)\n    return out,A,transcript\n\ndef rotate(b,words,control,reverse=False):\n    # multiply by x using a top-zero cyclic register, controlled by even chart.\n    for i in (range(1,len(words)) if reverse else reversed(range(1,len(words)))):\n        b.cswap(control,words[i-1],words[i])\n\ndef flag_zero(b,q,flag):\n    ladder=zero_hold(b,q);b.cx(ladder[-1],flag);zero_free(b,q,ladder)\n\ndef tail_scale_forward(b,p,q,d):\n    with inverse_context(b,q) as t:\n        out=b.alloc_qubits(12);mul_xor_noscratch(b,d,t,out)\n        mask=b.alloc_bits(12);b.hmr(d,mask);b.swap(d,out);b.free(out)\n    return mask\n\ndef tail_scale_reverse(b,p,q,d,mask):\n    with inverse_context(b,q) as t:\n        out=b.alloc_qubits(12);mul_xor_noscratch(b,d,q,out)\n        newer=b.alloc_bits(12);b.hmr(d,newer);b.swap(d,out);b.free(out)\n        ar.phase_product(b,d,t,newer);b.free(newer)\n    b.cz(d,mask);b.free(mask)\n\ndef synthesize(b,words,keep_leading=True):\n    """Return A,B, transcript. Exactly70datawords after forward completion."""\n    h=len(words)//2;assert h%2==1\n    cs,ks,_,_=constants(h);heads=[words[2*i:2*i+2] for i in range(h)]\n    p,q,d=[b.alloc_qubits(12) for _ in range(3)];b.x(q[0])\n    tail_recur(b,heads,p,q,d)\n    dmask=tail_scale_forward(b,p,q,d)\n    # Uniform tail: when q=0,d=0, [d/q,p]/q is x times the old even\n    # chart [p]/0. Every transfer is polynomial-linear, so the required\n    # padding x is already present; no parity bit or outer shift is needed.\n    A=[d,p];B=[q];history=[]\n    for i in reversed(range(h)):\n        oldA,oldB=A,B\n        A,B,m=absorb(b,heads[i],A,B,cs[i],ks[i])\n        history.append((heads[i],oldA,oldB,cs[i],ks[i],m))\n    # Highest A coefficient=0 and highest B coefficient=1 are determinant\n    # identities. The second A coefficient can be nonzero on a bad tail.\n    extra=[A[h],A[h+1],B[h]];b.x(B[h][0])\n    freed=extra[1:] if keep_leading else extra\n    for reg in freed:b.free(reg)\n    transcript=dict(h=h,heads=heads,p=p,q=q,d=d,dmask=dmask,history=history,A=A,B=B,extra=extra,freed=freed,keep_leading=keep_leading)\n    return A[:h],B[:h],transcript\n\ndef unsynthesize(b,transcript):\n    t=transcript;h=t[\'h\'];A=t[\'A\'];B=t[\'B\']\n    reclaim(b,t[\'freed\'])\n    b.x(B[h][0])\n    for head,oldA,oldB,c,kap,mask in reversed(t[\'history\']):absorb(b,head,oldA,oldB,c,kap,True,mask)\n    p,q,d=t[\'p\'],t[\'q\'],t[\'d\']\n    tail_scale_reverse(b,p,q,d,t[\'dmask\'])\n    tail_recur(b,t[\'heads\'],p,q,d,True);b.x(q[0])\n    b.free(p);b.free(q);b.free(d)\n', 'blocks/dqi/pro-q2-897/source/exp/fold_cached_norm/kernel.py': '"""Spend spare global width on a retained norm inside quadratic fold division."""\nfrom pathlib import Path\nimport sys\nfrom types import FunctionType\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.mul99.schedules import multiply_forward99,phase_inverse36_pivot\nfrom exp.noscratch.schedules import norm_xor_noscratch\nfrom exp.noscratch.kernels import mul_xor_noscratch\nfrom exp.inverse12_escape.joint import inverse_initialized\nfrom exp.arithmetic.division import phase_product24\nfrom exp.arithmetic.field import erase_product\nfrom exp.zero_padding_synthesis import kernel as prior\nfrom exp.inverse12_escape.zero1 import inverse_zero\nfrom exp.fold_cached_norm.phase34 import phase_inverse34\nfrom exp.inverse12_escape.zero1 import inverse6_zero20\nfrom exp.phase24_tradeoff.phase6small import phase_inverse6_classical6\nfrom exp.alternating_gauge import scalar as scalar_prior\n\ndef divide325(b,x,y,K=None):\n    """x/=y, preserve y,325 executed T/36 private; y!=0 or x=0 promise.\n\n    Inner synthesis explicitly clips its factor before this call. The norm\n    stays live through both base-field output scales, allowing immediate\n    old-half phase repayment and avoiding its42T reconstruction.\n    """\n    assert K is not None\n    b.cx(y[12:],y[:12]);old=multiply_forward99(b,x,y,K);b.cx(y[12:],y[:12])\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,y,N,K);u=inverse_initialized(b,N)\n    for off in (0,12):\n        half=x[off:off+12];out=b.alloc_qubits(12)\n        mul_xor_noscratch(b,half,u,out);erase_product(b,out,N,half)\n        b.swap(half,out);b.free(out)\n    mask=b.alloc_bits(12);b.hmr(u,mask);b.free(u)\n    phase_inverse36_pivot(b,N,mask);b.free(mask)\n    prior.ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,old,K);b.free(old)\n\ndef bind(fn,**changes):\n    return FunctionType(fn.__code__,dict(fn.__globals__,**changes),fn.__name__,fn.__defaults__,fn.__closure__)\n\ndivide323=bind(divide325,phase_inverse36_pivot=phase_inverse34)\ndivide323.__doc__=\'Unit factor division:323T/36private; retained norm and34T/21private inverse phase.\'\n\ndef divide_zero324(b,x,y,K=None,extra_zero_mask=None,after=None):\n    """Identity-at-zero division,324T/49 private, all HMR phases closed."""\n    assert K is not None\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,y,N,K);u,F=inverse_zero(b,N)\n    b.cx(y[12:],y[:12]);b.cx(F,y[0]);old=multiply_forward99(b,x,y,K)\n    b.cx(F,y[0]);b.cx(y[12:],y[:12]);b.cx(F,u[0])\n    for off in (0,12):\n        half=x[off:off+12];out=b.alloc_qubits(12);mul_xor_noscratch(b,half,u,out)\n        b.cx(F,N[0]);erase_product(b,out,N,half);b.cx(F,N[0])\n        b.swap(half,out);b.free(out)\n    b.cx(F,u[0]);um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    b.cx(F,y[0]);phase_product24(b,x,y,old,K);b.cx(F,y[0]);b.free(old)\n    if after is not None:after(F)\n    fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n    if extra_zero_mask is not None:b.cx(extra_zero_mask,fm[0])\n    phase_inverse34(b,N,um,fm[0]);b.free(um);b.free(fm)\n    prior.ar._erase_norm(b,y,N,K);b.free(N)\n\nabsorb=bind(prior.absorb_odd,divide_lowwidth369=divide323)\nabsorb_even=bind(prior.absorb_even,divide_lowwidth369=divide323)\nouter=bind(prior.outer,divide_zero369=divide_zero324)\n\ndef _inverse6_zero(b,x,with_zero):\n    assert with_zero\n    return inverse6_zero20(b,x)\n\ndivide_zero113=bind(scalar_prior.divide_zero117,inverse6_init19=_inverse6_zero,phase_inverse_generic6=phase_inverse6_classical6)\nmultiply_zero141=bind(scalar_prior.multiply_zero144,inverse_zero=inverse_zero,phase_inverse36_zero=phase_inverse34)\nprefix_reverse137=bind(scalar_prior.prefix_reverse141,divide_zero117=divide_zero113)\nodd_decode=bind(prior.synthesize.__globals__[\'odd_decode\'],prefix_reverse141=prefix_reverse137)\nsynthesize=bind(prior.synthesize,absorb=absorb,absorb_even=absorb_even,outer=outer,odd_decode=odd_decode,divide_zero117=divide_zero113)\nunsynthesize=bind(prior.unsynthesize,absorb=absorb,absorb_even=absorb_even,outer=outer,odd_decode=odd_decode,multiply_zero144=multiply_zero141)\n', 'blocks/dqi/pro-q2-897/source/exp/fold_cached_norm/phase34.py': '"""Use the held I6 DAG with a classical mask:34T,21 private, optional zero."""\nfrom exp.jet24.kernel import phase_inverse12_quantum64\n\ndef phase_inverse34(b,x,mask,zero_mask=None):\n    return phase_inverse12_quantum64(b,x,mask,zero_mask)\n', 'blocks/dqi/pro-q2-897/source/exp/fold_joint/divide.py': '"""Zero-aware streamed factor division24, no edits to arithmetic modules."""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.mul99.schedules import *\nfrom exp.safe_scalar.phase import phase_inverse36_zero\nfrom exp.noscratch.kernels import inverse3_noscratch\nfrom exp.arithmetic.lowwidth import classical_transpose\n\ndef inverse_stream_zero(b,x):\n    """Overwrite12-bit x with inv0(x); expose zero via bottom norm,68T.\n\n    Same recursive streamed schedule as old66T inverse, with2AND flag only\n    after deepest inverse\'s old-input wires are erased. Peak12+1 scratch.\n    """\n    n=len(x)\n    if n==3:\n        y=inverse3_noscratch(b,x);mask=b.alloc_bits(3);b.hmr(x,mask);b.swap(x,y);b.free(y)\n        b.x(x);u=b.alloc_qubits(1)[0];F=b.alloc_qubits(1)[0]\n        b.init_and(x[0],x[1],u);b.init_and(u,x[2],F);b.del_and(x[0],x[1],u);b.free(u);b.x(x)\n        return mask,F\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    N=b.alloc_qubits(k);norm_xor_noscratch(b,x,N,K)\n    nm,F=inverse_stream_zero(b,N);ar._phase_norm(b,x,nm,K);b.free(nm)\n    outb=b.alloc_qubits(k);mul_xor_noscratch(b,x[k:],N,outb)\n    raw=b.alloc_bits(n);b.hmr(x[k:],raw[k:])\n    mul_xor_noscratch(b,x[:k],N,x[k:]);b.cx(outb,x[k:]);b.hmr(x[:k],raw[:k])\n    b.swap(x[:k],x[k:]);b.swap(x[k:],outb);b.free(outb)\n    ar._erase_norm(b,x,N,K);b.free(N);linear_inplace(b,x,frm)\n    mask=classical_transpose(b,raw,to);b.free(raw);return mask,F\n\ndef divide_zero371(b,x,y,K=None,extra_zero_mask=None,after=None):\n    """x/=y for y!=0, identity when y0. <=371T,37 private, all-input safe.\n\n    Norm\'s zero flag costs2T in streamed inverse, not a23-bit factor test.\n    Its phase is fused into the existing36T inverse-norm phase correction.\n    WIDTH OBSTRUCTION: u12 remains live across a24-bit product target, +F.\n    """\n    if K is None:K=extension_columns()[2]\n    u=b.alloc_qubits(12);norm_xor_noscratch(b,y,u,K)\n    nm,F=inverse_stream_zero(b,u);ar._phase_norm(b,y,nm,K);b.free(nm)\n    b.cx(y[12:],y[:12]);b.cx(F,y[0]);oldmask=multiply_forward99(b,x,y,K);b.cx(F,y[0]);b.cx(y[12:],y[:12])\n    b.cx(F,u[0]);masks=[]\n    for off in(0,12):\n        half=x[off:off+12];out=b.alloc_qubits(12);mul_xor_noscratch(b,half,u,out)\n        mask=b.alloc_bits(12);b.hmr(half,mask);masks.append(mask);b.swap(half,out);b.free(out)\n    b.cx(F,u[0]);um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,y,N,K)\n    b.cx(F,N[0])\n    for off,mask in zip((0,12),masks):phase_product(b,x[off:off+12],N,mask);b.free(mask)\n    b.cx(F,N[0]);b.cx(F,y[0]);phase_product24(b,x,y,oldmask,K);b.cx(F,y[0]);b.free(oldmask)\n    if after is not None:after(F)\n    fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n    if extra_zero_mask is not None:b.cx(extra_zero_mask,fm[0])\n    phase_inverse36_zero(b,N,um,fm[0]);b.free(um);b.free(fm)\n    ar._erase_norm(b,y,N,K);b.free(N)\n', 'blocks/dqi/pro-q2-897/source/exp/fold_joint/kernel.py': '"""CRT joint remainder/Taylor fold. Live BEST modules are read only.\n\nThe factor flag remains live across quotient restoration, so its existing\ncompletion can be shared without another zero test or controlled word bank.\nThe polynomial and zero-factor completion match BEST for arbitrary heads.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.partial_gauge import synthesis_noflag as old\nfrom exp.mul99.kernels import mul24_xor99\nfrom exp.mul99.schedules import divide_lowwidth369\nfrom exp.fold_joint.divide import divide_zero371\nfrom exp.decoder.common import *\nbase=old.base;pg=old.pg;ck=old.ck;dv=old.dv\n\ndef epow(c,k):\n    ans=1;g=c|4096\n    for _ in range(k):ans=emul(ans,g)\n    return ans\ndef escale(b,R,c):ar.linear_inplace(b,combine(R),[emul(1<<i,c)for i in range(24)])\n\ndef q_times_low(b,B,kap,reverse=False):\n    n=len(B)\n    if not n:return\n    rows=[[kap if i==j else 1 if i-j in(1,2)else 0 for j in range(n)]for i in range(n)]\n    linear_words(b,B,rows,reverse)\n\ndef low_product(b,head,A,B):\n    """l*A mod x^k, k low Acoeffs; rank0/1/k+1 at k0/1/>=2."""\n    k=len(B)\n    if not k:return\n    V,W=base.vandermonde(k+1)\n    for t in range(k+1):\n        old.rank_affine(b,head,[1,t],A[:k],V[t][:k],B,[W[j][t]for j in range(k)])\n\ndef R_to_tops(b,R,B,c,reverse=False):\n    """CRT(E-remainder, lowTaylor)->(topcoeffs, unchangedlowTaylor)."""\n    k=len(B);powers=base.evaluation_coefficients(k,c)\n    if reverse:\n        cx(b,R[1],R[0],c);escale(b,R,epow(c,k))\n        for word,v in zip(B,powers):cx(b,word,R[0],v&4095);cx(b,word,R[1],v>>12)\n    else:\n        for word,v in zip(B,powers):cx(b,word,R[0],v&4095);cx(b,word,R[1],v>>12)\n        escale(b,R,einv(epow(c,k)));cx(b,R[1],R[0],c)\n\ndef divide414_callback(b,x,y,K,extra_zero_mask,after):\n    F=dv.zero_flag(b,y);b.cx(F,y[0]);divide_lowwidth369(b,x,y,K);b.cx(F,y[0])\n    after(F)\n    mask=b.alloc_bits(1);b.hmr(F,mask[0]);b.free(F);b.cx(extra_zero_mask,mask[0])\n    dv.phase_zero(b,y,mask[0]);b.free(mask)\n\ndef absorb(b,head,A,B,qi,p,virtual,pconst,c,kap,reverse=False,transcript=None,wide=False):\n    k=len(B);assert len(A)==k+1;out=B+head\n    if not reverse:\n        value=old.eval_implicit(b,A,qi,c);flag=dv.zero_flag(b,value)\n        spare=[b.alloc_qubits(12),b.alloc_qubits(12)];b.cx(flag,value[0]);cx(b,head[1],head[0],c)\n        mul24_xor99(b,combine(head),value,combine(spare),K)\n        cx(b,head[1],head[0],c);b.cx(flag,value[0]);old.eval_implicit(b,A,qi,c,True)\n        q_times_low(b,B,kap);b.cx(flag,A[0][0]);low_product(b,head,A,B);b.cx(flag,A[0][0])\n        v=epow(c,k+2);old.add_alias(b,p,spare,[v&4095,v>>12],virtual,pconst)\n        R_to_tops(b,spare,B,c)\n        hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n        for h,z in zip(head,spare):b.swap(h,z);b.free(z)\n        fm=b.alloc_bits(1);b.hmr(flag,fm[0]);b.free(flag)\n        return out,A,(hm,fm)\n    R_to_tops(b,head,B,c,True)\n    v=epow(c,k+2);old.add_alias(b,p,head,[v&4095,v>>12],virtual,pconst)\n    value=old.eval_implicit(b,A,qi,c);hm,fm=transcript\n    def after(F):\n        cx(b,head[1],head[0],c);old.eval_implicit(b,A,qi,c,True)\n        b.cx(F,A[0][0]);low_product(b,head,A,B);b.cx(F,A[0][0]);q_times_low(b,B,kap,True)\n        old.eval_implicit(b,A,qi,c);cx(b,head[1],head[0],c)\n    if wide:divide_zero371(b,combine(head),value,K=K,extra_zero_mask=fm[0],after=after)\n    else:divide414_callback(b,combine(head),value,K,fm[0],after)\n    b.free(fm)\n    cx(b,head[1],head[0],c);b.cz(combine(head),hm);b.free(hm)\n    old.eval_implicit(b,A,qi,c,True)\n    return out,A,transcript\n\ndef synthesize(b,words,keep_leading=True,wide=False):\n    h=len(words)//2;assert h>=3 and h%2==1\n    cs,ks,_,_=old.constants(h);heads=[words[2*i:2*i+2]for i in range(h)]\n    d=b.alloc_qubits(12);F=pg.zero_compact(b,heads[0][1]);masks=[]\n    for i in range(h):masks.append(old.prefix_encode(b,heads,i,F,d))\n    old.divide_safe138(b,d,heads[-1][1]);A=[d];B=[];hist=[]\n    for i in reversed(range(h)):\n        old.prefix_decode(b,heads,i,F,masks[i]);qi=heads[i-1][1]if i else None;p,v,pc=old.p_alias(heads,i,F)\n        oldA,oldB=A,B;A,B,t=absorb(b,heads[i],A,B,qi,p,v,pc,cs[i],ks[i],wide=wide)\n        hist.append((i,oldA,oldB,qi,p,v,pc,cs[i],ks[i],t))\n    extra=[A[h]];freed=[]if keep_leading else extra\n    for word in freed:b.free(word)\n    fmask=b.alloc_bits(1);b.hmr(F,fmask[0]);b.free(F)\n    t=dict(h=h,heads=heads,d=d,F=F,history=hist,A=A,B=B,extra=extra,freed=freed,keep_leading=keep_leading,fmask=fmask,wide=wide)\n    return A[:h],B,t\n\ndef unsynthesize(b,t):\n    if t[\'freed\']:reclaim(b,t[\'freed\'])\n    heads=t[\'heads\'];F=t[\'F\'];masks=[None]*t[\'h\']\n    for i,A,B,qi,p,v,pc,c,kap,token in reversed(t[\'history\']):\n        absorb(b,heads[i],A,B,qi,p,v,pc,c,kap,True,token,t[\'wide\']);masks[i]=old.prefix_encode(b,heads,i,F)\n        if i==0:\n            reclaim(b,[[F]]);lad=base.zero_hold(b,heads[0][1]);b.cx(lad[-1],F);base.zero_free(b,heads[0][1],lad)\n            b.cz(F,t[\'fmask\'][0]);b.free(t[\'fmask\'])\n    old.multiply_safe165(b,t[\'d\'],heads[-1][1])\n    for i in reversed(range(t[\'h\'])):old.prefix_decode(b,heads,i,F,masks[i],t[\'d\'])\n    pg.zero_compact_free(b,heads[0][1],F);b.free(t[\'d\'])\n', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/joint.py': '"""Correlated inverse-norm cache lowers the paired I12 reverse to67T/Q33.\nThe initialized inverse is unchanged64T/Q31; no <=63 isolated claim.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse19_joint.kernel import *\nfrom exp.inverse12_escape.zero1 import erase_raw_cache\n\ndef inverse6_keep_inner19(b,x):\n    """Preserve x; return inv0(x) and norm(inv0(x))3,19T/Q19."""\n    to,frm,K=tower(6);linear_inplace(b,x,to);g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,x,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(x)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms)\n    linear_inplace(b,g,LAY[\'raw_columns\']);R,E=g[:4],g[4:];z=list(E)+list(b.alloc_qubits(4));source=list(x)+list(R)+list(w)\n    for j,e in enumerate(LAY[\'outputs\']):ref.form_xor(b,source,e&65535,z[j])\n    for i,(l,r)in enumerate(LAY[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(LAY[\'outputs\']))\n        with ref.expose_pair(b,source,l,r)as(p,q):rankone_xor(b,p,q,z,v)\n    erase_raw_cache(b,x,R,LAY[\'raw_rows\'][:4]);linear_inplace(b,z,LAY[\'output_inverse_columns\'])\n    linear_inplace(b,x,frm);linear_inplace(b,z,frm);return z,w\n\ndef inverse12_reverse67(b,x,mask):\n    assert len(x)==12;to,frm,K=tower(12);linear_inplace(b,x,to)\n    N=b.alloc_qubits(6);norm_xor_noscratch(b,x,N,K);t,v=inverse6_keep_inner19(b,N);out=b.alloc_qubits(6)\n    mul_xor_noscratch(b,x[6:],t,out);erase_product(b,out,N,x[6:]);mul_xor_noscratch(b,x[:6],t,x[6:]);b.cx(out,x[6:])\n    b.cx(out,x[6:]);erase_product(b,x[6:],N,x[:6]);b.cx(out,x[6:]);b.swap(x[:6],x[6:]);b.swap(x[6:],out);b.free(out)\n    D=inverse3_noscratch(b,v);ar._erase_inverse_with_norm(b,t,N,D);b.free(N);erase_inverse(b,v,D);b.free(D)\n    to6,frm6,K6=tower(6);linear_inplace(b,t,to6);ar._erase_norm(b,t,v,K6);b.free(v);linear_inplace(b,t,frm6)\n    ar._erase_norm(b,x,t,K);b.free(t);linear_inplace(b,x,frm);b.cz(x,mask);b.free(mask)\n\ninverse12_forward64=inverse_inplace_forward20\n\ndef phase_inverse6_quantum19(b,u,L):\n    """Phase L dot inv0(u), bothquantum6;19T,9private, exactclosedHMR."""\n    to,frm,K=tower(6);linear_inplace(b,u,to);transpose=[sum(((frm[i]>>j)&1)<<i for i in range(6))for j in range(6)];linear_inplace(b,L,transpose)\n    g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,u,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(u)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms);source=prefix+list(w)\n    for j,e in enumerate(basic.DAG19[\'outputs\']):\n        if e&1:b.z(L[j])\n        for r in bits((e&65535)>>1):b.cz(L[j],source[r])\n    for i,(l,r)in enumerate(basic.DAG19[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(basic.DAG19[\'outputs\']));ix=list(bits(v));piv=ix[0]\n        for j in ix[1:]:b.cx(L[j],L[piv])\n        with ref.expose_pair(b,source,l,r)as(p,q):b.ccz(p,q,L[piv])\n        for j in reversed(ix[1:]):b.cx(L[j],L[piv])\n    m=b.alloc_bits(3);b.hmr(w,m);b.free(w);ref.phase_inverse3_virtual(b,prefix,forms,m);b.free(m)\n    for i,(l,r)in reversed(list(enumerate(basic.raw_forms19()))):\n        with ref.expose_pair(b,u,l,r)as(p,q):b.del_and(p,q,g[i])\n    b.free(g);linear_inplace(b,L,inverse_columns(transpose));linear_inplace(b,u,frm)\n\ndef inverse12_keep_inverse_norm64(b,x):\n    """Preserve x12, return t=inv(x),u=norm(t)6,64T;Q31."""\n    to,frm,K=tower(12);linear_inplace(b,x,to);D=b.alloc_qubits(6);norm_xor_noscratch(b,x,D,K);u=inverse6_init19(b,D);ar._erase_norm(b,x,D,K);b.free(D)\n    t=b.alloc_qubits(12);b.cx(x[6:],x[:6]);mul_xor_noscratch(b,x[:6],u,t[:6]);mul_xor_noscratch(b,x[6:],u,t[6:]);b.cx(x[6:],x[:6]);linear_inplace(b,x,frm);linear_inplace(b,t,frm);return t,u\n\ndef erase_norm_via_inverse_cache19(b,t,N,u):\n    """Erase N=inv0(t),u=norm(t)6;19T vsoldDcachecleanup24T.\n\n    Joint phase mask for N is linear in t and classical HMR outcomes; evaluate\n    inverse6 with that quantum mask, then erase u quadratically as norm(t).\n    """\n    to,frm,K=tower(12);linear_inplace(b,t,to);linear_inplace(b,N,to);m=b.alloc_bits(12);b.hmr(N,m);b.free(N);L=b.alloc_qubits(6);f=field(6)\n    def masks():\n        for j in range(6):\n            for i in range(6):\n                for r in bits(f.mul(1<<i,1<<j)):\n                    b.ccx(m[r],t[i],L[j]);b.ccx(m[r],t[i+6],L[j]);b.ccx(m[r+6],t[i+6],L[j])\n    masks();phase_inverse6_quantum19(b,u,L);masks();b.free(L);b.free(m);ar._erase_norm(b,t,u,K);b.free(u);linear_inplace(b,t,frm)\n\ndef inverse24_reverse209(b,x,mask,K=None):\n    cols,back,_=extension_columns(K);linear_inplace(b,x,cols);to,frm,K24=tower(24);linear_inplace(b,x,to)\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,x,N,K24);t,u=inverse12_keep_inverse_norm64(b,N);out=b.alloc_qubits(12)\n    mul_xor_noscratch(b,x[12:],t,out);erase_product(b,out,N,x[12:]);mul_xor_noscratch(b,x[:12],t,x[12:]);b.cx(out,x[12:]);b.cx(out,x[12:]);erase_product(b,x[12:],N,x[:12]);b.cx(out,x[12:]);b.swap(x[:12],x[12:]);b.swap(x[12:],out);b.free(out)\n    erase_norm_via_inverse_cache19(b,t,N,u);ar._erase_norm(b,x,t,K24);b.free(t);linear_inplace(b,x,frm);b.cz(x,mask);b.free(mask);linear_inplace(b,x,back)\n\ninverse24_forward190=basic.inverse24_forward19\n', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/zero1.py': '"""65T/Q31 zero-aware I12 from a one-AND correlated zero output.\nNot an improvement to isolated64T I12 inversion; no live module is changed.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse19_joint.kernel import *\nfrom types import FunctionType\nZLAY=json.loads(Path(__file__).with_name(\'zero_layout.json\').read_text())\n\ndef erase_raw_cache(b,x,cache,rows):\n    m=b.alloc_bits(len(cache));b.hmr(cache,m);b.free(cache)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,x,l,r)as(p,q):\n            for j,row in enumerate(rows):\n                if row>>i&1:b.ccz(p,q,m[j])\n    b.free(m)\n\ndef inverse6_zero20(b,x):\n    to,frm,K=tower(6);linear_inplace(b,x,to);g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,x,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(x)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms)\n    linear_inplace(b,g,LAY[\'raw_columns\']);R,E=g[:4],g[4:];z=list(E)+list(b.alloc_qubits(4));source=list(x)+list(R)+list(w)\n    for j,e in enumerate(LAY[\'outputs\']):ref.form_xor(b,source,e&65535,z[j])\n    for i,(l,r)in enumerate(LAY[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(LAY[\'outputs\']))\n        with ref.expose_pair(b,source,l,r)as(p,q):rankone_xor(b,p,q,z,v)\n    # One norm coordinate lies in the4D live raw-control span. Keep it, erase3.\n    linear_inplace(b,R,ZLAY[\'cache_columns\']);erase_raw_cache(b,x,R[1:],ZLAY[\'raw_cache_rows\'][1:])\n    source=list(x)+[R[0]]+list(w);left=1|(ZLAY[\'left_w\']<<8);right=1^ZLAY[\'input_form\']^(1<<7)^(ZLAY[\'right_w\']<<8)\n    F=b.alloc_qubits(1)[0]\n    with ref.expose_pair(b,source,left,right)as(p,q):b.init_and(p,q,F)\n    erase_raw_cache(b,x,R[:1],ZLAY[\'raw_cache_rows\'][:1]);linear_inplace(b,z,LAY[\'output_inverse_columns\']);ar._erase_norm(b,z,w,K);b.free(w)\n    linear_inplace(b,x,frm);linear_inplace(b,z,frm);return z,F\n\ndef inverse_zero(b,x):\n    to,frm,K=tower(12);linear_inplace(b,x,to);N=b.alloc_qubits(6);norm_xor_noscratch(b,x,N,K)\n    t,F=inverse6_zero20(b,N);ar._erase_norm(b,x,N,K);b.free(N)\n    out=b.alloc_qubits(12);b.cx(x[6:],x[:6]);mul_xor_noscratch(b,x[:6],t,out[:6]);mul_xor_noscratch(b,x[6:],t,out[6:]);b.cx(x[6:],x[:6])\n    ar._erase_norm(b,out,t,K);b.free(t);linear_inplace(b,x,frm);linear_inplace(b,out,frm);return out,F\n\nfrom exp.inverse19_joint import kernel as parentkernel\n_old=parentkernel.inverse_apply_safe\ninverse_apply_safe=FunctionType(_old.__code__,globals(),\'inverse_apply_safe\',_old.__defaults__,_old.__closure__)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/zero_layout.json': '{\n  "norm_coordinate": 3,\n  "raw_keep_coeff": 14,\n  "input_form": 58,\n  "left_w": 1,\n  "right_w": 4,\n  "pair_forms": [\n    2,\n    5\n  ],\n  "linear_correction": 3,\n  "cache_rows": [\n    14,\n    1,\n    2,\n    4\n  ],\n  "cache_columns": [\n    2,\n    5,\n    9,\n    1\n  ],\n  "raw_cache_rows": [\n    27,\n    41,\n    5,\n    14\n  ]\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse19_joint/kernel.py': '"""Width-audited19AND inverse substitutions; stable imports remain unchanged."""\nfrom pathlib import Path\nimport sys,json\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse_width_subspaces import kernel19 as basic\nfrom exp.inverse_width_subspaces.kernel19 import *\nfrom exp.safe_scalar import kernel as safe\nfrom exp.safe_scalar.phase import phase_inverse36_zero\nfrom exp.noscratch.schedules import norm_xor_noscratch\nLAY=json.loads(Path(__file__).with_name(\'layout.json\').read_text())\n\ndef inverse6_init19(b,x,with_zero=False):\n    """Fresh inverse6:19T/Q19, or inverse+zero21T/Q19; preserve x.\n\n    Reuse two redundant raw norm directions as output. Four other norm\n    directions remain until all final products are added. Compute zero only\n    AFTER those four wires are erased, using the still-live inverse norm3.\n    """\n    to,frm,K=tower(6);linear_inplace(b,x,to);g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,x,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(x)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms)\n    linear_inplace(b,g,LAY[\'raw_columns\']);R,E=g[:4],g[4:]\n    z=list(E)+list(b.alloc_qubits(4));source=list(x)+list(R)+list(w)\n    for j,e in enumerate(LAY[\'outputs\']):ref.form_xor(b,source,e&65535,z[j])\n    for i,(l,r)in enumerate(LAY[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(LAY[\'outputs\']));assert v\n        with ref.expose_pair(b,source,l,r)as(p,q):rankone_xor(b,p,q,z,v)\n    # Four retained raw products have only quadratic phase corrections.\n    rm=b.alloc_bits(4);b.hmr(R,rm);b.free(R)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,x,l,r)as(p,q):\n            for j,row in enumerate(LAY[\'raw_rows\'][:4]):\n                if row>>i&1:b.ccz(p,q,rm[j])\n    b.free(rm);linear_inplace(b,z,LAY[\'output_inverse_columns\'])\n    F=None\n    if with_zero:\n        b.x(w);u=b.alloc_qubits(1)[0];F=b.alloc_qubits(1)[0]\n        b.init_and(w[0],w[1],u);b.init_and(u,w[2],F);b.del_and(w[0],w[1],u);b.free(u);b.x(w)\n    ar._erase_norm(b,z,w,K);b.free(w);linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n    return(z,F)if with_zero else z\n\ndef inverse6_xor19(b,x,z=None,keep_norm=False):\n    if z is None and not keep_norm:return inverse6_init19(b,x)\n    return basic.inverse6_xor19(b,x,z,keep_norm)\n\n# Local recursive schedules, no mutation of imported globals.\nfrom types import FunctionType\ninverse6_xor20=inverse6_xor19\nfor _name in(\'inverse_initialized\',\'inverse12_keep_norm\',\'inverse_inplace_forward20\',\'inverse_inplace_reverse20\'):\n    _old=getattr(ref,_name);globals()[_name]=FunctionType(_old.__code__,globals(),_name,_old.__defaults__,_old.__closure__)\n\ndef inverse_zero(b,x):\n    """GF12 inverse0 +zero:66T/Q31, preserve input; closure unchanged36T."""\n    to,frm,K=tower(12);linear_inplace(b,x,to);N=b.alloc_qubits(6);norm_xor_noscratch(b,x,N,K)\n    t,F=inverse6_init19(b,N,True);ar._erase_norm(b,x,N,K);b.free(N)\n    out=b.alloc_qubits(12);b.cx(x[6:],x[:6]);mul_xor_noscratch(b,x[:6],t,out[:6]);mul_xor_noscratch(b,x[6:],t,out[6:]);b.cx(x[6:],x[:6])\n    ar._erase_norm(b,out,t,K);b.free(t);linear_inplace(b,x,frm);linear_inplace(b,out,frm);return out,F\n\nconsume=safe.consume;erase=safe.erase\n\ndef inverse_apply_safe(b,a,action,incoming=None,defer=False):\n    ai,F=inverse_zero(b,a);consume(b,ai,F,incoming)\n    b.cx(F,a[0]);b.cx(F,ai[0]);action(ai);b.cx(F,ai[0]);b.cx(F,a[0])\n    return erase(b,a,ai,F,defer)\n\nfrom exp.arithmetic.lowwidth import phase_inverse_generic6,phase_product_quantum\n\ndef inverse12_xor73(b,x,z,base6=inverse6_init19):\n    """Arbitrary-target XOR inverse12:73T/Q43, preserve x and target phase.\n\n    Initialized inverse schedules are NOT arbitrary-target oracles: they erase\n    t using norm(output), which is wrong for a preexisting target. Keep N and\n    close t by a9T phase inverse instead.\n    """\n    to,frm,K=tower(12);linear_inplace(b,x,to);linear_inplace(b,z,to)\n    N=b.alloc_qubits(6);norm_xor_noscratch(b,x,N,K);t=base6(b,N)\n    b.cx(x[6:],x[:6]);mul_xor_noscratch(b,x[:6],t,z[:6]);mul_xor_noscratch(b,x[6:],t,z[6:]);b.cx(x[6:],x[:6])\n    m=b.alloc_bits(6);b.hmr(t,m);b.free(t);phase_inverse_generic6(b,N,m);b.free(m)\n    ar._erase_norm(b,x,N,K);b.free(N);linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse19_joint/layout.json': '{\n  "raw_rows": [\n    41,\n    5,\n    14,\n    16,\n    1,\n    2\n  ],\n  "raw_columns": [\n    19,\n    36,\n    6,\n    5,\n    8,\n    1\n  ],\n  "output_columns": [\n    4,\n    8,\n    16,\n    1,\n    32,\n    6\n  ],\n  "output_inverse_columns": [\n    8,\n    33,\n    1,\n    2,\n    4,\n    16\n  ],\n  "gates": [\n    [\n      2,\n      2048\n    ],\n    [\n      48,\n      6144\n    ],\n    [\n      36,\n      4096\n    ],\n    [\n      2244,\n      4408\n    ],\n    [\n      2158,\n      12302\n    ],\n    [\n      2056,\n      8208\n    ],\n    [\n      6256,\n      8310\n    ],\n    [\n      6236,\n      10252\n    ],\n    [\n      4828,\n      9436\n    ],\n    [\n      4160,\n      8196\n    ]\n  ],\n  "outputs": [\n    54332300,\n    39463256,\n    60299450,\n    46469830,\n    11278712,\n    20656990\n  ]\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/baseline.py': '"""21T correlated inverse6 XOR seed, arbitrary target, clean private state."""\nimport sys\nfrom pathlib import Path\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.noscratch.kernels import *\n\ndef inverse6_xor21(b,x,z):\n    assert len(x)==len(z)==6\n    to,frm,K=tower(6);linear_inplace(b,x,to);linear_inplace(b,z,to)\n    N=b.alloc_qubits(3);f=field(3)\n    linear_xor(b,x[:3],N,square_columns(3))\n    linear_xor(b,x[3:],N,[f.mul(K,v)for v in square_columns(3)])\n    mul_xor_noscratch(b,x[:3],x[3:],N)\n    w=inverse3_noscratch(b,N)\n    b.cx(x[3:],x[:3]);mul_xor_noscratch(b,x[:3],w,z[:3]);b.cx(x[3:],x[:3])\n    mul_xor_noscratch(b,x[3:],w,z[3:])\n    erase_inverse(b,N,w);b.free(w);ar._erase_norm(b,x,N,K);b.free(N)\n    linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/inverse20.py': '"""Explicit 20-AND GF64 inverse XOR circuit with exact measurement cleanup.\n\nAll masks below describe physical linear forms. Prefix source order is x0..5,\nsix raw norm rank-products, then the three inverse-norm bits. Constant is bit0\nof a form mask; physical source i is bit(i+1).\n"""\ntry:\n    from .baseline import *\nexcept ImportError:\n    from baseline import *\nfrom contextlib import contextmanager\nimport json\n\nDAG=json.loads(Path(__file__).with_name(\'optimized.json\').read_text())\n\n@contextmanager\ndef expose_pair(b,registers,left,right):\n    u,v=left>>1,right>>1;p=next(bits(u));ops=[]\n    for j in bits(u^(1<<p)):b.cx(registers[j],registers[p]);ops.append((j,p))\n    if v>>p&1:v^=u^(1<<p)\n    q=next(bits(v&~(1<<p)))\n    for j in bits(v^(1<<q)):b.cx(registers[j],registers[q]);ops.append((j,q))\n    if left&1:b.x(registers[p])\n    if right&1:b.x(registers[q])\n    yield registers[p],registers[q]\n    if right&1:b.x(registers[q])\n    if left&1:b.x(registers[p])\n    for j,k in reversed(ops):b.cx(registers[j],registers[k])\n\ndef form_xor(b,registers,mask,target):\n    if mask&1:b.x(target)\n    for i in bits(mask>>1):b.cx(registers[i],target)\n\ndef norm_forms(K):\n    forms=[0]*3;f=field(3)\n    for i,v in enumerate(square_columns(3)):\n        for j in bits(v):forms[j]^=1<<(1+i)\n        for j in bits(f.mul(K,v)):forms[j]^=1<<(1+3+i)\n    for i,(_,v)in enumerate(tensor(3)):\n        for j in bits(v):forms[j]^=1<<(1+6+i)\n    return forms\n\ndef inverse3_virtual(b,registers,forms):\n    out=b.alloc_qubits(3)\n    for mask,vector in enumerate(inverse3_anf()):\n        if not vector:continue\n        ix=list(bits(mask))\n        if len(ix)==1:\n            for j in bits(vector):form_xor(b,registers,forms[ix[0]],out[j])\n        else:\n            assert len(ix)==2\n            with expose_pair(b,registers,forms[ix[0]],forms[ix[1]])as(p,q):rankone_xor(b,p,q,out,vector)\n    return out\n\ndef phase_inverse3_virtual(b,registers,forms,mask):\n    for monomial,vector in enumerate(inverse3_anf()):\n        if not vector:continue\n        ix=list(bits(monomial))\n        if len(ix)==1:\n            assert not(forms[ix[0]]&1)\n            for i in bits(forms[ix[0]]>>1):\n                for j in bits(vector):b.cz(registers[i],mask[j])\n        else:\n            with expose_pair(b,registers,forms[ix[0]],forms[ix[1]])as(p,q):\n                for j in bits(vector):b.ccz(p,q,mask[j])\n\ndef inverse6_xor20(b,x,z=None,keep_norm=False):\n    """z ^= inverse(x), inv0=0; 20T,21 total Q for input6+output6.\n\n    Arbitrary target is supported. If omitted, allocate clean six-bit target.\n    Exactly six norm ANDs, three inverse3 ANDs, eleven final ANDs; cleanup0T.\n    """\n    assert len(x)==6\n    to,frm,K=tower(6);linear_inplace(b,x,to)\n    if z is not None:linear_inplace(b,z,to)\n    g=b.alloc_qubits(6)\n    for i,(mask,_)in enumerate(tensor(3)):\n        with expose_pair(b,x,mask<<1,mask<<4)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(x)+list(g);forms=norm_forms(K)\n    w=inverse3_virtual(b,prefix,forms);source=prefix+list(w)\n    if z is None:z=b.alloc_qubits(6)\n    for j,expression in enumerate(DAG[\'outputs\']):form_xor(b,source,expression&65535,z[j])\n    for i,(left,right)in enumerate(DAG[\'gates\']):\n        vector=sum(((expression>>(16+i))&1)<<j for j,expression in enumerate(DAG[\'outputs\']))\n        assert vector\n        with expose_pair(b,source,left,right)as(p,q):rankone_xor(b,p,q,z,vector)\n    wm=b.alloc_bits(3);b.hmr(w,wm);b.free(w)\n    phase_inverse3_virtual(b,prefix,forms,wm);b.free(wm)\n    kept=None\n    if keep_norm:\n        kept=b.alloc_qubits(3)\n        for j,form in enumerate(forms):form_xor(b,prefix,form,kept[j])\n    for i,(mask,_)in reversed(list(enumerate(tensor(3)))):\n        with expose_pair(b,x,mask<<1,mask<<4)as(p,q):b.del_and(p,q,g[i])\n    b.free(g);linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n    return (z,kept)if keep_norm else z\n\ndef inverse_initialized(b,x,target=None):\n    """No-scratch recursive initialized inverse using the20T base at n6."""\n    n=len(x)\n    if n==3:return inverse3_noscratch(b,x,target)\n    if n==6:return inverse6_xor20(b,x,target)\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    norm=b.alloc_qubits(k);f=field(k)\n    linear_xor(b,x[:k],norm,square_columns(k))\n    linear_xor(b,x[k:],norm,[f.mul(K,v)for v in square_columns(k)])\n    mul_xor_noscratch(b,x[:k],x[k:],norm)\n    invnorm=inverse_initialized(b,norm)\n    ar._erase_norm(b,x,norm,K);b.free(norm)\n    if target is None:target=b.alloc_qubits(n)\n    b.cx(x[k:],x[:k]);mul_xor_noscratch(b,x[:k],invnorm,target[:k]);mul_xor_noscratch(b,x[k:],invnorm,target[k:]);b.cx(x[k:],x[:k])\n    ar._erase_norm(b,target,invnorm,K);b.free(invnorm)\n    linear_inplace(b,x,frm);linear_inplace(b,target,frm)\n    return target\n\ndef norm_xor(b,x,z,K):\n    k=len(z);f=field(k)\n    linear_xor(b,x[:k],z,square_columns(k))\n    linear_xor(b,x[k:],z,[f.mul(K,v)for v in square_columns(k)])\n    mul_xor_noscratch(b,x[:k],x[k:],z)\n\ndef inverse12_keep_norm(b,x):\n    """65T inverse12 keeping its6-bit norm; used by closed reverse24."""\n    assert len(x)==12\n    to,frm,K=tower(12);linear_inplace(b,x,to)\n    N=b.alloc_qubits(6);norm_xor(b,x,N,K);w=inverse6_xor20(b,N)\n    target=b.alloc_qubits(12);b.cx(x[6:],x[:6])\n    mul_xor_noscratch(b,x[:6],w,target[:6]);mul_xor_noscratch(b,x[6:],w,target[6:]);b.cx(x[6:],x[:6])\n    ar._erase_norm(b,target,w,K);b.free(w)\n    linear_inplace(b,x,frm);linear_inplace(b,target,frm)\n    return target,N\n\ndef inverse_inplace_forward20(b,x):\n    target=inverse_initialized(b,x);mask=b.alloc_bits(len(x))\n    b.hmr(x,mask);b.swap(x,target);b.free(target);return mask\n\ndef inverse_inplace_reverse20(b,x,mask):\n    """Closed reverse12/24 at71/215T,21/42 private qubits."""\n    n=len(x);assert n in(12,24);k=n//2\n    to,frm,K=tower(n);linear_inplace(b,x,to)\n    N=b.alloc_qubits(k);norm_xor(b,x,N,K)\n    t,D=inverse12_keep_norm(b,N)if n==24 else inverse6_xor20(b,N,keep_norm=True)\n    out=b.alloc_qubits(k);mul_xor_noscratch(b,x[k:],t,out)\n    erase_product(b,out,N,x[k:]);mul_xor_noscratch(b,x[:k],t,x[k:]);b.cx(out,x[k:])\n    b.cx(out,x[k:]);erase_product(b,x[k:],N,x[:k]);b.cx(out,x[k:])\n    b.swap(x[:k],x[k:]);b.swap(x[k:],out);b.free(out)\n    ar._erase_inverse_with_norm(b,t,N,D);b.free(N)\n    tok,frmk,Kk=tower(k);linear_inplace(b,t,tok)\n    u=b.alloc_qubits(k//2);norm_xor(b,t,u,Kk);erase_inverse(b,u,D);b.free(D)\n    ar._erase_norm(b,t,u,Kk);b.free(u);linear_inplace(b,t,frmk)\n    ar._erase_norm(b,x,t,K);b.free(t);linear_inplace(b,x,frm)\n    b.cz(x,mask);b.free(mask)\n\ndef inverse24_forward20(b,x,K=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols)\n    mask=inverse_inplace_forward20(b,x);linear_inplace(b,x,frm);return mask\n\ndef inverse24_reverse20(b,x,mask,K=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols)\n    inverse_inplace_reverse20(b,x,mask);linear_inplace(b,x,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/inverse20_iso.py': '"""Width-preserving20T GF64 inverse using the lambda2 norm isotopy.\n\nInitialized output reuses three redundant norm directions as output wires.\nNo core modules are edited or monkeypatched.\n"""\ntry:\n    from . import inverse20 as base\n    from .inverse20 import *\nexcept ImportError:\n    import inverse20 as base\n    from inverse20 import *\nDAG=json.loads(Path(__file__).with_name(\'optimized_isotopy2.json\').read_text())\nEXTRA_ROWS=[11,52,34]\nNORM_ROWS=[42,49,54]\nassert [(e>>7)&63 for e in DAG[\'outputs\']]==[11,0,52,34,52,52]\nassert all(not((l|r)&(63<<7))for l,r in DAG[\'gates\'])\n\ndef raw_forms():\n    f=field(3);scale=2;inv=f.invert(scale);forms=[]\n    for mask,_ in tensor(3):\n        left=sum(((mask&f.mul(scale,1<<i)).bit_count()&1)<<(1+i)for i in range(3))\n        right=sum(((mask&f.mul(inv,1<<i)).bit_count()&1)<<(4+i)for i in range(3))\n        forms.append((left,right))\n    return forms\n\ndef xw_form(mask):return(mask&127)|((mask>>6)&(7<<7))\n\ndef erase_extras(b,x,E):\n    masks=b.alloc_bits(3);b.hmr(E,masks);b.free(E)\n    for i,(left,right)in enumerate(raw_forms()):\n        with expose_pair(b,x,left,right)as(p,q):\n            for j,row in enumerate(EXTRA_ROWS):\n                if row>>i&1:b.ccz(p,q,masks[j])\n    b.free(masks)\n\ndef inverse6_xor20(b,x,z=None,keep_norm=False):\n    """20T exact inverse6: fresh-output Q15; arbitrary-target Q18.\n\n    keep_norm=True returns(input_norm3) too, initialized Q18. All private\n    erasures are explicit0T quadratic phase fixes. Zero maps to zero.\n    """\n    assert len(x)==6;fresh=z is None\n    to,frm,K=tower(6);linear_inplace(b,x,to)\n    if not fresh:linear_inplace(b,z,to)\n    g=b.alloc_qubits(6)\n    for i,(left,right)in enumerate(raw_forms()):\n        with expose_pair(b,x,left,right)as(p,q):b.init_and(p,q,g[i])\n    rows=NORM_ROWS+EXTRA_ROWS\n    cols=[sum(((r>>j)&1)<<i for i,r in enumerate(rows))for j in range(6)]\n    linear_inplace(b,g,cols);N,E=g[:3],g[3:]\n    for j,form in enumerate(norm_forms(K)):form_xor(b,x,form&127,N[j])\n    if not fresh:\n        b.cx(E[0],z[0]);b.cx(E[1],[z[2],z[4],z[5]]);b.cx(E[2],z[3])\n        erase_extras(b,x,E)\n    w=inverse3_noscratch(b,N)\n    if fresh:\n        if not keep_norm:ar._erase_norm(b,x,N,K);b.free(N)\n        extra=b.alloc_qubits(3)\n        # Fixed physical layout: these data wires are now output coordinates.\n        z=[E[0],extra[0],E[1],E[2],extra[1],extra[2]]\n        b.cx(z[2],[z[4],z[5]])\n    source=list(x)+list(w)\n    for j,e in enumerate(DAG[\'outputs\']):form_xor(b,source,xw_form(e),z[j])\n    for i,(left,right)in enumerate(DAG[\'gates\']):\n        vector=sum(((e>>(16+i))&1)<<j for j,e in enumerate(DAG[\'outputs\']))\n        with expose_pair(b,source,xw_form(left),xw_form(right))as(p,q):rankone_xor(b,p,q,z,vector)\n    if fresh:\n        ar._erase_norm(b,z,w,K);b.free(w)\n    else:\n        erase_inverse(b,N,w);b.free(w)\n        if not keep_norm:ar._erase_norm(b,x,N,K);b.free(N)\n    linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n    return(z,N)if keep_norm else z\n\n# These higher-level schedules are unchanged structurally. Rebind their private\n# globals to this plugin\'s inverse6 implementation; no imported module changes.\nfrom types import FunctionType\nfor _name in (\'inverse_initialized\',\'inverse12_keep_norm\',\'inverse_inplace_forward20\',\n              \'inverse_inplace_reverse20\',\'inverse24_forward20\',\'inverse24_reverse20\'):\n    _original=getattr(base,_name)\n    globals()[_name]=FunctionType(_original.__code__,globals(),_name,_original.__defaults__,_original.__closure__)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/optimized.json': '{\n  "prefix_AND": 9,\n  "final_AND": 11,\n  "total_AND": 20,\n  "valid_subspaces": 9,\n  "examined_subspaces": 286,\n  "constraints": [\n    3640\n  ],\n  "final_classes": [\n    1,\n    64,\n    130,\n    260,\n    520,\n    1040,\n    2080,\n    2,\n    4,\n    1664,\n    3072\n  ],\n  "gates": [\n    [\n      18,\n      8192\n    ],\n    [\n      16,\n      8192\n    ],\n    [\n      14,\n      57344\n    ],\n    [\n      8,\n      32768\n    ],\n    [\n      10,\n      40960\n    ],\n    [\n      12,\n      49152\n    ],\n    [\n      6,\n      24576\n    ],\n    [\n      126,\n      57344\n    ],\n    [\n      72,\n      32768\n    ],\n    [\n      11240,\n      24088\n    ],\n    [\n      13088,\n      50928\n    ]\n  ],\n  "outputs": [\n    106226362,\n    73483204,\n    98583492,\n    100721338,\n    67257284,\n    93078468\n  ],\n  "prefix_signal_names": [\n    "1",\n    "x0",\n    "x1",\n    "x2",\n    "x3",\n    "x4",\n    "x5",\n    "g0",\n    "g1",\n    "g2",\n    "g3",\n    "g4",\n    "g5",\n    "w0",\n    "w1",\n    "w2"\n  ]\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/optimized_isotopy2.json': '{\n  "prefix_AND": 9,\n  "final_AND": 11,\n  "total_AND": 20,\n  "isotopy_scale": 2,\n  "valid_subspaces": 15,\n  "examined_subspaces": 421,\n  "constraints": [\n    3598\n  ],\n  "final_classes": [\n    1,\n    32,\n    64,\n    520,\n    4095,\n    16,\n    128,\n    256,\n    3024,\n    642,\n    2308\n  ],\n  "gates": [\n    [\n      18,\n      8192\n    ],\n    [\n      54,\n      24576\n    ],\n    [\n      16,\n      8192\n    ],\n    [\n      10,\n      40960\n    ],\n    [\n      4,\n      16384\n    ],\n    [\n      108,\n      49152\n    ],\n    [\n      112,\n      57344\n    ],\n    [\n      64,\n      32768\n    ],\n    [\n      16396,\n      32812\n    ],\n    [\n      16478,\n      40974\n    ],\n    [\n      24584,\n      32816\n    ]\n  ],\n  "outputs": [\n    38430132,\n    2293760,\n    117881380,\n    27586910,\n    132102692,\n    119257636\n  ],\n  "prefix_signal_names": [\n    "1",\n    "x0",\n    "x1",\n    "x2",\n    "x3",\n    "x4",\n    "x5",\n    "g0",\n    "g1",\n    "g2",\n    "g3",\n    "g4",\n    "g5",\n    "w0",\n    "w1",\n    "w2"\n  ]\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_width/kernel.py': '"""39-private paired GF24 inverse via a three-bit norm cache.\n\nForward191T is unchanged iso20. Reverse220T adds five Toffolis while\nreducing reverse private workspace42->39. Explicit transcript inclinv0.\nNo stable module is edited or monkeypatched.\n"""\nimport sys,json\nfrom pathlib import Path\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse_search import inverse20_iso as iso\nfrom exp.inverse_search import inverse20 as ref\nfrom exp.inverse_search.inverse20_iso import *\nfrom exp.arithmetic.lowwidth import classical_transpose\nPART=json.loads(Path(__file__).with_name(\'partial5.json\').read_text())\nFULL=json.loads((ROOT/\'exp/inverse_search/optimized.json\').read_text())\n\ndef coefficient_xor(b,t,nm,e,anc):\n    """anc ^=dot(nm,conjugate(t)*e), e GF64 public, t tower12."""\n    f6=field(6)\n    for i in range(6):\n        v=f6.mul(e,1<<i)\n        for r in bits(v):\n            b.ccx(nm[r],t[i],anc);b.ccx(nm[r],t[i+6],anc);b.ccx(nm[r+6],t[i+6],anc)\n\ndef phase_partial_inverse6(b,u,t,nm,cm):\n    """14T: quantum-mask upper3(inv(u)), classical-mask lower3(inv(u)).\n\n    u canonicalGF64, t alreadytower12; nm12 classical, cm3classical.\n    Upper/lower refer to the standardGF8 tower6, not polynomialGF64 bits.\n    All u/norm/inverse prefix state and the one parity ancilla are restored.\n    """\n    to,frm,K=tower(6);linear_inplace(b,u,to);g=b.alloc_qubits(6)\n    for i,(mask,_) in enumerate(tensor(3)):\n        with ref.expose_pair(b,u,mask<<1,mask<<4)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(u)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms);source=prefix+list(w)\n    # Retained lower3 cache\'s ordinary classical HMR correction costs0T.\n    for j,e in enumerate(FULL[\'outputs\'][:3]):\n        assert not(e&1)\n        for r in bits(e&65535):b.cz(source[r-1],cm[j])\n    cb=b.alloc_bits(1)[0];b.bit_store(cb,False)\n    for i,(left,right)in enumerate(FULL[\'gates\']):\n        js=[j for j,e in enumerate(FULL[\'outputs\'][:3]) if e>>(16+i)&1]\n        if not js:continue\n        for j in js:b.cx(cm[j],cb)\n        with ref.expose_pair(b,source,left,right)as(p,q):b.ccz(p,q,cb)\n        for j in reversed(js):b.cx(cm[j],cb)\n    b.free(cb)\n    # Exactly five products span the three missing output functions.\n    anc=b.alloc_qubits(1)[0]\n    def coeff(v):coefficient_xor(b,t,nm,apply(frm,v<<3),anc)\n    for r in range(16):\n        v=sum(((e>>r)&1)<<j for j,e in enumerate(PART[\'outputs\']))\n        if not v:continue\n        coeff(v)\n        if r:b.cz(anc,source[r-1])\n        else:b.z(anc)\n        coeff(v)\n    for i,(left,right)in enumerate(PART[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(PART[\'outputs\']))\n        assert v\n        coeff(v)\n        with ref.expose_pair(b,source,left,right)as(p,q):b.ccz(p,q,anc)\n        coeff(v)\n    b.free(anc)\n    wm=b.alloc_bits(3);b.hmr(w,wm);b.free(w);ref.phase_inverse3_virtual(b,prefix,forms,wm);b.free(wm)\n    for i,(mask,_)in reversed(list(enumerate(tensor(3)))):\n        with ref.expose_pair(b,u,mask<<1,mask<<4)as(p,q):b.del_and(p,q,g[i])\n    b.free(g);linear_inplace(b,u,frm)\n\ndef compress_D(b,N,D):\n    """Retain lower3 tower6 coordinates, eraseupper3 with quadratic norm phase."""\n    to6,frm6,K6=tower(6);linear_inplace(b,D,to6);m=b.alloc_bits(3);b.hmr(D[3:],m);b.free(D[3:])\n    mt=b.alloc_bits(6);b.bit_store(mt,False);b.cx(m,mt[3:]);canon=classical_transpose(b,mt,to6);b.free(mt);b.free(m)\n    to12,frm12,K12=tower(12);linear_inplace(b,N,to12);ar._phase_norm(b,N,canon,K12);linear_inplace(b,N,frm12);b.free(canon)\n    return D[:3]\n\ndef erase_N_cache(b,t,N,C):\n    """Joint cleanup29T: norm(t)15 +sharedprefix9 +fivequantumCCZ."""\n    to12,frm12,K12=tower(12);linear_inplace(b,t,to12);linear_inplace(b,N,to12)\n    nm=b.alloc_bits(12);b.hmr(N,nm);b.free(N);frm6=tower(6)[1];f6=field(6)\n    for j in range(3):\n        for i in range(6):\n            for r in bits(f6.mul(frm6[j],1<<i)):\n                b.ccz(C[j],t[i],nm[r]);b.ccz(C[j],t[i+6],nm[r]);b.ccz(C[j],t[i+6],nm[r+6])\n    cm=b.alloc_bits(3);b.hmr(C,cm);b.free(C)\n    u=b.alloc_qubits(6);iso.norm_xor(b,t,u,K12);phase_partial_inverse6(b,u,t,nm,cm);b.free(nm);b.free(cm)\n    ar._erase_norm(b,t,u,K12);b.free(u);linear_inplace(b,t,frm12)\n\ndef inverse_inplace_reverse39(b,x,mask):\n    assert len(x)==24;k=12;to,frm,K=tower(24);linear_inplace(b,x,to)\n    N=b.alloc_qubits(12);iso.norm_xor(b,x,N,K);t,D=iso.inverse12_keep_norm(b,N);C=compress_D(b,N,D)\n    out=b.alloc_qubits(12);mul_xor_noscratch(b,x[12:],t,out)\n    erase_product(b,out,N,x[12:]);mul_xor_noscratch(b,x[:12],t,x[12:]);b.cx(out,x[12:])\n    b.cx(out,x[12:]);erase_product(b,x[12:],N,x[:12]);b.cx(out,x[12:])\n    b.swap(x[:12],x[12:]);b.swap(x[12:],out);b.free(out)\n    erase_N_cache(b,t,N,C);ar._erase_norm(b,x,t,K);b.free(t);linear_inplace(b,x,frm);b.cz(x,mask);b.free(mask)\n\ndef inverse24_forward39(b,x,K=None):return iso.inverse24_forward20(b,x,K)\n\ndef inverse24_reverse39(b,x,mask,K=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols);inverse_inplace_reverse39(b,x,mask);linear_inplace(b,x,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_width/partial5.json': '{\n  "prefix_AND": 9,\n  "final_AND": 5,\n  "outputs": [\n    1630906,\n    1131460,\n    458752\n  ],\n  "gates": [\n    [\n      16,\n      8192\n    ],\n    [\n      32,\n      16384\n    ],\n    [\n      80,\n      40960\n    ],\n    [\n      11240,\n      24088\n    ],\n    [\n      13088,\n      50928\n    ]\n  ],\n  "classes": [\n    64,\n    4032,\n    512,\n    1664,\n    3072\n  ],\n  "input_basis": "tower6 GF8 pair",\n  "output_bits": [\n    3,\n    4,\n    5\n  ],\n  "truth_table_values": 64,\n  "truth_table_errors": 0\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_width_subspaces/inverse19.json': '{\n  "total_AND": 19,\n  "prefix_AND": 9,\n  "final_AND": 10,\n  "scale": 6,\n  "classes": [\n    65,\n    2048,\n    63,\n    1885,\n    1154,\n    324,\n    2208,\n    1552,\n    1690,\n    3839\n  ],\n  "gates": [\n    [\n      2,\n      8192\n    ],\n    [\n      48,\n      24576\n    ],\n    [\n      36,\n      16384\n    ],\n    [\n      13508,\n      17080\n    ],\n    [\n      8302,\n      49166\n    ],\n    [\n      8200,\n      32784\n    ],\n    [\n      24688,\n      32886\n    ],\n    [\n      24668,\n      40972\n    ],\n    [\n      21468,\n      40156\n    ],\n    [\n      16448,\n      32772\n    ]\n  ],\n  "outputs": [\n    29546338,\n    46486470,\n    11297528,\n    54342028,\n    20694494,\n    39494616\n  ],\n  "truth_table_points": 64,\n  "truth_table_errors": 0,\n  "raw_norm_control_rank": 4,\n  "raw_norm_control_basis": [\n    41,\n    5,\n    14,\n    16\n  ],\n  "controls_only_through_norm": false\n}\n', 'blocks/dqi/pro-q2-897/source/exp/inverse_width_subspaces/kernel19.py': '"""New19-AND GF64 inverse at norm isotopy6; I24 pair190/219T,39private.\n\nNo imported module is changed. The six-bit XOR inverse supports arbitrary\noutput targets; initialized totalQ21 is wider than iso20, but nested I24 peaks\nremain36forward/39reverse because the larger subcall is not the global peak.\n"""\nimport sys,json\nfrom pathlib import Path\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse_search import inverse20 as ref\nfrom exp.inverse_search import inverse20_iso as iso\nfrom exp.inverse_search.inverse20_iso import *\nfrom exp.inverse_width.kernel import compress_D,erase_N_cache\nDAG19=json.loads(Path(__file__).with_name(\'inverse19.json\').read_text())\n\ndef raw_forms19():\n    f=field(3);scale=6;inv=f.invert(scale);out=[]\n    for mask,v in tensor(3):\n        left=sum(((mask&f.mul(scale,1<<i)).bit_count()&1)<<(1+i)for i in range(3))\n        right=sum(((mask&f.mul(inv,1<<i)).bit_count()&1)<<(4+i)for i in range(3));out.append((left,right))\n    return out\n\ndef inverse6_xor19(b,x,z=None,keep_norm=False):\n    assert len(x)==6;to,frm,K=tower(6);linear_inplace(b,x,to)\n    if z is not None:linear_inplace(b,z,to)\n    g=b.alloc_qubits(6)\n    for i,(left,right)in enumerate(raw_forms19()):\n        with ref.expose_pair(b,x,left,right)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(x)+list(g);forms=ref.norm_forms(K);w=ref.inverse3_virtual(b,prefix,forms);source=prefix+list(w)\n    if z is None:z=b.alloc_qubits(6)\n    for j,e in enumerate(DAG19[\'outputs\']):ref.form_xor(b,source,e&65535,z[j])\n    for i,(left,right)in enumerate(DAG19[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(DAG19[\'outputs\']));assert v\n        with ref.expose_pair(b,source,left,right)as(p,q):rankone_xor(b,p,q,z,v)\n    wm=b.alloc_bits(3);b.hmr(w,wm);b.free(w);ref.phase_inverse3_virtual(b,prefix,forms,wm);b.free(wm)\n    kept=None\n    if keep_norm:\n        kept=b.alloc_qubits(3)\n        for j,f in enumerate(forms):ref.form_xor(b,prefix,f,kept[j])\n    for i,(left,right)in reversed(list(enumerate(raw_forms19()))):\n        with ref.expose_pair(b,x,left,right)as(p,q):b.del_and(p,q,g[i])\n    b.free(g);linear_inplace(b,x,frm);linear_inplace(b,z,frm)\n    return (z,kept)if keep_norm else z\n\n# Rebind only this module\'s private copies of the recursive schedules.\nfrom types import FunctionType\ninverse6_xor20=inverse6_xor19\nfor _name in(\'inverse_initialized\',\'inverse12_keep_norm\',\'inverse_inplace_forward20\'):\n    _old=getattr(ref,_name);globals()[_name]=FunctionType(_old.__code__,globals(),_name,_old.__defaults__,_old.__closure__)\n\ndef inverse24_forward19(b,x,K=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols)\n    mask=inverse_inplace_forward20(b,x);linear_inplace(b,x,frm);return mask\n\ndef inverse_inplace_reverse19(b,x,mask):\n    to,frm,K=tower(24);linear_inplace(b,x,to);N=b.alloc_qubits(12);iso.norm_xor(b,x,N,K)\n    t,D=inverse12_keep_norm(b,N);C=compress_D(b,N,D);out=b.alloc_qubits(12);mul_xor_noscratch(b,x[12:],t,out)\n    erase_product(b,out,N,x[12:]);mul_xor_noscratch(b,x[:12],t,x[12:]);b.cx(out,x[12:])\n    b.cx(out,x[12:]);erase_product(b,x[12:],N,x[:12]);b.cx(out,x[12:])\n    b.swap(x[:12],x[12:]);b.swap(x[12:],out);b.free(out)\n    erase_N_cache(b,t,N,C);ar._erase_norm(b,x,t,K);b.free(t);linear_inplace(b,x,frm);b.cz(x,mask);b.free(mask)\n\ndef inverse24_reverse19(b,x,mask,K=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols);inverse_inplace_reverse19(b,x,mask);linear_inplace(b,x,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/jet24/kernel.py': '"""Completed dual-number inversion reference; allinput, explicitextension basis.\nJ(u,v)=(inv0u,v*(inv0u+[u0])²). Forward290T; closedpair686T/49private.\nNo persistent quantum cache; return only two classical masks.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse12_escape import joint as base\nfrom exp.inverse12_escape import zero1\nfrom exp.inverse12_escape.joint import *\nfrom exp.mul99.kernels import mul_xor99_zero,phase_product_quantum99\nfrom exp.arithmetic.lowwidth import classical_transpose,phase_product_quantum\n\n\ndef phase_zero_virtual(b,prefix,forms,w,mask):\n    """zero(prefix norm) phase,0T while its norm3 andinverse3 forms live."""\n    b.z(mask);source=list(prefix)+list(w);f=field(3)\n    for i in range(3):\n        for j in range(3):\n            if f.mul(1<<i,1<<j)&1:\n                with ref.expose_pair(b,source,forms[i],1<<(1+len(prefix)+j))as(p,q):b.ccz(p,q,mask)\n\n\ndef phase_outputs_classical(b,source,outputs,gates,m):\n    for j,e in enumerate(outputs):\n        if e&1:b.z(m[j])\n        for r in bits((e&65535)>>1):b.cz(source[r],m[j])\n    cb=b.alloc_bits(1)[0];b.bit_store(cb,False)\n    for i,(l,r)in enumerate(gates):\n        js=[j for j,e in enumerate(outputs)if e>>(16+i)&1]\n        for j in js:b.cx(m[j],cb)\n        with ref.expose_pair(b,source,l,r)as(p,q):b.ccz(p,q,cb)\n        for j in reversed(js):b.cx(m[j],cb)\n    b.free(cb)\n\n\ndef phase_inverse12_quantum64(b,x,L,zero_mask=None):\n    """Phase L.inv0(x),quantumL12,64T withpaid-innerprefixretained."""\n    to12,frm12,K12=tower(12);linear_inplace(b,x,to12);T=[sum(((frm12[i]>>j)&1)<<i for i in range(12))for j in range(12)];linear_inplace(b,L,T)\n    D=b.alloc_qubits(6);norm_xor_noscratch(b,x,D,K12);to6,frm6,K6=tower(6);linear_inplace(b,D,to6);g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,D,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(D)+list(g);forms=ref.norm_forms(K6);w=ref.inverse3_virtual(b,prefix,forms);source=prefix+list(w);out=b.alloc_qubits(6)\n    for j,e in enumerate(basic.DAG19[\'outputs\']):ref.form_xor(b,source,e&65535,out[j])\n    for i,(l,r)in enumerate(basic.DAG19[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(basic.DAG19[\'outputs\']))\n        with ref.expose_pair(b,source,l,r)as(p,q):rankone_xor(b,p,q,out,v)\n    linear_inplace(b,out,frm6);b.cx(x[6:],x[:6]);phase_product_quantum(b,x[:6],out,L[:6]);phase_product_quantum(b,x[6:],out,L[6:]);b.cx(x[6:],x[:6])\n    if zero_mask is not None:phase_zero_virtual(b,prefix,forms,w,zero_mask)\n    m=b.alloc_bits(6);b.hmr(out,m);b.free(out);mt=classical_transpose(b,m,frm6);b.free(m);phase_outputs_classical(b,source,basic.DAG19[\'outputs\'],basic.DAG19[\'gates\'],mt);b.free(mt)\n    wm=b.alloc_bits(3);b.hmr(w,wm);b.free(w);ref.phase_inverse3_virtual(b,prefix,forms,wm);b.free(wm)\n    for i,(l,r)in reversed(list(enumerate(basic.raw_forms19()))):\n        with ref.expose_pair(b,D,l,r)as(p,q):b.del_and(p,q,g[i])\n    b.free(g);linear_inplace(b,D,frm6);ar._erase_norm(b,x,D,K12);b.free(D);linear_inplace(b,L,inverse_columns(T));linear_inplace(b,x,frm12)\n\n\ndef phase_inverse24_classical106(b,x,mask,zero_mask=None):\n    """Phase mask.inv0(x)+zero_mask*[x0],106T/45private."""\n    to,frm,K=tower(24);linear_inplace(b,x,to);mt=classical_transpose(b,mask,frm);N=b.alloc_qubits(12);norm_xor_noscratch(b,x,N,K);L=b.alloc_qubits(12);f=field(12)\n    def coeff():\n        for j in range(12):\n            for i in range(12):\n                for r in bits(f.mul(1<<i,1<<j)):\n                    b.ccx(mt[r],x[i],L[j]);b.ccx(mt[r],x[i+12],L[j]);b.ccx(mt[r+12],x[i+12],L[j])\n    coeff();phase_inverse12_quantum64(b,N,L,zero_mask);coeff();b.free(L);ar._erase_norm(b,x,N,K);b.free(N);b.free(mt);linear_inplace(b,x,frm)\n\n\ndef inverse24_initialized_zero191(b,x):\n    to,frm,K=tower(24);linear_inplace(b,x,to);N=b.alloc_qubits(12);norm_xor_noscratch(b,x,N,K);t,F=zero1.inverse_zero(b,N);ar._erase_norm(b,x,N,K);b.free(N)\n    z=b.alloc_qubits(24);b.cx(x[12:],x[:12]);mul_xor_noscratch(b,x[:12],t,z[:12]);mul_xor_noscratch(b,x[12:],t,z[12:]);b.cx(x[12:],x[:12]);ar._erase_norm(b,z,t,K);b.free(t);linear_inplace(b,x,frm);linear_inplace(b,z,frm);return z,F\n\n\ndef safe_square(b,x,F,undo=False):\n    sq=square_columns(24)\n    if undo:linear_inplace(b,x,inverse_columns(sq));b.cx(F,x[0])\n    else:b.cx(F,x[0]);linear_inplace(b,x,sq)\n\n\ndef forward_canonical(b,u,v):\n    z,F=inverse24_initialized_zero191(b,u);out=b.alloc_qubits(24);safe_square(b,z,F);mul_xor99_zero(b,v,z,out);safe_square(b,z,F,True)\n    m=b.alloc_bits(24);b.hmr(v,m);safe_square(b,u,F);phase_product_quantum99(b,out,u,m);safe_square(b,u,F,True);b.free(m);b.swap(v,out);b.free(out)\n    mu=b.alloc_bits(24);fz=b.alloc_bits(1);b.hmr(u,mu);b.swap(u,z);b.free(z);b.hmr(F,fz[0]);b.free(F);return mu,fz\n\n\ndef jet_forward(b,u,v,K=None):\n    """290T/49private;returns classical(mu24,zero1),noquantumcache."""\n    cols,frm,_=extension_columns(K);linear_inplace(b,u,cols);linear_inplace(b,v,cols);token=forward_canonical(b,u,v);linear_inplace(b,u,frm);linear_inplace(b,v,frm);return token\n\n\ndef jet_reverse(b,u,v,token,K=None):\n    """396T/49private closedreverse ofjet_forward;106T inversephase."""\n    cols,frm,_=extension_columns(K);linear_inplace(b,u,cols);linear_inplace(b,v,cols);mu2,fz2=forward_canonical(b,u,v);mu,fz=token;b.cz(u,mu);b.free(mu);b.cx(fz,fz2);b.free(fz);phase_inverse24_classical106(b,u,mu2,fz2[0]);b.free(mu2);b.free(fz2);linear_inplace(b,u,frm);linear_inplace(b,v,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/jet24/pro_jet.py': '"""Complete jet kernel:269 forward+389 reverse=658 Toffolis,43 private qubits.\nOnly supplied-library imports and the documented builder API are used.\nTokens contain only classical bits; no per-call state survives elsewhere.\nStatic masks are little endian, with affine bit0=constant; linear maps list columns.\nPlans r/g/p change and retire a source basis, grow outputs, or add one product.\nEvery retired function\'s exact quadratic measurement phase is repaid immediately.\n"""\nfrom field import bits, field, extension_columns, linear_inplace, square_columns, inverse_columns, tower\nfrom exp.mul99.kernels import mul_xor99_zero, phase_product_quantum99\nfrom exp.noscratch.schedules import norm_xor_noscratch\nfrom exp.noscratch.kernels import rankone_xor, mul_xor_noscratch\nfrom field import tensor, apply, linear_xor\nfrom exp.jet24.kernel import safe_square\nimport field as ar\nfrom exp.inverse_search import inverse20 as ref\nLEAF_OLD12 = ([3334, 28416, 16128, 9744], [[86, 118], [84, 112], [76, 96], [70, 126], [64, 122]], 7, 10)\nLEAF1 = ([15939726, 1292428, 48148704, 42382372, 51177556, 58773552], [[130, 530], [326, 658], [474, 542], [220, 840], [456, 664], [176, 304], [318, 512], [310, 690], [236, 776], [394, 638]], 10, 16)\n\ndef forward_canonical(b, u, v):\n    z, F = inverse24(b, u)\n    mu = b.alloc_bits(24)\n    b.hmr(u, mu)\n    b.swap(u, z)\n    b.free(z)\n    out = b.alloc_qubits(24)\n    safe_square(b, u, F)\n    with b.mark(\'multiply v\'): mul_xor99_zero(b, v, u, out)\n    safe_square(b, u, F, True)\n    mv = b.alloc_bits(24)\n    b.hmr(v, mv)\n    b.swap(v, out)\n    b.free(out)\n    fz = b.alloc_bits(1)\n    b.hmr(F, fz[0])\n    b.free(F)\n    return (mu, mv, fz)\n\ndef scale_halves(b, x, t, N):\n    """Scale each half by nonzero t; N is its multiplicative inverse."""\n    for offset in (0, 12):\n        half = x[offset:offset + 12]\n        out = b.alloc_qubits(12)\n        mul_xor_noscratch(b, half, t, out)\n        ar.erase_product(b, out, N, half)\n        b.swap(half, out)\n        b.free(out)\n\ndef erase_inverse_cached(b, t, N, k, F, fz):\n    """Erase N=inv0(t), k=Norm(t), F=[t=0]; include forward flag debt."""\n    to, frm, K = tower(12)\n    linear_inplace(b, t, to)\n    linear_inplace(b, N, to)\n    m = b.alloc_bits(12)\n    b.hmr(N, m)\n    b.free(N)\n    L = b.alloc_qubits(6)\n    f = field(6)\n\n    def coeff():\n        for j in range(6):\n            for i in range(6):\n                for r in bits(f.mul(1 << i, 1 << j)):\n                    b.ccx(m[r], t[i], L[j])\n                    b.ccx(m[r], t[i + 6], L[j])\n                    b.ccx(m[r + 6], t[i + 6], L[j])\n    coeff()\n    phase_inverse6_heldflag17(b, k, L, F, fz[0])\n    coeff()\n    b.free(L)\n    b.free(m)\n    b.free(fz)\n    ar._erase_norm(b, t, k, K)\n    b.free(k)\n    linear_inplace(b, t, frm)\n\ndef start_inverse(b, N):\n    """Keep N; return k=inv0(Norm(N)) and F=[N=0]."""\n    to, frm, K = tower(12)\n    linear_inplace(b, N, to)\n    D = b.alloc_qubits(6)\n    norm_xor_noscratch(b, N, D, K)\n    k, F = inverse6_zero18(b, D)\n    ar._erase_norm(b, N, D, K)\n    b.free(D)\n    linear_inplace(b, N, frm)\n    return (k, F)\n\ndef finish_inverse(b, N, k):\n    """For safe nonzero N, finish t=inv(N), retaining k=Norm(t)."""\n    to, frm, K = tower(12)\n    linear_inplace(b, N, to)\n    t = b.alloc_qubits(12)\n    b.cx(N[6:], N[:6])\n    mul_xor_noscratch(b, N[:6], k, t[:6])\n    mul_xor_noscratch(b, N[6:], k, t[6:])\n    b.cx(N[6:], N[:6])\n    linear_inplace(b, N, frm)\n    linear_inplace(b, t, frm)\n    return t\n\ndef reverse_canonical(b, u, v, token):\n    mu, mv, fz = token\n    to, frm, K = tower(24)\n    linear_inplace(b, u, to)\n    N = b.alloc_qubits(12)\n    with b.mark(\'reverse norm\'): norm_xor_noscratch(b, u, N, K)\n    with b.mark(\'reverse start inverse with flag\'): k, F = start_inverse(b, N)\n    b.cx(u[12:], u[:12])\n    linear_inplace(b, u, frm)\n    safe_square(b, u, F)\n    out = b.alloc_qubits(24)\n    with b.mark(\'reverse multiply v\'): mul_xor99_zero(b, v, u, out)\n    safe_square(b, u, F, True)\n    linear_inplace(b, u, to)\n    b.cx(u[12:], u[:12])\n    vm = b.alloc_bits(24)\n    b.hmr(v, vm)\n    b.swap(v, out)\n    b.free(out)\n    b.cx(F, N[0])\n    b.cx(F, k[0])\n    with b.mark(\'reverse finish inverse\'): t = finish_inverse(b, N, k)\n    linear_inplace(b, v, to)\n    sq = square_columns(12)\n    linear_inplace(b, t, sq)\n    linear_inplace(b, N, sq)\n    with b.mark(\'reverse scale v\'): scale_halves(b, v, t, N)\n    linear_inplace(b, N, inverse_columns(sq))\n    linear_inplace(b, t, inverse_columns(sq))\n    linear_inplace(b, v, frm)\n    linear_inplace(b, u, frm)\n    safe_square(b, u, F)\n    phase_product_quantum99(b, v, u, vm)\n    safe_square(b, u, F, True)\n    b.free(vm)\n    linear_inplace(b, u, to)\n    b.cx(F, u[0])\n    b.cx(u[12:], u[:12])\n    with b.mark(\'reverse scale u\'): scale_halves(b, u, t, N)\n    b.cx(F, u[0])\n    b.cx(F, N[0])\n    b.cx(F, t[0])\n    b.cx(F, k[0])\n    with b.mark(\'reverse cached phase inverse\'): erase_inverse_cached(b, t, N, k, F, fz)\n    ar._erase_norm(b, u, t, K)\n    b.free(t)\n    linear_inplace(b, u, frm)\n    b.cz(u, mu)\n    b.cz(v, mv)\n    b.free(mu)\n    b.free(mv)\n\ndef jet_forward(b, u, v):\n    """Apply J in the decoder basis; return classical (mu, mv, fz)."""\n    cols, frm, _ = extension_columns(None)\n    linear_inplace(b, u, cols)\n    linear_inplace(b, v, cols)\n    token = forward_canonical(b, u, v)\n    linear_inplace(b, u, frm)\n    linear_inplace(b, v, frm)\n    return token\n\ndef jet_reverse(b, u, v, token):\n    """Restore the input and repay every deferred measurement sign."""\n    cols, frm, _ = extension_columns(None)\n    linear_inplace(b, u, cols)\n    linear_inplace(b, v, cols)\n    reverse_canonical(b, u, v, token)\n    linear_inplace(b, u, frm)\n    linear_inplace(b, v, frm)\n\ndef inverse6_zero18_plan(b, x):\n    return inverse6_fixed18(b, x, True)\nPLAN12 = {\'n\': 12, \'terms\': [(33, 2112, 35), (56, 3584, 905), (25, 1600, 362), (55, 3520, 3), (32, 2048, 334), (23, 1472, 397), (21, 1344, 40), (6, 384, 918), (19, 1216, 190), (3, 192, 50), (30, 1920, 656), (29, 1856, 34), (48, 3072, 18), (28, 1792, 87), (44, 2816, 197)], \'affine\': [1, 4, 16, 3, 12, 48, 63, 57, 33, 2, 8, 32], \'reduce\': [65, 129, 3, 256, 1, 512, 4, 8, 26, 55], \'retire\': [24619, 15261, 374, 2885], \'out\': [3216, 2004, 529, 338, 206, 123, 1, 2, 4, 8, 16, 32], \'size\': 10, \'leaves\': [([33, 56, 2112, 3584, 135168, 229376], [9, 42, 585, 2730], LEAF_OLD12), ([55, 32, 3520, 2048, 225280, 131072], [14, 13, 910, 845], LEAF_OLD12), ([21, 6, 1344, 384, 86016, 24576], [22, 62, 1430, 4030], LEAF_OLD12), ([3, 30, 192, 1920, 12288, 122880], [16, 34, 1040, 2210], LEAF_OLD12), ([48, 28, 3072, 1792, 196608, 114688], [23, 5, 1495, 325], LEAF_OLD12)]}\nPLAN24 = {\'n\': 24, \'terms\': [(1776, 1114384, 419755), (1969, 6951184, 1511959), (1697, 332048, 3917), (321, 8257536, 1627580), (2016, 8065296, 1628913), (272, 1315137, 3418), (592, 15793935, 1270639), (3439, 2494223, 177757), (608, 200463, 2329), (3903, 14020608, 628018), (3423, 14089999, 629803), (3855, 15990591, 3908), (3726, 9349354, 1770621), (1207, 12966122, 538163), (3165, 2963690, 1869), (2617, 6701056, 1626702), (1636, 4946154, 1627395), (2282, 10721849, 382), (1877, 9365742, 1084187), (3879, 8165614, 1481953), (1993, 641262, 1299), (2162, 16494592, 604154), (4027, 15890670, 602857), (2286, 8857714, 2546), (119, 2089470, 1642329), (714, 3359230, 238442), (820, 3420670, 1033), (701, 1609728, 39987), (393, 2925054, 38970), (510, 2871997, 1891), (2300, 16154471, 1179523), (233, 16314215, 1165379), (3982, 7810919, 412), (2069, 7974912, 903104), (1947, 958311, 902748), (3943, 8476693, 2527), (888, 8947848, 1473400), (984, 11864200, 1664433), (2896, 8554632, 415), (160, 12517376, 917193), (3056, 4032648, 917334), (2184, 655520, 1070)], \'affine\': [755, 2929, 2347, 572, 2191, 2631, 3604, 1538, 3661, 1633, 237, 499, 3208, 725, 2900, 3574, 1837, 3303, 873, 3492, 1637, 2503, 1978, 3771], \'reduce\': [1, 2, 4, 9, 22, 61, 127, 251, 394, 632, 2036, 3457, 4096, 12288, 20480, 61440, 77824, 143360, 286720, 524288, 1056768], \'retire\': [], \'out\': [4910144, 9817152, 3062784, 1540096, 111680, 186368, 266304, 535616, 46080, 25600, 7232, 10304, 65, 1027, 2053, 15, 18, 34, 134, 256, 515, 1, 2, 4], \'size\': 21, \'leaves\': [([321, 272, 2016, 8257536, 6950912, 1114112, 4563402752, 29796335616, 28470935552], [1468, 2801, 3418, 9332966, 6014396, 11475697], LEAF1), ([3903, 3855, 3423, 14020608, 2490368, 15790080, 64676167680, 9932111872, 10200547328], [1330, 3115, 3908, 10971766, 5449010, 12762155], LEAF1), ([2617, 2282, 1636, 6701056, 12963840, 9347072, 38285606912, 62511906816, 53099888640], [590, 1283, 382, 3343152, 2417230, 5256451], LEAF1), ([2162, 2286, 4027, 16494592, 8163328, 9363456, 38352715776, 31490834432, 33436991488], [2042, 745, 2546, 14716424, 8366074, 3052265], LEAF1), ([701, 510, 393, 1609728, 3358720, 2088960, 8556380160, 1996488704, 13757317120], [3123, 2106, 1891, 11864912, 12794931, 8628282], LEAF1), ([2069, 3943, 1947, 7974912, 16310272, 16150528, 66152562688, 38587596800, 66806874112], [1984, 1628, 2527, 14810655, 8128448, 6669916], LEAF1), ([160, 2184, 3056, 12517376, 11862016, 8945664, 36641439744, 14898167808, 48586817536], [3785, 3926, 1070, 11434727, 15507145, 16084822], LEAF1)]}\n\ndef cached_inverse(b, x, plan, with_zero=True):\n    n = plan[\'n\']\n    h = n // 2\n    to, frm, K = tower(n)\n    linear_inplace(b, x, to)\n    terms = plan[\'terms\']\n    with b.mark(\'inverse%d norm\' % n): raw = raw_cache(b, x, plan)\n    D = raw[:h]\n    linear_xor(b, x, D, plan[\'affine\'])\n    with b.mark(\'inverse%d scalar\' % n):\n        value = inverse6_fixed18(b, D, with_zero) if h == 6 else cached_inverse(b, D, PLAN12, with_zero)\n    k, F = value\n    linear_xor(b, x, D, plan[\'affine\'])\n    linear_inplace(b, raw, plan[\'reduce\'])\n    kept = plan[\'size\'] - len(plan[\'retire\'])\n    for q, row in zip(raw[kept:], plan[\'retire\']):\n        mask = b.alloc_bits(1)[0]\n        b.hmr(q, mask)\n        for j in bits(row):\n            left, right, _ = terms[j]\n            with ref.expose_pair(b, x, left << 1, right << 1) as (p, r): b.ccz(p, r, mask)\n        b.free(mask)\n        b.free(q)\n    out = list(raw[:kept]) + list(b.alloc_qubits(n - kept))\n    linear_inplace(b, out, plan[\'out\'])\n    source = list(x) + list(k)\n    with b.mark(\'inverse%d products\' % n):\n        for forms, vectors, leaf in plan[\'leaves\']:\n            outputs, gates, low, offset = leaf\n\n            def remap(form):\n                result = form & 1\n                for j in bits(form >> 1): result ^= forms[j] << 1\n                return result\n            for e, v in zip(outputs, vectors):\n                if e & 1:\n                    for j in bits(v): b.x(out[j])\n                for a in bits((e & (1 << low) - 1) >> 1):\n                    for i in bits(forms[a]):\n                        for j in bits(v): b.cx(source[i], out[j])\n            for i, (left, right) in enumerate(gates):\n                vec = 0\n                for e, v in zip(outputs, vectors):\n                    if e >> offset + i & 1: vec ^= v\n                with ref.expose_pair(b, source, remap(left), remap(right)) as (p, q): rankone_xor(b, p, q, out, vec)\n    ar._erase_norm(b, out, k, K)\n    b.free(k)\n    linear_inplace(b, x, frm)\n    linear_inplace(b, out, frm)\n    return (out, F)\n\ndef inverse24(b, x):\n    return cached_inverse(b, x, PLAN24, True)\ninverse6_zero18 = inverse6_zero18_plan\nTO6 = [1, 17, 37, 54, 49, 30]\nFROM6 = [1, 14, 23, 58, 3, 18]\n\ndef raw_cache(b, x, plan):\n    raw = b.alloc_qubits(plan[\'size\'])\n    for left, right, vec in plan[\'terms\']:\n        with ref.expose_pair(b, x, left << 1, right << 1) as (p, q): rankone_xor(b, p, q, raw, vec)\n    return raw\nW14_JOINT18 = {\'K\': 1, \'scale\': 1, \'raw_gates\': [[2, 16], [14, 112], [8, 64], [10, 80], [12, 96], [6, 48]], \'gates\': [[63688, 65242], [47190, 50256], [42244, 64160], [42064, 65236], [37240, 62320], [11090, 43290], [9360, 33288], [48208, 51212], [8560, 51442]], \'outputs\': [31410304, 13183220, 23895034, 16089926, 14693266, 6608146, 67091], \'AND\': 18}\n\ndef phase_inverse6_heldflag17(b, x, L, F, zero_mask=None):\n    """Apply inverse phase; consume the supplied exact F=[x=0]."""\n    dag = W14_JOINT18\n    frm = FROM6\n    linear_inplace(b, x, TO6)\n    g = b.alloc_qubits(6)\n    for j, (l, r) in enumerate(dag[\'raw_gates\']):\n        with ref.expose_pair(b, x, l, r) as (p, q): b.init_and(p, q, g[j])\n    prefix = list(x) + list(g)\n    forms = ref.norm_forms(1)\n    w = ref.inverse3_virtual(b, prefix, forms)\n    source = prefix + list(w)\n    trans = [sum(((frm[i] >> j & 1) << i for i in range(6))) for j in range(6)]\n    linear_inplace(b, L, trans)\n    for j, e in enumerate(dag[\'outputs\'][:6]):\n        uses_g0 = e >> 16 & 1\n        lin = e & 65535 ^ (1555 if uses_g0 else 0)\n        if lin & 1: b.z(L[j])\n        for r in bits(lin >> 1): b.cz(L[j], source[r])\n        if uses_g0: b.cz(F, L[j])\n    for i, (l, r) in enumerate(dag[\'gates\'][1:], 1):\n        vector = sum(((e >> 16 + i & 1) << j for j, e in enumerate(dag[\'outputs\'][:6])))\n        js = list(bits(vector))\n        if not js: continue\n        pL = js[0]\n        for j in js[1:]: b.cx(L[j], L[pL])\n        with ref.expose_pair(b, source, l, r) as (p, q): b.ccz(p, q, L[pL])\n        for j in reversed(js[1:]): b.cx(L[j], L[pL])\n    if zero_mask is not None: b.cz(F, zero_mask)\n    fm = b.alloc_bits(1)[0]\n    b.hmr(F, fm)\n    b.free(F)\n    b.z(fm)\n    for r in bits(1555 >> 1): b.cz(source[r], fm)\n    with ref.expose_pair(b, source, *dag[\'gates\'][0]) as (p, q): b.ccz(p, q, fm)\n    b.free(fm)\n    m = b.alloc_bits(3)\n    b.hmr(w, m)\n    b.free(w)\n    ref.phase_inverse3_virtual(b, prefix, forms, m)\n    b.free(m)\n    for i, (l, r) in reversed(list(enumerate(dag[\'raw_gates\']))):\n        with ref.expose_pair(b, x, l, r) as (p, q): b.del_and(p, q, g[i])\n    b.free(g)\n    linear_inplace(b, L, inverse_columns(trans))\n    linear_inplace(b, x, frm)\nRAW = [[2, 16], [14, 112], [8, 64], [10, 80], [12, 96], [6, 48]]\nPLAN = [(\'r\', [397, 80, 58, 2, 153, 25, 9, 5, 1], 9, []), (\'g\', 1, None, None), (\'p\', 2948, 4256, 1), (\'r\', [1, 2, 4, 8, 16, 64, 304, 32, 128], 9, []), (\'g\', 1, None, None), (\'p\', 7928, 12528, 2), (\'r\', [1, 2, 4, 8, 16, 32, 160, 64, 256], 9, []), (\'g\', 1, None, None), (\'p\', 19728, 9864, 4), (\'r\', [1, 2, 4, 8, 16, 32, 64, 128, 256], 9, []), (\'p\', 17360, 46220, 7), (\'r\', [1, 2, 4, 8, 16, 32, 64, 256, 128], 8, [None]), (\'p\', 27760, 21746, 13), (\'r\', [1, 2, 4, 8, 16, 32, 64, 128], 7, [None]), (\'p\', 7250, 15002, 27), (\'r\', [1, 2, 4, 8, 16, 32, 64], 5, [(11, [(4, 2), (4, 3), (5, 1), (6, 1), (6, 3)]), None]), (\'g\', 1, None, None), (\'p\', 200, 474, 64), (\'r\', [17, 14, 3, 10, 12], 4, [(104, [(4, 1), (4, 2), (5, 1), (5, 3), (6, 2), (7, 0), (8, 0), (10, 0)])]), (\'p\', 214, 464, 41), (\'r\', [5, 10, 1, 2], 2, [(62, [(4, 2), (4, 3), (5, 1), (5, 3), (6, 1), (6, 2), (7, 0)]), (45, [(1, 0), (3, 0), (3, 2), (4, 0), (4, 1), (4, 2), (4, 3), (5, 0), (5, 3), (5, 4), (6, 0), (6, 1), (6, 2), (6, 5), (7, 0), (7, 2), (7, 3), (7, 6), (8, 0), (8, 1), (8, 2), (8, 3), (8, 7), (9, 5), (9, 6), (9, 7), (10, 0), (10, 1), (10, 4), (10, 5), (10, 7), (10, 8), (11, 0), (11, 2), (11, 3), (11, 5), (11, 9), (11, 10)])]), (\'p\', 208, 468, 59), (\'r\', [1, 2], 0, [(126, [(2, 0), (2, 1), (3, 0), (4, 1), (4, 2), (4, 3), (5, 0), (5, 4), (6, 0), (6, 1), (7, 2), (7, 4), (8, 0), (8, 7), (9, 2), (9, 4), (10, 0), (10, 1), (10, 6), (10, 7), (10, 9), (11, 7), (11, 8), (12, 8), (12, 9)]), (75, [(2, 0), (2, 1), (4, 2), (5, 1), (5, 2), (6, 2), (6, 3), (6, 4), (7, 0), (7, 2), (7, 3), (7, 4), (8, 2), (8, 5), (9, 1), (9, 6), (9, 8), (10, 2), (10, 8)])])]\nOUTPUT = [45, 9, 63, 23, 15, 59, 75]\nLINEAR = [0, 116, 122, 70, 18, 18, 19]\n\ndef inverse6_joint18_12extra(b, x):\n    frm = FROM6\n    linear_inplace(b, x, TO6)\n    raw = b.alloc_qubits(6)\n    for j, (l, r) in enumerate(RAW):\n        with ref.expose_pair(b, x, l, r) as (p, q): b.init_and(p, q, raw[j])\n    w = ref.inverse3_virtual(b, list(x) + list(raw), ref.norm_forms(1))\n    raw = list(raw) + list(w)\n    out = []\n    for kind, a, c, d in PLAN:\n        if kind == \'r\':\n            linear_inplace(b, raw, a)\n            for j, action in enumerate(d, c):\n                if action is None:\n                    out.append(raw[j])\n                    continue\n                coeff, phase = action\n                for k in bits(coeff): b.cx(raw[j], out[k])\n                m = b.alloc_bits(1)[0]\n                b.hmr(raw[j], m)\n                live = list(x) + raw[:c] + raw[j + 1:] + out\n                for i, k in phase:\n                    if i == 0: b.z(m)\n                    elif k == 0 or i == k: b.cz(live[i - 1], m)\n                    else:\n                        b.ccz(live[i - 1], live[k - 1], m)\n                b.free(m)\n                b.free(raw[j])\n            raw = list(raw[:c])\n        elif kind == \'g\': out.extend(b.alloc_qubits(a))\n        else:\n            with ref.expose_pair(b, list(x) + raw, a, c) as (p, q): rankone_xor(b, p, q, out, d)\n    linear_inplace(b, out, OUTPUT)\n    for j, form in enumerate(LINEAR): ref.form_xor(b, x, form, out[j])\n    linear_inplace(b, x, frm)\n    linear_inplace(b, out[:6], frm)\n    return (out[:6], out[6])\n\ndef inverse6_fixed18(b, x, joint=True):\n    return inverse6_joint18_12extra(b, x)\n', 'blocks/dqi/pro-q2-897/source/exp/mask_projection/kernel.py': '"""Classical mask projection onto the70 syndrome coordinates,0 Toffolis.\n\nInterpolate a degree<=70 polynomial F with F(0)=0 through70 selected masks\nin the trace-dual field basis. Pay Tr(sum S_k F_(k+1)) directly, subtract\nF\'s evaluations from every mask, and omit the70 now-zero scan positions.\nAn additive FFT avoids a dense4095x70 classical CNOT transform.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.decoder.common import *\n\nSELECTED=tuple(range(70))\n\ndef ev(poly,x):\n    y=0\n    for a in reversed(poly):y=mul(y,x)^a\n    return y\n\ndef tr(x):\n    y=0\n    for _ in range(12):y^=x;x=mul(x,x)\n    assert y in(0,1)\n    return y\n\n@lru_cache(None)\ndef trace_columns():\n    return tuple(sum(tr(mul(1<<i,1<<j))<<i for i in range(12))for j in range(12))\n\n@lru_cache(None)\ndef subspace_polynomials():\n    out=[[0,1]]\n    for j in range(12):\n        s=out[-1];d=ev(s,1<<j);assert d\n        z=[0]*(2*len(s)-1)\n        for i,c in enumerate(s):z[2*i]^=mul(c,c);z[i]^=mul(d,c)\n        out.append(z)\n    return out\n\n@lru_cache(None)\ndef novel_basis(active):\n    ss=subspace_polynomials();out=[]\n    for i in range(active):\n        p=[1]\n        for j in range(i.bit_length()):\n            if i>>j&1:p=pmul(p,ss[j])\n        assert len(p)==i+1 and p[-1]==1\n        out.append(tuple(p))\n    return tuple(out)\n\ndef to_novel(b,words,active,reverse=False):\n    basis=novel_basis(active)\n    for i in(range(1,active)if reverse else range(active-1,0,-1)):\n        for j,c in enumerate(basis[i][:-1]):cx(b,words[i],words[j],c)\n\ndef additive_fft(b,words,m=12,offset=0,active=71,reverse=False):\n    """Novel-basis coefficients -> values at offset+span(1,...,2^(m-1)).\n\n    Only the first active coefficients may be nonzero on entry. The inverse\n    restores the unused classical wires tozero. Output order is numeric field\n    representation, not primitive-element exponent order.\n    """\n    assert len(words)==1<<m and 0<active<=len(words)\n    if not m:return\n    h=1<<(m-1);lo,hi=words[:h],words[h:];high=max(0,active-h)\n    nextactive=min(active,h);ss=subspace_polynomials()[m-1]\n    c=ev(ss,offset);d=ev(ss,1<<(m-1));assert d\n    if reverse:\n        additive_fft(b,lo,m-1,offset,nextactive,True)\n        additive_fft(b,hi,m-1,offset^(1<<(m-1)),nextactive,True)\n    for j in range(nextactive):\n        if j>=high:b.cx(lo[j],hi[j])\n        elif not reverse:\n            cx(b,hi[j],lo[j],c);scale(b,hi[j],d);b.cx(lo[j],hi[j])\n        else:\n            b.cx(lo[j],hi[j]);scale(b,hi[j],inv(d));cx(b,hi[j],lo[j],c)\n    if not reverse:\n        additive_fft(b,lo,m-1,offset,nextactive)\n        additive_fft(b,hi,m-1,offset^(1<<(m-1)),nextactive)\n\n@lru_cache(None)\ndef interpolation_matrix():\n    alpha=gf.primitive_element\n    return tuple(tuple(power(power(alpha,j),k+1)for k in range(70))for j in SELECTED)\n\ndef residual_masks(b,E,F):\n    beta=1;T=trace_columns()\n    for mask in E:\n        ar.linear_xor(b,F[beta],mask,T);beta=mul(beta,gf.primitive_element)\n    assert beta==1\n\ndef project(b,S,E):\n    assert len(S)==70 and len(E)==4095\n    F=[b.alloc_bits(12)for _ in range(4096)]\n    b.bit_store(combine(F),False)\n    Ti=ar.inverse_columns(trace_columns())\n    for k,j in enumerate(SELECTED):\n        b.cx(E[j],F[k+1]);ar.linear_inplace(b,F[k+1],Ti)\n    linear_words(b,F[1:71],interpolation_matrix(),reverse=True)\n    for s,f in zip(S,F[1:71]):\n        ar.linear_inplace(b,f,trace_columns());b.cz(s,f);ar.linear_inplace(b,f,Ti)\n    to_novel(b,F,71);additive_fft(b,F)\n    residual_masks(b,E,F)\n    return F\n\ndef restore(b,E,F):\n    residual_masks(b,E,F)\n    # F is classical deterministic working memory, not quantum garbage.\n    for word in F:b.free(word)\n', 'blocks/dqi/pro-q2-897/source/exp/mul99/kernels.py': '"""GF24 multiplication: GF16 degree6, 11 evaluations times rank9 =99.\n\nSeparate experimental implementation; does not edit the stable arithmetic API.\n"""\nfrom functools import lru_cache\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2]\nsys.path.insert(0,str(ROOT/\'exp/arithmetic\'))\nimport field as ar\nfrom field import *\n\n\ndef extension_tensor(n,k,d,subterms):\n    f=field(n);sub=field(k);gen=f.pow(f.primitive_element,(2**n-1)//(2**k-1))\n    def ev(v,p):\n        out=0\n        for i in reversed(range(k+1)):out=f.mul(out,p)^((v>>i)&1)\n        return out\n    root=next(f.pow(gen,i) for i in range(1,2**k) if ev(sub.modulus,f.pow(gen,i))==0)\n    embed=[f.pow(root,i) for i in range(k)]\n    cols=[f.mul(v,f.pow(f.primitive_element,j)) for j in range(d) for v in embed]\n    inv=inverse_columns(cols);masks=[]\n    for point in list(range(2*d-2))+[None]:\n        values=[]\n        for co in inv:\n            if point is None:value=(co>>(k*(d-1)))&((1<<k)-1)\n            else:\n                value=0\n                for j in reversed(range(d)):value=sub.mul(value,point)^((co>>(k*j))&((1<<k)-1))\n            values.append(value)\n        for mask,_ in subterms:\n            masks.append(sum(((v&mask).bit_count()%2)<<j for j,v in enumerate(values)))\n    equations=[]\n    for i in range(n):\n        for j in range(i,n):\n            row=sum((((m>>i)&1)&((m>>j)&1))<<r for r,m in enumerate(masks))\n            equations.append((row,f.mul(1<<i,1<<j)))\n    vectors=ar._solve_output_vectors(equations,len(masks));assert all(vectors)\n    return tuple(zip(masks,vectors))\n\n\n@lru_cache(None)\ndef tensor4():return extension_tensor(4,2,2,tensor(2))\n\n\n@lru_cache(None)\ndef tensor99():\n    out=extension_tensor(24,4,6,tensor4());assert len(out)==99\n    return out\n\n\ndef multiplier(b,x,y,z,terms):\n    assert not(set(map(str,x))&set(map(str,y)) or set(map(str,x))&set(map(str,z)) or set(map(str,y))&set(map(str,z)))\n    for mask,v in terms:\n        ix=list(bits(mask));iz=list(bits(v));p=ix[0];t=iz[0]\n        for i in ix[1:]:b.cx(x[i],x[p]);b.cx(y[i],y[p])\n        for j in iz[1:]:b.cx(z[t],z[j])\n        b.ccx(x[p],y[p],z[t])\n        for j in reversed(iz[1:]):b.cx(z[t],z[j])\n        for i in reversed(ix[1:]):b.cx(y[i],y[p]);b.cx(x[i],x[p])\n\n\ndef mul_xor99_zero(b,x,y,z):return multiplier(b,x,y,z,tensor99())\n\n\n@lru_cache(None)\ndef tensor99_extension(K=None):\n    cols,invcols,_=extension_columns(K)\n    return tuple((sum(((mask&col).bit_count()%2)<<j for j,col in enumerate(cols)),apply(invcols,v))\n                 for mask,v in tensor99())\n\n\ndef mul24_xor99_zero(b,x,y,z,K=None):return multiplier(b,x,y,z,tensor99_extension(K))\n\n\ndef phase_product_quantum99(b,x,y,mask):\n    for m,v in tensor99():\n        ix=list(bits(m));iz=list(bits(v));p=ix[0];t=iz[0]\n        for i in ix[1:]:b.cx(x[i],x[p]);b.cx(y[i],y[p])\n        for j in iz[1:]:b.cx(mask[j],mask[t])\n        b.ccz(x[p],y[p],mask[t])\n        for j in reversed(iz[1:]):b.cx(mask[j],mask[t])\n        for i in reversed(ix[1:]):b.cx(y[i],y[p]);b.cx(x[i],x[p])\n\n\n# Stable public spellings: no zero-target promise is required.\nmul_xor99=mul_xor99_zero\nmul24_xor99=mul24_xor99_zero\n', 'blocks/dqi/pro-q2-897/source/exp/mul99/phase36.py': '"""Inverse12 phase with12 scratch and at most36 EXECUTED Toffolis.\n\nThe coefficient mask is classical. A classical first-nonzero pivot exposes\nits parity in an existing input bit, avoiding the13th quantum wire.\n"""\ntry:\n    from .kernels import *\nexcept ImportError:\n    from kernels import *\nsys.path.insert(0,str(ROOT))\nfrom exp.noscratch.kernels import inverse3_noscratch\nfrom lowwidth import classical_transpose\n\n\ndef phase_classical_parity(b,a,c,x,co):\n    """CCZ(a,c,parity(co*x)), zero quantum scratch, <=1 executed CCZ."""\n    first=b.alloc_bits(len(x));b.bit_store(first,False)\n    seen=b.alloc_bits(1)[0];b.bit_store(seen,False)\n    for j in range(len(x)):\n        b.x(seen);b.push_condition(seen)\n        b.bit_store(first[j],True,control=co[j])\n        b.pop_condition();b.x(seen)\n        b.bit_store(seen,True,control=co[j])\n    for j in range(len(x)):\n        b.push_condition(first[j])\n        for k in range(len(x)):\n            if k!=j:b.ccx(co[k],x[k],x[j])\n        b.ccz(a,c,x[j])\n        for k in reversed(range(len(x))):\n            if k!=j:b.ccx(co[k],x[k],x[j])\n        b.pop_condition()\n    b.free(seen);b.free(first)\n\n\ndef phase_inverse36_pivot(b,x,mask):\n    """Phase mask dot inverse0(x): <=36 mean T /12 quantum scratch.\n\n    Static count168T due to12 mutually exclusive classical branches per CCZ.\n    No quantum input controls branch selection; all branches use HMR-mask bits.\n    """\n    assert len(x)==len(mask)==12\n    to12,frm12,K12=tower(12);linear_inplace(b,x,to12)\n    mt=classical_transpose(b,mask,frm12)\n    D=b.alloc_qubits(6);ar._norm_xor(b,x,D,K12)\n    to6,frm6,K6=tower(6);linear_inplace(b,D,to6)\n    E=b.alloc_qubits(3);ar._norm_xor(b,D,E,K6);w=inverse3_noscratch(b,E)\n    f6=field(6)\n    for segment in(0,1):\n        if segment==0:b.cx(D[3:],D[:3])\n        controls=D[:3] if segment==0 else D[3:]\n        for m,v in tensor(3):\n            h=apply(frm6,v<<(3*segment));co=b.alloc_bits(12);b.bit_store(co,False)\n            for j in range(6):\n                for r in bits(f6.mul(h,1<<j)):\n                    b.cx(mt[r],co[j]);b.cx(mt[r],co[j+6]);b.cx(mt[r+6],co[j+6])\n            ix=list(bits(m));p=ix[0]\n            for j in ix[1:]:b.cx(controls[j],controls[p]);b.cx(w[j],w[p])\n            phase_classical_parity(b,controls[p],w[p],x,co)\n            for j in reversed(ix[1:]):b.cx(w[j],w[p]);b.cx(controls[j],controls[p])\n            b.free(co)\n        if segment==0:b.cx(D[3:],D[:3])\n    erase_inverse(b,E,w);b.free(w)\n    ar._erase_norm(b,D,E,K6);b.free(E);linear_inplace(b,D,frm6)\n    ar._erase_norm(b,x,D,K12);b.free(D);linear_inplace(b,x,frm12);b.free(mt)\n', 'blocks/dqi/pro-q2-897/source/exp/mul99/schedules.py': '"""Explicit compositions; no patching or edits of existing arithmetic modules."""\ntry:\n    from .kernels import *\nexcept ImportError:\n    from kernels import *\nsys.path.insert(0,str(ROOT))\nfrom exp.noscratch.schedules import norm_xor_noscratch,inverse_stream_noscratch\nfrom exp.noscratch.phase39 import phase_inverse39\nfrom exp.noscratch.kernels import mul_xor_noscratch\nfrom division import phase_product24,zero_flag,phase_zero\ntry:\n    from .phase36 import phase_inverse36_pivot\nexcept ImportError:\n    from phase36 import phase_inverse36_pivot\n\n\ndef multiply_forward99(b,x,y,K=None):\n    """x*=y:99T/24scratch, old-x phase deferred. Promise x=0 when y=0."""\n    z=b.alloc_qubits(24);mul24_xor99(b,x,y,z,K)\n    m=b.alloc_bits(24);b.hmr(x,m);b.swap(x,z);b.free(z);return m\n\n\ndef divide_lowwidth372(b,x,y,K=None):\n    """x/=y, preserves y:372T/24scratch. Promise x=0 when y=0."""\n    return _divide_lowwidth(b,x,y,K,phase_inverse39)\n\n\ndef divide_lowwidth369(b,x,y,K=None):\n    """Same division, <=369 EXECUTED T /24scratch; static count501T."""\n    return _divide_lowwidth(b,x,y,K,phase_inverse36_pivot)\n\n\ndef _divide_lowwidth(b,x,y,K,phase_kernel):\n    if K is None:K=extension_columns()[2]\n    b.cx(y[12:],y[:12]);oldmask=multiply_forward99(b,x,y,K);b.cx(y[12:],y[:12])\n    u=b.alloc_qubits(12);norm_xor_noscratch(b,y,u,K)\n    nm=inverse_stream_noscratch(b,u);ar._phase_norm(b,y,nm,K);b.free(nm)\n    masks=[]\n    for offset in(0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor_noscratch(b,half,u,out)\n        m=b.alloc_bits(12);b.hmr(half,m);masks.append(m);b.swap(half,out);b.free(out)\n    um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,y,N,K)\n    for offset,m in zip((0,12),masks):phase_product(b,x[offset:offset+12],N,m);b.free(m)\n    phase_kernel(b,N,um);b.free(um)\n    ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,oldmask,K);b.free(oldmask)\n\n\ndef divide_zero_bridge417(b,x,y,K=None,extra_zero_mask=None):\n    """x/=y if y!=0, else identity:417T/25scratch, all inputs exact."""\n    flag=zero_flag(b,y);b.cx(flag,y[0])\n    divide_lowwidth372(b,x,y,K)\n    b.cx(flag,y[0])\n    mask=b.alloc_bits(1);b.hmr(flag,mask[0]);b.free(flag)\n    if extra_zero_mask is not None:b.cx(extra_zero_mask,mask[0])\n    phase_zero(b,y,mask[0]);b.free(mask)\n\n\ndef divide_zero_bridge414(b,x,y,K=None,extra_zero_mask=None):\n    """All-input exact zero bridge: <=414 EXECUTED T /25scratch."""\n    flag=zero_flag(b,y);b.cx(flag,y[0])\n    divide_lowwidth369(b,x,y,K)\n    b.cx(flag,y[0])\n    mask=b.alloc_bits(1);b.hmr(flag,mask[0]);b.free(flag)\n    if extra_zero_mask is not None:b.cx(extra_zero_mask,mask[0])\n    phase_zero(b,y,mask[0]);b.free(mask)\n\n\ndef multiply_zero_forward99(b,x,y,K=None):\n    """x*=y if y!=0, else identity:146T/25scratch; deferred masks."""\n    flag=zero_flag(b,y)\n    z=b.alloc_qubits(24);mul24_xor99(b,x,y,z,K);b.ccx(flag,x,z)\n    old=b.alloc_bits(24);b.hmr(x,old);b.swap(x,z);b.free(z)\n    mask=b.alloc_bits(1);b.hmr(flag,mask[0]);b.free(flag)\n    return old,mask[0]\n', 'blocks/dqi/pro-q2-897/source/exp/noscratch/kernels.py': '"""Scratch-free rankone bilinear accumulation; no coremodule edits."""\nimport sys\nfrom pathlib import Path\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT/\'exp/arithmetic\'))\nimport field as ar\nfrom field import *\n\ndef rankone_xor(b,x,y,z,vector):\n    """z^=vector*(xANDy), arbitraryz, zeroancilla, oneCCX.\n\n    x,y arebits; vector is nonzero bitmaskoverz. Inputsdisjointfromtarget.\n    """\n    out=list(bits(vector));p=out[0]\n    for j in out[1:]:b.cx(z[p],z[j])\n    b.ccx(x,y,z[p])\n    for j in reversed(out[1:]):b.cx(z[p],z[j])\n\ndef tensor_xor_noscratch(b,x,y,z,terms):\n    assert len(x)==len(y)==len(z)\n    assert not(set(map(str,x))&set(map(str,y))or set(map(str,x))&set(map(str,z))or set(map(str,y))&set(map(str,z)))\n    for mask,vector in terms:\n        ix=list(bits(mask));p=ix[0]\n        for j in ix[1:]:b.cx(x[j],x[p]);b.cx(y[j],y[p])\n        rankone_xor(b,x[p],y[p],z,vector)\n        for j in reversed(ix[1:]):b.cx(y[j],y[p]);b.cx(x[j],x[p])\n\ndef mul_xor_noscratch(b,x,y,z):\n    """DisjointGF2^n productaccumulate; n2/3/6/12, rank3/6/15/42,0scratch."""\n    tensor_xor_noscratch(b,x,y,z,tensor(len(x)))\n\ndef mul24_xor_noscratch(b,x,y,z,K=None):\n    """ExplicitquadraticGF24 productaccumulate, arbitrarytarget,105T/0scratch."""\n    assert len(x)==24;tensor_xor_noscratch(b,x,y,z,ar.tensor24_extension(K))\n\ndef inverse3_xor_noscratch(b,x,z):\n    """z^=inverseGF8(x), inverse0=0, arbitrarytarget,3T/0scratch."""\n    assert len(x)==len(z)==3 and not(set(map(str,x))&set(map(str,z)))\n    for mask,vector in enumerate(inverse3_anf()):\n        if not vector:continue\n        ix=list(bits(mask))\n        if len(ix)==1:\n            for j in bits(vector):b.cx(x[ix[0]],z[j])\n        elif len(ix)==2:rankone_xor(b,x[ix[0]],x[ix[1]],z,vector)\n        else:raise ValueError((mask,vector))\n\ndef inverse3_noscratch(b,x,target=None):\n    """Freshcleantargetinverse, or XORintogiventarget,3T/0privatequbits."""\n    if target is None:target=b.alloc_qubits(3)\n    inverse3_xor_noscratch(b,x,target);return target\n', 'blocks/dqi/pro-q2-897/source/exp/noscratch/phase39.py': '"""Inverse GF4096 phase with twelve private qubits, 39 Toffolis.\n\nOnly imports core modules; this plugin never patches or edits them.\n"""\ntry:\n    from .kernels import *\nexcept ImportError:\n    from kernels import *\nfrom lowwidth import classical_transpose\n\n\ndef norm_bit_xor(b,x,z,K,bit=0):\n    """XOR one GF8 norm coordinate into arbitrary single-qubit z, 3T."""\n    assert len(x)==6\n    a,c=x[:3],x[3:];f=field(3)\n    for j,v in enumerate(square_columns(3)):\n        if v>>bit&1:b.cx(a[j],z)\n        if f.mul(K,v)>>bit&1:b.cx(c[j],z)\n    for mask,vector in tensor(3):\n        if not(vector>>bit&1):continue\n        ix=list(bits(mask));p=ix[0]\n        for j in ix[1:]:b.cx(a[j],a[p]);b.cx(c[j],c[p])\n        b.ccx(a[p],c[p],z)\n        for j in reversed(ix[1:]):b.cx(c[j],c[p]);b.cx(a[j],a[p])\n\n\ndef phase_inverse39(b,x,mask):\n    """Apply (-1)^dot(mask,inverse(x)); inv(0)=0, x unchanged.\n\n    E[0] is HMR-erased and its quadratic norm phase repaid before the\n    mask-parity qubit is allocated. Once that qubit is freed, E[0] is\n    reconstructed with 3 CCX, enabling the ordinary inverse erasure.\n    Peak D6 + E2 + w3 + parity1 = 12 private quantum wires.\n    """\n    assert len(x)==len(mask)==12\n    to12,frm12,K12=tower(12);linear_inplace(b,x,to12)\n    mt=classical_transpose(b,mask,frm12)\n    D=b.alloc_qubits(6);ar._norm_xor(b,x,D,K12)\n    to6,frm6,K6=tower(6);linear_inplace(b,D,to6)\n    E=b.alloc_qubits(3);ar._norm_xor(b,D,E,K6)\n    w=inverse3_noscratch(b,E)\n    em=b.alloc_bits(3);b.bit_store(em,False);b.hmr(E[:1],em[:1])\n    ar._phase_norm(b,D,em,K6);b.free(em);b.free(E[:1])\n    anc=b.alloc_qubits(1)[0];f6=field(6)\n    def coefficient(h):\n        for j in range(6):\n            for r in bits(f6.mul(h,1<<j)):\n                b.ccx(mt[r],x[j],anc)\n                b.ccx(mt[r],x[j+6],anc)\n                b.ccx(mt[r+6],x[j+6],anc)\n    for segment in (0,1):\n        if segment==0:b.cx(D[3:],D[:3])\n        controls=D[:3] if segment==0 else D[3:]\n        for m,v in tensor(3):\n            h=apply(frm6,v<<(3*segment));ix=list(bits(m));p=ix[0]\n            coefficient(h)\n            for j in ix[1:]:b.cx(controls[j],controls[p]);b.cx(w[j],w[p])\n            b.ccz(controls[p],w[p],anc)\n            for j in reversed(ix[1:]):b.cx(w[j],w[p]);b.cx(controls[j],controls[p])\n            coefficient(h)\n        if segment==0:b.cx(D[3:],D[:3])\n    b.free(anc)\n    e0=b.alloc_qubits(1);norm_bit_xor(b,D,e0[0],K6)\n    E=[e0[0],E[1],E[2]]\n    erase_inverse(b,E,w);b.free(w)\n    ar._erase_norm(b,D,E,K6);b.free(E);linear_inplace(b,D,frm6)\n    ar._erase_norm(b,x,D,K12);b.free(D)\n    linear_inplace(b,x,frm12);b.free(mt)\n', 'blocks/dqi/pro-q2-897/source/exp/noscratch/schedules.py': '"""Reference compositions of the plugins; no changes to arithmetic modules."""\ntry:\n    from .phase39 import *\nexcept ImportError:\n    from phase39 import *\nfrom division import phase_product24\n\n\ndef norm_xor_noscratch(b,x,z,K):\n    k=len(z);f=field(k)\n    linear_xor(b,x[:k],z,square_columns(k))\n    linear_xor(b,x[k:],z,[f.mul(K,v)for v in square_columns(k)])\n    mul_xor_noscratch(b,x[:k],x[k:],z)\n\n\ndef inverse_stream_noscratch(b,x):\n    """Same streamed inverse T3/6/12/24=3/21/66/192, n scratch."""\n    n=len(x)\n    if n==3:\n        y=inverse3_noscratch(b,x);mask=b.alloc_bits(n)\n        b.hmr(x,mask);b.swap(x,y);b.free(y);return mask\n    k=n//2;to,frm,K=tower(n);linear_inplace(b,x,to)\n    N=b.alloc_qubits(k);norm_xor_noscratch(b,x,N,K)\n    nm=inverse_stream_noscratch(b,N);ar._phase_norm(b,x,nm,K);b.free(nm)\n    outb=b.alloc_qubits(k);mul_xor_noscratch(b,x[k:],N,outb)\n    raw=b.alloc_bits(n);b.hmr(x[k:],raw[k:])\n    mul_xor_noscratch(b,x[:k],N,x[k:]);b.cx(outb,x[k:]);b.hmr(x[:k],raw[:k])\n    b.swap(x[:k],x[k:]);b.swap(x[k:],outb);b.free(outb)\n    ar._erase_norm(b,x,N,K);b.free(N);linear_inplace(b,x,frm)\n    mask=classical_transpose(b,raw,to);b.free(raw);return mask\n\n\ndef multiply_forward_noscratch(b,x,y,K=None):\n    z=b.alloc_qubits(24);mul24_xor_noscratch(b,x,y,z,K)\n    m=b.alloc_bits(24);b.hmr(x,m);b.swap(x,z);b.free(z);return m\n\n\ndef divide_lowwidth378(b,x,y,K=None):\n    """378T/24 scratch, promise x=0 when y=0; preserves y."""\n    if K is None:K=extension_columns()[2]\n    b.cx(y[12:],y[:12]);oldmask=multiply_forward_noscratch(b,x,y,K);b.cx(y[12:],y[:12])\n    u=b.alloc_qubits(12);norm_xor_noscratch(b,y,u,K)\n    nm=inverse_stream_noscratch(b,u);ar._phase_norm(b,y,nm,K);b.free(nm)\n    masks=[]\n    for offset in(0,12):\n        half=x[offset:offset+12];out=b.alloc_qubits(12);mul_xor_noscratch(b,half,u,out)\n        m=b.alloc_bits(12);b.hmr(half,m);masks.append(m);b.swap(half,out);b.free(out)\n    um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,y,N,K)\n    for offset,m in zip((0,12),masks):phase_product(b,x[offset:offset+12],N,m);b.free(m)\n    phase_inverse39(b,N,um);b.free(um)\n    ar._erase_norm(b,y,N,K);b.free(N)\n    phase_product24(b,x,y,oldmask,K);b.free(oldmask)\n', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/prefix.py': '"""Closed local q-prefix step with the historical swap zero chart.\n\nInputs (b,p,q) -> (q*b+p,p,q) when q!=0, or (p,b,q) whenq=0.\nForward65T, reverse152T using divide117; pair217T,29private peak.\nThe reverse is exact for all (b,p,q), not only adjacent-nonzero continuants.\n"""\ntry:\n    from .scalar import *\nexcept ImportError:\n    from scalar import *\n\ndef zero_hold(b,x):\n    b.x(x);lad=b.alloc_qubits(len(x)-1);b.init_and(x[0],x[1],lad[0])\n    for j in range(2,len(x)):b.init_and(lad[j-2],x[j],lad[j-1])\n    b.x(x);return lad\n\ndef zero_free(b,x,lad):\n    b.x(x)\n    for j in reversed(range(2,len(x))):b.del_and(lad[j-2],x[j],lad[j-1])\n    b.del_and(x[0],x[1],lad[0]);b.free(lad);b.x(x)\n\ndef forward(b,x,p,q,virtual_zero=None,after_product=None):\n    out=b.alloc_qubits(12);mul_xor0(b,x,q,out);b.cx(p,out)\n    if virtual_zero is not None:b.ccx(virtual_zero,p,out)\n    lad=zero_hold(b,q);b.cswap(lad[-1],x,p)\n    m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out)\n    if after_product is not None:after_product()\n    zero_free(b,q,lad);return m\n\ndef reverse(b,x,p,q,m):\n    lad=zero_hold(b,q);f=lad[-1];b.cswap(f,x,p);b.cx(p,x);b.ccx(f,p,x)\n    b.cx(f,q[0]);divide117(b,x,q);b.cx(f,q[0])\n    b.cz(x,m)\n    for j in range(12):b.ccz(f,x[j],m[j]);b.ccz(f,p[j],m[j])\n    b.free(m);zero_free(b,q,lad)\n\ndef zero_compact(b,x):\n    lad=zero_hold(b,x);b.x(x)\n    for j in reversed(range(2,len(x)-1)):b.del_and(lad[j-2],x[j],lad[j-1]);b.free(lad[j-1])\n    b.del_and(x[0],x[1],lad[0]);b.free(lad[0]);b.x(x)\n    return lad[-1]\n\ndef zero_compact_free(b,x,f):\n    b.x(x);lad=b.alloc_qubits(len(x)-2);b.init_and(x[0],x[1],lad[0])\n    for j in range(2,len(x)-1):b.init_and(lad[j-2],x[j],lad[j-1])\n    b.del_and(lad[-1],x[-1],f);b.free(f)\n    for j in reversed(range(2,len(x)-1)):b.del_and(lad[j-2],x[j],lad[j-1])\n    b.del_and(x[0],x[1],lad[0]);b.free(lad);b.x(x)\n\ndef reverse_compact(b,x,p,q,m,virtual_zero=None):\n    """162T reverse,19private; paired with forward gives227T/23private."""\n    f=zero_compact(b,q);b.cswap(f,x,p);b.cx(p,x);b.ccx(f,p,x)\n    if virtual_zero is not None:b.ccx(virtual_zero,p,x)\n    b.cx(f,q[0]);divide117(b,x,q);b.cx(f,q[0]);b.cz(x,m)\n    for j in range(12):b.ccz(f,x[j],m[j]);b.ccz(f,p[j],m[j])\n    b.free(m);zero_compact_free(b,q,f)\n\ndef boundary_forward(b,x,q,F,after_product=None):\n    """i=1 special chart; F=zero(original q1),66T,12private."""\n    b.cx(F,q[0]);out=b.alloc_qubits(12);mul_xor0(b,x,q,out)\n    m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out);b.cx(F,q[0]);b.x(x[0])\n    if after_product is not None:after_product()\n    b.ccx(F,x,q);b.cx(F,q[0]);b.ccx(F,q,x)\n    return m\n\ndef boundary_reverse(b,x,q,F,m):\n    """129T,18private; exact inverse, closes boundary_forward\'s mask."""\n    b.cswap(F,q,x);b.cx(F,q[0]);b.x(x[0]);b.cx(F,x[0])\n    b.cx(F,q[0]);divide117(b,x,q);b.cx(F,q[0]);b.cz(x,m);b.free(m)\n', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/scalar.py': '"""Retained-norm GF12 division:117T,18 private, complete local HMR phase."""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT/\'exp/arithmetic\'))\nfrom field import *\nimport field as ar\nfrom lowwidth import phase_inverse_generic6\n\ndef divide117(b,x,y):\n    """x/=y, y preserved; y=0 requires x=0. 117T/18 private qubits."""\n    f=field(12);to,frm,K=tower(12);conj=[f.frobenius(1<<i,6) for i in range(12)]\n    linear_inplace(b,y,conj);out=b.alloc_qubits(12);mul_xor0(b,x,y,out)\n    old=b.alloc_bits(12);b.hmr(x,old);b.swap(x,out);b.free(out);linear_inplace(b,y,conj)\n    N=b.alloc_qubits(6);linear_inplace(b,y,to);ar._norm_xor(b,y,N,K);linear_inplace(b,y,frm)\n    u=inverse(b,N);linear_inplace(b,x,to)\n    for off in(0,6):\n        half=x[off:off+6];out=b.alloc_qubits(6);mul_xor0(b,half,u,out)\n        m=b.alloc_bits(6);b.hmr(half,m);b.swap(half,out);b.free(out);phase_product(b,half,N,m);b.free(m)\n    um=b.alloc_bits(6);b.hmr(u,um);b.free(u);phase_inverse_generic6(b,N,um);b.free(um)\n    linear_inplace(b,y,to);ar._erase_norm(b,y,N,K);linear_inplace(b,y,frm);b.free(N)\n    linear_inplace(b,x,frm);phase_product(b,x,y,old);b.free(old)\n', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/synthesis_noflag.py': '"""Partial-q-gauge synthesis. Same public A,B,transcript API as decoder.\n\n70 input field words plus one rolling d field word and one boundary flag.\nOnly lower A/B coefficients are stored; implicit high coefficients are aliases\nof earlier q-prefix slots. extra[0] is the physical A[h] zero on valid heads.\nNo edits to the live decoder; local measurement transcript is fully closed.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT));sys.path.insert(0,str(ROOT/\'exp/decoder\'))\nfrom common import *\nfrom extract import constants\nfrom exp.decoder import synthesis as base\nfrom exp.partial_gauge import prefix as pg\nfrom exp.partial_gauge.scalar import divide117\nfrom exp.partial_gauge.tail import divide_safe138,multiply_safe165,controlled_linear_xor\nfrom exp.mul99.schedules import divide_zero_bridge414\nimport division as dv\nck=base.ck\n\ndef xword(b,w,v):\n    for j in ar.bits(v):b.x(w[j])\n\ndef add_alias(b,src,targets,coeffs,virtual_zero=None,constant=0):\n    if src is None:\n        for z,c in zip(targets,coeffs):xword(b,z,mul(constant,c))\n    else:\n        for z,c in zip(targets,coeffs):cx(b,src,z,c)\n        if virtual_zero is not None:controlled_linear_xor(b,virtual_zero,src,targets,coeffs)\n\ndef eval_implicit(b,A,qi,c,reverse=False):\n    if len(A)==1:\n        assert qi is not None\n        return base.expose_eval(b,A+[qi],c,reverse)\n    v=base.evaluation_coefficients(len(A)+1,c)[-1]\n    if reverse:\n        add_alias(b,qi,A[:2],[v&4095,v>>12],constant=1)\n        return base.expose_eval(b,A,c,True)\n    out=base.expose_eval(b,A,c)\n    add_alias(b,qi,A[:2],[v&4095,v>>12],constant=1);return out\n\ndef rank_affine(b,head,ac,words,bc,target,tc,constant=0):\n    if not any(tc):return\n    x,ap=ck.expose(b,head,ac);y,bp=ck.expose(b,words,bc)\n    xword(b,y,constant)\n    p=next(i for i,c in enumerate(tc) if c);v=tc[p];vi=inv(v)\n    for i,c in enumerate(tc):\n        if i!=p:cx(b,target[p],target[i],mul(c,vi))\n    scale(b,target[p],vi);ar.mul_xor0(b,x,y,target[p]);scale(b,target[p],v)\n    for i in reversed(range(len(tc))):\n        if i!=p:cx(b,target[p],target[i],mul(tc[i],vi))\n    xword(b,y,constant);ck.unexpose(b,words,bc,bp);ck.unexpose(b,head,ac,ap)\n\ndef projected_product(b,head,A,qi,target,kap):\n    k=len(A)-1;n=k+3;V,W=base.vandermonde(n)\n    words=A+([qi] if qi is not None else [])\n    for t in range(n):\n        tc=[W[j][t] for j in range(k+2)];top=W[k+2][t]\n        tc[k]^=mul(kap,top);tc[k+1]^=top\n        rank_affine(b,head,[1,t],words,V[t][:len(words)],target,tc,V[t][k+1] if qi is None else 0)\n\ndef truncated_quotient(b,head,A,qi,B,kap):\n    k=len(B)\n    if not k:return\n    words=A[1:]+([qi] if qi is not None else [])\n    for ac,bc,tc in reversed(base.quotient_terms(k+2,kap)):\n        rank_affine(b,head,ac,words,bc[:len(words)],B,tc[:k],bc[k] if qi is None else 0)\n\ndef prefix_encode(b,heads,i,F,d=None):\n    a,x=heads[i]\n    if not i:\n        if d is not None:b.cx(a,d);b.cx(x,d)\n        return None\n    q=heads[i-1][1]\n    def after():\n        ar.linear_inplace(b,q,ar.square_columns(12));ar.mul_xor0(b,a,q,d)\n        ar.linear_inplace(b,q,[power(1<<j,2048) for j in range(12)]);ar.mul_xor0(b,x,q,d)\n    callback=after if d is not None else None\n    if i==1:return pg.boundary_forward(b,x,q,F,callback)\n    return pg.forward(b,x,heads[i-2][1],q,F if i==2 else None,callback)\n\ndef prefix_decode(b,heads,i,F,m,d=None):\n    a,x=heads[i]\n    if not i:\n        if d is not None:b.cx(a,d);b.cx(x,d)\n        return\n    q=heads[i-1][1]\n    if d is not None:\n        ar.mul_xor0(b,x,q,d)\n        if i==1:b.ccx(F,q,d)\n    if i==1:pg.boundary_reverse(b,x,q,F,m)\n    else:pg.reverse_compact(b,x,heads[i-2][1],q,m,F if i==2 else None)\n    if d is not None:\n        ar.linear_inplace(b,q,ar.square_columns(12));ar.mul_xor0(b,a,q,d)\n        ar.linear_inplace(b,q,[power(1<<j,2048) for j in range(12)])\n\ndef p_alias(heads,i,F):\n    return (None,None,0) if i==0 else (None,None,1) if i==1 else (heads[i-2][1],F if i==2 else None,0)\n\ndef absorb(b,head,A,B,qi,p,virtual,pconst,c,kap,reverse=False,transcript=None):\n    k=len(B);assert len(A)==k+1;out=head+B\n    if not reverse:\n        value=eval_implicit(b,A,qi,c);flag=dv.zero_flag(b,value);eval_implicit(b,A,qi,c,True)\n        b.cx(flag,A[0][0]);spare=[b.alloc_qubits(12),b.alloc_qubits(12)];target=spare+B\n        linear_words(b,target,base.embedding_matrix(k+2,kap));projected_product(b,head,A,qi,target,kap)\n        b.cx(flag,A[0][0]);cs=[0]*(k+2);cs[k]=kap;cs[k+1]=1\n        add_alias(b,p,target,cs,virtual,pconst)\n        hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n        for h,z in zip(head,spare):b.swap(h,z);b.free(z)\n        fm=b.alloc_bits(1);b.hmr(flag,fm[0]);b.free(flag)\n        return out,A,(hm,fm)\n    linear_words(b,out,base.embedding_matrix(k+2,kap),True)\n    quo,rem=divmodp([0]*(k+2)+[1],[kap,1,1]);coeffs=[rem[j] if j<len(rem) else 0 for j in range(2)]+[quo[j] if j<len(quo) else 0 for j in range(k)]\n    add_alias(b,p,out,coeffs,virtual,pconst)\n    value=eval_implicit(b,A,qi,c);cx(b,head[1],head[0],c);hm,fm=transcript\n    divide_zero_bridge414(b,combine(head),value,K=K,extra_zero_mask=fm[0]);b.free(fm)\n    cx(b,head[1],head[0],c);b.cz(combine(head),hm);b.free(hm)\n    eval_implicit(b,A,qi,c,True);truncated_quotient(b,head,A,qi,B,kap)\n    return out,A,transcript\n\ndef synthesize(b,words,keep_leading=True):\n    h=len(words)//2;assert h>=3 and h%2==1\n    cs,ks,_,_=constants(h);heads=[words[2*i:2*i+2] for i in range(h)]\n    d=b.alloc_qubits(12);F=pg.zero_compact(b,heads[0][1]);masks=[]\n    for i in range(h):masks.append(prefix_encode(b,heads,i,F,d))\n    divide_safe138(b,d,heads[-1][1]);A=[d];B=[];history=[]\n    for i in reversed(range(h)):\n        prefix_decode(b,heads,i,F,masks[i]);qi=heads[i-1][1] if i else None;p,virtual,pc=p_alias(heads,i,F)\n        oldA,oldB=A,B;A,B,m=absorb(b,heads[i],A,B,qi,p,virtual,pc,cs[i],ks[i]);history.append((i,oldA,oldB,qi,p,virtual,pc,cs[i],ks[i],m))\n    extra=[A[h]];freed=[] if keep_leading else extra\n    for w in freed:b.free(w)\n    fmask=b.alloc_bits(1);b.hmr(F,fmask[0]);b.free(F)\n    t=dict(fmask=fmask,kind=\'partial_q_gauge_noflag\',h=h,heads=heads,d=d,F=F,history=history,A=A,B=B,extra=extra,freed=freed,keep_leading=keep_leading)\n    return A[:h],B,t\n\ndef unsynthesize(b,t):\n    if t[\'freed\']:reclaim(b,t[\'freed\'])\n    heads=t[\'heads\'];F=t[\'F\'];masks=[None]*t[\'h\']\n    for i,A,B,qi,p,virtual,pc,c,kap,m in reversed(t[\'history\']):\n        absorb(b,heads[i],A,B,qi,p,virtual,pc,c,kap,True,m);masks[i]=prefix_encode(b,heads,i,F)\n        if i==0:\n            reclaim(b,[[F]]);lad=base.zero_hold(b,heads[0][1]);b.cx(lad[-1],F);base.zero_free(b,heads[0][1],lad)\n            b.cz(F,t[\'fmask\'][0]);b.free(t[\'fmask\'])\n    multiply_safe165(b,t[\'d\'],heads[-1][1])\n    for i in reversed(range(t[\'h\'])):prefix_decode(b,heads,i,F,masks[i],t[\'d\'])\n    pg.zero_compact_free(b,heads[0][1],F);b.free(t[\'d\'])\n', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/tail.py': '"""Closed valid-tail scale pair; q=0 requires d=0 throughout."""\ntry:\n    from .scalar import *\nexcept ImportError:\n    from scalar import *\nfrom lowwidth import phase_inverse36\n\ndef multiply_closed144(b,x,q):\n    """x*=q, 144T/19private, q0 requires x0; all HMR phases closed."""\n    out=b.alloc_qubits(12);mul_xor0(b,x,q,out);m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out)\n    iq=inverse(b,q);phase_product(b,x,iq,m);b.free(m)\n    im=b.alloc_bits(12);b.hmr(iq,im);b.free(iq);phase_inverse36(b,q,im);b.free(im)\n\ndef divide_safe138(b,x,q):\n    """All-input x/=q when q!=0, identity whenq0;138T/20private."""\n    try:from .prefix import zero_compact,zero_compact_free\n    except ImportError:from prefix import zero_compact,zero_compact_free\n    f=zero_compact(b,q);b.cx(f,q[0]);divide117(b,x,q);b.cx(f,q[0]);zero_compact_free(b,q,f)\n\ndef multiply_safe165(b,x,q):\n    """Inverse all-input tail map,165T/20private, no deferred phase."""\n    try:from .prefix import zero_compact,zero_compact_free\n    except ImportError:from prefix import zero_compact,zero_compact_free\n    f=zero_compact(b,q);b.cx(f,q[0]);multiply_closed144(b,x,q);b.cx(f,q[0]);zero_compact_free(b,q,f)\n\ndef controlled_linear_xor(b,F,x,targets,coeffs):\n    """targets[j]^=F*coeffs[j]*x,12T,0scratch; GF12 coefficients classical."""\n    f=field(12);p=next(i for i,v in enumerate(coeffs) if v);v=coeffs[p];vi=f.invert(v)\n    for i,c in enumerate(coeffs):\n        if i!=p:linear_xor(b,targets[p],targets[i],[f.mul(f.mul(c,vi),1<<j) for j in range(12)])\n    linear_inplace(b,targets[p],[f.mul(vi,1<<j) for j in range(12)])\n    b.ccx(F,x,targets[p]);linear_inplace(b,targets[p],[f.mul(v,1<<j) for j in range(12)])\n    for i in reversed(range(len(coeffs))):\n        if i!=p:linear_xor(b,targets[p],targets[i],[f.mul(f.mul(coeffs[i],vi),1<<j) for j in range(12)])\n', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/kernel.py': '"""Inverse24 classical-mask phase tradeoff; frozen jet source is read-only."""\nfrom exp.jet24.kernel import *\nfrom exp.mul99.phase36 import phase_classical_parity\n\n\ndef rank(rows):\n    p={}\n    for v in rows:\n        while v:\n            j=v.bit_length()-1\n            if j in p:v^=p[j]\n            else:p[j]=v;break\n    return len(p)\n\n\ndef cols_from_rows(rows):return[sum(((r>>j)&1)<<i for i,r in enumerate(rows))for j in range(len(rows))]\n\n\ndef layout():\n    rows=[]\n    for l,r in LAY[\'gates\']:\n        for v in(l>>1,r>>1):\n            if rank(rows+[v])>len(rows):rows.append(v)\n    assert len(rows)==12\n    pivot=next(j for j in range(6)if rank(rows+[1<<j])==13);rows.append(1<<pivot);cols=cols_from_rows(rows);inv=inverse_columns(cols)\n    def form(e):return(e&1)|sum((((((e>>1)&65535)&inv[j]).bit_count()&1)<<(j+1))for j in range(13))\n    outputs=[(e&~65535)|form(e&65535)for e in LAY[\'outputs\']];gates=[tuple(form(e)for e in g)for g in LAY[\'gates\']]\n    assert all(not(e>>13&1)for g in gates for e in g)\n    v=sum(((e>>13)&1)<<j for j,e in enumerate(outputs));assert rank([1,2,v])==3\n    rest=[]\n    for j in range(2,6):\n        if rank([1,2,v]+rest+[1<<j])>3+len(rest):rest.append(1<<j)\n    B=[1,2]+rest+[v];assert len(B)==6;U=inverse_columns(B)\n    def outputmix(values):return[sum(0 for _ in[])for j in[]]\n    newoutputs=[]\n    for j in range(6):\n        e=0\n        for i in range(6):\n            if U[i]>>j&1:e^=outputs[i]\n        newoutputs.append(e)\n    assert[(e>>13)&1 for e in newoutputs]==[0,0,0,0,0,1]\n    return dict(pivot=pivot,source_columns=cols,source_inverse=inv,output_B=B,outputs=newoutputs,gates=gates)\n\nL30=layout()\n\n\ndef norm_bit_xor(b,N,target,K12,to6,j):\n    """One tower6 component ofnorm12(N),6CCX andnoquantumscratch.\n    N is12towercoordinates; outputcanonical6isthenmappedbyto6.\n    """\n    f=field(6)\n    for i in range(6):\n        if apply(to6,f.square(1<<i))>>j&1:b.cx(N[i],target)\n        if apply(to6,f.mul(K12,f.square(1<<i)))>>j&1:b.cx(N[i+6],target)\n        row=sum(((apply(to6,f.mul(1<<i,1<<k))>>j)&1)<<k for k in range(6));ix=list(bits(row));p=ix[0]\n        for k in ix[1:]:b.cx(N[6+k],N[6+p])\n        b.ccx(N[i],N[6+p],target)\n        for k in reversed(ix[1:]):b.cx(N[6+k],N[6+p])\n\n\ndef virtual_mask_phase(b,x24,mask24,N12,out6,frm12):\n    """30 executedCCZ,0scratch; L12 virtual inx24 andclassicalmask."""\n    f=field(12)\n    for segment in(0,1):\n        if segment==0:b.cx(N12[6:],N12[:6])\n        xx=N12[:6]if segment==0 else N12[6:]\n        for m,v in tensor(6):\n            h=apply(frm12,v<<(6*segment));co=b.alloc_bits(24);b.bit_store(co,False)\n            for i in range(12):\n                for r in bits(f.mul(h,1<<i)):\n                    b.cx(mask24[r],co[i]);b.cx(mask24[r],co[i+12]);b.cx(mask24[r+12],co[i+12])\n            ix=list(bits(m));p=ix[0]\n            for k in ix[1:]:b.cx(xx[k],xx[p]);b.cx(out6[k],out6[p])\n            phase_classical_parity(b,xx[p],out6[p],x24,co)\n            for k in reversed(ix[1:]):b.cx(out6[k],out6[p]);b.cx(xx[k],xx[p])\n            b.free(co)\n        if segment==0:b.cx(N12[6:],N12[:6])\n\n\ndef phase_inverse24_canonical118(b,x,mask,zero_mask=None):\n    """<=118 executedT/30private, inverse0phase; inputcanonical24."""\n    to24,frm24,K24=tower(24);linear_inplace(b,x,to24);mt=classical_transpose(b,mask,frm24)\n    N=b.alloc_qubits(12);norm_xor_noscratch(b,x,N,K24);to12,frm12,K12=tower(12);linear_inplace(b,N,to12)\n    D=b.alloc_qubits(6);norm_xor_noscratch(b,N,D,K12);to6,frm6,K6=tower(6);linear_inplace(b,D,to6);g=b.alloc_qubits(6)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,D,l,r)as(p,q):b.init_and(p,q,g[i])\n    prefix=list(D)+list(g);forms=ref.norm_forms(K6);w=ref.inverse3_virtual(b,prefix,forms)\n    linear_inplace(b,g,LAY[\'raw_columns\']);R,E=g[:4],g[4:];source=list(D)+list(R)+list(w);linear_inplace(b,source,L30[\'source_columns\']);C=source[:12];z=list(E)+list(b.alloc_qubits(3))+source[12:]\n    for j,e in enumerate(L30[\'outputs\']):ref.form_xor(b,C,e&8191,z[j])\n    for i,(l,r)in enumerate(L30[\'gates\']):\n        v=sum(((e>>(16+i))&1)<<j for j,e in enumerate(L30[\'outputs\']))\n        with ref.expose_pair(b,C,l,r)as(p,q):rankone_xor(b,p,q,z,v)\n    linear_inplace(b,z,L30[\'output_B\']);linear_inplace(b,z,LAY[\'output_inverse_columns\']);linear_inplace(b,z,frm6)\n    virtual_mask_phase(b,x,mt,N,z,frm12);b.free(mt)\n    # OutputHMR first frees6 before restoring the one originalD coordinate.\n    zm=b.alloc_bits(6);b.hmr(z,zm);b.free(z);ym=classical_transpose(b,zm,frm6);b.free(zm);mm=classical_transpose(b,ym,LAY[\'output_inverse_columns\']);b.free(ym)\n    dj=b.alloc_qubits(1)[0];norm_bit_xor(b,N,dj,K12,to6,L30[\'pivot\']);S=list(C)+[dj];linear_inplace(b,S,L30[\'source_inverse\']);DD=S[:6];RR=S[6:10];ww=S[10:]\n    phase_outputs_classical(b,S,LAY[\'outputs\'],LAY[\'gates\'],mm)\n    for i,(l,r)in enumerate(basic.raw_forms19()):\n        with ref.expose_pair(b,DD,l,r)as(p,q):\n            for j,row in enumerate(LAY[\'raw_rows\'][4:]):\n                if row>>i&1:b.ccz(p,q,mm[j])\n    b.free(mm)\n    EE=b.alloc_qubits(3);norm_xor_noscratch(b,DD,EE,K6)\n    if zero_mask is not None:\n        b.z(zero_mask)\n        for i in range(3):\n            for j in range(3):\n                if field(3).mul(1<<i,1<<j)&1:b.ccz(EE[i],ww[j],zero_mask)\n    erase_inverse(b,EE,ww);b.free(ww);ar._erase_norm(b,DD,EE,K6);b.free(EE);zero1.erase_raw_cache(b,DD,RR,LAY[\'raw_rows\'][:4]);linear_inplace(b,DD,frm6)\n    dm=b.alloc_bits(6);b.hmr(DD,dm);b.free(DD);ar._phase_norm(b,N,dm,K12);b.free(dm);linear_inplace(b,N,frm12);ar._erase_norm(b,x,N,K24);b.free(N);linear_inplace(b,x,frm24)\n\n\ndef phase_inverse24_extension118(b,x,mask,K=None,zero_mask=None):\n    cols,frm,_=extension_columns(K);linear_inplace(b,x,cols);mt=classical_transpose(b,mask,frm);phase_inverse24_canonical118(b,x,mt,zero_mask);b.free(mt);linear_inplace(b,x,frm)\n', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/phase6_six_search.json': '{\n  "seed": 202610262337,\n  "profiles": [\n    {\n      "hyper": 1,\n      "tested": 7640\n    },\n    {\n      "hyper": 2,\n      "tested": 7784\n    },\n    {\n      "hyper": 3,\n      "tested": 7672\n    },\n    {\n      "hyper": 4,\n      "tested": 7008\n    }\n  ],\n  "shears_tested": 30104,\n  "candidate": {\n    "p": 1321,\n    "q": 372,\n    "w": 29,\n    "output_bit": 5,\n    "terms": [\n      [\n        4\n      ],\n      [\n        6\n      ],\n      [\n        7\n      ],\n      [\n        9\n      ],\n      [\n        0,\n        2\n      ],\n      [\n        0,\n        3\n      ],\n      [\n        0,\n        5\n      ],\n      [\n        0,\n        6\n      ],\n      [\n        0,\n        7\n      ],\n      [\n        0,\n        8\n      ],\n      [\n        0,\n        10\n      ],\n      [\n        2,\n        6\n      ],\n      [\n        2,\n        8\n      ],\n      [\n        3,\n        4\n      ],\n      [\n        3,\n        10\n      ],\n      [\n        4,\n        5\n      ],\n      [\n        4,\n        6\n      ],\n      [\n        4,\n        7\n      ],\n      [\n        4,\n        8\n      ],\n      [\n        5,\n        10\n      ],\n      [\n        6,\n        7\n      ],\n      [\n        6,\n        9\n      ],\n      [\n        6,\n        10\n      ],\n      [\n        7,\n        8\n      ],\n      [\n        7,\n        10\n      ]\n    ],\n    "zero_terms": [\n      [\n        0\n      ],\n      [\n        1\n      ],\n      [\n        2\n      ],\n      [\n        3\n      ],\n      [\n        4\n      ],\n      [\n        9\n      ],\n      [\n        10\n      ],\n      [\n        0,\n        1\n      ],\n      [\n        0,\n        2\n      ],\n      [\n        0,\n        6\n      ],\n      [\n        0,\n        7\n      ],\n      [\n        0,\n        8\n      ],\n      [\n        0,\n        9\n      ],\n      [\n        1,\n        2\n      ],\n      [\n        1,\n        4\n      ],\n      [\n        1,\n        6\n      ],\n      [\n        1,\n        9\n      ],\n      [\n        1,\n        10\n      ],\n      [\n        2,\n        8\n      ],\n      [\n        3,\n        4\n      ],\n      [\n        3,\n        6\n      ],\n      [\n        3,\n        7\n      ],\n      [\n        4,\n        7\n      ],\n      [\n        4,\n        8\n      ],\n      [\n        5,\n        6\n      ],\n      [\n        6,\n        8\n      ],\n      [\n        6,\n        9\n      ],\n      [\n        6,\n        10\n      ],\n      [\n        7,\n        10\n      ],\n      [\n        8,\n        9\n      ]\n    ],\n    "raw_forms": [\n      [\n        12,\n        80\n      ],\n      [\n        4,\n        64\n      ],\n      [\n        2,\n        48\n      ],\n      [\n        8,\n        16\n      ],\n      [\n        10,\n        32\n      ]\n    ],\n    "raw_coefficients": [\n      1,\n      2,\n      8,\n      16,\n      32\n    ],\n    "truth_points": 64,\n    "errors": 0,\n    "prefix_AND": 6,\n    "scratch": 5,\n    "hyper": 4,\n    "attempt": 902\n  },\n  "seconds": 3.8534505367279053\n}\n', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/phase6small.py': '"""Classical-mask inverse6 phase:<=6executedT/5scratch; exact64pointSLP.\nClassicalmask-dependent constantfieldscaling maps63nonzero components toone.\n"""\nfrom exp.phase24_tradeoff.kernel import *\nimport json\nP6=json.loads(Path(__file__).with_name(\'phase6_six_search.json\').read_text())[\'candidate\']\n\ndef classical_equal(b,x,value):\n    chain=b.alloc_bits(len(x)+1);b.bit_store(chain,False);b.x(chain[0])\n    for j,q in enumerate(x):\n        if not(value>>j&1):b.x(q)\n        b.push_condition(chain[j]);b.bit_store(chain[j+1],True,control=q);b.pop_condition()\n        if not(value>>j&1):b.x(q)\n    b.free(chain[:-1]);return chain[-1]\n\n\ndef fixed_phase6(b,x,do_inverse=True,zero_mask=None):\n    to,frm,K=tower(6);linear_inplace(b,x,to);ng=len(P6[\'raw_forms\']);g=b.alloc_qubits(ng)\n    for i,(l,r)in enumerate(P6[\'raw_forms\']):\n        with ref.expose_pair(b,x,l,r)as(p,q):b.init_and(p,q,g[i])\n    w=P6[\'w\'];base=[]\n    for j in range(ng):\n        if rank(base+[w,1<<j])==len(base)+2:base.append(1<<j)\n    B=base+[w];assert len(B)==ng;iv=inverse_columns(B);linear_inplace(b,g,iv)\n    def form(v):\n        return(v&63)|sum((((v>>6)&co).bit_count()&1)<<(6+j)for j,co in enumerate(B))\n    p,q=form(P6[\'p\']),form(P6[\'q\']);assert not(((p|q)>>(6+ng-1))&1);source=list(x)+list(g[:ng-1])\n    with ref.expose_pair(b,source,p<<1,q<<1)as(a,c):b.ccx(a,c,g[ng-1])\n    linear_inplace(b,g,B);S=list(x)+list(g)\n    for term in(P6[\'terms\']if do_inverse else[]):\n        if len(term)==1:b.z(S[term[0]])\n        else:b.cz(S[term[0]],S[term[1]])\n    if zero_mask is not None:\n        b.z(zero_mask);b.push_condition(zero_mask)\n        for term in P6[\'zero_terms\']:\n            if len(term)==1:b.z(S[term[0]])\n            else:b.cz(S[term[0]],S[term[1]])\n        b.pop_condition()\n    linear_inplace(b,g,iv);hm=b.alloc_bits(1)[0];b.hmr(g[ng-1],hm)\n    with ref.expose_pair(b,source,p<<1,q<<1)as(a,c):b.ccz(a,c,hm)\n    rows=[sum(((iv[i]>>j)&1)<<i for i in range(ng))for j in range(ng)]\n    for i,(l,r)in enumerate(P6[\'raw_forms\']):\n        if rows[ng-1]>>i&1:\n            with ref.expose_pair(b,x,l,r)as(a,c):b.ccz(a,c,hm)\n    b.free(hm);b.free(g[ng-1]);mm=b.alloc_bits(ng-1);b.hmr(g[:ng-1],mm);b.free(g[:ng-1])\n    for i,(l,r)in enumerate(P6[\'raw_forms\']):\n        with ref.expose_pair(b,x,l,r)as(a,c):\n            for j in range(ng-1):\n                if rows[j]>>i&1:b.ccz(a,c,mm[j])\n    b.free(mm);linear_inplace(b,x,frm)\n\n\ndef mask_factors():\n    f=field(6);out={}\n    for s in range(1,64):\n        si=f.invert(s);m=sum(((f.mul(si,1<<j)>>P6[\'output_bit\'])&1)<<j for j in range(6));assert m not in out;out[m]=s\n    return out\nFACTORS=mask_factors()\n\n\ndef phase_inverse6_classical6(b,x,mask,zero_mask=None):\n    """Exactmask·inv0(x),0→0,<=6executedT/5scratch,arbitraryclassicalmask."""\n    for m,s in sorted(FACTORS.items()):\n        eq=classical_equal(b,mask,m);b.push_condition(eq);co=[field(6).mul(s,1<<j)for j in range(6)];linear_inplace(b,x,co);fixed_phase6(b,x,zero_mask=zero_mask);linear_inplace(b,x,inverse_columns(co));b.pop_condition();b.free(eq)\n\n    if zero_mask is not None:\n        eq=classical_equal(b,mask,0);b.push_condition(eq);b.push_condition(zero_mask);fixed_phase6(b,x,do_inverse=False,zero_mask=zero_mask);b.pop_condition();b.pop_condition();b.free(eq)\n', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/converter.py': '# Correct residues on coprime degree profiles with gaps at most two.\n# Zero-clipped scalar permutations preserve restoration on unsupported profiles.\n"""Guarded history-free multi-h2 Bezout lookahead, with borrowed monic-root word.\n\nOn proper coprime half-degree charts with drops1or2, parent gap flags are\nrecomputed from the deliberatelyzero child quotient metadata. No persistent\nquantum history register. Crosspass token is classical and phaseclosed after\nreverse around a diagonal oracle. See RESTORATION.md for the all-input\nrestoration argument; larger gaps may still produce wrong decoded phases.\n"""\nimport sys\nfrom pathlib import Path\nsys.path.insert(0,str(Path(__file__).resolve().parents[1]/\'convert\'))\nfrom convert import ar,frob\nfrom repair import rotate\nfrom kernels import hankel,convolution_xor\nfrom lowwidth import phase_inverse36\nfrom contextlib import contextmanager\n\ndef scale(b,x,a,ai,divide=True):\n    out=b.alloc_qubits(12)\n    ar.mul_xor0(b,x,ai if divide else a,out)\n    ar.erase_product(b,a if divide else ai,out,x)\n    b.swap(x,out);b.free(out)\n\ndef runops(b,ops,a,ai,undo=False):\n    for op in (reversed(ops)if undo else ops):\n        if op[0]==\'s\':scale(b,op[1],a,ai,op[2]^undo)\n        elif op[0]==\'m\':ar.mul_xor0(b,op[1],op[2],op[3])\n        elif op[0]==\'c\':convolution_xor(b,op[1],op[2],op[3],undo)\n        else:raise ValueError(op[0])\n\ndef zero_toggle(b,a,target):\n    # Holdthe11ANDladderuntiltheflagcopy, thenuncomputeallforfree.\n    for q in a:b.x(q)\n    ladder=list(b.alloc_qubits(len(a)-1));b.init_and(a[0],a[1],ladder[0])\n    for i in range(1,len(ladder)):b.init_and(ladder[i-1],a[i+1],ladder[i])\n    b.cx(ladder[-1],target)\n    for i in reversed(range(len(ladder))):\n        if i:b.del_and(ladder[i-1],a[i+1],ladder[i])\n        else:b.del_and(a[0],a[1],ladder[0])\n    b.free(ladder)\n    for q in a:b.x(q)\n\nfrom exp.inverse12_escape.zero1 import inverse_zero,consume,erase,inverse_apply_safe\n\n@contextmanager\ndef q_inverse(b,a,m,reverse,phase):\n    ai,F=inverse_zero(b,a)\n    if reverse:consume(b,ai,F,phase[\'q\'].pop(m))\n    b.cx(F,a[0]);b.cx(F,ai[0]);yield ai;b.cx(F,ai[0]);b.cx(F,a[0])\n    token=erase(b,a,ai,F,not reverse)\n    if not reverse:phase[\'q\'][m]=token\n\ndef inverse_apply(b,a,action,incoming=None,defer=False):\n    return inverse_apply_safe(b,a,action,incoming,defer)\n\ndef pre(b,f,g,payloads,undo=False,incoming=None,defer=False):\n    m=len(f)-1;n=m-1;a=g[-1];q0=f[-2];q1=f[-1]\n    def action(ai):\n        ops=[(\'s\',v[-1],True)for v in payloads]\n        ops +=[(\'m\',g[i],v[-1],v[i])for v in payloads for i in range(n)]\n        ops +=[(\'s\',q1,True),(\'m\',q1,g[-2],q0),(\'s\',q0,True),(\'c\',g[:-1],[q0,q1],f[:-2])]\n        runops(b,ops,a,ai,undo)\n    return inverse_apply(b,a,action,incoming,defer)\n\ndef canonical_permute(b,f,g,F,reverse=False):\n    # [G0,F0,G1,F1,...,G_(m-2),F_(m-2),G_(m-1)] leftcyclic.\n    cyc=[]\n    for i in range(len(g)-1):cyc +=[g[i],f[i]]\n    cyc +=[g[-1]]\n    seq=range(len(cyc)-2,-1,-1) if reverse else range(len(cyc)-1)\n    for i in seq:b.cswap(F,cyc[i],cyc[i+1])\n\ndef masked_product(b,F,x,c,target):\n    u=b.alloc_qubits(12)\n    for j in range(12):b.init_and(F,x[j],u[j])\n    ar.mul_xor0(b,u,c,target)\n    mask=b.alloc_bits(12);b.hmr(u,mask);b.free(u)\n    for j in range(12):b.ccz(F,x[j],mask[j])\n    b.free(mask)\n\ndef fresh_zero_flag(b,a):\n    F=b.alloc_qubits(1)[0];zero_toggle(b,a,F);return F\n\ndef clear_flag(b,a,F):zero_toggle(b,a,F);b.free(F)\n\ndef q_post(b,q,payloads,H,m,parent,reverse=False,phase=None):\n    if parent is None:\n        with q_inverse(b,q,m,reverse,phase) as qi:\n            for v in payloads:scale(b,v[-1],q,qi,not reverse)\n        return\n    tops,p,c=parent;F=fresh_zero_flag(b,q)\n    b.cswap(F,q,p)\n    if not reverse:\n        for v,t in zip(payloads,tops):b.cswap(F,v[-1],t)\n    with q_inverse(b,q,m,reverse,phase) as qi:\n        if not reverse:\n            for v in payloads:scale(b,v[-1],q,qi,True)\n            for v,t in zip(payloads,tops):masked_product(b,F,v[-1],c,t)\n        else:\n            for v,t in zip(payloads,tops):masked_product(b,F,v[-1],c,t)\n            for v in payloads:scale(b,v[-1],q,qi,False)\n    if reverse:\n        for v,t in zip(payloads,tops):b.cswap(F,v[-1],t)\n    b.cswap(F,q,p);clear_flag(b,q,F)\n\ndef post_look(b,f,g,payloads,H,parent,reverse=False,incoming=None,defer=False,phase=None):\n    m=len(f)-1;n=m-1;a=g[-1];q0=f[-2];q1=f[-1]\n    def dots():\n        for v in payloads:\n            for i in range(n):ar.mul_xor0(b,g[i],v[i],v[-1])\n    def last(undo,inc=None,delay=False):\n        def action(ai):\n            ops=[(\'s\',v[-1],True)for v in payloads]\n            ops +=[(\'c\',g[:-1],[q0,q1],f[:-2]),(\'s\',q0,False),(\'m\',q1,g[-2],q0),(\'s\',q1,False)]\n            runops(b,ops,a,ai,undo)\n        return inverse_apply(b,a,action,inc,delay)\n    if not reverse:q_post(b,q1,payloads,H,m,parent,phase=phase);dots();return last(False,incoming,defer)\n    else:\n        carry=last(True,incoming,defer);dots();q_post(b,q1,payloads,H,m,parent,True,phase);return carry\n\ndef base(b,f,g,payloads,H,parent,reverse=False,phase=None):\n    a=g[0];q=f[-1]\n    def first(undo,inc=None,delay=False):\n        def action(ai):runops(b,[(\'s\',v[0],True)for v in payloads]+[(\'s\',q,True)],a,ai,undo)\n        return inverse_apply(b,a,action,inc,delay)\n    def last(undo,inc=None,delay=False):\n        def action(ai):runops(b,[(\'s\',v[0],True)for v in payloads]+[(\'s\',q,False)],a,ai,undo)\n        return inverse_apply(b,a,action,inc,delay)\n    if not reverse:\n        carry=first(False,delay=True);q_post(b,q,payloads,H,1,parent,phase=phase);phase[\'a\'][1]=last(False,carry,True)\n    else:\n        carry=last(True,phase[\'a\'].pop(1),True);q_post(b,q,payloads,H,1,parent,True,phase);first(True,carry)\n\ndef look_solve(b,f,g,payloads,H=None,parent=None,reverse=False,phase=None):\n    m=len(f)-1\n    if m==1:return base(b,f,g,payloads,H,parent,reverse,phase)\n    F=fresh_zero_flag(b,g[-1]);rotate(b,g,F)\n    if not reverse:carry=pre(b,f,g,payloads,defer=True)\n    else:carry=post_look(b,f,g,payloads,H,parent,True,incoming=phase[\'a\'].pop(m),defer=True,phase=phase)\n    canonical_permute(b,f,g,F);clear_flag(b,g[-1],F)\n    childparent=([v[-1]for v in payloads],f[-1],f[-2])\n    look_solve(b,g,f[:-2],[v[:-1]for v in payloads],None,childparent,reverse,phase)\n    F=fresh_zero_flag(b,g[-1]);canonical_permute(b,f,g,F,True)\n    if not reverse:phase[\'a\'][m]=post_look(b,f,g,payloads,H,parent,incoming=carry,defer=True,phase=phase)\n    else:pre(b,f,g,payloads,True,incoming=carry)\n    rotate(b,g,F,True);clear_flag(b,g[-1],F)\n\ndef bez_lookahead(b,O,E,payloads,reverse=False,history_scratch12=None,phase=None):\n    def reduce():\n        for i in range(len(O)):ar.mul_xor0(b,E[-1],O[i],E[i])\n    reduce()\n    lead=list(b.alloc_qubits(12))if history_scratch12 is None else list(history_scratch12)\n    assert len(lead)==12\n    b.x(lead[0])\n    look_solve(b,O+[lead],E[:-1],payloads,reverse=reverse,phase=phase)\n    b.x(lead[0])\n    if history_scratch12 is None:b.free(lead)\n    reduce()\n\ndef convert_safe(b,A,B,reverse=False,history_scratch12=None,phase_token=None):\n    """Optional12zero wires borrowed as monic rootleadingword; returnedzero."""\n    phase={\'a\':{},\'q\':{}}if phase_token is None else phase_token\n    assert not reverse or phase_token is not None\n    E=B[::2];O=B[1::2];u=A[1::2];w=A[:34:2];top=A[-1]\n    if reverse:hankel(b,B,A,True)\n    frob(b,B,2048);frob(b,A,2048)\n    if reverse:\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n        for i in range(17):ar.mul_xor0(b,O[i],u[i],top)\n        bez_lookahead(b,O,E,[u,w],True,history_scratch12,phase)\n        for i in range(17):ar.mul_xor0(b,top,O[i],w[i])\n    else:\n        for i in range(17):ar.mul_xor0(b,top,O[i],w[i])\n        bez_lookahead(b,O,E,[u,w],False,history_scratch12,phase)\n        for i in range(17):ar.mul_xor0(b,O[i],u[i],top)\n        for i in range(17):b.swap(A[2*i],A[2*i+1])\n    frob(b,A,2);frob(b,B,2)\n    if not reverse:hankel(b,B,A)\n\n    if reverse:assert not phase[\'a\']and not phase[\'q\']\n    return phase if not reverse else None\n', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/kernel.py': '"""Zero-aware inverse12: expose zero cheaply at the bottom tower norm."""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.inverse_search.inverse20_iso import *\nfrom exp.noscratch.schedules import norm_xor_noscratch\nfrom exp.safe_scalar.phase import phase_inverse36_zero\n\ndef inverse_zero(b,x):\n    """Preserve x, return inverse0(x) and [x=0],67T with clean scratch."""\n    to,frm,K=tower(12);linear_inplace(b,x,to)\n    N=b.alloc_qubits(6);norm_xor_noscratch(b,x,N,K)\n    t,D=inverse6_xor20(b,N,keep_norm=True)\n    b.x(D);u=b.alloc_qubits(1)[0];F=b.alloc_qubits(1)[0]\n    b.init_and(D[0],D[1],u);b.init_and(u,D[2],F);b.del_and(D[0],D[1],u);b.free(u);b.x(D)\n    to6,frm6,K6=tower(6);linear_inplace(b,N,to6)\n    ar._erase_norm(b,N,D,K6);b.free(D);linear_inplace(b,N,frm6)\n    ar._erase_norm(b,x,N,K);b.free(N)\n    out=b.alloc_qubits(12);b.cx(x[6:],x[:6])\n    mul_xor_noscratch(b,x[:6],t,out[:6]);mul_xor_noscratch(b,x[6:],t,out[6:]);b.cx(x[6:],x[:6])\n    ar._erase_norm(b,out,t,K);b.free(t)\n    linear_inplace(b,x,frm);linear_inplace(b,out,frm)\n    return out,F\n\ndef consume(b,ai,F,token):\n    if token is not None:\n        m,z=token;b.cz(ai,m);b.cz(F,z[0]);b.free(m);b.free(z)\n\ndef erase(b,a,ai,F,defer):\n    m=b.alloc_bits(12);z=b.alloc_bits(1)\n    b.hmr(ai,m);b.hmr(F,z[0]);b.free(ai);b.free(F)\n    if defer:return m,z\n    phase_inverse36_zero(b,a,m,z[0]);b.free(m);b.free(z)\n\ndef inverse_apply_safe(b,a,action,incoming=None,defer=False):\n    ai,F=inverse_zero(b,a);consume(b,ai,F,incoming)\n    b.cx(F,a[0]);b.cx(F,ai[0]);action(ai);b.cx(F,ai[0]);b.cx(F,a[0])\n    return erase(b,a,ai,F,defer)\n', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/phase.py': '"""Inverse12 phase with12 scratch and at most36 EXECUTED Toffolis.\n\nThe coefficient mask is classical. A classical first-nonzero pivot exposes\nits parity in an existing input bit, avoiding the13th quantum wire.\n"""\ntry:\n    from exp.mul99.kernels import *\nexcept ImportError:\n    from exp.mul99.kernels import *\nsys.path.insert(0,str(ROOT))\nfrom exp.noscratch.kernels import inverse3_noscratch\nfrom lowwidth import classical_transpose\n\n\ndef phase_classical_parity(b,a,c,x,co):\n    """CCZ(a,c,parity(co*x)), zero quantum scratch, <=1 executed CCZ."""\n    first=b.alloc_bits(len(x));b.bit_store(first,False)\n    seen=b.alloc_bits(1)[0];b.bit_store(seen,False)\n    for j in range(len(x)):\n        b.x(seen);b.push_condition(seen)\n        b.bit_store(first[j],True,control=co[j])\n        b.pop_condition();b.x(seen)\n        b.bit_store(seen,True,control=co[j])\n    for j in range(len(x)):\n        b.push_condition(first[j])\n        for k in range(len(x)):\n            if k!=j:b.ccx(co[k],x[k],x[j])\n        b.ccz(a,c,x[j])\n        for k in reversed(range(len(x))):\n            if k!=j:b.ccx(co[k],x[k],x[j])\n        b.pop_condition()\n    b.free(seen);b.free(first)\n\n\ndef phase_inverse36_zero(b,x,mask,zero_mask):\n    """Phase mask dot inverse0(x): <=36 mean T /12 quantum scratch.\n\n    Static count168T due to12 mutually exclusive classical branches per CCZ.\n    No quantum input controls branch selection; all branches use HMR-mask bits.\n    """\n    assert len(x)==len(mask)==12\n    to12,frm12,K12=tower(12);linear_inplace(b,x,to12)\n    mt=classical_transpose(b,mask,frm12)\n    D=b.alloc_qubits(6);ar._norm_xor(b,x,D,K12)\n    to6,frm6,K6=tower(6);linear_inplace(b,D,to6)\n    E=b.alloc_qubits(3);ar._norm_xor(b,D,E,K6);w=inverse3_noscratch(b,E)\n    b.z(zero_mask);zm=b.alloc_bits(3);b.bit_store(zm,False);b.cx(zero_mask,zm[0])\n    phase_product(b,E,w,zm);b.free(zm)\n    f6=field(6)\n    for segment in(0,1):\n        if segment==0:b.cx(D[3:],D[:3])\n        controls=D[:3] if segment==0 else D[3:]\n        for m,v in tensor(3):\n            h=apply(frm6,v<<(3*segment));co=b.alloc_bits(12);b.bit_store(co,False)\n            for j in range(6):\n                for r in bits(f6.mul(h,1<<j)):\n                    b.cx(mt[r],co[j]);b.cx(mt[r],co[j+6]);b.cx(mt[r+6],co[j+6])\n            ix=list(bits(m));p=ix[0]\n            for j in ix[1:]:b.cx(controls[j],controls[p]);b.cx(w[j],w[p])\n            phase_classical_parity(b,controls[p],w[p],x,co)\n            for j in reversed(ix[1:]):b.cx(w[j],w[p]);b.cx(controls[j],controls[p])\n            b.free(co)\n        if segment==0:b.cx(D[3:],D[:3])\n    erase_inverse(b,E,w);b.free(w)\n    ar._erase_norm(b,D,E,K6);b.free(E);linear_inplace(b,D,frm6)\n    ar._erase_norm(b,x,D,K12);b.free(D);linear_inplace(b,x,frm12);b.free(mt)\n', 'blocks/dqi/pro-q2-897/source/exp/scan/scan.py': '"""Exact45045-Toffoli Chien/residue phase scan,11workspacequbits.\n\nA,B:35GF12 words each, degree<35 numerator and monicdegree35 denominator.\nE:4095classical12-bit masks, indexed beta=alpha**j. Bothbanks and E restored.\nA is already residue-converted; phase is dot(Ej,A(beta)/(beta*P(beta)))\nwhen B(beta)=0, and0otherwise. P is the node polynomial from extract.constants.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2]\nsys.path.insert(0,str(ROOT/\'exp/decoder\'))\nfrom common import *\nfrom extract import constants\n\n\ndef scan(b,A,B,E,*,points=4095,skip_points=(),node_polynomial=None):\n    if len(A)!=35 or len(B)!=35 or len(E)<points:raise ValueError(\'need35+35words and one maskperpoint\')\n    if points!=4095:raise ValueError(\'fullcycle4095points required for bank restoration\')\n    alpha=gf.primitive_element;P=constants()[2]if node_polynomial is None else node_polynomial\n    powers=[power(alpha,k) for k in range(35)]\n    skip_points=frozenset(skip_points)\n    assert all(0<=j<4095 for j in skip_points)\n    beta=1;lead=1;leadstep=power(alpha,35);m=b.alloc_bits(12)\n    for j in range(4095):\n        # The original coefficientk now holds Ak*beta**k (likewiseB).\n        if j not in skip_points:\n            for k in range(1,35):b.cx(A[k],A[0]);b.cx(B[k],B[0])\n            b.gf2_iadd(lead,target=B[0])\n            ladder=zero_hold(b,B[0]);flag=ladder[-1]\n            pv=0\n            for v in reversed(P):pv=mul(pv,beta)^v\n            c=inv(mul(beta,pv));assert c\n            # Classically apply transpose of multiplicationc to maskEj.\n            b.bit_store(m,False)\n            for t in range(12):\n                for i in ar.bits(mul(c,1<<t)):b.cx(E[j][i],m[t])\n            b.ccz(flag,A[0],m)\n            zero_free(b,B[0],ladder)\n            b.gf2_iadd(lead,target=B[0])\n            for k in range(1,35):b.cx(A[k],A[0]);b.cx(B[k],B[0])\n        for k in range(1,35):scale(b,A[k],powers[k]);scale(b,B[k],powers[k])\n        beta=mul(beta,alpha);lead=mul(lead,leadstep)\n    assert beta==lead==1\n    b.free(m)\n', 'blocks/dqi/pro-q2-897/source/exp/tail_alias/divide.py': '"""Zero-aware initialized inverse fits below the product peak:369T safe divide."""\nfrom pathlib import Path\nimport sys\nfrom types import FunctionType\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.fold_joint import divide as prior\nfrom exp.inverse19_joint.kernel import inverse_zero\n\ndef inverse_overwrite_zero66(b,x):\n    """Overwrite inverse0+zero,66T,31 localQ; old input mask deferred."""\n    out,F=inverse_zero(b,x);m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out)\n    return m,F\n\n_globals=dict(prior.divide_zero371.__globals__,inverse_stream_zero=inverse_overwrite_zero66)\n_fn=prior.divide_zero371\ndivide_zero369=FunctionType(_fn.__code__,_globals,\'divide_zero369\',_fn.__defaults__,_fn.__closure__)\ndivide_zero369.__doc__=\'All-input safe GF24 division: <=369 executed T,37 private. Old initialized inverse+flag replaces streamed68T with66T below37-bit product peak.\'\n', 'blocks/dqi/pro-q2-897/source/exp/tail_alias/kernel.py': '"""Tail d aliases raw a0; quadratic companion action closes outer fold.\n\nNo additional persistent field word. Inputs and returned A/B use70 words.\nAll maps are total: factors zero are replaced by factor1 consistently.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.fold_joint.kernel import *\nfrom exp.tail_alias.divide import divide_zero369\nfrom types import FunctionType\n# Private binding: imported CRT arithmetic modules are never mutated.\n_absorb=absorb\nabsorb=FunctionType(_absorb.__code__,dict(_absorb.__globals__,divide_zero371=divide_zero369),\n    \'absorb\',_absorb.__defaults__,_absorb.__closure__)\n\ndef expose_quotient(b,C,D):\n    """Top C pair -> quotient l of q*C by monic D, h>=3; involution."""\n    b.cx(C[-1],C[-2]);ar.mul_xor0(b,C[-1],D[-1],C[-2])\n\ndef outer(b,C,D,c,kap,reverse=False,transcript=None,wide=True):\n    """C <- q*C mod (D+[D(gamma)=0]), D monic degree h unchanged.\n\n    D lists its h low coefficients, C its h coefficients. Constant completion\n    leaves quotient l unaffected for h>=3. CRT output is lowerC then top pair.\n    """\n    h=len(C);assert len(D)==h and h>=3;head=C[-2:];low=C[:-2]\n    if not reverse:\n        expose_quotient(b,C,D)\n        value=old.eval_implicit(b,D,None,c);F=dv.zero_flag(b,value)\n        spare=[b.alloc_qubits(12),b.alloc_qubits(12)]\n        b.cx(F,value[0]);cx(b,head[1],head[0],c)\n        mul24_xor99(b,combine(head),value,combine(spare),K)\n        cx(b,head[1],head[0],c);b.cx(F,value[0]);old.eval_implicit(b,D,None,c,True)\n        q_times_low(b,low,kap);b.cx(F,D[0][0]);low_product(b,head,D,low);b.cx(F,D[0][0])\n        R_to_tops(b,spare,low,c)\n        hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n        for w,z in zip(head,spare):b.swap(w,z);b.free(z)\n        fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n        return hm,fm\n    R_to_tops(b,head,low,c,True)\n    value=old.eval_implicit(b,D,None,c);hm,fm=transcript\n    def after(F):\n        cx(b,head[1],head[0],c);old.eval_implicit(b,D,None,c,True)\n        b.cx(F,D[0][0]);low_product(b,head,D,low);b.cx(F,D[0][0]);q_times_low(b,low,kap,True)\n        old.eval_implicit(b,D,None,c);cx(b,head[1],head[0],c)\n    if wide:divide_zero369(b,combine(head),value,K=K,extra_zero_mask=fm[0],after=after)\n    else:divide414_callback(b,combine(head),value,K,fm[0],after)\n    b.free(fm);cx(b,head[1],head[0],c);b.cz(combine(head),hm);b.free(hm)\n    old.eval_implicit(b,D,None,c,True);expose_quotient(b,C,D)\n    return transcript\n\ndef synthesize(b,words,wide=True):\n    h=len(words)//2;assert h>=3 and h%2==1\n    cs,ks,_,_=old.constants(h);heads=[words[2*i:2*i+2]for i in range(h)]\n    d=heads[0][0];F=pg.zero_compact(b,heads[0][1]);b.cx(heads[0][1],d);masks=[None]*h\n    for i in range(1,h):masks[i]=old.prefix_encode(b,heads,i,F,d)\n    old.divide_safe138(b,d,heads[-1][1]);A=[d];B=[];hist=[]\n    for i in reversed(range(1,h)):\n        old.prefix_decode(b,heads,i,F,masks[i]);qi=heads[i-1][1];p,v,pc=old.p_alias(heads,i,F)\n        oldA,oldB=A,B;A,B,t=absorb(b,heads[i],A,B,qi,p,v,pc,cs[i],ks[i],wide=wide)\n        hist.append((i,oldA,oldB,qi,p,v,pc,cs[i],ks[i],t))\n    C=B+[heads[0][1]];ot=outer(b,C,A,cs[0],ks[0],wide=wide)\n    fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n    return C,A,dict(h=h,heads=heads,d=d,F=F,history=hist,A=C,B=A,outer=ot,wide=wide,fmask=fm)\n\ndef unsynthesize(b,t):\n    h=t[\'h\'];heads=t[\'heads\'];F=t[\'F\'];cs,ks,_,_=old.constants(h)\n    outer(b,t[\'A\'],t[\'B\'],cs[0],ks[0],True,t[\'outer\'],t[\'wide\'])\n    reclaim(b,[[F]]);lad=base.zero_hold(b,heads[0][1]);b.cx(lad[-1],F);base.zero_free(b,heads[0][1],lad)\n    b.cz(F,t[\'fmask\'][0]);b.free(t[\'fmask\']);masks=[None]*h\n    for i,A,B,qi,p,v,pc,c,kap,token in reversed(t[\'history\']):\n        absorb(b,heads[i],A,B,qi,p,v,pc,c,kap,True,token,t[\'wide\'])\n        masks[i]=old.prefix_encode(b,heads,i,F)\n    old.multiply_safe165(b,t[\'d\'],heads[-1][1])\n    for i in reversed(range(1,h)):old.prefix_decode(b,heads,i,F,masks[i],t[\'d\'])\n    b.cx(heads[0][1],t[\'d\']);pg.zero_compact_free(b,heads[0][1],F)\n', 'blocks/dqi/pro-q2-897/source/exp/zero_padding_synthesis/kernel.py': '"""One-flag Taylor folds with common x-power padding on valid heads."""\nfrom pathlib import Path\nimport sys\nfrom types import FunctionType\nROOT=Path(__file__).resolve().parents[2];sys.path.insert(0,str(ROOT))\nfrom exp.tail_alias.kernel import *\nfrom exp.alternating_gauge import kernel as alt\nfrom exp.alternating_gauge.even_fold import full_product,full_quotient\nfrom exp.confluent_synthesis import kernel as ca\n\n\ndef zero_finish(b,value,F,fm):\n m=b.alloc_bits(1);b.hmr(F,m[0]);b.free(F);b.cx(fm[0],m[0]);b.free(fm)\n dv.phase_zero(b,value,m[0]);b.free(m)\n\n\ndef absorb_odd(b,head,A,B,Q,p,virtual,pconst,c,kap,reverse=False,transcript=None,wide=False):\n k=len(B);assert len(A)==k+1 and virtual is None and pconst==0\n out=B+head;v=epow(c,k+2);co=[v&4095,v>>12]\n if not reverse:\n  value=old.eval_implicit(b,A,Q,c);F=dv.zero_flag(b,value)\n  spare=[b.alloc_qubits(12),b.alloc_qubits(12)];b.cx(F,value[0]);cx(b,head[1],head[0],c)\n  mul24_xor99(b,combine(head),value,combine(spare),K)\n  cx(b,head[1],head[0],c);b.cx(F,value[0]);old.eval_implicit(b,A,Q,c,True)\n  q_times_low(b,B,kap);b.cx(F,A[0][0]);low_product(b,head,A,B);b.cx(F,A[0][0])\n  old.add_alias(b,p,spare,co);old.controlled_linear_xor(b,F,p,spare,co);R_to_tops(b,spare,B,c)\n  hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n  for x,z in zip(head,spare):b.swap(x,z);b.free(z)\n  fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n  return out,A,(hm,fm)\n R_to_tops(b,head,B,c,True);old.add_alias(b,p,head,co)\n value=old.eval_implicit(b,A,Q,c);F=dv.zero_flag(b,value);hm,fm=transcript\n old.controlled_linear_xor(b,F,p,head,co);b.cx(F,value[0]);divide_lowwidth369(b,combine(head),value,K);b.cx(F,value[0])\n cx(b,head[1],head[0],c);old.eval_implicit(b,A,Q,c,True)\n b.cx(F,A[0][0]);low_product(b,head,A,B);b.cx(F,A[0][0]);q_times_low(b,B,kap,True)\n value=old.eval_implicit(b,A,Q,c);zero_finish(b,value,F,fm);old.eval_implicit(b,A,Q,c,True)\n b.cz(combine(head),hm);b.free(hm)\n return out,A,transcript\n\n\ndef absorb_even(b,head,A,B,Q,r,c,kap,reverse=False,transcript=None):\n k=len(B);out=head+B;target=out+[r]\n if not reverse:\n  value=old.eval_implicit(b,A,Q,c);F=dv.zero_flag(b,value);old.eval_implicit(b,A,Q,c,True)\n  b.cx(F,A[0][0]);spare=[b.alloc_qubits(12),b.alloc_qubits(12)];full=spare+B+[r]\n  linear_words(b,full,base.embedding_matrix(k+3,kap));full_product(b,head,A,Q,full);b.cx(F,A[0][0])\n  old.controlled_linear_xor(b,F,r,full[-3:-1],[kap,1])\n  hm=b.alloc_bits(24);b.hmr(combine(head),hm)\n  for x,z in zip(head,spare):b.swap(x,z);b.free(z)\n  fm=b.alloc_bits(1);b.hmr(F,fm[0]);b.free(F)\n  return out,A,(hm,fm)\n value=old.eval_implicit(b,A,Q,c);F=dv.zero_flag(b,value);old.eval_implicit(b,A,Q,c,True)\n old.controlled_linear_xor(b,F,r,out[-2:],[kap,1]);linear_words(b,target,base.embedding_matrix(k+3,kap),True)\n value=old.eval_implicit(b,A,Q,c);cx(b,head[1],head[0],c);hm,fm=transcript\n b.cx(F,value[0]);divide_lowwidth369(b,combine(head),value,K);b.cx(F,value[0]);cx(b,head[1],head[0],c)\n b.cz(combine(head),hm);b.free(hm);old.eval_implicit(b,A,Q,c,True)\n full_quotient(b,head,A,Q,B,r,kap)\n value=old.eval_implicit(b,A,Q,c);zero_finish(b,value,F,fm);old.eval_implicit(b,A,Q,c,True)\n return out,A,transcript\n\nsynthesize=FunctionType(alt.synthesize.__code__,dict(alt.synthesize.__globals__,old=ca.old,absorb=absorb_odd,absorb_even=absorb_even),\'synthesize\',alt.synthesize.__defaults__,alt.synthesize.__closure__)\nunsynthesize=FunctionType(alt.unsynthesize.__code__,dict(alt.unsynthesize.__globals__,old=ca.old,absorb=absorb_odd,absorb_even=absorb_even),\'unsynthesize\',alt.unsynthesize.__defaults__,alt.unsynthesize.__closure__)\n', 'blocks/dqi/pro-q2-897/source/final_decoder/conversion.py': '"""Reuse Euclid gap predicates with a sparse quantum flag bank.\n\nKEEP_FULL holds nine flags through the diagonal middle stages. KEEP_CHILD\nalso holds level3 through its child recursion; that flag is measured at\nforward return and reconstructed with its matching phase at reverse entry.\nOther levels stream masks between identical adjacent predicates. Every\nfinal flag measurement is repaid by its exact zero-predicate phase.\n"""\nfrom pathlib import Path\nimport sys\nfrom types import SimpleNamespace\nfrom contextlib import contextmanager\nsys.path.insert(0,str(Path(__file__).resolve().parents[1]))\nfrom exp.convert_short_hankel import phase34 as base\nfrom types import FunctionType\nfrom exp.fold_cached_norm.phase34 import phase_inverse34\nfrom exp.decoder.common import ar\n\ndef clone(fn,scope):\n    out=FunctionType(fn.__code__,scope,fn.__name__,fn.__defaults__,fn.__closure__)\n    out.__kwdefaults__=fn.__kwdefaults__\n    return out\n\ndef clear_flag(b,a,F):\n    m=b.alloc_bits(1);b.hmr(F,m[0]);b.free(F)\n    b.push_condition(m[0]);b.x(a)\n    nodes=b.alloc_qubits(len(a)-2)\n    b.init_and(a[0],a[1],nodes[0])\n    for i in range(1,len(nodes)):b.init_and(nodes[i-1],a[i+1],nodes[i])\n    b.ccz(nodes[-1],a[-1],m[0])\n    for i in reversed(range(len(nodes))):\n        if i:b.del_and(nodes[i-1],a[i+1],nodes[i])\n        else:b.del_and(a[0],a[1],nodes[0])\n    b.free(nodes);b.x(a);b.pop_condition();b.free(m)\n\ndef hold_flag(b,F):\n    m=b.alloc_bits(1);b.hmr(F,m[0]);b.free(F);return m\n\ndef consume_flag(b,F,m):\n    b.cz(F,m[0]);b.free(m)\n\nKEEP_FULL=frozenset((2,4,6,8,10,12,14,16,17))\nKEEP_CHILD=KEEP_FULL|{3}\n\n\ndef q_post(b,q,payloads,H,m,parent,reverse=False,phase=None):\n    if parent is None:\n        with q_inverse(b,q,m,reverse,phase) as qi:\n            for v in payloads:scale(b,v[-1],q,qi,not reverse)\n        return\n    tops,p,c=parent\n    if reverse or m+1 in KEEP_CHILD:F=phase[\'given_gap\'].pop(m+1)\n    else:\n        F=fresh_zero_flag(b,q)\n        consume_flag(b,F,phase[\'incoming_gap\'].pop(m+1))\n    b.cswap(F,q,p)\n    if not reverse:\n        for v,t in zip(payloads,tops):b.cswap(F,v[-1],t)\n    with q_inverse(b,q,m,reverse,phase) as qi:\n        if not reverse:\n            for v in payloads:scale(b,v[-1],q,qi,True)\n            for v,t in zip(payloads,tops):masked_product(b,F,v[-1],c,t)\n        else:\n            for v,t in zip(payloads,tops):masked_product(b,F,v[-1],c,t)\n            for v in payloads:scale(b,v[-1],q,qi,False)\n    if reverse:\n        for v,t in zip(payloads,tops):b.cswap(F,v[-1],t)\n    b.cswap(F,q,p)\n    if reverse and m+1 not in KEEP_CHILD:phase[\'return_mask\'][m+1]=hold_flag(b,F)\n    else:phase[\'returned_gap\'][m+1]=F\n\n\ndef look_solve(b,f,g,payloads,H=None,parent=None,reverse=False,phase=None):\n    m=len(f)-1\n    if m==1:return base_case(b,f,g,payloads,H,parent,reverse,phase)\n    keep=m in KEEP_CHILD;full=m in KEEP_FULL\n    F=phase[\'held_gap\'].pop(m) if reverse and full else fresh_zero_flag(b,g[-1])\n    if reverse and not full:consume_flag(b,F,phase[\'gap_flags\'].pop(m))\n    rotate(b,g,F)\n    if not reverse:carry=pre(b,f,g,payloads,defer=True)\n    else:carry=post_look(b,f,g,payloads,H,parent,True,incoming=phase[\'a\'].pop(m),defer=True,phase=phase)\n    canonical_permute(b,f,g,F)\n    if reverse or keep:phase[\'given_gap\'][m]=F\n    else:phase[\'incoming_gap\'][m]=hold_flag(b,F)\n    childparent=([v[-1]for v in payloads],f[-1],f[-2])\n    look_solve(b,g,f[:-2],[v[:-1]for v in payloads],None,childparent,reverse,phase)\n    if reverse and not keep:\n        F=fresh_zero_flag(b,g[-1]);consume_flag(b,F,phase[\'return_mask\'].pop(m))\n    else:F=phase[\'returned_gap\'].pop(m)\n    canonical_permute(b,f,g,F,True)\n    if not reverse:phase[\'a\'][m]=post_look(b,f,g,payloads,H,parent,incoming=carry,defer=True,phase=phase)\n    else:pre(b,f,g,payloads,True,incoming=carry)\n    rotate(b,g,F,True)\n    if reverse:clear_flag(b,g[-1],F)\n    elif full:phase[\'held_gap\'][m]=F\n    else:phase[\'gap_flags\'][m]=hold_flag(b,F)\n\n\ndef erase_product_zero(b, a, ai, F):\n    mask = b.alloc_bits(1)\n    b.hmr(F, mask[0]); b.free(F)\n    b.z(mask[0])\n    word = b.alloc_bits(12)\n    b.bit_store(word, False); b.cx(mask[0], word[0])\n    ar.phase_product(b, a, ai, word)\n    b.free(word); b.free(mask)\n\n\ndef cached_root_action(b, f, g, payloads, ai, Z, post, undo):\n    a = g[-1]; q0, q1 = f[-2:]\n    n = len(g)-1\n    if post:\n        operations = [(\'s\', v[-1], False) for v in payloads]\n        operations += [(\'m\', g[i], v[i], v[-1]) for v in payloads for i in range(n)]\n        operations += [(\'s\', v[-1], True) for v in payloads]\n        operations += [(\'c\', g[:-1], [q0, q1], f[:-2]),\n                       (\'s\', q0, False), (\'m\', q1, g[-2], q0)]\n    else:\n        operations = [(\'s\', v[-1], True) for v in payloads]\n        operations += [(\'m\', g[i], v[-1], v[i]) for v in payloads for i in range(n)]\n        operations += [(\'m\', q1, g[-2], q0), (\'s\', q0, True),\n                       (\'c\', g[:-1], [q0, q1], f[:-2])]\n    b.cx(Z, a[0])\n    old.runops(b, operations, a, ai, undo)\n    b.cx(Z, a[0])\n\n\ndef look_root_cached(b, O, g, payloads, reverse, phase):\n    m = len(O); keep = m in KEEP_CHILD; full = m in KEEP_FULL\n    F = phase[\'held_gap\'].pop(m) if reverse and full else old.fresh_zero_flag(b, g[-1])\n    if reverse and not full:\n        consume_flag(b, F, phase[\'gap_flags\'].pop(m))\n    old.rotate(b, g, F)\n    a = g[-1]\n    ai, Z = old.inverse_zero(b, a)\n    b.cx(Z, ai[0])\n    if reverse:\n        mask = phase.pop(\'_root_inverse_mask\')\n        b.cz(ai, mask); b.free(mask)\n    f = list(O) + [ai]\n    cached_root_action(b, f, g, payloads, ai, Z, post=reverse, undo=reverse)\n    erase_product_zero(b, a, ai, Z)\n    old.canonical_permute(b, f, g, F)\n    if reverse or keep:\n        phase[\'given_gap\'][m] = F\n    else:\n        phase[\'incoming_gap\'][m] = hold_flag(b, F)\n    parent = ([v[-1] for v in payloads], f[-1], f[-2])\n    old.look_solve(b, g, f[:-2], [v[:-1] for v in payloads], None,\n                   parent, reverse, phase)\n    if reverse and not keep:\n        F = old.fresh_zero_flag(b, g[-1])\n        consume_flag(b, F, phase[\'return_mask\'].pop(m))\n    else:\n        F = phase[\'returned_gap\'].pop(m)\n    old.canonical_permute(b, f, g, F, True)\n    Z = old.fresh_zero_flag(b, a)\n    cached_root_action(b, f, g, payloads, ai, Z, post=not reverse, undo=reverse)\n    erase_product_zero(b, a, ai, Z)\n    mask = b.alloc_bits(12)\n    b.hmr(ai, mask); b.free(ai)\n    if reverse:\n        phase_inverse34(b, a, mask, zero_mask=mask[0]); b.free(mask)\n    else:\n        phase[\'_root_inverse_mask\'] = mask\n    old.rotate(b, g, F, True)\n    if reverse:\n        clear_flag(b, g[-1], F)\n    elif full:\n        phase[\'held_gap\'][m] = F\n    else:\n        phase[\'gap_flags\'][m] = hold_flag(b, F)\n\n\ndef _bez_monic(b, O, E, payloads, reverse, history_scratch12, phase):\n    def reduce():\n        for i in range(len(O)):\n            ar.mul_xor0(b, E[-1], O[i], E[i])\n    reduce()\n    look_root_cached(b, O, E[:-1], payloads, reverse, phase)\n    reduce()\n\n\n\nscope = dict(vars(base._private), KEEP_CHILD=KEEP_CHILD, KEEP_FULL=KEEP_FULL, hold_flag=hold_flag,\n             consume_flag=consume_flag, clear_flag=clear_flag)\nfor name in [\'scale\', \'runops\', \'zero_toggle\', \'inverse_apply\', \'pre\',\n             \'canonical_permute\', \'masked_product\', \'fresh_zero_flag\',\n             \'post_look\', \'base\', \'bez_lookahead\']:\n    scope[name] = clone(getattr(base._private, name), scope)\nscope[\'base_case\'] = scope[\'base\']\nscope[\'q_inverse\'] = contextmanager(clone(base._private.q_inverse.__wrapped__, scope))\nscope[\'q_post\'] = clone(q_post, scope)\nscope[\'look_solve\'] = clone(look_solve, scope)\nshort = dict(base._short_scope, KEEP_CHILD=KEEP_CHILD, KEEP_FULL=KEEP_FULL, old=SimpleNamespace(**scope),\n             hold_flag=hold_flag, consume_flag=consume_flag, clear_flag=clear_flag,\n             ar=ar, erase_product_zero=erase_product_zero,\n             phase_inverse34=phase_inverse34)\nfor name, fn in [(\'cached_root_action\', cached_root_action),\n                 (\'look_root_cached\', look_root_cached), (\'_bez_monic\', _bez_monic)]:\n    short[name] = clone(fn, short)\n_core_convert = clone(base._short_scope[\'convert_safe\'], short)\n\n\ndef convert_safe(b, A, B, reverse=False, history_scratch12=None,\n                 phase_token=None, *, monic_root=False):\n    keys = [\'gap_flags\', \'given_gap\', \'incoming_gap\', \'return_mask\',\n            \'returned_gap\', \'held_gap\']\n    if phase_token is None:\n        phase_token = {\'a\': {}, \'q\': {}, \'_short_monic\': bool(monic_root)}\n        for key in keys:\n            phase_token[key] = {}\n    result = _core_convert(b, A, B, reverse, history_scratch12, phase_token,\n                          monic_root=monic_root)\n    if reverse:\n        assert all(not phase_token[key] for key in keys)\n        assert \'_root_inverse_mask\' not in phase_token\n    return result\n', 'blocks/dqi/pro-q2-897/source/final_decoder/decode.py': '"""Nine homogeneous external chains, a trailing jet, and exact phase replay.\n\nThe only new dependencies are final_decoder/synthesis.py and conversion.py.\nEvery other import is an unchanged decoderkit module.\n"""\nfrom pathlib import Path\nimport sys\nROOT=Path(__file__).resolve().parents[1]\nsys.path[:0]=[str(ROOT),str(ROOT.parent)]\nimport decode as baseline\nfrom exp.decoder.common import *\nfrom exp.confluent_extract.model import synthetic\nfrom exp.inverse12_escape.joint import inverse24_forward190,inverse24_reverse209\nfrom exp.jet24.pro_jet import jet_forward,jet_reverse\nfrom exp.inverse19_joint.kernel import inverse_initialized\nfrom exp.fold_cached_norm.phase34 import phase_inverse34\nfrom exp.convert_short_hankel.phase34 import _private as guarded\nfrom final_decoder import synthesis as folds\nfrom final_decoder.conversion import convert_safe,clear_flag\n\n\ndef geometry():\n    groups=[];qs=[];P=[0,0,1]\n    for i in range(17):\n        c=2*i;kap=mul(c,c)^c^K\n        groups.append(dict(zero=False,c=c,kap=kap,gamma=c|4096,\n                           count=2,head=2*i,word=4*i,t=1))\n        qs.extend([(kap,1)]*2)\n        for _ in range(2):P=pmul(P,[kap,1,1])\n    groups.append(dict(zero=True,c=0,kap=0,gamma=0,\n                       count=1,head=34,word=68,t=0))\n    qs.append((0,0));matrix=[]\n    for g in groups[:-1]:\n        Q=P\n        for _ in range(2):\n            Q,r=synthetic(Q,g[\'gamma\']);assert r==0\n            q=Q+[0]*(70-len(Q))\n            matrix.extend([tuple(x&4095 for x in q),tuple(x>>12 for x in q)])\n    matrix.extend([tuple(P[1:]+[0]*(71-len(P))),\n                   tuple(P[2:]+[0]*(72-len(P)))])\n    return groups,qs,P,matrix\n\n\ngroups,qs,P,matrix=geometry()\ncs=[g[\'c\']for g in groups[:-1]for _ in range(2)]\nks=[kap for kap,_ in qs[:-1]]\nPscan,rem=divmodp(P,[ks[0],1,1]);assert rem==[0]\ncorrections=[];c=0\nfor kap in ks:corrections.append(c);c^=inv(kap)\nassert c==0\n\n\ndef square12(b,x,reverse=False):\n    ar.linear_inplace(b,x,[power(1<<j,2048 if reverse else 2)for j in range(12)])\n\n\ndef scale_word(b,x,f,fi):\n    out=b.alloc_qubits(12);ar.mul_xor0(b,x,f,out)\n    ar.erase_product(b,out,fi,x);b.swap(x,out);b.free(out)\n\n\ndef masked_update(b,x,f,F):\n    """Return the phase mask of F*x_old, after x += F*x_old*(f+1)."""\n    u=b.alloc_qubits(12)\n    for a,z in zip(x,u):b.init_and(F,a,z)\n    b.x(f[0]);ar.mul_xor0(b,u,f,x);b.x(f[0])\n    m=b.alloc_bits(12);b.hmr(u,m);b.free(u);return m\n\n\ndef phase_masked_product(b,x,f,F,m):\n    t=b.alloc_qubits(1)[0];field=ar.field(12)\n    for i,a in enumerate(x):\n        b.init_and(F,a,t)\n        for j,c in enumerate(f):\n            for k in ar.bits(field.mul(1<<i,1<<j)):b.ccz(t,c,m[k])\n        b.del_and(F,a,t)\n    b.free(t);b.free(m)\n\n\ndef reverse_masked_multiply(b,x,f,fi,F,m):\n    fresh=masked_update(b,x,fi,F)\n    for a,bit in zip(x,m):b.ccz(F,a,bit)\n    b.free(m);phase_masked_product(b,x,f,F,fresh)\n\n\ndef deferred_multiply(b,x,f):\n    out=b.alloc_qubits(12);ar.mul_xor0(b,x,f,out)\n    m=b.alloc_bits(12);b.hmr(x,m);b.swap(x,out);b.free(out);return m\n\n\ndef reverse_deferred_multiply(b,x,f,fi,m):\n    fresh=deferred_multiply(b,x,fi)\n    b.cz(x,m);b.free(m);ar.phase_product(b,x,f,fresh);b.free(fresh)\n\n\ndef controlled_inverse(b,x,F,reverse=False,token=None):\n    xi=inverse_initialized(b,x)\n    if reverse:b.cz(xi,token);b.free(token)\n    b.cswap(F,x,xi)\n    m=b.alloc_bits(12);b.hmr(xi,m);b.free(xi)\n    if reverse:phase_inverse34(b,x,m);b.free(m)\n    else:return m\n\n\ndef term_correction_with_inverse(b,p,q,d,u,v,pole,qi,reverse=False,token=None,Z=None):\n    if reverse:\n        Z=guarded.fresh_zero_flag(b,d);b.cz(Z,token[0]);b.free(token)\n    F=b.alloc_qubits(1)[0];b.cx(Z,F);b.ccx(Z,pole,F)\n    def add_product(x,target):\n        w=b.alloc_qubits(12);ar.mul_xor0(b,x,qi,w)\n        b.ccx(F,w,target);ar.erase_product(b,x,qi,w);b.free(w)\n    if reverse:add_product(v,u);add_product(p,v)\n    else:add_product(p,v);add_product(v,u)\n    b.ccx(Z,pole,F);b.cx(Z,F);b.free(F)\n    if reverse:return Z\n    zm=b.alloc_bits(1);b.hmr(Z,zm[0]);b.free(Z);return zm\n\n\ndef tail_finish(b,p,q,d,u,v,pole,reverse=False,token=None,Z=None):\n    """Share q\'s inverse0 across correction and clipped pole products."""\n    if reverse:b.cswap(pole,v,d);b.cswap(pole,u,v)\n    qi,Zq=guarded.inverse_zero(b,q)\n    if reverse:guarded.consume(b,qi,Zq,token[\'inverse\'])\n    if not reverse:tt=term_correction_with_inverse(b,p,q,d,u,v,pole,qi,Z=Z)\n    b.cx(Zq,q[0]);b.cx(Zq,qi[0])\n    if reverse:\n        reverse_masked_multiply(b,d,qi,q,pole,token[\'d\'])\n        reverse_masked_multiply(b,v,q,qi,pole,token[\'v\'])\n    else:\n        vm=masked_update(b,v,q,pole);dm=masked_update(b,d,qi,pole)\n    b.cx(Zq,qi[0]);b.cx(Zq,q[0])\n    if reverse:Z=term_correction_with_inverse(b,p,q,d,u,v,pole,qi,True,token[\'term\'])\n    qi_token=guarded.erase(b,q,qi,Zq,not reverse)\n    if reverse:return Z\n    if not reverse:\n        b.cswap(pole,u,v);b.cswap(pole,v,d)\n        return dict(inverse=qi_token,term=tt,v=vm,d=dm)\n\n\ndef tail_inplace(b,p,q,d,u,v,pole,reverse=False,token=None):\n    """(d,u,v) <-> (delta,N0,N1), with p,q the implicit leaders."""\n    if not reverse:\n        ar.mul_xor0(b,q,v,p)\n        square12(b,q);ar.mul_xor0(b,q,u,d);square12(b,q,True)\n        masks=[]\n        def divide(pi):\n            b.x(pole);masks.append(masked_update(b,d,pi,pole));b.x(pole)\n        tp=guarded.inverse_apply(b,p,divide,defer=True);tp=(tp,masks[0]);ar.mul_xor0(b,q,v,p)\n        def multiply(di,m):\n            b.x(pole);vm=masked_update(b,v,d,pole);b.x(pole)\n            ar.mul_xor0(b,q,u,v);um=deferred_multiply(b,u,d)\n            return vm,um\n        Z=guarded.fresh_zero_flag(b,d);b.cx(Z,d[0]);td=multiply(None,None);b.cx(Z,d[0])\n        tf=tail_finish(b,p,q,d,u,v,pole,Z=Z)\n        return dict(p=tp,d=td,finish=tf)\n    Z=tail_finish(b,p,q,d,u,v,pole,True,token[\'finish\'])\n    def divide(di,m):\n        vm,um=m\n        reverse_deferred_multiply(b,u,d,di,um);ar.mul_xor0(b,q,u,v)\n        b.x(pole);reverse_masked_multiply(b,v,d,di,pole,vm);b.x(pole)\n    di=inverse_initialized(b,d);b.cx(Z,d[0]);b.cx(Z,di[0])\n    divide(di,token[\'d\']);b.cx(Z,di[0]);b.cx(Z,d[0])\n    guarded.erase(b,d,di,Z,False);ar.mul_xor0(b,q,v,p)\n    def multiply(pi):\n        b.x(pole);reverse_masked_multiply(b,d,pi,p,pole,token[\'p\'][1]);b.x(pole)\n    guarded.inverse_apply(b,p,multiply,incoming=token[\'p\'][0]);ar.mul_xor0(b,q,v,p)\n    square12(b,q);ar.mul_xor0(b,q,u,d);square12(b,q,True)\n\n\ndef synthesize(b,words,pole):\n    heads=[words[2*i:2*i+2]for i in range(34)];u,v=words[-2:]\n    d,border=heads[0];r=b.alloc_qubits(12)\n    b.cx(border,r);b.cx(border,d);b.cx(r,border);b.x(border[0])\n    for i in range(1,34):\n        qi,target=(r,border)if i%2 else(border,r)\n        ar.mul_xor0(b,heads[i][1],qi,target)\n        folds.add_tail(b,heads[i],qi,d,bool(i%2))\n    ar.mul_xor0(b,r,border,d);tt=tail_inplace(b,r,border,d,u,v,pole)\n    # The exact chart remains live through the diagonal middle stages.\n    A,B,hist=[u,v],[d],[]\n    for i in reversed(range(1,34)):\n        aa,bb=A,B\n        if i%2:\n            ar.mul_xor0(b,heads[i][1],r,border);qi,p=r,border\n            A,B,t=folds.sg[\'absorb\'](b,heads[i],A,B,qi,p,None,0,cs[i],ks[i],wide=False)\n        else:\n            qi,p=border,r\n            A,B,t=folds.sg[\'absorb_even\'](b,heads[i],A,B,qi,r,cs[i],ks[i])\n        hist.append((i,aa,bb,qi,p,t))\n    b.x(border[0]);b.free(border);C=B+[r]\n    def undo():\n        folds.ug[\'reclaim\'](b,[border]);b.x(border[0])\n        for i,aa,bb,qi,p,t in reversed(hist):\n            if i%2:\n                folds.ug[\'absorb\'](b,heads[i],aa,bb,qi,p,None,0,cs[i],ks[i],True,t,False)\n                ar.mul_xor0(b,heads[i][1],r,border)\n            else:folds.ug[\'absorb_even\'](b,heads[i],aa,bb,qi,r,cs[i],ks[i],True,t)\n        tail_inplace(b,r,border,d,u,v,pole,True,tt);ar.mul_xor0(b,r,border,d)\n        for i in reversed(range(1,34)):\n            qi,target=(r,border)if i%2 else(border,r)\n            folds.add_tail(b,heads[i],qi,d,bool(i%2));ar.mul_xor0(b,heads[i][1],qi,target)\n        b.cx(r,d);b.x(border[0]);b.cx(r,border);b.cx(border,r);b.free(r)\n    return C,A,undo\n\n\ndef step(b,head,N,D,L,kap,c,reverse=False):\n    a,h=head\n    if reverse:scale(b,N,inv(kap));b.swap(N,D);ar.mul_xor0(b,a,D,N);scale(b,L,inv(kap))\n    cx(b,a,h,c);square12(b,D);ar.mul_xor0(b,h,D,L);square12(b,D,True);cx(b,a,h,c)\n    if not reverse:scale(b,L,kap);ar.mul_xor0(b,a,D,N);b.swap(N,D);scale(b,N,kap)\n\n\ndef first_step(b,head,N,D,L,kap):\n    b.cx(head[0],N);b.swap(N,D);scale(b,N,kap)\n    b.cx(head[1],L);scale(b,L,kap)\n\n\ndef scalar_step(b,head,N,D,kap,reverse=False,first=False):\n    if reverse:scale(b,N,inv(kap));b.swap(N,D)\n    if first:b.cx(head[0],N)\n    else:ar.mul_xor0(b,head[0],D,N)\n    if not reverse:b.swap(N,D);scale(b,N,kap)\n\n\ndef normal_form(b,N,D,L):\n    """Normalize by the guaranteed nonzero D; defer only its inverse mask."""\n    di=inverse_initialized(b,D);scale_word(b,N,di,D)\n    square12(b,D);square12(b,di);scale_word(b,L,di,D)\n    square12(b,di,True);square12(b,D,True)\n    m=b.alloc_bits(12);b.hmr(di,m);b.free(di);return m\n\n\ndef extraction(b,words,reverse=False,token=None):\n    heads=[words[2*i:2*i+2]for i in range(34)];N,L=words[-2:]\n    if not reverse:\n        ext=external_extraction(b,words);D=b.alloc_qubits(12);b.x(D[0])\n        first_step(b,heads[0],N,D,L,ks[0])\n        for head,kap,c in zip(heads[1:],ks[1:],corrections[1:]):step(b,head,N,D,L,kap,c)\n        pole=b.alloc_qubits(1)[0];guarded.zero_toggle(b,D,pole);b.cswap(pole,N,D)\n        norm=normal_form(b,N,D,L);ci=controlled_inverse(b,L,pole)\n        gauge=b.alloc_bits(12);b.hmr(D,gauge);b.free(D)\n        return dict(external=ext,pole=pole,normalization=norm,gauge_mask=gauge,chart_inverse=ci)\n    pole=token[\'pole\'];controlled_inverse(b,L,pole,True,token[\'chart_inverse\'])\n    D=b.alloc_qubits(12);b.x(D[0]);b.cswap(pole,N,D)\n    for head,kap,c in reversed(list(zip(heads,ks,corrections))[1:]):step(b,head,N,D,L,kap,c,True)\n    scale(b,N,inv(ks[0]));b.swap(N,D);scale(b,L,inv(ks[0]))\n    extra_z_mask=normal_form(b,N,D,L)\n    b.cx(heads[0][0],N);b.cx(heads[0][1],L)\n    b.cz(D,token[\'normalization\']);b.free(token[\'normalization\'])\n    inverse_mask=b.alloc_bits(12);b.hmr(D,inverse_mask);b.free(D)\n    z_mask=token[\'gauge_mask\'];b.cx(extra_z_mask,z_mask);b.free(extra_z_mask)\n    D=b.alloc_qubits(12);b.x(D[0]);scalar_step(b,heads[0],N,D,ks[0],first=True)\n    for head,kap in zip(heads[1:],ks[1:]):scalar_step(b,head,N,D,kap)\n    b.cswap(pole,N,D);b.cz(D,z_mask);b.free(z_mask)\n    phase_inverse34(b,D,inverse_mask);b.free(inverse_mask);b.cswap(pole,N,D)\n    guarded.zero_toggle(b,D,pole)\n    for head,kap in reversed(list(zip(heads[1:],ks[1:]))):scalar_step(b,head,N,D,kap,True)\n    scalar_step(b,heads[0],N,D,ks[0],True,first=True)\n    b.x(D[0]);b.free(D);b.free(pole);external_extraction(b,words,True,token[\'external\'])\n\n\ndef decode(b,S,E):\n    with b.mark(\'mask_projection\'):mt=baseline.project(b,S,E)\n    with b.mark(\'initial\'):linear_words(b,S,matrix)\n    with b.mark(\'extract\'):et=extraction(b,S)\n    with b.mark(\'synthesis\'):A,B,undo=synthesize(b,S,et[\'pole\'])\n    with b.mark(\'convert\'):ct=convert_safe(b,A,B,monic_root=True)\n    with b.mark(\'scan\'):baseline.scan(b,A,B,E,skip_points=baseline.SELECTED,node_polynomial=Pscan)\n    with b.mark(\'unconvert\'):convert_safe(b,A,B,True,phase_token=ct)\n    with b.mark(\'unsynthesis\'):undo()\n    with b.mark(\'unextract\'):extraction(b,S,True,et)\n    with b.mark(\'uninitial\'):linear_words(b,S,matrix,True)\n    with b.mark(\'restore_masks\'):baseline.restore(b,E,mt)\n\n\nfrom functools import lru_cache\nfrom exp.mul99.kernels import mul24_xor99\nfrom exp.mul99.schedules import multiply_forward99\nfrom exp.arithmetic.division import phase_product24,zero_flag\nfrom exp.noscratch.schedules import norm_xor_noscratch,inverse_stream_noscratch\nfrom exp.noscratch.kernels import mul_xor_noscratch\nfrom exp.mul99.phase36 import phase_inverse36_pivot\n\nEXTERNAL_HOMOGENEOUS=frozenset(range(8,17))\n\n@lru_cache(None)\ndef external_square_columns(reverse=False):\n    def epow(x,n):\n        out=1\n        while n:\n            if n&1:out=emul(out,x)\n            x=emul(x,x);n>>=1\n        return out\n    return tuple(epow(1<<i,1<<23 if reverse else 2)for i in range(24))\n\n\ndef external_square24(b,x,reverse=False):ar.linear_inplace(b,x,external_square_columns(reverse))\n\n\ndef external_mul_head(b,head,D,N,ct):\n    a,h=head;cx(b,h,a,ct)\n    mul24_xor99(b,combine(head),D,N,K)\n    cx(b,h,a,ct)\n\n\ndef external_corrected_product(b,head,D,L,ct,c):\n    a,h=head\n    if not c:\n        ar.mul_xor0(b,h,D[:12],L[:12]);ar.mul_xor0(b,h,D[12:],L[12:])\n    else:\n        scale(b,a,c);cx(b,h,a,mul(c,ct)^1);scale(b,h,c)\n        mul24_xor99(b,combine(head),D,L,K)\n        scale(b,h,inv(c));cx(b,h,a,mul(c,ct)^1);scale(b,a,inv(c))\n\n\ndef external_chain_step(b,head,N,D,L,q,ct,c,reverse=False,first=False):\n    if reverse:\n        scalepair(b,N,inv(q));b.swap(N,D);external_mul_head(b,head,D,N,ct);scalepair(b,L,inv(q))\n    if first and not reverse:\n        b.cx(head[1],L[:12]);scalepair(b,L,q)\n        a,h=head;cx(b,h,a,ct);b.cx(combine(head),N);cx(b,h,a,ct)\n        b.swap(N,D);scalepair(b,N,q)\n        return\n    external_square24(b,D);external_corrected_product(b,head,D,L,ct,c);external_square24(b,D,True)\n    if not reverse:\n        scalepair(b,L,q);external_mul_head(b,head,D,N,ct);b.swap(N,D);scalepair(b,N,q)\n\n\ndef external_scalar_step(b,head,N,D,q,ct,reverse=False,first=False):\n    if reverse:scalepair(b,N,inv(q));b.swap(N,D)\n    if first:\n        a,h=head;cx(b,h,a,ct);b.cx(combine(head),N);cx(b,h,a,ct)\n    else:external_mul_head(b,head,D,N,ct)\n    if not reverse:b.swap(N,D);scalepair(b,N,q)\n\n\ndef external_ordinary_node(b,words,gi,reverse=False,token=None):\n    g=groups[gi];u=combine(words[g[\'word\']:g[\'word\']+2]);v=combine(words[g[\'word\']+2:g[\'word\']+4])\n    history=[]if not reverse else list(token)\n    for h in reversed(range(2*gi))if reverse else range(2*gi):\n        a,d=words[2*h:2*h+2];q=ks[h]^g[\'kap\']\n        def translate():\n            b.cx(a,u[:12]);cx(b,d,u[:12],g[\'c\']);b.cx(d,u[12:]);b.cx(d,v[:12])\n        if not reverse:\n            translate();history.append(jet_forward(b,u,v));scalepair(b,v,q);b.cx(u,v);scalepair(b,u,q)\n        else:\n            scalepair(b,u,inv(q));b.cx(u,v);scalepair(b,v,inv(q));jet_reverse(b,u,v,history.pop());translate()\n    if reverse:assert not history\n    return history\n\n\ndef external_normalize_paired(b,N,D,L,reverse=False,token=None):\n    b.cx(D[12:],D[:12]);nm=multiply_forward99(b,N,D,K)\n    external_square24(b,D);lm=multiply_forward99(b,L,D,K);external_square24(b,D,True)\n    b.cx(D[12:],D[:12])\n    u=b.alloc_qubits(12);norm_xor_noscratch(b,D,u,K)\n    if reverse:b.cz(u,token);b.free(token)\n    m=inverse_stream_noscratch(b,u);ar._phase_norm(b,D,m,K);b.free(m)\n    masks=[]\n    for word,squared in [(N,False),(L,True)]:\n        if squared:square12(b,u)\n        row=[]\n        for off in (0,12):\n            half=word[off:off+12];out=b.alloc_qubits(12)\n            mul_xor_noscratch(b,half,u,out);hm=b.alloc_bits(12)\n            b.hmr(half,hm);b.swap(half,out);b.free(out);row.append(hm)\n        masks.append(row)\n        if squared:square12(b,u,True)\n    if reverse:um=u\n    else:\n        um=b.alloc_bits(12);b.hmr(u,um);b.free(u)\n    norm=b.alloc_qubits(12);norm_xor_noscratch(b,D,norm,K)\n    for word,row,squared in [(N,masks[0],False),(L,masks[1],True)]:\n        if squared:square12(b,norm)\n        for off,hm in zip((0,12),row):\n            ar.phase_product(b,word[off:off+12],norm,hm);b.free(hm)\n        if squared:square12(b,norm,True)\n    if not reverse:ar._erase_norm(b,D,norm,K);b.free(norm)\n    phase_product24(b,N,D,nm,K);b.free(nm)\n    external_square24(b,D);phase_product24(b,L,D,lm,K);b.free(lm);external_square24(b,D,True)\n    return (norm,um) if reverse else um\n\n\ndef external_inverse_phase_norm_pair(b,x,mask,reciprocal,norm):\n    b.cx(x[12:],x[:12])\n    ar.phase_product(b,x[:12],reciprocal,mask[:12]);ar.phase_product(b,x[12:],reciprocal,mask[12:])\n    b.cx(x[12:],x[:12])\n    im=b.alloc_bits(12);b.hmr(reciprocal,im);b.free(reciprocal)\n    phase_inverse36_pivot(b,norm,im);b.free(im)\n    nm=b.alloc_bits(12);b.hmr(norm,nm);b.free(norm)\n    ar._phase_norm(b,x,nm,K);b.free(nm)\n\n\ndef external_node_chain(b,words,gi,reverse=False,token=None,normalizer=None):\n    g=groups[gi];N=combine(words[g[\'word\']:g[\'word\']+2]);L=combine(words[g[\'word\']+2:g[\'word\']+4])\n    heads=[words[2*i:2*i+2]for i in range(2*gi)]\n    params=[];c=0\n    for i in range(2*gi):\n        q=ks[i]^g[\'kap\'];assert q\n        params.append((q,c));c^=inv(q)\n    assert c==0\n    if not reverse:\n        D=b.alloc_qubits(24);b.x(D[0])\n        for i,(head,(q,c))in enumerate(zip(heads,params)):\n            external_chain_step(b,head,N,D,L,q,g[\'c\'],c,first=i==0)\n        pole=zero_flag(b,D);b.cswap(pole,N,D)\n        norm=external_normalize_paired(b,N,D,L)\n        gauge=b.alloc_bits(24);b.hmr(D,gauge);b.free(D)\n        return dict(pole=pole,gauge=gauge,norm=norm)\n    pole=token[\'pole\'];D=b.alloc_qubits(24);b.x(D[0]);b.cswap(pole,N,D)\n    for head,(q,c)in reversed(list(zip(heads,params))[1:]):external_chain_step(b,head,N,D,L,q,g[\'c\'],c,True)\n    q0=params[0][0];scalepair(b,N,inv(q0));b.swap(N,D);scalepair(b,L,inv(q0))\n    norm=external_normalize_paired(b,N,D,L,True,token[\'norm\'])\n    a,h=heads[0];cx(b,h,a,g[\'c\']);b.cx(combine(heads[0]),N);cx(b,h,a,g[\'c\']);b.cx(h,L[:12])\n    invmask=b.alloc_bits(24);b.hmr(D,invmask);b.free(D)\n    D=b.alloc_qubits(24);b.x(D[0])\n    for i,(head,(q,c))in enumerate(zip(heads,params)):external_scalar_step(b,head,N,D,params[i][0],g[\'c\'],first=i==0)\n    b.cswap(pole,N,D);b.cz(D,token[\'gauge\']);b.free(token[\'gauge\'])\n    reciprocal,direct=norm\n    external_inverse_phase_norm_pair(b,D,invmask,reciprocal,direct);b.free(invmask);b.cswap(pole,N,D)\n    clear_flag(b,D,pole)\n    for i in reversed(range(len(heads))):external_scalar_step(b,heads[i],N,D,params[i][0],g[\'c\'],True,first=i==0)\n    b.x(D[0]);b.free(D)\n\n\ndef external_extraction(b,words,reverse=False,token=None):\n    history=[]if not reverse else list(token)\n    for gi in reversed(range(17))if reverse else range(17):\n        g=groups[gi];u=combine(words[g[\'word\']:g[\'word\']+2]);v=combine(words[g[\'word\']+2:g[\'word\']+4])\n        a,d=words[g[\'word\']:g[\'word\']+2]\n        if not reverse:\n            nt=external_node_chain(b,words,gi,normalizer=None)if gi in EXTERNAL_HOMOGENEOUS else external_ordinary_node(b,words,gi)\n            cx(b,d,a,g[\'c\']);b.cx(d,v[:12]);local=inverse24_forward190(b,v,K)\n            cx(b,words[g[\'word\']+3],words[g[\'word\']+2],g[\'c\'])\n            history.append((nt,local))\n        else:\n            nt,local=history.pop();cx(b,words[g[\'word\']+3],words[g[\'word\']+2],g[\'c\'])\n            inverse24_reverse209(b,v,local,K);b.cx(d,v[:12]);cx(b,d,a,g[\'c\'])\n            if gi in EXTERNAL_HOMOGENEOUS:external_node_chain(b,words,gi,True,nt,None)\n            else:external_ordinary_node(b,words,gi,True,nt)\n    assert not reverse or not history\n    return history\n', 'blocks/dqi/pro-q2-897/source/final_decoder/synthesis.py': '"""Synthesis with additive continuants, shared norms, and no outer fold.\n\nEvery imported dependency is an unchanged file in decoderkit.\nThe caller compensates for the omitted fixed quadratic in the scan.\nAll arithmetic uses circ builder gates.\n"""\nimport sys\nfrom pathlib import Path\nHERE = Path(__file__).resolve().parents[1]\nsys.path[:0] = [str(HERE), str(HERE.parent)]\nimport decode as baseline\nfrom exp.decoder.common import ar\nfrom exp.mul99.schedules import multiply_forward99\nfrom exp.noscratch.schedules import norm_xor_noscratch\nfrom exp.noscratch.kernels import mul_xor_noscratch\nfrom exp.arithmetic.division import phase_product24\nfrom exp.inverse_search.inverse20_iso import inverse6_xor20\nfrom exp.phase24_tradeoff.phase6small import phase_inverse6_classical6\nfrom exp.zero_padding_synthesis.kernel import *\nfrom exp.fold_cached_norm import kernel as FK\n\n\n\ndef divide_fused(b, x, y, F, fm, after, K=None):\n    """Consume F=[y=0] and prior phase fm; divide with identity at zero.\n\n    The callback sees restored y and the live flag.  It may change x, but\n    must restore y before returning.  Every norm/flag measurement is closed.\n    """\n    b.cx(F, y[0])\n    b.cx(y[12:], y[:12])\n    oldmask = multiply_forward99(b, x, y, K)\n    b.cx(y[12:], y[:12])\n    N = b.alloc_qubits(12)\n    norm_xor_noscratch(b, y, N, K)\n    conj = [ar.field(12).frobenius(1 << j, 6) for j in range(12)]\n    ar.linear_inplace(b, N, conj)\n    masks = []\n    for off in (0, 12):\n        half = x[off:off + 12]\n        out = b.alloc_qubits(12)\n        ar.mul_xor0(b, half, N, out)\n        m = b.alloc_bits(12)\n        b.hmr(half, m)\n        b.swap(half, out)\n        b.free(out)\n        masks.append(m)\n    ar.linear_inplace(b, N, conj)\n    to, frm, k6 = ar.tower(12)\n    ar.linear_inplace(b, N, to)\n    M = b.alloc_qubits(6)\n    norm_xor_noscratch(b, N, M, k6)\n    ar.linear_inplace(b, N, frm)\n    t = inverse6_xor20(b, M)\n    for off in (0, 12):\n        half = x[off:off + 12]\n        ar.linear_inplace(b, half, to)\n        for j in (0, 6):\n            part = half[j:j + 6]\n            out = b.alloc_qubits(6)\n            mul_xor_noscratch(b, part, t, out)\n            ar.erase_product(b, out, M, part)\n            b.swap(part, out)\n            b.free(out)\n        ar.linear_inplace(b, half, frm)\n    for off, m in zip((0, 12), masks):\n        ar.phase_product(b, x[off:off + 12], N, m)\n        b.free(m)\n    phase_product24(b, x, y, oldmask, K)\n    b.free(oldmask)\n\n    # Norms of a clipped zero factor differ only in their constant bit.\n    b.cx(F, y[0])\n    b.cx(F, N[0])\n    b.cx(F, M[0])\n    after()\n    fmask = b.alloc_bits(1)\n    b.hmr(F, fmask[0])\n    b.free(F)\n    b.cx(fm[0], fmask[0])\n    b.free(fm)\n    b.z(fmask[0])\n    pm = b.alloc_bits(6)\n    b.bit_store(pm, False)\n    b.cx(fmask[0], pm[0])\n    ar.phase_product(b, M, t, pm)\n    b.free(pm)\n    b.free(fmask)\n    tm = b.alloc_bits(6)\n    b.hmr(t, tm)\n    b.free(t)\n    # t=inv0(M)+[M=0], so its constant output bit also masks the zero phase.\n    phase_inverse6_classical6(b, M, tm, zero_mask=tm[0])\n    b.free(tm)\n    ar.linear_inplace(b, N, to)\n    ar._erase_norm(b, N, M, k6)\n    b.free(M)\n    ar.linear_inplace(b, N, frm)\n    ar._erase_norm(b, y, N, K)\n    b.free(N)\n\n\ndef absorb_odd_fused(b, head, A, B, Q, p, virtual, pconst, c, kap,\n                     reverse=False, transcript=None, wide=False):\n    if not reverse:\n        return FK.prior.absorb_odd(b, head, A, B, Q, p, virtual, pconst,\n                                   c, kap, reverse, transcript, wide)\n    k = len(B)\n    v = epow(c, k + 2)\n    co = [v & 4095, v >> 12]\n    R_to_tops(b, head, B, c, True)\n    old.add_alias(b, p, head, co)\n    value = old.eval_implicit(b, A, Q, c)\n    F = dv.zero_flag(b, value)\n    hm, fm = transcript\n    old.controlled_linear_xor(b, F, p, head, co)\n    def after():\n        cx(b, head[1], head[0], c)\n        old.eval_implicit(b, A, Q, c, True)\n        b.cx(F, A[0][0])\n        low_product(b, head, A, B)\n        b.cx(F, A[0][0])\n        q_times_low(b, B, kap, True)\n        old.eval_implicit(b, A, Q, c)\n    divide_fused(b, combine(head), value, F, fm, after, K)\n    b.cz(combine(head), hm)\n    b.free(hm)\n    old.eval_implicit(b, A, Q, c, True)\n    return B + head, A, transcript\n\n\ndef absorb_even_fused(b, head, A, B, Q, r, c, kap,\n                      reverse=False, transcript=None):\n    if not reverse:\n        return FK.prior.absorb_even(b, head, A, B, Q, r, c, kap,\n                                    reverse, transcript)\n    k = len(B)\n    out = head + B\n    target = out + [r]\n    value = old.eval_implicit(b, A, Q, c)\n    F = dv.zero_flag(b, value)\n    old.eval_implicit(b, A, Q, c, True)\n    old.controlled_linear_xor(b, F, r, out[-2:], [kap, 1])\n    linear_words(b, target, base.embedding_matrix(k + 3, kap), True)\n    value = old.eval_implicit(b, A, Q, c)\n    cx(b, head[1], head[0], c)\n    hm, fm = transcript\n    def after():\n        cx(b, head[1], head[0], c)\n        old.eval_implicit(b, A, Q, c, True)\n        full_quotient(b, head, A, Q, B, r, kap)\n        old.eval_implicit(b, A, Q, c)\n    divide_fused(b, combine(head), value, F, fm, after, K)\n    b.cz(combine(head), hm)\n    b.free(hm)\n    old.eval_implicit(b, A, Q, c, True)\n    return out, A, transcript\n\n\n# Bind only the two inner fold reverses into the original private graph.\n_shared_unsynthesize = FK.bind(baseline.unsynthesize,\n    absorb=absorb_odd_fused, absorb_even=absorb_even_fused)\n\nsg = baseline.synthesize.__globals__\nug = _shared_unsynthesize.__globals__\nar = sg[\'ar\']\n\n\ndef add_tail(b, head, q, d, odd):\n    a, x = head\n    if not odd:\n        b.cx(x, a)\n    sg[\'add_tail_a\'](b, a, q, d)\n    if not odd:\n        b.cx(x, a)\n', 'blocks/dqi/pro-q2-897/source/playbook/exp/cf_check/audit.py': '"""Standalone classical finite-field EXTRACT/SYNTHESIZE audit (no circuits/verifier)."""\nimport argparse, json, random, time\nfrom collections import Counter\n\nclass Field:\n    def __init__(self, m, modulus):\n        self.m=m; self.n=1<<m; self.mask=self.n-1\n        self.exp=[]; self.log=[-1]*self.n; x=1\n        for i in range(self.n-1):\n            assert self.log[x]<0\n            self.log[x]=i; self.exp.append(x)\n            x<<=1\n            if x&self.n:x^=modulus\n        assert x==1\n        self.exp*=2\n        self.K=next(x for x in range(1,self.n) if self.trace(x)==1)\n    def mul(self,a,b):return self.exp[self.log[a]+self.log[b]] if a and b else 0\n    def inv(self,a):return self.exp[self.n-1-self.log[a]] if a else 0\n    def div(self,a,b):return self.mul(a,self.inv(b))\n    def trace(self,a):\n        t=0\n        for _ in range(self.m):t^=a;a=self.mul(a,a)\n        return t\n    def emul(self,u,v):\n        a=u&self.mask;b=u>>self.m;c=v&self.mask;d=v>>self.m\n        bd=self.mul(b,d)\n        return (self.mul(a,c)^self.mul(self.K,bd)) | ((self.mul(a,d)^self.mul(b,c)^bd)<<self.m)\n    def einv(self,u):\n        a=u&self.mask;b=u>>self.m\n        ni=self.inv(self.mul(a,a)^self.mul(a,b)^self.mul(self.K,self.mul(b,b)))\n        return self.mul(a^b,ni)|(self.mul(b,ni)<<self.m)\n    def escale(self,u,a):return self.mul(u&self.mask,a)|(self.mul(u>>self.m,a)<<self.m)\n    def ev(self,p,u):\n        out=0\n        for a in reversed(p):out=self.emul(out,u)^a\n        return out\n    def trim(self,p):\n        while len(p)>1 and not p[-1]:p.pop()\n        return p\n    def add(self,a,b):\n        c=[0]*max(len(a),len(b))\n        for i,x in enumerate(a):c[i]^=x\n        for i,x in enumerate(b):c[i]^=x\n        return self.trim(c)\n    def pmul(self,a,b):\n        c=[0]*(len(a)+len(b)-1)\n        for i,x in enumerate(a):\n            for j,y in enumerate(b):c[i+j]^=self.mul(x,y)\n        return self.trim(c)\n    def scale(self,p,c):return self.trim([self.mul(x,c) for x in p])\n    def pdiv(self,a,b):\n        a=a.copy();out=[0]*max(1,len(a)-len(b)+1);bi=self.inv(b[-1])\n        assert b!=[0]\n        while a!=[0] and len(a)>=len(b):\n            d=len(a)-len(b);v=self.mul(a[-1],bi);out[d]=v\n            for i,x in enumerate(b):a[i+d]^=self.mul(v,x)\n            self.trim(a)\n        return self.trim(out),a\n    def gcd(self,a,b):\n        while b!=[0]:a,b=b,self.pdiv(a,b)[1]\n        return self.scale(a,self.inv(a[-1]))\n    def derivative(self,p):return self.trim([p[i] if i%2 else 0 for i in range(1,len(p))] or [0])\n\ndef data(f, cs, betas, ws):\n    kappas=[f.mul(c,c)^c^f.K for c in cs]\n    nodes=[c|f.n for c in cs]\n    D=[1]\n    for b in betas:D=f.pmul(D,[b,1])\n    N=[0]\n    for b,w in zip(betas,ws):\n        quotient,rem=f.pdiv(D,[b,1]);assert rem==[0]\n        N=f.add(N,f.scale(quotient,w))\n    vals=[f.emul(f.ev(N,g),f.einv(f.ev(D,g))) for g in nodes]\n    return kappas,nodes,N,D,vals\n\ndef extract(f,cs,kappas,vals):\n    vals=vals.copy();heads=[];zeros=[]\n    for i,c in enumerate(cs):\n        b=vals[i]>>f.m;a=(vals[i]&f.mask)^f.mul(b,c);heads.append((a,b))\n        for j in range(i+1,len(cs)):\n            u=vals[j]^a^f.mul(b,cs[j])^(b<<f.m)\n            if not u:zeros.append((i,j))\n            vals[j]=f.escale(f.einv(u),kappas[i]^kappas[j])\n    return heads,zeros\n\ndef synth(f,kappas,heads):\n    p,q,d=0,1,0\n    for a,b in heads:p,q,d=q,p^f.mul(b,q),d^f.mul(p,q)^f.mul(a^b,f.mul(q,q))\n    tail=(p,q,d)\n    if q:A,B=[f.div(d,q),p],[q]\n    else:A,B=[p],[0]\n    for (a,b),kappa in reversed(list(zip(heads,kappas))):\n        A,B=f.add(f.pmul([a,b],A),f.pmul([kappa,1,1],B)),A\n    if not q:A,B=[0]+A,[0]+B\n    return f.trim(A),f.trim(B),tail\n\ndef rational_audit(f,kappas,nodes,N,D,heads):\n    """Follow exact rational transforms until the first genuine pole or termination."""\n    first_poles=[];termination=None\n    for i,(a,b) in enumerate(heads):\n        if f.ev(D,nodes[i])==0:return {\'head_pole\':i,\'poles\':first_poles}\n        assert f.ev(N,nodes[i])==f.emul(a^f.mul(b,nodes[i]&f.mask)^(b<<f.m),f.ev(D,nodes[i]))\n        H=f.add(N,f.pmul([a,b],D))\n        if H==[0]:termination=i;break\n        H,rem=f.pdiv(H,[kappas[i],1,1]);assert rem==[0]\n        poles=[j for j in range(i+1,len(nodes)) if f.ev(H,nodes[j])==0]\n        if poles:return {\'first_nonterminal_zero_step\':i,\'poles\':poles,\'residual_degree\':len(H)-1}\n        N,D=D,H\n    return {\'termination\':termination,\'poles\':first_poles}\n\ndef validate(m,modulus,h,repeats,seed):\n    f=Field(m,modulus);rng=random.Random(seed);cs=list(range(0,2*h,2));assert max(cs)<f.n\n    results=Counter();examples=[]\n    for t in range(h+1):\n        for trial in range(repeats):\n            betas=rng.sample(range(1,f.n),t);ws=[rng.randrange(1,f.n) for _ in betas]\n            kappas,nodes,N,D,vals=data(f,cs,betas,ws)\n            heads,zeros=extract(f,cs,kappas,vals);A,B,tail=synth(f,kappas,heads)\n            ra=rational_audit(f,kappas,nodes,N,D,heads)\n            correct=f.pmul(A,D)==f.pmul(B,N)\n            sqfree=f.gcd(B,f.derivative(B))==[1]\n            degree=len(B)-1\n            result={\'t\':t,\'trial\':trial,\'correct\':correct,\'degree\':degree,\'monic\':B[-1]==1,\'squarefree\':sqfree,\'tail\':tail,\'rational\':ra,\'zeros\':len(zeros)}\n            results[\'cases\']+=1;results[\'correct\']+=correct;results[\'monic\']+=B[-1]==1;results[\'degree_h\']+=degree==h;results[\'squarefree\']+=sqfree\n            if \'termination\' in ra and ra[\'termination\'] is not None:\n                term=ra[\'termination\'];results[\'terminated\']+=1\n                assert all(a==b==0 for a,b in heads[term+1:])\n                results[\'later_heads_zero\']+=1\n            if not correct or not sqfree or degree!=h or not result[\'monic\']:\n                if len(examples)<12:examples.append(dict(result,betas=betas,ws=ws,heads=heads,A=A,B=B))\n            if correct:\n                pad,rem=f.pdiv(B,D);assert rem==[0]\n                expected=D.copy()\n                for j in range(t+1,h,2):expected=f.pmul(expected,[kappas[j],1,1])\n                if not t%2:expected=[0]+expected\n                assert expected==B,(\'padding index pattern\',t,B,expected)\n                allowed=[kappas[i] for i in range(h)]\n                if not tail[1]:\n                    pad,rem=f.pdiv(pad,[0,1]);assert rem==[0]\n                for k in allowed:\n                    quotient,rem=f.pdiv(pad,[k,1,1])\n                    if rem==[0]:pad=quotient\n                assert pad==[1],pad\n                assert all(f.ev(B,b)!=0 for b in range(1,f.n) if b not in betas)\n    return {\'m\':m,\'K\':f.K,\'h\':h,\'repeats_per_weight\':repeats,\'counts\':dict(results),\'examples\':examples}\n\nif __name__==\'__main__\':\n    parser=argparse.ArgumentParser();parser.add_argument(\'--m\',type=int,default=12);parser.add_argument(\'--h\',type=int,default=35);parser.add_argument(\'--repeats\',type=int,default=3);parser.add_argument(\'--seed\',type=int,default=1)\n    args=parser.parse_args();mods={3:0b1011,4:0b10011,6:0b1000011,12:0x1053}\n    start=time.time();result=validate(args.m,mods[args.m],args.h,args.repeats,args.seed);result[\'seconds\']=time.time()-start\n    print(json.dumps(result,indent=2))\n', 'blocks/dqi/pro-q2-897/source/source_sha256.json': '{\n "build_897.py": "82026453bac5923fd9f5cb36236b644106a5bdd4b3a07b19c3f6a7c1b29fba0d",\n "decode.py": "c5c62693cbe65147fc63c3b75c3b04c4d785487a2dcdffed1cf103e90efabde4",\n "exp/alternating_gauge/even_fold.py": "784cf9b2873b5fa04da4233a528d7ffd8acf9d30fb2e39dc9992d46b1438cd16",\n "exp/alternating_gauge/kernel.py": "d99aa5eaef8a0df416270d8eee0fa55a9847ee7c08c3aa6ee437352e1abdd05a",\n "exp/alternating_gauge/scalar.py": "2fb3e5c4ec62df647de519647a57f51c765907ce1b6ecfe2d512035a91523b15",\n "exp/arithmetic/division.py": "28567073545c805e32636668cc0bb33f0dbc441a1de0dee2825774b3a33d2832",\n "exp/arithmetic/field.py": "c1b8b2ff583ffd5997856605c35d37f7116aa5811e52a3ce04a1ae7215d381c3",\n "exp/arithmetic/lowwidth.py": "c536fa2b42393b83aae2bef80eeda22d83a6c799ef555178e94e09f1f5abc22a",\n "exp/confluent_extract/kernel.py": "3acfc52dbf5994a85671b975be506b7967d2d075a3e1fc39f7c7cd16a8586ff3",\n "exp/confluent_extract/model.py": "9b8d6fbaad6aec5da87d5247a11276c2fa177883adff9eca030a9bc2f3286387",\n "exp/confluent_synthesis/kernel.py": "2eaabffce88e50ab3ce4719e764b3ed43c8bd6d8a6acf22a4461574268691c54",\n "exp/convert/convert.py": "7afdafa60806df5afbc0af8fd01417a3e09fc15c5dc83fdd28d68079e78df7a0",\n "exp/convert/hankel_check.py": "6d0ac6f55a70418049f46300863eeabb90a7c38b476b84db6e83d395edc3d72c",\n "exp/convert/kernels.py": "f5efa2f2b008cce86a0b751d400ee75777b77b6872d1875d6fae281cd7f09ef9",\n "exp/convert/repair.py": "8beceea9091a9656901290498fbdc0e4373963374b5ee8f7e98d257ece1daeaf",\n "exp/convert_short_hankel/kernel.py": "6dd64a123e4bd668bbf30e7d3c6449e738c8bf693f9a31bce1c9c39810ba7ddb",\n "exp/convert_short_hankel/phase34.py": "620985ba489a03cef51095157af60b7459732b72b68f85a10265f34905ed203f",\n "exp/decoder/common.py": "60f9db94c70dd746a429147b25121381934737fe6a576a5526ef70e1b12007a5",\n "exp/decoder/extract.py": "6597bd52d2bebedae8860acaa4aaeaf980c3d21977781382ba8926833c8f7696",\n "exp/decoder/safe_bridge.py": "023c81c775967a71ef44c571200bbe3ffb8b3da443fc2816e751856fc3f0d8d9",\n "exp/decoder/synthesis.py": "fdd7de226b7445007689ce2e50dbabbe77fdbc20e994ec41ac821a6ad2474c49",\n "exp/fold_cached_norm/kernel.py": "6a148ebe23eae9194ec0ddb921ff6da1d362565218913170a36691c45dd7c084",\n "exp/fold_cached_norm/phase34.py": "974d76939483c061e7b7995600fcdcb05e91529297725acfba41ce5ce808661f",\n "exp/fold_joint/divide.py": "8456426c171f36a716a39b457a0de5206b5d4db6c8553f1bfc919248ba1de851",\n "exp/fold_joint/kernel.py": "d4764e763c6ed1ff538b58c1cc628df0c9577300aae5a4d83a5338e0290e018f",\n "exp/inverse12_escape/joint.py": "4bcc5614693d646bd910b9ed4748b278d2ec8dc1584bf8a45d3e465dd757cdc1",\n "exp/inverse12_escape/zero1.py": "c9274f825d452082d88a98ded25a269aab247626a17032287c26ca1d8a71c724",\n "exp/inverse12_escape/zero_layout.json": "45d1361fcf582282d15cc6fabeab290f8e9fd9dc4c68a3e14ffff86e9f9a537f",\n "exp/inverse19_joint/kernel.py": "035152593a105b8178522a6edbf7cda3b709d8a86f51ae2148e71b01a1274913",\n "exp/inverse19_joint/layout.json": "0adee49170ae597ed5a7df4d4d0df0e495fba49013f6acd9ae6a3f06b9ad4ab4",\n "exp/inverse_search/baseline.py": "7aecb36f951122861fe0443eee92f1d65a10484fc86f2e535f3db44f5dd553da",\n "exp/inverse_search/inverse20.py": "e36162777e51361a7ceb47ae6b86a34ce492b23c2fde8083113b03985b98060d",\n "exp/inverse_search/inverse20_iso.py": "1b4ffd1d35b12ca4f327c8885063fb76eef459eeabed1fddae371e2e4bb4f635",\n "exp/inverse_search/optimized.json": "b18cfd6025658d02e38adfbe0ea7eca09960bbbdd458abe1ee1d539e4d3f6ca7",\n "exp/inverse_search/optimized_isotopy2.json": "28424aa2f6b0b92b7b3b4f6bc62feda566746feb9a9f21660ed5103420eae606",\n "exp/inverse_width/kernel.py": "accdd344715e6972ebe22e1679ca872d455a75466f1b1236b83bb636f6466409",\n "exp/inverse_width/partial5.json": "202ddf3a0a62530ac4f4028774e87c729cba6cc85dd465570a59e882a83cf4f7",\n "exp/inverse_width_subspaces/inverse19.json": "c868b9fc56f05ec89d34375ad086490691237211ecdc99b4a0c0b6e86b468a3f",\n "exp/inverse_width_subspaces/kernel19.py": "2cb588e60d5fe22ea59d625164e0a0d285cbcd352a80f5bf86f53f76050c9f4b",\n "exp/jet24/kernel.py": "f490e6205c678135dc2273d1f02e5ce9540087995f52c37e174b51dce0b4a5f6",\n "exp/jet24/pro_jet.py": "168975c442003084f644852b5bc012c44f7ea681d224ca736bf438fffffd377b",\n "exp/mask_projection/kernel.py": "80148ee048d1fd9aaace9637d2e652a7f46dbcb4306913f053437af4a2119088",\n "exp/mul99/kernels.py": "8f93d0a23f465cf72045e02f702b0b40d629b72d1618952a8fc79c22e23ee6ec",\n "exp/mul99/phase36.py": "d976ab9f6fbb9e2972b9518576b0135b07181d1b6cbfb74a61c094c211e1dccf",\n "exp/mul99/schedules.py": "9d16c576747f9d556b857f846c4029131f977cd6eb15d8a37cff4b84b0919c66",\n "exp/noscratch/kernels.py": "fa64bb5d5c359c507ac9a488c3f73363d5390bd43a4697daa67df57c322d7370",\n "exp/noscratch/phase39.py": "a47a3ad3ce8787f868b93b40156e3496eead5200b187244b5db13ec0f0c79f1a",\n "exp/noscratch/schedules.py": "6871354632e828b1b3efed382d84c39bac42e278f5ac700728efc3f1745cfa8d",\n "exp/partial_gauge/prefix.py": "dba10093b9314daf696ebd8fbc85543dbcf73534eb45d2303dc9ce53189b0652",\n "exp/partial_gauge/scalar.py": "7b3cb8eece8ccbc42462f13dcab9ef6b253581a06ce4d14906d88a541774de29",\n "exp/partial_gauge/synthesis_noflag.py": "cb316d80cdcfa636735156171a2b4d0bb57be22e4c5495806709f224fd0ea8ae",\n "exp/partial_gauge/tail.py": "cef30a29c419aef478fbf9c98f4277ab4a1fc63ca52e95d6025d592f16c7aff3",\n "exp/phase24_tradeoff/kernel.py": "984fc30f4c834b8f3cc1a7276f26f54801c01c8fd53b757e9ba767325059711a",\n "exp/phase24_tradeoff/phase6_six_search.json": "afb39dcf8748b1b1419a5e8db6b5440be74db6687464e6e0a9052a08b168fdb0",\n "exp/phase24_tradeoff/phase6small.py": "28590db3c6a0843fedc2894bb9899df45a7b99ea4544ce6b7517e739e672fe32",\n "exp/safe_scalar/converter.py": "771a242b7910c3e59f27bc82c72e8f7b86f9cb201d84ca3712ba7fd7e28a2841",\n "exp/safe_scalar/kernel.py": "72fbedc923066741e50cf4372de75ade749844c3554005b2f72ea5594190e84a",\n "exp/safe_scalar/phase.py": "30fdcd43d2234c6476344c1608449684199c8f3e8fb403f96aca80ae2379b048",\n "exp/scan/scan.py": "bab543d7fbaa33666e8c718a40ec47ac2b9a52e06e59de27bbbe45e68c754b55",\n "exp/tail_alias/divide.py": "46d4c7c4c94a06e9e244c030934ab45092ad6793b530c501226e2edb62a3ee5c",\n "exp/tail_alias/kernel.py": "3452ed2c039c73d3b1f45743f1ddeb40b88ffc32178b4052115f17bf18ba48ff",\n "exp/zero_padding_synthesis/kernel.py": "f2b84166a1ef1fb131ead3d907b21625375bd45a24187e74b440ff45647fc185",\n "final_decoder/conversion.py": "b3d05d20180fec5ab2f484cdca9e69da537bc1b5988ebf61984be68ef1383462",\n "final_decoder/decode.py": "a5f5dcb7be195b8a5e345c3d475c0eb807dbdf5e62192d0753088442f12627ff",\n "final_decoder/synthesis.py": "56ca15de2471de8dbe583294069283c25e306393e9f2630ad9420ec3eb8999a2",\n "playbook/exp/cf_check/audit.py": "c3211bbccebce2183b1a24385812dc505d1041ca7244274bbdffa75a95f37b33"\n}\n'}
SOURCE_HASHES = {'notebooks/launch.py': '72179eb4e59baa4adbcab0fa3f8b56cccde7d6ca359a7f06e17d45efaeb14d7b', 'notebooks/take.py': '64b527642010d9b07d9ad3e9b1aeec112feecac14a1053b208cfb46033652e8a', 'notebooks/check-submission': 'b1a0e5190fb0d96df85cc10184c834d33e100f1627b2fbbc4f5604e0f240284b', 'notebooks/extra-format.py': '6593250be3d89ee0942e01daf7251d0e5b351148786aa851d7f07dfe5dcac588', 'LICENSE': 'cfc7749b96f63bd31c3c42b5c471bf756814053e847c10f3eb003417bc523d30', 'blocks/dqi/pro-q2-897/source/README.md': '028361739e72e57ddc6b6bc563b317b3320420c06c78396c33152e7394d80271', 'blocks/dqi/pro-q2-897/source/build_897.py': '82026453bac5923fd9f5cb36236b644106a5bdd4b3a07b19c3f6a7c1b29fba0d', 'blocks/dqi/pro-q2-897/source/decode.py': 'c5c62693cbe65147fc63c3b75c3b04c4d785487a2dcdffed1cf103e90efabde4', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/even_fold.py': '784cf9b2873b5fa04da4233a528d7ffd8acf9d30fb2e39dc9992d46b1438cd16', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/kernel.py': 'd99aa5eaef8a0df416270d8eee0fa55a9847ee7c08c3aa6ee437352e1abdd05a', 'blocks/dqi/pro-q2-897/source/exp/alternating_gauge/scalar.py': '2fb3e5c4ec62df647de519647a57f51c765907ce1b6ecfe2d512035a91523b15', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/division.py': '28567073545c805e32636668cc0bb33f0dbc441a1de0dee2825774b3a33d2832', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/field.py': 'c1b8b2ff583ffd5997856605c35d37f7116aa5811e52a3ce04a1ae7215d381c3', 'blocks/dqi/pro-q2-897/source/exp/arithmetic/lowwidth.py': 'c536fa2b42393b83aae2bef80eeda22d83a6c799ef555178e94e09f1f5abc22a', 'blocks/dqi/pro-q2-897/source/exp/confluent_extract/kernel.py': '3acfc52dbf5994a85671b975be506b7967d2d075a3e1fc39f7c7cd16a8586ff3', 'blocks/dqi/pro-q2-897/source/exp/confluent_extract/model.py': '9b8d6fbaad6aec5da87d5247a11276c2fa177883adff9eca030a9bc2f3286387', 'blocks/dqi/pro-q2-897/source/exp/confluent_synthesis/kernel.py': '2eaabffce88e50ab3ce4719e764b3ed43c8bd6d8a6acf22a4461574268691c54', 'blocks/dqi/pro-q2-897/source/exp/convert/convert.py': '7afdafa60806df5afbc0af8fd01417a3e09fc15c5dc83fdd28d68079e78df7a0', 'blocks/dqi/pro-q2-897/source/exp/convert/hankel_check.py': '6d0ac6f55a70418049f46300863eeabb90a7c38b476b84db6e83d395edc3d72c', 'blocks/dqi/pro-q2-897/source/exp/convert/kernels.py': 'f5efa2f2b008cce86a0b751d400ee75777b77b6872d1875d6fae281cd7f09ef9', 'blocks/dqi/pro-q2-897/source/exp/convert/repair.py': '8beceea9091a9656901290498fbdc0e4373963374b5ee8f7e98d257ece1daeaf', 'blocks/dqi/pro-q2-897/source/exp/convert_short_hankel/kernel.py': '6dd64a123e4bd668bbf30e7d3c6449e738c8bf693f9a31bce1c9c39810ba7ddb', 'blocks/dqi/pro-q2-897/source/exp/convert_short_hankel/phase34.py': '620985ba489a03cef51095157af60b7459732b72b68f85a10265f34905ed203f', 'blocks/dqi/pro-q2-897/source/exp/decoder/common.py': '60f9db94c70dd746a429147b25121381934737fe6a576a5526ef70e1b12007a5', 'blocks/dqi/pro-q2-897/source/exp/decoder/extract.py': '6597bd52d2bebedae8860acaa4aaeaf980c3d21977781382ba8926833c8f7696', 'blocks/dqi/pro-q2-897/source/exp/decoder/safe_bridge.py': '023c81c775967a71ef44c571200bbe3ffb8b3da443fc2816e751856fc3f0d8d9', 'blocks/dqi/pro-q2-897/source/exp/decoder/synthesis.py': 'fdd7de226b7445007689ce2e50dbabbe77fdbc20e994ec41ac821a6ad2474c49', 'blocks/dqi/pro-q2-897/source/exp/fold_cached_norm/kernel.py': '6a148ebe23eae9194ec0ddb921ff6da1d362565218913170a36691c45dd7c084', 'blocks/dqi/pro-q2-897/source/exp/fold_cached_norm/phase34.py': '974d76939483c061e7b7995600fcdcb05e91529297725acfba41ce5ce808661f', 'blocks/dqi/pro-q2-897/source/exp/fold_joint/divide.py': '8456426c171f36a716a39b457a0de5206b5d4db6c8553f1bfc919248ba1de851', 'blocks/dqi/pro-q2-897/source/exp/fold_joint/kernel.py': 'd4764e763c6ed1ff538b58c1cc628df0c9577300aae5a4d83a5338e0290e018f', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/joint.py': '4bcc5614693d646bd910b9ed4748b278d2ec8dc1584bf8a45d3e465dd757cdc1', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/zero1.py': 'c9274f825d452082d88a98ded25a269aab247626a17032287c26ca1d8a71c724', 'blocks/dqi/pro-q2-897/source/exp/inverse12_escape/zero_layout.json': '45d1361fcf582282d15cc6fabeab290f8e9fd9dc4c68a3e14ffff86e9f9a537f', 'blocks/dqi/pro-q2-897/source/exp/inverse19_joint/kernel.py': '035152593a105b8178522a6edbf7cda3b709d8a86f51ae2148e71b01a1274913', 'blocks/dqi/pro-q2-897/source/exp/inverse19_joint/layout.json': '0adee49170ae597ed5a7df4d4d0df0e495fba49013f6acd9ae6a3f06b9ad4ab4', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/baseline.py': '7aecb36f951122861fe0443eee92f1d65a10484fc86f2e535f3db44f5dd553da', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/inverse20.py': 'e36162777e51361a7ceb47ae6b86a34ce492b23c2fde8083113b03985b98060d', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/inverse20_iso.py': '1b4ffd1d35b12ca4f327c8885063fb76eef459eeabed1fddae371e2e4bb4f635', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/optimized.json': 'b18cfd6025658d02e38adfbe0ea7eca09960bbbdd458abe1ee1d539e4d3f6ca7', 'blocks/dqi/pro-q2-897/source/exp/inverse_search/optimized_isotopy2.json': '28424aa2f6b0b92b7b3b4f6bc62feda566746feb9a9f21660ed5103420eae606', 'blocks/dqi/pro-q2-897/source/exp/inverse_width/kernel.py': 'accdd344715e6972ebe22e1679ca872d455a75466f1b1236b83bb636f6466409', 'blocks/dqi/pro-q2-897/source/exp/inverse_width/partial5.json': '202ddf3a0a62530ac4f4028774e87c729cba6cc85dd465570a59e882a83cf4f7', 'blocks/dqi/pro-q2-897/source/exp/inverse_width_subspaces/inverse19.json': 'c868b9fc56f05ec89d34375ad086490691237211ecdc99b4a0c0b6e86b468a3f', 'blocks/dqi/pro-q2-897/source/exp/inverse_width_subspaces/kernel19.py': '2cb588e60d5fe22ea59d625164e0a0d285cbcd352a80f5bf86f53f76050c9f4b', 'blocks/dqi/pro-q2-897/source/exp/jet24/kernel.py': 'f490e6205c678135dc2273d1f02e5ce9540087995f52c37e174b51dce0b4a5f6', 'blocks/dqi/pro-q2-897/source/exp/jet24/pro_jet.py': '168975c442003084f644852b5bc012c44f7ea681d224ca736bf438fffffd377b', 'blocks/dqi/pro-q2-897/source/exp/mask_projection/kernel.py': '80148ee048d1fd9aaace9637d2e652a7f46dbcb4306913f053437af4a2119088', 'blocks/dqi/pro-q2-897/source/exp/mul99/kernels.py': '8f93d0a23f465cf72045e02f702b0b40d629b72d1618952a8fc79c22e23ee6ec', 'blocks/dqi/pro-q2-897/source/exp/mul99/phase36.py': 'd976ab9f6fbb9e2972b9518576b0135b07181d1b6cbfb74a61c094c211e1dccf', 'blocks/dqi/pro-q2-897/source/exp/mul99/schedules.py': '9d16c576747f9d556b857f846c4029131f977cd6eb15d8a37cff4b84b0919c66', 'blocks/dqi/pro-q2-897/source/exp/noscratch/kernels.py': 'fa64bb5d5c359c507ac9a488c3f73363d5390bd43a4697daa67df57c322d7370', 'blocks/dqi/pro-q2-897/source/exp/noscratch/phase39.py': 'a47a3ad3ce8787f868b93b40156e3496eead5200b187244b5db13ec0f0c79f1a', 'blocks/dqi/pro-q2-897/source/exp/noscratch/schedules.py': '6871354632e828b1b3efed382d84c39bac42e278f5ac700728efc3f1745cfa8d', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/prefix.py': 'dba10093b9314daf696ebd8fbc85543dbcf73534eb45d2303dc9ce53189b0652', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/scalar.py': '7b3cb8eece8ccbc42462f13dcab9ef6b253581a06ce4d14906d88a541774de29', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/synthesis_noflag.py': 'cb316d80cdcfa636735156171a2b4d0bb57be22e4c5495806709f224fd0ea8ae', 'blocks/dqi/pro-q2-897/source/exp/partial_gauge/tail.py': 'cef30a29c419aef478fbf9c98f4277ab4a1fc63ca52e95d6025d592f16c7aff3', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/kernel.py': '984fc30f4c834b8f3cc1a7276f26f54801c01c8fd53b757e9ba767325059711a', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/phase6_six_search.json': 'afb39dcf8748b1b1419a5e8db6b5440be74db6687464e6e0a9052a08b168fdb0', 'blocks/dqi/pro-q2-897/source/exp/phase24_tradeoff/phase6small.py': '28590db3c6a0843fedc2894bb9899df45a7b99ea4544ce6b7517e739e672fe32', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/converter.py': '771a242b7910c3e59f27bc82c72e8f7b86f9cb201d84ca3712ba7fd7e28a2841', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/kernel.py': '72fbedc923066741e50cf4372de75ade749844c3554005b2f72ea5594190e84a', 'blocks/dqi/pro-q2-897/source/exp/safe_scalar/phase.py': '30fdcd43d2234c6476344c1608449684199c8f3e8fb403f96aca80ae2379b048', 'blocks/dqi/pro-q2-897/source/exp/scan/scan.py': 'bab543d7fbaa33666e8c718a40ec47ac2b9a52e06e59de27bbbe45e68c754b55', 'blocks/dqi/pro-q2-897/source/exp/tail_alias/divide.py': '46d4c7c4c94a06e9e244c030934ab45092ad6793b530c501226e2edb62a3ee5c', 'blocks/dqi/pro-q2-897/source/exp/tail_alias/kernel.py': '3452ed2c039c73d3b1f45743f1ddeb40b88ffc32178b4052115f17bf18ba48ff', 'blocks/dqi/pro-q2-897/source/exp/zero_padding_synthesis/kernel.py': 'f2b84166a1ef1fb131ead3d907b21625375bd45a24187e74b440ff45647fc185', 'blocks/dqi/pro-q2-897/source/final_decoder/conversion.py': 'b3d05d20180fec5ab2f484cdca9e69da537bc1b5988ebf61984be68ef1383462', 'blocks/dqi/pro-q2-897/source/final_decoder/decode.py': 'a5f5dcb7be195b8a5e345c3d475c0eb807dbdf5e62192d0753088442f12627ff', 'blocks/dqi/pro-q2-897/source/final_decoder/synthesis.py': '56ca15de2471de8dbe583294069283c25e306393e9f2630ad9420ec3eb8999a2', 'blocks/dqi/pro-q2-897/source/playbook/exp/cf_check/audit.py': 'c3211bbccebce2183b1a24385812dc505d1041ca7244274bbdffa75a95f37b33', 'blocks/dqi/pro-q2-897/source/source_sha256.json': '3ae6477c166779d42c87197c23c1e3be3d0837a807e402d777adc80e49cea392'}
INPUT_SOURCES = {}
MEMBERS = {'invsqrt.kmb': {'sha256': '0766a111a76f241aa57fbbed466436d743be13436943ddb0628c291ca07d67d6', 'size': 459891}, 'dqi.kmb': {'sha256': 'b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6', 'size': 505134467}, 'dfthcsos/angle_k_0.kmb': {'sha256': 'e5573a285d20d8855ed18361793cbd982b3f34b437020e744b9e8547524d54de', 'size': 440187}, 'dfthcsos/angle_k_1.kmb': {'sha256': '0b8c81a4ab79057616fb4a3091b0f57daea0e5f030f75b98d2af3fa925ac72db', 'size': 444374}, 'dfthcsos/angle_k_2.kmb': {'sha256': '2c3eec48fe095cd9a8881e305aef760cbc26e883025416bcfe0adf3a19d9b54e', 'size': 440687}, 'dfthcsos/angle_k_3.kmb': {'sha256': 'b00eaf4e42260665c0c3b2e5019d2f9bb99f665941dc888628fabf8f8209d393', 'size': 443650}, 'dfthcsos/angle_k_4.kmb': {'sha256': 'e313df4b29f8080cfad7d3f17c50664f53bfd4db2b3dc684e9890465e19b4157', 'size': 442863}, 'dfthcsos/angle_k_5.kmb': {'sha256': 'd8ad6e6611242d1ca3b94824c0fc3e43d67974da5a643cc8a7aa2c7a6eeecfb8', 'size': 443001}, 'dfthcsos/angle_k_6.kmb': {'sha256': '9a24ac09e5685f1cce54dff4a2d67137cf1cfeaca38a2f2a8731690c137d9a33', 'size': 445680}, 'dfthcsos/angle_k_7.kmb': {'sha256': '22cd7e1a3ee27828c5bd1803a4f29f940648f8cb8302b42780988c18aa8f89b7', 'size': 443090}, 'dfthcsos/angle_k_8.kmb': {'sha256': 'b3aff0f13e13b5c463d92f55ef238d043c91ed662361cb1dc6370de4e6ac179a', 'size': 440914}, 'dfthcsos/angle_k_9.kmb': {'sha256': '90a751b6c8aaa9bc82ccd33972bf402a1d870bba9136897e0a7dc25b24707d31', 'size': 447437}, 'dfthcsos/angle_k_10.kmb': {'sha256': '46293ddbfcfdeb8d8e1da8d52632b88bb62f1ab0281353db8c92a799cfb620ef', 'size': 445623}, 'dfthcsos/angle_k_11.kmb': {'sha256': 'b593ea6e62efc8a64cb00ba8b810befac7335952e320115e008be6a73e639041', 'size': 442670}, 'dfthcsos/angle_k_12.kmb': {'sha256': '5a92bfb5046d740053e6875da6db24ef524d4389b185178c8bebd80478e8ce19', 'size': 443531}, 'dfthcsos/angle_k_13.kmb': {'sha256': 'a39add5ddfe85460258dffa193d9613bd63c7bdcd34886efc1c5289a7ef42299', 'size': 442191}, 'dfthcsos/angle_k_14.kmb': {'sha256': '30f99f850fc23987686df1b65ec72d57abd5eb49ab48543c9d145b0be7970f11', 'size': 442207}, 'dfthcsos/submission.npz': {'sha256': '5d014b2e18fa4298b99a96ae1831231c9b848ddae0f20412c4681df322792d19', 'size': 2885708}}
EXPECTED = {'invsqrt': {'status': 'PASS', 'score': 103546.168164559, 'toffoli_equivalent': 1643.5899708660158, 'toffoli_count': 1348.4991469771198, 'max_toffoli_count': 1416, 't_count': 1.0, 'rotation_count': 41.0, 'rotation_eps': 2.434041480874084e-13, 'num_qubits': 63, 'reaction_depth': 792, 'num_ops': 46610, 'verified_shots': 200540928, 'worst_error_over_tolerance': 0.9993237694056958, 'worst_x': 161, 'circuit_sha256': '0766a111a76f241aa57fbbed466436d743be13436943ddb0628c291ca07d67d6'}, 'dqi': {'status': 'PASS', 'score': 422346240055.8793, 'm': 4095, 'n': 70, 'degree': 12, 'r': 2016, 'verified_shots': 102400, 'value_errors': 0, 'phase_errors': 0, 'empirical_error_rate': 0.0, 'effective_error_rate': 0.0008660587301657926, 'log2_delta': -128.0, 'repetition_factor': 1.1912604648115832, 'mean_error_weight': 31.046737331746634, 'toffoli_count': 440632.99798828125, 'max_toffoli_count': 454477, 'reaction_depth': 9339, 'num_ops': 55778145, 'num_qubits': 897, 'num_bits': 103164, 'circuit_sha256': 'b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6'}, 'dfthcsos': {'status': 'PASS', 'score': 349339717.3328993, 'lambda_eff': 47.49402243356817, 'toffoli_count': 17066.0, 'num_qubits': 431, 'brot': 8, 'batch_size': 5, 'num_circuits': 15, 'reaction_depth': 165, 'num_ops': 734988, 'verified_shots': 87424, 'submission_sha256': 'ab4d2288ede68da488c0dda35e321fea8755a50737a8491b8414e7447809bfb3'}}
PARTS = {'invsqrt': '0766a111a76f241aa57fbbed466436d743be13436943ddb0628c291ca07d67d6', 'dqi': 'b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6', 'dfthcsos': 'ab4d2288ede68da488c0dda35e321fea8755a50737a8491b8414e7447809bfb3'}
REFERENCE = {'invsqrt': 2076774.233715, 'dqi': 24439463583672.93, 'dfthcsos': 3355599588.3155375}
BLOCKS = {'invsqrt': {'id': 'mrmeows-21', 'title': "mrmeows' 63-qubit hybrid circuit", 'part': '0766a111a76f241aa57fbbed466436d743be13436943ddb0628c291ca07d67d6'}, 'dqi': {'id': 'pro-q2-897', 'title': "GPT-6 Astra Pro's 897-qubit decoder with the 658-Toffoli jet", 'part': 'b3354f60e459dcebe67910363e96498f8bb0aa7f699bc016ef962015c3732fd6'}, 'dfthcsos': {'id': 'heliosli-v8', 'title': "heliosli's J5 chemistry circuits on Ahmed Berat Özer's C46 table, with refit weights", 'part': 'ab4d2288ede68da488c0dda35e321fea8755a50737a8491b8414e7447809bfb3'}}
BUILDS = {'dqi': {'member': 'dqi.kmb', 'script': 'blocks/dqi/pro-q2-897/source/build_897.py'}}
NOTEBOOK = {'assembled_from_commit': '98286bd634320b11254f9b3c2d83948eb3d18f3c', 'blocks_changed_since_commit': True}

import hashlib, importlib.metadata, json, math, os, pathlib, platform, shutil, subprocess, sys, tempfile, time, zipfile
sys.dont_write_bytecode = True
START_TIME = time.monotonic()
INPUT = pathlib.Path(os.environ.get('QCOMP_INPUT', '/kaggle/input'))
WORKING = pathlib.Path(os.environ.get('QCOMP_WORKING', '/kaggle/working'))
WORKING.mkdir(parents=True, exist_ok=True)
SESSION = pathlib.Path(tempfile.mkdtemp(prefix='qcomp-'))
ROOT, OUT = SESSION / 'source', SESSION / 'build'
SUB = OUT / 'submission'
(ROOT / 'data').mkdir(parents=True)
SUB.mkdir(parents=True)
# Use at most the Kaggle CPU budget, one numerical-library thread per process.
if hasattr(os, 'sched_getaffinity'):
    affinity = sorted(os.sched_getaffinity(0))[:4]
    os.sched_setaffinity(0, affinity)
else:
    affinity = list(range(min(4, os.cpu_count() or 1)))
WORKERS = len(affinity)
for key in ['OMP_NUM_THREADS', 'OPENBLAS_NUM_THREADS', 'MKL_NUM_THREADS', 'NUMEXPR_NUM_THREADS']:
    os.environ[key] = '1'
TIMINGS = []


def sha256(path):
    with open(path, 'rb') as stream:
        return hashlib.file_digest(stream, 'sha256').hexdigest()


folders = [p.parents[1] for p in sorted(INPUT.rglob('kickmix.so')) if p.parent.name == 'kickmix']
# A dataset may carry its own copies of Kickmix: the official inputs are the first folder whose files all match
# their pinned SHA-256.
COMP = next((folder for folder in folders if all(
    (folder / name).is_file() and sha256(folder / name) == digest for name, digest in COMPETITION_FILES.items())), None)
assert COMP, f'No folder holds the official inputs with their pinned SHA-256; Kickmix found in {folders}'
assert sys.version_info[:2] == (3, 13), 'Select the pinned Python 3.13 CPU image for kickmix.so'
(ROOT / 'data/comp').symlink_to(COMP)
for name, body in SOURCES.items():
    path = ROOT / name
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(body, encoding='utf-8')
    assert sha256(path) == SOURCE_HASHES[name], name
# Block files too large to embed come from an attached dataset or notebook output, public or private: each is found
# among the inputs by name, size and SHA-256.
for name, want in INPUT_SOURCES.items():
    path = ROOT / name
    found = next((p for p in sorted(INPUT.rglob(path.name)) if p.is_file() and p.stat().st_size == want['size']
                  and sha256(p) == want['sha256']), None)
    assert found, f'{name}: no input file has SHA-256 {want["sha256"]}; attach the inputs the Run section names'
    path.parent.mkdir(parents=True, exist_ok=True)
    shutil.copyfile(found, path)
    print(f'{path.name} taken from {found}')
(ROOT / 'members.json').write_text(json.dumps(MEMBERS))


def run(*args, record):
    """Run one embedded script or official verifier in its own Python, keeping stdout as a record."""
    started = time.monotonic()
    result = subprocess.run([sys.executable, '-I', '-B', str(ROOT / 'notebooks/launch.py'), *map(str, args)],
                            cwd=ROOT, capture_output=True, text=True)
    TIMINGS.append({'command': list(map(str, args)), 'seconds': time.monotonic() - started,
                    'returncode': result.returncode})
    (OUT / record).write_text(result.stdout)
    (OUT / (record + '.stderr')).write_text(result.stderr)
    print(f'{pathlib.Path(str(args[0])).name}: {TIMINGS[-1]["seconds"]:.1f} s', flush=True)
    if result.returncode:
        print(result.stdout[-5000:], result.stderr[-5000:])
        raise RuntimeError(f'{args[0]} exited {result.returncode}')
    return result.stdout


BUILT = OUT / 'built'


def build(task):
    """Build a block's member, one file or a folder of them, from its embedded source, the way its authors run it,
    and check the SHA-256 of every file."""
    spec = BUILDS[task]
    target = BUILT / spec['member']
    target.parent.mkdir(parents=True, exist_ok=True)
    started = time.monotonic()
    # Our own code, run as a script: its folder leads sys.path, and Kickmix comes from the authenticated files.
    env = {**os.environ, 'PYTHONPATH': str(ROOT / 'data/comp/kickmix'), 'PYTHONNOUSERSITE': '1'}
    result = subprocess.run([sys.executable, '-B', str(ROOT / spec['script']), str(target)], cwd=SESSION, env=env,
                            capture_output=True, text=True)
    TIMINGS.append({'command': ['build', spec['script']], 'seconds': time.monotonic() - started,
                    'returncode': result.returncode})
    (OUT / f'{task}.build.txt').write_text(result.stdout)
    (OUT / f'{task}.build.stderr').write_text(result.stderr)
    if result.returncode:
        print(result.stdout[-5000:], result.stderr[-5000:])
        raise RuntimeError(f'{spec["script"]} exited {result.returncode}')
    names = [name for name in MEMBERS if name == spec['member'] or name.startswith(spec['member'] + '/')]
    assert names, spec['member']
    for name in names:
        digest = sha256(BUILT / name)
        assert digest == MEMBERS[name]['sha256'], f'{name}: built {digest}, not the verified file'
    record = json.loads(target.with_suffix('.build.json').read_text())
    if 'summary' in record:
        print(f'{spec["member"]} built in {TIMINGS[-1]["seconds"]:.0f} s: {record["summary"]}')
    else:
        print(f'{spec["member"]} built in {TIMINGS[-1]["seconds"]:.0f} s: Q {record["Q"]}, static Toffolis '
              f'{record["T_static"]:,}, SHA-256 {digest[:12]}..., the circuit the verifier passed')
    return record


VERIFIERS = {'invsqrt': ['data/comp/verifiers/invsqrt/verify_submission_invsqrt.py', 'invsqrt.kmb', '--max-workers'],
             'dqi': ['data/comp/verifiers/dqi/verify_submission_dqi.py', 'dqi.kmb', '--max-workers'],
             'dfthcsos': ['data/comp/verifiers/dfthcsos/verify.py', 'dfthcsos']}
CHECKED = ['status', 'score', 'num_qubits', 'toffoli_count', 'verified_shots', 'toffoli_equivalent', 't_count',
           'rotation_count', 'repetition_factor', 'value_errors', 'phase_errors', 'lambda_eff']


def verify(task):
    """Run the unmodified official verifier at its defaults and compare with the recorded run."""
    script, target, *workers = VERIFIERS[task]
    report = json.loads(run(script, SUB / target, *([*workers, WORKERS] if workers else []),
                            record=f'{task}.official.json'))
    expected = EXPECTED[task]
    assert report['status'] == 'PASS', report
    for key in CHECKED:
        if key in expected:
            assert math.isclose(report[key], expected[key], rel_tol=1e-9, abs_tol=1e-12) if isinstance(
                expected[key], float) else report[key] == expected[key], (task, key, report[key], expected[key])
    part = report.get('circuit_sha256', report.get('submission_sha256'))
    assert part == PARTS[task], (task, part)
    print(f'{task}: PASS | Q {report["num_qubits"]} | cost {report["score"]:,.6g} | '
          f'{REFERENCE[task] / report["score"]:.4f}x below the reference | same as the recorded run')
    return report


ENVIRONMENT = {'python': platform.python_version(), 'platform': platform.platform(),
               'packages': {name: importlib.metadata.version(name) for name in ['numpy', 'scipy']},
               'workers': WORKERS, 'affinity': affinity, 'numeric_threads': 1,
               'kickmix_sha256': sha256(COMP / 'kickmix/kickmix.so'),
               'seed': 'official default: derived from each circuit SHA-256, empty salt', **NOTEBOOK}
print(f'{len(COMPETITION_FILES)} official input files authenticated; {len(SOURCES)} embedded files written; '
      f'workers {WORKERS}; no network, installs or training')
```

```python
BUILT_RECORDS = {task: build(task) for task in BUILDS}
```

```python
taken = json.loads(run('notebooks/take.py', INPUT, SUB, ROOT / 'members.json', BUILT, record='take.json'))
for name, where in taken['taken'].items():
    print(f'{name:26} {MEMBERS[name]["sha256"][:12]}  from {where}')
```

```python
VERIFY = True  # False: report the recorded official runs instead of rerunning them
REPORTS = {}
for task in REFERENCE:
    REPORTS[task] = verify(task) if VERIFY else EXPECTED[task]
```

```python
print(run('notebooks/check-submission', SUB, record='check-submission.txt').splitlines()[-1])
extra = json.loads(run('notebooks/extra-format.py', SUB, record='extra-format.json'))
names = sorted(str(p.relative_to(SUB)) for p in SUB.rglob('*') if p.is_file())
assert names == sorted(MEMBERS), names
actual = {name: sha256(SUB / name) for name in names}
assert all(actual[n] == MEMBERS[n]['sha256'] for n in names), 'Output member hash mismatch'
with zipfile.ZipFile(WORKING / 'submission.zip', 'w', zipfile.ZIP_DEFLATED, compresslevel=6) as archive:
    # Fixed container metadata, so the same members always give the same archive.
    for name in names:
        info = zipfile.ZipInfo(name, date_time=(2026, 10, 27, 0, 0, 0))
        info.compress_type = zipfile.ZIP_DEFLATED
        info.create_system = 3
        info.external_attr = 0o100644 << 16
        info._compresslevel = 6
        with (SUB / name).open('rb') as source, archive.open(info, 'w', force_zip64=True) as target:
            shutil.copyfileobj(source, target, length=1 << 20)
with zipfile.ZipFile(WORKING / 'submission.zip') as archive:
    assert sorted(archive.namelist()) == names
    for name in names:
        with archive.open(name) as stream:
            assert hashlib.file_digest(stream, 'sha256').hexdigest() == actual[name], name
for name, digest in COMPETITION_FILES.items():
    assert sha256(COMP / name) == digest, f'Official input changed during the run: {name}'
gm = math.prod(REFERENCE[t] / REPORTS[t]['score'] for t in REFERENCE) ** (1 / 3)
summary = {'status': 'PASS', 'blocks': BLOCKS, 'verifiers': REPORTS, 'member_sha256': actual,
           'geometric_mean': gm, 'archive_sha256': sha256(WORKING / 'submission.zip'),
           'archive_bytes': (WORKING / 'submission.zip').stat().st_size, 'environment': ENVIRONMENT,
           'timings': TIMINGS, 'elapsed_seconds': time.monotonic() - START_TIME}
(WORKING / 'summary.json').write_text(json.dumps(summary, indent=2) + '\n')
(WORKING / 'submission-members.sha256').write_text(''.join(f'{actual[n]}  {n}\n' for n in names))
for task, r in REPORTS.items():
    print(f'{task:9} {BLOCKS[task]["title"]}: cost {r["score"]:,.6g}, ratio {REFERENCE[task] / r["score"]:.4f}')
print(f'{len(names)} members, each byte-identical to its block; format checks and ZIP read-back PASS.')
print(f'Geometric mean of the three ratios: {gm:.5f} (official verifiers at their defaults; '
      f'the leaderboard measures again)')
print(f'Elapsed {summary["elapsed_seconds"]:.0f} s | {WORKING / "submission.zip"}')
```
