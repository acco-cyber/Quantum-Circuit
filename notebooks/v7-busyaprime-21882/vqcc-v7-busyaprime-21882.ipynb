Our v6 scored 19.579. The best verified public portfolio is now Akmal Xodarev (Busya PRIME)'s
[LB 21.88210 - 21x Cheaper Than Google's Circuits](https://www.kaggle.com/code/busyaprime/lb-21-88210-21x-cheaper-than-google-s-circuits):
invsqrt 63q (cost 103,546), DQI R18 on mrmeows' round17 Q870 chain (4.421e11), and a chemistry weight refit of
Ahmed Berat Ozer's V9 factorization (3.551e8). Every member carries a FULL official-verifier PASS receipt in that output
(rerun of the hosts' metric code, cost within 1e-9 of the record).

No new circuit here. The notebook attaches that output, checks the archive against its pinned SHA-256, checks member
layout and the receipts, and writes `/kaggle/working/submission.zip`. CPU only, no GPU.


```python
import glob, hashlib, shutil, zipfile, json
PIN = 'b28c1e4ff9e49b6a57f08c3ff0b0ba6ba42d5ee56a0282d3063d0eb643c6ce43'
src = next(p for p in glob.glob('/kaggle/input/**/submission.zip', recursive=True) if 'busyaprime' in p.lower() or '21-88210' in p)
h = hashlib.sha256(open(src, 'rb').read()).hexdigest(); print(src, h)
assert h == PIN, 'archive changed upstream - stop, do not submit'
with zipfile.ZipFile(src) as z:
    names = z.namelist(); print(len(names), 'members'); print('\n'.join(sorted(names)))
    assert 'invsqrt.kmb' in names and 'dqi.kmb' in names and 'dfthcsos/submission.npz' in names
    assert sum(n.startswith('dfthcsos/angle_k_') for n in names) == 15
    assert z.testzip() is None
for r in sorted(glob.glob('/kaggle/input/**/receipts/*.json', recursive=True)):
    j = json.load(open(r)); print(r.split('/')[-1], j.get('status'), j.get('score'), j.get('num_qubits'))
    assert j.get('status') == 'PASS'
run = glob.glob('/kaggle/input/**/qcc33_v8_run.json', recursive=True)
if run: print(json.dumps({k: v for k, v in json.load(open(run[0])).items() if k in ('score', 'costs', 'zip_sha256', 'gates')}, indent=1))
shutil.copy(src, '/kaggle/working/submission.zip')
print('written', hashlib.sha256(open('/kaggle/working/submission.zip', 'rb').read()).hexdigest())
```
